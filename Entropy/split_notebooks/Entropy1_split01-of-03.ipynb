### Multiscale Entropy


<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — Multiscale entropy</h3><p>Multiscale entropy first coarse-grains the signal at a scale $\tau$ and then evaluates an entropy statistic, commonly Sample Entropy, on each coarse-grained series. For the standard non-overlapping mean coarse-graining, each new sample is the mean of $\tau$ consecutive original samples.</p><p><b>Cell role.</b> This code applies that theory to the notebook's existing EEG-derived data and preserves the surrounding notebook's plotting/output conventions. Numerical shortcuts selected by a speed mode change sampling density, iteration counts, or visualization resolution; they do not change the defining formula above.</p></div>

$$y_j^{(\\tau)}=\\frac1{\\tau}\\sum_{i=(j-1)\\tau+1}^{j\\tau}x_i.$$

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Multiscale entropy asks how signal irregularity changes when the data are viewed at progressively coarser time scales. At scale one the original samples are analyzed; at larger scales, groups of samples are combined to create a coarse-grained series before an entropy statistic is computed. A signal can therefore look irregular at short scales but more predictable at long scales, or vice versa. Because coarse-graining shortens the series, estimates at large scales generally have greater uncertainty.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance. The symbol τ (tau) commonly denotes a time delay, scale, or characteristic time. Here it should be read according to the definition immediately preceding the formula.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values. The visualization is a way of displaying already-computed numbers. Color, ordering, or ranking can make patterns easier to see but does not create additional statistical evidence beyond the values and uncertainty measures being plotted.</p></div>


```python
import sys
import time
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import nolds

# -------------------------------------------------------
# CONFIG
# -------------------------------------------------------
# Reuse the project-wide runtime modes. ``full`` preserves the original cell exactly;
# faster modes only thin the embedding trajectory before the same MSE calculation.
FAST_OPTION = globals().get("FAST_OPTION", "full")
if FAST_OPTION not in {"full", "balanced", "fast", "ultra"}:
    raise ValueError("FAST_OPTION must be one of: full, balanced, fast, ultra")
ENTROPY_LEGACY_SPEED = {
    "full":     {"embedding_stride": 1},
    "balanced": {"embedding_stride": 2},
    "fast":     {"embedding_stride": 4},
    "ultra":    {"embedding_stride": 8},
}[FAST_OPTION]

base_dir = Path("/home/a/projects/Complete-Neural-Signal-Analysis")
embedding_3d_dir = base_dir / "embedding_data" / "3dembedding_data"

output_directory = base_dir / "multiscale_entropy_3d"
plots_directory = output_directory / "plots"

output_directory.mkdir(parents=True, exist_ok=True)
plots_directory.mkdir(parents=True, exist_ok=True)

max_scale = 5
emb_dim = 2
tolerance_mode = "default"   # "default" or float
SHOW_PLOTS = True            # ALWAYS display as well as save
SAVE_PARTIALS = True         # save CSVs after each channel
HEARTBEAT = True             # print frequent progress updates

BG = "black"
ACCENT = "cyan"

# -------------------------------------------------------
# GLOBAL PLOT STYLE
# -------------------------------------------------------
plt.rcParams.update({
    "figure.facecolor": BG,
    "axes.facecolor": BG,
    "savefig.facecolor": BG,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT,
})

# -------------------------------------------------------
# HEARTBEAT HELPERS
# -------------------------------------------------------
t0_global = time.perf_counter()

def hb(msg):
    if HEARTBEAT:
        elapsed = time.perf_counter() - t0_global
        print(f"[{elapsed:8.1f}s] {msg}", flush=True)

# -------------------------------------------------------
# STYLE HELPER
# -------------------------------------------------------
def style_ax(ax):
    ax.set_facecolor(BG)
    ax.tick_params(colors=ACCENT)
    for spine in ax.spines.values():
        spine.set_color(ACCENT)
    ax.grid(True, alpha=0.20, color=ACCENT)

# -------------------------------------------------------
# HELPERS
# -------------------------------------------------------
def extract_channel_name(file_path):
    name = Path(file_path).stem
    prefixes = ["3dembedded_", "embedded_"]
    suffixes = ["_data"]

    for prefix in prefixes:
        if name.startswith(prefix):
            name = name[len(prefix):]

    for suffix in suffixes:
        if name.endswith(suffix):
            name = name[:-len(suffix)]

    return name

def coarse_grain(time_series, scale):
    """
    Fast coarse-graining by averaging non-overlapping windows.
    Keeps the final partial window if needed.
    """
    x = np.asarray(time_series, dtype=float)
    n = len(x)

    if scale <= 0:
        raise ValueError("scale must be >= 1")
    if n == 0:
        return np.array([], dtype=float)

    full_len = (n // scale) * scale
    coarse_parts = []

    if full_len > 0:
        coarse_parts.append(x[:full_len].reshape(-1, scale).mean(axis=1))

    if full_len < n:
        coarse_parts.append(np.array([x[full_len:].mean()], dtype=float))

    if not coarse_parts:
        return np.array([], dtype=float)

    return np.concatenate(coarse_parts)

def compute_tolerance(x, tolerance_mode="default"):
    x = np.asarray(x, dtype=float)
    if tolerance_mode == "default":
        return 0.1 * np.std(x)
    return float(tolerance_mode)

def safe_sampen(x, emb_dim=2, tolerance=0.2):
    x = np.asarray(x, dtype=float)

    if len(x) <= emb_dim + 1:
        return np.nan

    try:
        return float(nolds.sampen(x, emb_dim=emb_dim, tolerance=tolerance))
    except Exception as e:
        hb(f"[WARN] sampen failed: {e}")
        return np.nan

def multiscale_entropy_verbose(time_series, max_scale, emb_dim=2, tolerance_mode="default",
                               channel_name=None, dim_idx=None):
    """
    Compute multiscale entropy across scales 1..max_scale with heartbeat output.
    """
    x = np.asarray(time_series, dtype=float)
    tol = compute_tolerance(x, tolerance_mode=tolerance_mode)

    mse = np.full(max_scale, np.nan, dtype=float)

    hb(
        f"Starting MSE | channel={channel_name} | dim={dim_idx} | "
        f"len={len(x)} | tol={tol:.6f}"
    )

    for scale in range(1, max_scale + 1):
        t_scale = time.perf_counter()

        coarse_ts = coarse_grain(x, scale)
        mse[scale - 1] = safe_sampen(coarse_ts, emb_dim=emb_dim, tolerance=tol)

        dt = time.perf_counter() - t_scale
        hb(
            f"  scale={scale}/{max_scale} | channel={channel_name} | dim={dim_idx} | "
            f"coarse_len={len(coarse_ts)} | mse={mse[scale - 1]:.6f} | dt={dt:.2f}s"
        )

    return mse

def save_and_show(fig, out_path):
    plt.tight_layout()
    plt.savefig(out_path, dpi=180, bbox_inches="tight")
    hb(f"Saved plot: {out_path}")

    if SHOW_PLOTS:
        plt.show()
        hb(f"Displayed plot: {out_path.name}")

    plt.close(fig)

# -------------------------------------------------------
# LOAD FILES
# -------------------------------------------------------
file_paths = sorted(embedding_3d_dir.glob("*.npy"))

if not file_paths:
    raise FileNotFoundError(f"No .npy files found in: {embedding_3d_dir}")

hb(f"Found {len(file_paths)} 3D embedding files in: {embedding_3d_dir}")

# -------------------------------------------------------
# MAIN COMPUTATION
# -------------------------------------------------------
long_rows = []
summary_rows = []

long_csv_path = output_directory / "mse_3d_long.csv"
summary_csv_path = output_directory / "mse_3d_summary.csv"

n_files = len(file_paths)

for file_idx, file_path in enumerate(file_paths, start=1):
    t_file = time.perf_counter()

    hb(f"Loading file {file_idx}/{n_files}: {file_path.name}")
    data = np.load(file_path)
    channel_name = extract_channel_name(file_path)

    data = np.asarray(data, dtype=float)
    _speed_stride = int(ENTROPY_LEGACY_SPEED["embedding_stride"])
    if _speed_stride > 1:
        data = data[::_speed_stride]
    if data.ndim != 2:
        hb(f"[SKIP] {file_path.name}: expected 2D array, got shape {data.shape}")
        continue

    if data.shape[1] < 1:
        hb(f"[SKIP] {file_path.name}: no dimensions found")
        continue

    hb(f"Processing {file_path.name} | channel={channel_name} | shape={data.shape}")

    dim_curves = []

    for dim in range(data.shape[1]):
        t_dim = time.perf_counter()

        mse = multiscale_entropy_verbose(
            data[:, dim],
            max_scale=max_scale,
            emb_dim=emb_dim,
            tolerance_mode=tolerance_mode,
            channel_name=channel_name,
            dim_idx=dim + 1
        )
        dim_curves.append(mse)

        for scale_idx, value in enumerate(mse, start=1):
            long_rows.append({
                "channel": channel_name,
                "file": file_path.name,
                "dimension": dim + 1,
                "scale": scale_idx,
                "mse": value,
            })

        hb(
            f"Finished dimension {dim + 1}/{data.shape[1]} | channel={channel_name} | "
            f"curve={np.array2string(mse, precision=5, separator=', ')} | "
            f"dt={time.perf_counter() - t_dim:.2f}s"
        )

    dim_curves = np.asarray(dim_curves, dtype=float)
    mean_curve = np.nanmean(dim_curves, axis=0)

    for scale_idx, value in enumerate(mean_curve, start=1):
        summary_rows.append({
            "channel": channel_name,
            "scale": scale_idx,
            "mean_mse_across_dims": value,
        })

    # ---------------------------------------------------
    # PER-CHANNEL PLOT
    # ---------------------------------------------------
    hb(f"Plotting channel summary: {channel_name}")

    fig, ax = plt.subplots(figsize=(9, 6), facecolor=BG)
    style_ax(ax)

    scales = np.arange(1, max_scale + 1)

    for dim in range(dim_curves.shape[0]):
        ax.plot(
            scales,
            dim_curves[dim],
            marker="o",
            linewidth=1.6,
            label=f"Dim {dim + 1}"
        )

    ax.plot(
        scales,
        mean_curve,
        marker="s",
        linewidth=2.2,
        linestyle="--",
        label="Mean"
    )

    ax.set_title(f"Multiscale Entropy | {channel_name}")
    ax.set_xlabel("Scale")
    ax.set_ylabel("Sample Entropy")
    ax.set_xticks(scales)

    legend = ax.legend(facecolor=BG, edgecolor=ACCENT, labelcolor=ACCENT)
    for text in legend.get_texts():
        text.set_color(ACCENT)

    out_path = plots_directory / f"mse_{channel_name}.png"
    save_and_show(fig, out_path)

    # ---------------------------------------------------
    # SAVE PARTIAL RESULTS
    # ---------------------------------------------------
    if SAVE_PARTIALS:
        pd.DataFrame(long_rows).to_csv(long_csv_path, index=False)
        pd.DataFrame(summary_rows).to_csv(summary_csv_path, index=False)
        hb(f"Saved partial CSVs after channel {channel_name}")

    hb(
        f"Finished file {file_idx}/{n_files}: {file_path.name} | "
        f"dt={time.perf_counter() - t_file:.2f}s"
    )

# -------------------------------------------------------
# FINAL SAVE TABLES
# -------------------------------------------------------
long_df = pd.DataFrame(long_rows)
summary_df = pd.DataFrame(summary_rows)

long_df.to_csv(long_csv_path, index=False)
summary_df.to_csv(summary_csv_path, index=False)

hb(f"Saved long-form MSE results: {long_csv_path}")
hb(f"Saved summary MSE results:   {summary_csv_path}")

# -------------------------------------------------------
# GRAND SUMMARY PLOT
# -------------------------------------------------------
if not summary_df.empty:
    hb("Building grand summary plot")

    pivot_df = summary_df.pivot(index="scale", columns="channel", values="mean_mse_across_dims")

    fig, ax = plt.subplots(figsize=(12, 7), facecolor=BG)
    style_ax(ax)

    for channel in pivot_df.columns:
        ax.plot(
            pivot_df.index.to_numpy(),
            pivot_df[channel].to_numpy(),
            linewidth=1.6,
            marker="o",
            label=channel
        )

    ax.set_title("Multiscale Entropy Across Channels (mean across 3D embedding dimensions)")
    ax.set_xlabel("Scale")
    ax.set_ylabel("Mean Sample Entropy")
    ax.set_xticks(np.arange(1, max_scale + 1))

    legend = ax.legend(
        facecolor=BG,
        edgecolor=ACCENT,
        labelcolor=ACCENT,
        ncol=4,
        fontsize=8
    )
    for text in legend.get_texts():
        text.set_color(ACCENT)

    summary_plot_path = plots_directory / "mse_3d_all_channels_mean.png"
    save_and_show(fig, summary_plot_path)

hb("Done.")
```

### Approximate Entropy

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — Approximate entropy</h3><p>Approximate entropy compares average logarithmic match frequencies for templates of length $m$ and $m+1$. Unlike Sample Entropy, the classical ApEn definition includes self-matches, which biases small samples downward but prevents zero counts in many cases.</p><p><b>Cell role.</b> This code applies that theory to the notebook's existing EEG-derived data and preserves the surrounding notebook's plotting/output conventions. Numerical shortcuts selected by a speed mode change sampling density, iteration counts, or visualization resolution; they do not change the defining formula above.</p></div>

$$\\operatorname{ApEn}(m,r)=\\phi_m(r)-\\phi_{m+1}(r),$$

$$\\phi_m(r)=\\frac1{N-m+1}\\sum_i\\log C_i^{(m)}(r).$$

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Approximate entropy is also based on repeated short patterns, but unlike Sample Entropy it includes self-matches in its probability estimates. Larger values usually indicate that patterns are less predictable from one step to the next. Because self-matches introduce finite-sample bias, Approximate Entropy and Sample Entropy should not be treated as numerically interchangeable even when they are computed with the same embedding length and tolerance.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values. The visualization is a way of displaying already-computed numbers. Color, ordering, or ranking can make patterns easier to see but does not create additional statistical evidence beyond the values and uncertainty measures being plotted.</p></div>


```python
import time
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from numpy.lib.stride_tricks import sliding_window_view

# -------------------------------------------------------
# CONFIG
# -------------------------------------------------------
base_dir = Path("/home/a/projects/Complete-Neural-Signal-Analysis")
eeg_path = base_dir / "eeg_data_with_channels.npy"

output_directory = base_dir / "approximate_entropy_fast"
plots_directory = output_directory / "plots"

output_directory.mkdir(parents=True, exist_ok=True)
plots_directory.mkdir(parents=True, exist_ok=True)

eeg_channels = [
    'Fp1', 'Fpz', 'Fp2', 'F7', 'F3', 'Fz', 'F4', 'F8', 'FC5', 'FC1', 'FC2', 'FC6',
    'M1', 'T7', 'C3', 'Cz', 'C4', 'T8', 'M2', 'CP5', 'CP1', 'CP2', 'CP6',
    'P7', 'P3', 'Pz', 'P4', 'P8', 'POz', 'O1', 'Oz', 'O2'
]

sampling_rate = 1000
start_time, end_time = 814.571, 921.515
start_index, end_index = int(start_time * sampling_rate), int(end_time * sampling_rate)

# Approximate Entropy parameters
m = 2
r_factor = 0.2

# SPEED / HEARTBEAT CONTROLS
FAST_OPTION = globals().get("FAST_OPTION", "full")
if FAST_OPTION not in {"full", "balanced", "fast", "ultra"}:
    raise ValueError("FAST_OPTION must be one of: full, balanced, fast, ultra")
_APEN_SPEED = {
    "full":     {"max_points": 2000, "chunk": 400},
    "balanced": {"max_points": 1500, "chunk": 350},
    "fast":     {"max_points": 1000, "chunk": 300},
    "ultra":    {"max_points": 600,  "chunk": 200},
}[FAST_OPTION]
MAX_POINTS_PER_CHANNEL = _APEN_SPEED["max_points"]
CHUNK_SIZE = _APEN_SPEED["chunk"]                # chunk size for pairwise Chebyshev comparisons
SHOW_PLOTS = True               # ALWAYS display + save
SAVE_PARTIALS = True
HEARTBEAT = True
HEARTBEAT_EVERY_CHUNK = 1       # print every chunk

BG = "black"
ACCENT = "cyan"

# -------------------------------------------------------
# GLOBAL STYLE
# -------------------------------------------------------
plt.rcParams.update({
    "figure.facecolor": BG,
    "axes.facecolor": BG,
    "savefig.facecolor": BG,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT,
})

# -------------------------------------------------------
# HEARTBEAT
# -------------------------------------------------------
t0_global = time.perf_counter()

def hb(msg):
    if HEARTBEAT:
        dt = time.perf_counter() - t0_global
        print(f"[{dt:8.1f}s] {msg}", flush=True)

# -------------------------------------------------------
# STYLE
# -------------------------------------------------------
def style_ax(ax):
    ax.set_facecolor(BG)
    ax.tick_params(colors=ACCENT)
    for spine in ax.spines.values():
        spine.set_color(ACCENT)
    ax.grid(True, alpha=0.20, color=ACCENT)

def save_and_show(fig, out_path):
    plt.tight_layout()
    plt.savefig(out_path, dpi=180, bbox_inches="tight")
    hb(f"Saved plot: {out_path.name}")
    if SHOW_PLOTS:
        plt.show()
        hb(f"Displayed plot: {out_path.name}")
    plt.close(fig)

# -------------------------------------------------------
# HELPERS
# -------------------------------------------------------
def reduce_series_length(x, max_points=2000):
    """
    Fast stride-based reduction to keep ApEn tractable.
    """
    x = np.asarray(x, dtype=float)
    n = len(x)

    if n <= max_points:
        return x, 1

    stride = int(np.ceil(n / max_points))
    return x[::stride], stride

def apen_phi_chunked(x, emb_dim, r, chunk_size=400, label="phi"):
    """
    Compute phi(m) in chunks using Chebyshev distance.
    This keeps memory bounded and gives heartbeat progress.
    """
    x = np.asarray(x, dtype=np.float32)

    if len(x) <= emb_dim:
        return np.nan

    X = sliding_window_view(x, window_shape=emb_dim).astype(np.float32, copy=False)
    n_vectors = len(X)

    hb(f"Starting {label} | emb_dim={emb_dim} | vectors={n_vectors} | r={r:.6f}")

    counts = np.empty(n_vectors, dtype=np.float32)
    n_chunks = int(np.ceil(n_vectors / chunk_size))

    for chunk_idx, start in enumerate(range(0, n_vectors, chunk_size), start=1):
        end = min(start + chunk_size, n_vectors)
        Xi = X[start:end]  # shape: (chunk, emb_dim)

        # Chebyshev distance chunk against all vectors
        # result shape: (chunk, n_vectors)
        dist = np.max(np.abs(Xi[:, None, :] - X[None, :, :]), axis=2)
        counts[start:end] = np.sum(dist <= r, axis=1)

        if HEARTBEAT and (chunk_idx % HEARTBEAT_EVERY_CHUNK == 0 or chunk_idx == n_chunks):
            hb(f"  {label} chunk {chunk_idx}/{n_chunks} complete")

    C = counts / n_vectors
    phi_val = float(np.mean(np.log(C + 1e-12)))

    hb(f"Finished {label} | emb_dim={emb_dim} | phi={phi_val:.6f}")
    return phi_val

def approximate_entropy_chunked(x, m=2, r=0.2, chunk_size=400):
    """
    ApEn(m, r) = |phi(m) - phi(m+1)|
    """
    x = np.asarray(x, dtype=float)

    if len(x) <= m + 1:
        return np.nan

    phi_m = apen_phi_chunked(x, emb_dim=m, r=r, chunk_size=chunk_size, label="phi(m)")
    phi_m1 = apen_phi_chunked(x, emb_dim=m + 1, r=r, chunk_size=chunk_size, label="phi(m+1)")

    if not np.isfinite(phi_m) or not np.isfinite(phi_m1):
        return np.nan

    apen = abs(phi_m - phi_m1)
    hb(f"Computed ApEn = {apen:.6f}")
    return apen

# -------------------------------------------------------
# LOAD EEG
# -------------------------------------------------------
hb(f"Loading EEG data from: {eeg_path}")
EEG_data = np.load(eeg_path, allow_pickle=True)

if EEG_data.ndim != 2:
    raise ValueError(f"Expected 2D EEG array, got shape {EEG_data.shape}")

# Handle either (time, channels) or (channels, time)
if EEG_data.shape[1] == len(eeg_channels):
    EEG_segment = EEG_data[start_index:end_index, :]
elif EEG_data.shape[0] == len(eeg_channels):
    EEG_segment = EEG_data[:, start_index:end_index].T
else:
    raise ValueError(f"Unexpected EEG shape: {EEG_data.shape}")

hb(f"Loaded EEG segment: shape={EEG_segment.shape}")
hb(f"Time window: {start_time:.3f}s to {end_time:.3f}s")

# -------------------------------------------------------
# MAIN COMPUTATION
# -------------------------------------------------------
rows = []
csv_path = output_directory / "approximate_entropy_results.csv"

for i, channel in enumerate(eeg_channels):
    t_channel = time.perf_counter()

    hb(f"Starting channel {i+1}/{len(eeg_channels)}: {channel}")

    x_full = EEG_segment[:, i].astype(float)
    x_used, stride = reduce_series_length(x_full, max_points=MAX_POINTS_PER_CHANNEL)

    # keep original formulation: tolerance from the channel std
    r = r_factor * np.std(x_full)

    hb(
        f"Prepared channel {channel} | raw_len={len(x_full)} | "
        f"used_len={len(x_used)} | stride={stride} | r={r:.6f}"
    )

    apen = approximate_entropy_chunked(
        x_used,
        m=m,
        r=r,
        chunk_size=CHUNK_SIZE
    )

    rows.append({
        "Channel": channel,
        "Approximate Entropy": apen,
        "Raw Length": len(x_full),
        "Used Length": len(x_used),
        "Stride Used": stride,
        "m": m,
        "r": r,
        "Chunk Size": CHUNK_SIZE,
    })

    hb(
        f"Finished channel {channel} | ApEn={apen:.6f} | "
        f"dt={time.perf_counter() - t_channel:.2f}s"
    )

    if SAVE_PARTIALS:
        pd.DataFrame(rows).to_csv(csv_path, index=False)
        hb(f"Saved partial CSV after {channel}")

# -------------------------------------------------------
# SAVE RESULTS
# -------------------------------------------------------
approx_entropy_df = pd.DataFrame(rows)
approx_entropy_df["Channel"] = pd.Categorical(
    approx_entropy_df["Channel"],
    categories=eeg_channels,
    ordered=True
)
approx_entropy_df = approx_entropy_df.sort_values("Channel").reset_index(drop=True)
approx_entropy_df.to_csv(csv_path, index=False)

hb(f"Saved final CSV: {csv_path}")
print("\nApproximate Entropy results:")
print(approx_entropy_df)

# -------------------------------------------------------
# PLOT: CHANNEL ORDER
# -------------------------------------------------------
fig, ax = plt.subplots(figsize=(14, 7), facecolor=BG)
style_ax(ax)

ax.bar(
    approx_entropy_df["Channel"],
    approx_entropy_df["Approximate Entropy"].to_numpy(dtype=float)
)

ax.set_title("Approximate Entropy of EEG Channels")
ax.set_xlabel("EEG Channel")
ax.set_ylabel("Approximate Entropy")
ax.tick_params(axis="x", rotation=90)

bar_plot_path = plots_directory / "approximate_entropy_by_channel.png"
save_and_show(fig, bar_plot_path)

# -------------------------------------------------------
# PLOT: SORTED
# -------------------------------------------------------
sorted_df = approx_entropy_df.sort_values("Approximate Entropy", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(14, 7), facecolor=BG)
style_ax(ax)

ax.bar(
    sorted_df["Channel"],
    sorted_df["Approximate Entropy"].to_numpy(dtype=float)
)

ax.set_title("Approximate Entropy of EEG Channels (Sorted)")
ax.set_xlabel("EEG Channel")
ax.set_ylabel("Approximate Entropy")
ax.tick_params(axis="x", rotation=90)

sorted_plot_path = plots_directory / "approximate_entropy_sorted.png"
save_and_show(fig, sorted_plot_path)

print("\nTop channels by Approximate Entropy:")
print(sorted_df[["Channel", "Approximate Entropy"]].head(10))

hb("Done.")
```

### Fractal Dimension of Network Connectivity

# Transfer Entropy

# Hemispheric 

<div style="font-size: 13px; font-family: 'Times New Roman', Times, serif; background-color: #181818; color: #D0D0D0; padding: 20px; border-radius: 8px; margin: 10px; display: flex; flex-wrap: nowrap; justify-content: space-between;">
    <!-- Column 1 -->
    <div style="flex: 1; margin-right: 10px;">
        <h2>Introduction</h2>
        <p>
            This script performs analysis on EEG data using the concept of Transfer Entropy (TE). Specifically, it investigates the directionality of information flow between different EEG channels based on their 2D embeddings.
        </p>
        <h2>Data Preparation</h2>
        <p>
            Initially, the code unzips a dataset of 2D embedded EEG data for multiple channels. The embedded data is then loaded into Python, trimmed to a desired length, and grouped by hemisphere (left and right).
        </p>
    </div>
    <!-- Column 2 -->
    <div style="flex: 1; margin-left: 10px;">
        <h2>Analysis</h2>
        <p>
            Transfer Entropy calculations are conducted to quantify the amount of information flowing from one EEG channel to another. This is done after binning the data and applying the PyInform library’s TE function.
        </p>
        <p>
            The TE formula is represented as:
            \[
            TE_{X \rightarrow Y} = \sum_{y_{t+1}, y_t, x_t} p(y_{t+1}, y_t, x_t) \log \frac{p(y_{t+1} | y_t, x_t)}{p(y_{t+1} | y_t)}
            \]
            Here \(X\) and \(Y\) are random variables, \(p(y_{t+1}, y_t, x_t)\) is the joint probability distribution function, and \(p(y_{t+1} | y_t, x_t)\) and \(p(y_{t+1} | y_t)\) are conditional probability distribution functions.
        </p>
        </p>
    </div>
</div>

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Transfer entropy measures whether the past of one signal provides information about the future of another beyond the information already contained in the target signal's own past. Mathematically it is a conditional information measure, so direction matters: transfer from X to Y is not generally equal to transfer from Y to X. In finite EEG recordings, reliable estimation can require many samples because the method estimates conditional probability structure in a multidimensional history space.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values.</p></div>


<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — Transfer entropy</h3><p>Transfer entropy is a conditional mutual information that measures whether the past of a source $X$ improves prediction of the next target value $Y_{t+1}$ beyond the target's own past. It is directional and depends on the history lengths, lag, estimator, and any discretization used by the implementation.</p><p><b>Cell role.</b> This code applies that theory to the notebook's existing EEG-derived data and preserves the surrounding notebook's plotting/output conventions. Numerical shortcuts selected by a speed mode change sampling density, iteration counts, or visualization resolution; they do not change the defining formula above.</p></div>

$$T_{X\\to Y}=\\sum p(y_{t+1},y_t^{(k)},x_t^{(l)})\\log\\frac{p(y_{t+1}\\mid y_t^{(k)},x_t^{(l)})}{p(y_{t+1}\\mid y_t^{(k)})}.$$

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Mutual information measures how much knowing one variable reduces uncertainty about another. Unlike linear correlation, it can capture nonlinear statistical dependence. In delay-selection problems, mutual information between a signal and a delayed copy is examined as the delay changes; a local minimum is often used as a compromise between coordinates that are too redundant and coordinates that are almost unrelated. The estimate is sensitive to finite sample size and to how probability densities are estimated.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values. The visualization is a way of displaying already-computed numbers. Color, ordering, or ranking can make patterns easier to see but does not create additional statistical evidence beyond the values and uncertainty measures being plotted.</p></div>


```python
import time
import zipfile
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pyinform import transfer_entropy

# -------------------------------------------------------
# CONFIG
# -------------------------------------------------------
base_dir = Path("/home/a/projects/Complete-Neural-Signal-Analysis")
zipped_file_path = base_dir / "2d_embedded_data.zip"

output_directory = base_dir / "transfer_entropy_2d"
plots_directory = output_directory / "plots"

output_directory.mkdir(parents=True, exist_ok=True)
plots_directory.mkdir(parents=True, exist_ok=True)

eeg_channels = [
    'Fp1', 'Fpz', 'Fp2', 'F7', 'F3', 'Fz', 'F4', 'F8', 'FC5', 'FC1', 'FC2', 'FC6',
    'M1', 'T7', 'C3', 'Cz', 'C4', 'T8', 'M2', 'CP5', 'CP1', 'CP2', 'CP6',
    'P7', 'P3', 'Pz', 'P4', 'P8', 'POz', 'O1', 'Oz', 'O2'
]

left_channels = ['Fp1', 'F7', 'F3', 'FC5', 'M1', 'T7', 'C3', 'CP5', 'P7', 'P3', 'O1']
right_channels = ['Fp2', 'F8', 'F4', 'FC6', 'M2', 'T8', 'C4', 'CP6', 'P8', 'P4', 'O2']

# central_channels existed in your original script, but was not used there either
central_channels = ['Fpz', 'Fz', 'FC1', 'FC2', 'Cz', 'CP1', 'CP2', 'Pz', 'POz', 'Oz']

# TE settings
K_HISTORY = 1
NUM_BINS = 32

FAST_OPTION = globals().get("FAST_OPTION", "full")
if FAST_OPTION not in {"full", "balanced", "fast", "ultra"}:
    raise ValueError("FAST_OPTION must be one of: full, balanced, fast, ultra")
_TE_PAIR_SPEED = {
    "full":     {"max_points": 20000, "preview": 3000},
    "balanced": {"max_points": 12000, "preview": 2500},
    "fast":     {"max_points": 7000,  "preview": 1800},
    "ultra":    {"max_points": 3500,  "preview": 1200},
}[FAST_OPTION]

# SPEED / OUTPUT CONTROLS
MAX_POINTS_PER_SERIES = _TE_PAIR_SPEED["max_points"]
PREVIEW_POINTS = _TE_PAIR_SPEED["preview"]
SHOW_PLOTS = True               # ALWAYS display + save
HEARTBEAT = True
SAVE_PARTIALS = True

BG = "black"
ACCENT = "cyan"

# -------------------------------------------------------
# GLOBAL STYLE
# -------------------------------------------------------
plt.rcParams.update({
    "figure.facecolor": BG,
    "axes.facecolor": BG,
    "savefig.facecolor": BG,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT,
})

# -------------------------------------------------------
# HEARTBEAT
# -------------------------------------------------------
t0_global = time.perf_counter()

def hb(msg):
    if HEARTBEAT:
        dt = time.perf_counter() - t0_global
        print(f"[{dt:8.1f}s] {msg}", flush=True)

# -------------------------------------------------------
# STYLE
# -------------------------------------------------------
def style_ax(ax):
    ax.set_facecolor(BG)
    ax.tick_params(colors=ACCENT)
    for spine in ax.spines.values():
        spine.set_color(ACCENT)
    ax.grid(True, alpha=0.20, color=ACCENT)

def save_and_show(fig, out_path):
    plt.tight_layout()
    plt.savefig(out_path, dpi=180, bbox_inches="tight")
    hb(f"Saved plot: {out_path.name}")
    if SHOW_PLOTS:
        plt.show()
        hb(f"Displayed plot: {out_path.name}")
    plt.close(fig)

# -------------------------------------------------------
# HELPERS
# -------------------------------------------------------
def load_embedded_channel_from_zip(zipf, channel_name):
    """
    Supports both:
      - 2dembedded_<channel>.npy
      - 2dembedded_<channel>_data.npy
    """
    candidates = [
        f"2dembedded_{channel_name}.npy",
        f"2dembedded_{channel_name}_data.npy",
    ]

    for name in candidates:
        try:
            with zipf.open(name) as f:
                arr = np.load(f)
            arr = np.asarray(arr, dtype=float)
            if arr.ndim != 2 or arr.shape[1] != 2:
                raise ValueError(f"{name} must have shape (N, 2), got {arr.shape}")
            return arr
        except KeyError:
            continue

    raise FileNotFoundError(f"No 2D embedding file found for channel {channel_name}")

def reduce_series_length(x, max_points=20000):
    x = np.asarray(x, dtype=float)
    n = len(x)

    if n <= max_points:
        return x, 1

    stride = int(np.ceil(n / max_points))
    return x[::stride], stride

def standardize_1d(x):
    x = np.asarray(x, dtype=float)
    return (x - np.mean(x)) / (np.std(x) + 1e-12)

def joint_bin_equal_width(x, y, num_bins=32):
    """
    Standardize each series, then bin both using the same edges
    derived from the combined value range.
    """
    x = standardize_1d(x)
    y = standardize_1d(y)

    joint = np.concatenate([x, y])
    xmin, xmax = np.min(joint), np.max(joint)

    if not np.isfinite(xmin) or not np.isfinite(xmax) or xmax <= xmin:
        return np.zeros_like(x, dtype=np.int32), np.zeros_like(y, dtype=np.int32)

    edges = np.linspace(xmin, xmax, num_bins + 1)

    xb = np.digitize(x, edges[1:-1], right=False).astype(np.int32)
    yb = np.digitize(y, edges[1:-1], right=False).astype(np.int32)

    xb = np.clip(xb, 0, num_bins - 1)
    yb = np.clip(yb, 0, num_bins - 1)

    return xb, yb

def compute_te_pair(source, target, k=1):
    """
    Transfer entropy from source -> target.
    """
    return float(transfer_entropy(source, target, k))

# -------------------------------------------------------
# LOAD ONLY NEEDED CHANNELS
# -------------------------------------------------------
channels_needed = sorted(set(left_channels + right_channels))
embedded = {}

hb(f"Opening zip: {zipped_file_path}")

with zipfile.ZipFile(zipped_file_path, "r") as zipf:
    for idx, ch in enumerate(channels_needed, start=1):
        hb(f"Loading channel {idx}/{len(channels_needed)}: {ch}")
        embedded[ch] = load_embedded_channel_from_zip(zipf, ch)
        hb(f"Loaded {ch}: shape={embedded[ch].shape}")

# -------------------------------------------------------
# ALIGN LENGTHS
# -------------------------------------------------------
common_length = min(len(embedded[ch]) for ch in channels_needed)
hb(f"Common length across required channels: {common_length}")

for ch in channels_needed:
    embedded[ch] = embedded[ch][:common_length]

# -------------------------------------------------------
# BUILD HEMISPHERE AVERAGES
# -------------------------------------------------------
hb("Computing left/right average embeddings")

left_stack = np.stack([embedded[ch] for ch in left_channels], axis=0)
right_stack = np.stack([embedded[ch] for ch in right_channels], axis=0)

left_embedding_avg = np.mean(left_stack, axis=0)
right_embedding_avg = np.mean(right_stack, axis=0)

hb(f"Left average shape:  {left_embedding_avg.shape}")
hb(f"Right average shape: {right_embedding_avg.shape}")

# -------------------------------------------------------
# REDUCE LENGTH FOR SPEED
# -------------------------------------------------------
left_reduced = np.zeros((0, 2), dtype=float)
right_reduced = np.zeros((0, 2), dtype=float)

left_reduced_dim0, stride0 = reduce_series_length(left_embedding_avg[:, 0], MAX_POINTS_PER_SERIES)
right_reduced_dim0, _ = reduce_series_length(right_embedding_avg[:, 0], MAX_POINTS_PER_SERIES)

left_reduced_dim1, stride1 = reduce_series_length(left_embedding_avg[:, 1], MAX_POINTS_PER_SERIES)
right_reduced_dim1, _ = reduce_series_length(right_embedding_avg[:, 1], MAX_POINTS_PER_SERIES)

left_reduced = np.column_stack([left_reduced_dim0, left_reduced_dim1])
right_reduced = np.column_stack([right_reduced_dim0, right_reduced_dim1])

hb(
    f"Reduced dimension 1 | left_len={len(left_reduced_dim0)} | right_len={len(right_reduced_dim0)} | stride={stride0}"
)
hb(
    f"Reduced dimension 2 | left_len={len(left_reduced_dim1)} | right_len={len(right_reduced_dim1)} | stride={stride1}"
)

# -------------------------------------------------------
# COMPUTE TE BY EMBEDDING DIMENSION
# -------------------------------------------------------
rows = []
csv_path = output_directory / "transfer_entropy_results.csv"

for dim in range(2):
    t_dim = time.perf_counter()

    hb(f"Starting TE for embedding dimension {dim + 1}/2")

    left_data = left_reduced[:, dim]
    right_data = right_reduced[:, dim]

    hb(f"Binning dimension {dim + 1} with {NUM_BINS} bins")
    left_binned, right_binned = joint_bin_equal_width(left_data, right_data, num_bins=NUM_BINS)

    hb(f"Computing TE Left -> Right | dim={dim + 1}")
    te_left_to_right = compute_te_pair(left_binned, right_binned, k=K_HISTORY)

    hb(f"Computing TE Right -> Left | dim={dim + 1}")
    te_right_to_left = compute_te_pair(right_binned, left_binned, k=K_HISTORY)

    rows.append({
        "dimension": dim + 1,
        "TE_left_to_right": te_left_to_right,
        "TE_right_to_left": te_right_to_left,
        "TE_asymmetry_LR_minus_RL": te_left_to_right - te_right_to_left,
        "left_length_used": len(left_data),
        "right_length_used": len(right_data),
        "stride_used": stride0 if dim == 0 else stride1,
        "num_bins": NUM_BINS,
        "k_history": K_HISTORY,
    })

    hb(
        f"Finished dimension {dim + 1} | "
        f"TE L->R={te_left_to_right:.6f} | TE R->L={te_right_to_left:.6f} | "
        f"dt={time.perf_counter() - t_dim:.2f}s"
    )

    if SAVE_PARTIALS:
        pd.DataFrame(rows).to_csv(csv_path, index=False)
        hb(f"Saved partial CSV after dimension {dim + 1}")

# -------------------------------------------------------
# SAVE RESULTS
# -------------------------------------------------------
results_df = pd.DataFrame(rows)
results_df.to_csv(csv_path, index=False)

hb(f"Saved final CSV: {csv_path}")
print("\nTransfer Entropy results:")
print(results_df)

# -------------------------------------------------------
# SUMMARY PLOT
# -------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 6), facecolor=BG)
style_ax(ax)

x = np.arange(len(results_df))
width = 0.35

ax.bar(
    x - width / 2,
    results_df["TE_left_to_right"].to_numpy(dtype=float),
    width=width,
    color=ACCENT,
    alpha=0.85,
    label="Left → Right"
)

ax.bar(
    x + width / 2,
    results_df["TE_right_to_left"].to_numpy(dtype=float),
    width=width,
    color=ACCENT,
    alpha=0.35,
    label="Right → Left"
)

ax.set_title("Transfer Entropy Between Left and Right Hemisphere Averages")
ax.set_xlabel("Embedding Dimension")
ax.set_ylabel("Transfer Entropy")
ax.set_xticks(x)
ax.set_xticklabels([f"Dim {d}" for d in results_df["dimension"]])

legend = ax.legend(facecolor=BG, edgecolor=ACCENT, labelcolor=ACCENT)
for text in legend.get_texts():
    text.set_color(ACCENT)

summary_plot_path = plots_directory / "transfer_entropy_summary.png"
save_and_show(fig, summary_plot_path)

# -------------------------------------------------------
# ASYMMETRY PLOT
# -------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 6), facecolor=BG)
style_ax(ax)

ax.bar(
    [f"Dim {d}" for d in results_df["dimension"]],
    results_df["TE_asymmetry_LR_minus_RL"].to_numpy(dtype=float),
    color=ACCENT,
    alpha=0.85
)

ax.axhline(0.0, color=ACCENT, linewidth=1.0, linestyle="--")
ax.set_title("Transfer Entropy Asymmetry (Left→Right minus Right→Left)")
ax.set_xlabel("Embedding Dimension")
ax.set_ylabel("TE Asymmetry")

asymmetry_plot_path = plots_directory / "transfer_entropy_asymmetry.png"
save_and_show(fig, asymmetry_plot_path)

# -------------------------------------------------------
# PREVIEW PLOT OF REDUCED HEMISPHERE SIGNALS
# -------------------------------------------------------
for dim in range(2):
    fig, ax = plt.subplots(figsize=(12, 5), facecolor=BG)
    style_ax(ax)

    n_preview = min(PREVIEW_POINTS, len(left_reduced), len(right_reduced))
    t = np.arange(n_preview)

    ax.plot(
        t,
        left_reduced[:n_preview, dim],
        color=ACCENT,
        linewidth=1.4,
        linestyle="-",
        label="Left hemisphere average"
    )

    ax.plot(
        t,
        right_reduced[:n_preview, dim],
        color=ACCENT,
        linewidth=1.4,
        linestyle="--",
        label="Right hemisphere average"
    )

    ax.set_title(f"Hemisphere-Averaged Embedded Signals | Dimension {dim + 1}")
    ax.set_xlabel("Sample Index")
    ax.set_ylabel("Amplitude")

    legend = ax.legend(facecolor=BG, edgecolor=ACCENT, labelcolor=ACCENT)
    for text in legend.get_texts():
        text.set_color(ACCENT)

    preview_plot_path = plots_directory / f"hemisphere_average_preview_dim{dim + 1}.png"
    save_and_show(fig, preview_plot_path)

# -------------------------------------------------------
# PRINT SUMMARY
# -------------------------------------------------------
mean_lr = results_df["TE_left_to_right"].mean()
mean_rl = results_df["TE_right_to_left"].mean()

print(f"\nMean TE Left → Right: {mean_lr:.6f}")
print(f"Mean TE Right → Left: {mean_rl:.6f}")
print(f"Mean asymmetry (L→R - R→L): {(mean_lr - mean_rl):.6f}")

hb("Done.")
```

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — Transfer entropy</h3><p>Transfer entropy is a conditional mutual information that measures whether the past of a source $X$ improves prediction of the next target value $Y_{t+1}$ beyond the target's own past. It is directional and depends on the history lengths, lag, estimator, and any discretization used by the implementation.</p><p><b>Cell role.</b> This code applies that theory to the notebook's existing EEG-derived data and preserves the surrounding notebook's plotting/output conventions. Numerical shortcuts selected by a speed mode change sampling density, iteration counts, or visualization resolution; they do not change the defining formula above.</p></div>

$$T_{X\\to Y}=\\sum p(y_{t+1},y_t^{(k)},x_t^{(l)})\\log\\frac{p(y_{t+1}\\mid y_t^{(k)},x_t^{(l)})}{p(y_{t+1}\\mid y_t^{(k)})}.$$

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Mutual information measures how much knowing one variable reduces uncertainty about another. Unlike linear correlation, it can capture nonlinear statistical dependence. In delay-selection problems, mutual information between a signal and a delayed copy is examined as the delay changes; a local minimum is often used as a compromise between coordinates that are too redundant and coordinates that are almost unrelated. The estimate is sensitive to finite sample size and to how probability densities are estimated.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values. The visualization is a way of displaying already-computed numbers. Color, ordering, or ranking can make patterns easier to see but does not create additional statistical evidence beyond the values and uncertainty measures being plotted.</p></div>


```python
import time
import zipfile
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pyinform import transfer_entropy

# -------------------------------------------------------
# CONFIG
# -------------------------------------------------------
base_dir = Path("/home/a/projects/Complete-Neural-Signal-Analysis")
zipped_file_path = base_dir / "2d_embedded_data.zip"

output_directory = base_dir / "transfer_entropy_2d_exploratory"
plots_directory = output_directory / "plots"

output_directory.mkdir(parents=True, exist_ok=True)
plots_directory.mkdir(parents=True, exist_ok=True)

eeg_channels = [
    'Fp1', 'Fpz', 'Fp2', 'F7', 'F3', 'Fz', 'F4', 'F8', 'FC5', 'FC1', 'FC2', 'FC6',
    'M1', 'T7', 'C3', 'Cz', 'C4', 'T8', 'M2', 'CP5', 'CP1', 'CP2', 'CP6',
    'P7', 'P3', 'Pz', 'P4', 'P8', 'POz', 'O1', 'Oz', 'O2'
]

left_channels = ['Fp1', 'F7', 'F3', 'FC5', 'M1', 'T7', 'C3', 'CP5', 'P7', 'P3', 'O1']
right_channels = ['Fp2', 'F8', 'F4', 'FC6', 'M2', 'T8', 'C4', 'CP6', 'P8', 'P4', 'O2']

# TE settings
K_HISTORY = 1
NUM_BINS = 32

FAST_OPTION = globals().get("FAST_OPTION", "full")
if FAST_OPTION not in {"full", "balanced", "fast", "ultra"}:
    raise ValueError("FAST_OPTION must be one of: full, balanced, fast, ultra")
_TE_DYNAMIC_SPEED = {
    "full":     {"max_points": 12000, "preview": 3000, "step": 300, "lag_step": 2},
    "balanced": {"max_points": 9000,  "preview": 2500, "step": 450, "lag_step": 2},
    "fast":     {"max_points": 6000,  "preview": 1800, "step": 600, "lag_step": 4},
    "ultra":    {"max_points": 3500,  "preview": 1200, "step": 900, "lag_step": 5},
}[FAST_OPTION]

# SPEED / EXPLORATION CONTROLS
MAX_POINTS_PER_SERIES = _TE_DYNAMIC_SPEED["max_points"]
PREVIEW_POINTS = _TE_DYNAMIC_SPEED["preview"]
ROLLING_WINDOW = 1500
ROLLING_STEP = _TE_DYNAMIC_SPEED["step"]
LAG_VALUES = np.arange(-20, 21, _TE_DYNAMIC_SPEED["lag_step"])

# OUTPUT CONTROLS
SHOW_PLOTS = True          # ALWAYS display + save
HEARTBEAT = True
SAVE_PARTIALS = True

# -------------------------------------------------------
# COLOR PALETTE
# -------------------------------------------------------
BG = "black"
ACCENT = "#00E5FF"      # axes/ticks/text/grid cyan
C_PRIMARY = "#00E5FF"   # bright cyan
C_SECONDARY = "#14B8A6" # teal
C_DIFF = "#A78BFA"      # soft violet
C_HIGHLIGHT = "#7DD3FC" # pale aqua
C_MUTED = "#67E8F9"     # light cyan

# -------------------------------------------------------
# GLOBAL STYLE
# -------------------------------------------------------
plt.rcParams.update({
    "figure.facecolor": BG,
    "axes.facecolor": BG,
    "savefig.facecolor": BG,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT,
})

# -------------------------------------------------------
# HEARTBEAT
# -------------------------------------------------------
t0_global = time.perf_counter()

def hb(msg):
    if HEARTBEAT:
        dt = time.perf_counter() - t0_global
        print(f"[{dt:8.1f}s] {msg}", flush=True)

# -------------------------------------------------------
# STYLING HELPERS
# -------------------------------------------------------
def style_ax(ax):
    ax.set_facecolor(BG)
    ax.tick_params(colors=ACCENT)
    for spine in ax.spines.values():
        spine.set_color(ACCENT)
    ax.grid(True, alpha=0.18, color=ACCENT)

def style_legend(legend):
    if legend is None:
        return
    frame = legend.get_frame()
    frame.set_facecolor(BG)
    frame.set_edgecolor(ACCENT)
    for text in legend.get_texts():
        text.set_color(ACCENT)

def save_and_show(fig, out_path):
    plt.tight_layout()
    plt.savefig(out_path, dpi=180, bbox_inches="tight")
    hb(f"Saved plot: {out_path.name}")
    if SHOW_PLOTS:
        plt.show()
        hb(f"Displayed plot: {out_path.name}")
    plt.close(fig)

# -------------------------------------------------------
# HELPERS
# -------------------------------------------------------
def load_embedded_channel_from_zip(zipf, channel_name):
    candidates = [
        f"2dembedded_{channel_name}.npy",
        f"2dembedded_{channel_name}_data.npy",
    ]
    for name in candidates:
        try:
            with zipf.open(name) as f:
                arr = np.load(f)
            arr = np.asarray(arr, dtype=float)
            if arr.ndim != 2 or arr.shape[1] != 2:
                raise ValueError(f"{name} must have shape (N, 2), got {arr.shape}")
            return arr
        except KeyError:
            continue
    raise FileNotFoundError(f"No 2D embedding file found for channel {channel_name}")

def reduce_series_length(x, max_points=12000):
    x = np.asarray(x, dtype=float)
    n = len(x)
    if n <= max_points:
        return x, 1
    stride = int(np.ceil(n / max_points))
    return x[::stride], stride

def standardize_1d(x):
    x = np.asarray(x, dtype=float)
    return (x - np.mean(x)) / (np.std(x) + 1e-12)

def joint_bin_equal_width(x, y, num_bins=32):
    """
    Standardize each series, then bin them using shared edges.
    """
    x = standardize_1d(x)
    y = standardize_1d(y)

    joint = np.concatenate([x, y])
    xmin, xmax = np.min(joint), np.max(joint)

    if not np.isfinite(xmin) or not np.isfinite(xmax) or xmax <= xmin:
        return np.zeros_like(x, dtype=np.int32), np.zeros_like(y, dtype=np.int32)

    edges = np.linspace(xmin, xmax, num_bins + 1)

    xb = np.digitize(x, edges[1:-1], right=False).astype(np.int32)
    yb = np.digitize(y, edges[1:-1], right=False).astype(np.int32)

    xb = np.clip(xb, 0, num_bins - 1)
    yb = np.clip(yb, 0, num_bins - 1)

    return xb, yb

def compute_te_pair(source, target, k=1):
    return float(transfer_entropy(source, target, k))

def rolling_te(x, y, window=1500, step=300, num_bins=32, k=1, heartbeat_label=""):
    """
    Rolling TE for exploratory time-localized directionality.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    rows = []
    n = len(x)

    if n < window:
        return pd.DataFrame(columns=["start", "center", "TE_LR", "TE_RL", "TE_diff"])

    starts = list(range(0, n - window + 1, step))
    hb(f"Starting rolling TE {heartbeat_label} | windows={len(starts)}")

    for idx, start in enumerate(starts, start=1):
        end = start + window
        xs = x[start:end]
        ys = y[start:end]

        xb, yb = joint_bin_equal_width(xs, ys, num_bins=num_bins)
        te_lr = compute_te_pair(xb, yb, k=k)
        te_rl = compute_te_pair(yb, xb, k=k)

        rows.append({
            "start": start,
            "center": start + window // 2,
            "TE_LR": te_lr,
            "TE_RL": te_rl,
            "TE_diff": te_lr - te_rl,
        })

        if idx == 1 or idx % 5 == 0 or idx == len(starts):
            hb(f"  rolling TE {heartbeat_label} | window {idx}/{len(starts)}")

    return pd.DataFrame(rows)

def align_for_lag(x, y, lag):
    """
    Positive lag means x leads y by 'lag' samples.
    Returns aligned (x_aligned, y_aligned).
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    if lag > 0:
        return x[:-lag], y[lag:]
    if lag < 0:
        lag = abs(lag)
        return x[lag:], y[:-lag]
    return x, y

def lag_sweep_te(x, y, lag_values, num_bins=32, k=1, heartbeat_label=""):
    """
    Exploratory TE as a function of alignment lag.
    """
    rows = []
    hb(f"Starting lag sweep {heartbeat_label} | n_lags={len(lag_values)}")

    for idx, lag in enumerate(lag_values, start=1):
        xa, ya = align_for_lag(x, y, lag)

        xb, yb = joint_bin_equal_width(xa, ya, num_bins=num_bins)
        te_lr = compute_te_pair(xb, yb, k=k)
        te_rl = compute_te_pair(yb, xb, k=k)

        rows.append({
            "lag": lag,
            "TE_LR": te_lr,
            "TE_RL": te_rl,
            "TE_diff": te_lr - te_rl,
            "N_used": len(xa),
        })

        if idx == 1 or idx % 5 == 0 or idx == len(lag_values):
            hb(f"  lag sweep {heartbeat_label} | lag {idx}/{len(lag_values)}")

    return pd.DataFrame(rows)

# -------------------------------------------------------
# LOAD ONLY NEEDED CHANNELS
# -------------------------------------------------------
channels_needed = sorted(set(left_channels + right_channels))
embedded = {}

hb(f"Opening zip: {zipped_file_path}")

with zipfile.ZipFile(zipped_file_path, "r") as zipf:
    for idx, ch in enumerate(channels_needed, start=1):
        hb(f"Loading channel {idx}/{len(channels_needed)}: {ch}")
        embedded[ch] = load_embedded_channel_from_zip(zipf, ch)
        hb(f"Loaded {ch}: shape={embedded[ch].shape}")

# -------------------------------------------------------
# ALIGN LENGTHS
# -------------------------------------------------------
common_length = min(len(embedded[ch]) for ch in channels_needed)
hb(f"Common length across required channels: {common_length}")

for ch in channels_needed:
    embedded[ch] = embedded[ch][:common_length]

# -------------------------------------------------------
# BUILD HEMISPHERE AVERAGES
# -------------------------------------------------------
hb("Computing hemisphere-average embeddings")

left_stack = np.stack([embedded[ch] for ch in left_channels], axis=0)
right_stack = np.stack([embedded[ch] for ch in right_channels], axis=0)

left_embedding_avg = np.mean(left_stack, axis=0)
right_embedding_avg = np.mean(right_stack, axis=0)

hb(f"Left average shape:  {left_embedding_avg.shape}")
hb(f"Right average shape: {right_embedding_avg.shape}")

# -------------------------------------------------------
# REDUCE LENGTH FOR SPEED
# -------------------------------------------------------
left_dim0, stride0 = reduce_series_length(left_embedding_avg[:, 0], MAX_POINTS_PER_SERIES)
right_dim0, _ = reduce_series_length(right_embedding_avg[:, 0], MAX_POINTS_PER_SERIES)

left_dim1, stride1 = reduce_series_length(left_embedding_avg[:, 1], MAX_POINTS_PER_SERIES)
right_dim1, _ = reduce_series_length(right_embedding_avg[:, 1], MAX_POINTS_PER_SERIES)

left_reduced = np.column_stack([left_dim0, left_dim1])
right_reduced = np.column_stack([right_dim0, right_dim1])

hb(f"Reduced Dim 1 | len={len(left_dim0)} | stride={stride0}")
hb(f"Reduced Dim 2 | len={len(left_dim1)} | stride={stride1}")

# -------------------------------------------------------
# BASIC TE SUMMARY
# -------------------------------------------------------
summary_rows = []
summary_csv_path = output_directory / "transfer_entropy_summary.csv"

for dim in range(2):
    hb(f"Computing basic TE summary | dim={dim + 1}")

    left_data = left_reduced[:, dim]
    right_data = right_reduced[:, dim]

    xb, yb = joint_bin_equal_width(left_data, right_data, num_bins=NUM_BINS)
    te_lr = compute_te_pair(xb, yb, k=K_HISTORY)
    te_rl = compute_te_pair(yb, xb, k=K_HISTORY)

    summary_rows.append({
        "dimension": dim + 1,
        "TE_left_to_right": te_lr,
        "TE_right_to_left": te_rl,
        "TE_asymmetry_LR_minus_RL": te_lr - te_rl,
        "length_used": len(left_data),
        "stride_used": stride0 if dim == 0 else stride1,
        "num_bins": NUM_BINS,
        "k_history": K_HISTORY,
    })

    if SAVE_PARTIALS:
        pd.DataFrame(summary_rows).to_csv(summary_csv_path, index=False)
        hb(f"Saved partial summary CSV after dim {dim + 1}")

summary_df = pd.DataFrame(summary_rows)
summary_df.to_csv(summary_csv_path, index=False)
hb(f"Saved summary CSV: {summary_csv_path}")
print("\nBasic Transfer Entropy summary:")
print(summary_df)

# -------------------------------------------------------
# PLOT 1: BASIC TE SUMMARY
# -------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 6), facecolor=BG)
style_ax(ax)

x = np.arange(len(summary_df))
width = 0.35

ax.bar(
    x - width / 2,
    summary_df["TE_left_to_right"].to_numpy(dtype=float),
    width=width,
    color=C_PRIMARY,
    alpha=0.90,
    label="Left → Right"
)

ax.bar(
    x + width / 2,
    summary_df["TE_right_to_left"].to_numpy(dtype=float),
    width=width,
    color=C_SECONDARY,
    alpha=0.90,
    label="Right → Left"
)

ax.set_title("Transfer Entropy Between Left and Right Hemisphere Averages")
ax.set_xlabel("Embedding Dimension")
ax.set_ylabel("Transfer Entropy")
ax.set_xticks(x)
ax.set_xticklabels([f"Dim {d}" for d in summary_df["dimension"]])

legend = ax.legend()
style_legend(legend)

save_and_show(fig, plots_directory / "te_summary.png")

# -------------------------------------------------------
# PLOT 2: TE ASYMMETRY
# -------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 6), facecolor=BG)
style_ax(ax)

ax.bar(
    [f"Dim {d}" for d in summary_df["dimension"]],
    summary_df["TE_asymmetry_LR_minus_RL"].to_numpy(dtype=float),
    color=C_DIFF,
    alpha=0.92
)

ax.axhline(0.0, color=C_MUTED, linewidth=1.0, linestyle="--")
ax.set_title("Transfer Entropy Asymmetry (Left→Right minus Right→Left)")
ax.set_xlabel("Embedding Dimension")
ax.set_ylabel("TE Asymmetry")

save_and_show(fig, plots_directory / "te_asymmetry.png")

# -------------------------------------------------------
# PLOT 3: Z-SCORED PREVIEWS
# -------------------------------------------------------
for dim in range(2):
    hb(f"Plotting z-scored preview | dim={dim + 1}")

    left_z = standardize_1d(left_reduced[:, dim])
    right_z = standardize_1d(right_reduced[:, dim])

    n_preview = min(PREVIEW_POINTS, len(left_z), len(right_z))
    t = np.arange(n_preview)

    fig, ax = plt.subplots(figsize=(12, 5), facecolor=BG)
    style_ax(ax)

    ax.plot(
        t,
        left_z[:n_preview],
        linewidth=1.4,
        linestyle="-",
        color=C_PRIMARY,
        label="Left hemisphere z-score"
    )
    ax.plot(
        t,
        right_z[:n_preview],
        linewidth=1.4,
        linestyle="--",
        color=C_SECONDARY,
        label="Right hemisphere z-score"
    )

    ax.set_title(f"Z-Scored Hemisphere-Averaged Embedded Signals | Dimension {dim + 1}")
    ax.set_xlabel("Sample Index")
    ax.set_ylabel("Z-Score")

    legend = ax.legend()
    style_legend(legend)

    save_and_show(fig, plots_directory / f"zscore_preview_dim{dim + 1}.png")

# -------------------------------------------------------
# PLOT 4: LEFT VS RIGHT SCATTER
# -------------------------------------------------------
for dim in range(2):
    hb(f"Plotting left-vs-right scatter | dim={dim + 1}")

    left_z = standardize_1d(left_reduced[:, dim])
    right_z = standardize_1d(right_reduced[:, dim])

    n_scatter = min(PREVIEW_POINTS, len(left_z), len(right_z))
    xs = left_z[:n_scatter]
    ys = right_z[:n_scatter]

    fig, ax = plt.subplots(figsize=(7, 7), facecolor=BG)
    style_ax(ax)

    ax.scatter(xs, ys, s=6, alpha=0.55, color=C_HIGHLIGHT)

    lo = min(xs.min(), ys.min())
    hi = max(xs.max(), ys.max())
    ax.plot([lo, hi], [lo, hi], linestyle="--", linewidth=1.0, color=C_DIFF, alpha=0.85)

    ax.set_title(f"Left vs Right Hemisphere Scatter | Dimension {dim + 1}")
    ax.set_xlabel("Left hemisphere z-score")
    ax.set_ylabel("Right hemisphere z-score")

    save_and_show(fig, plots_directory / f"left_vs_right_scatter_dim{dim + 1}.png")

# -------------------------------------------------------
# ROLLING TE
# -------------------------------------------------------
rolling_frames = []
rolling_csv_path = output_directory / "transfer_entropy_rolling.csv"

for dim in range(2):
    left_data = left_reduced[:, dim]
    right_data = right_reduced[:, dim]

    df_roll = rolling_te(
        left_data,
        right_data,
        window=ROLLING_WINDOW,
        step=ROLLING_STEP,
        num_bins=NUM_BINS,
        k=K_HISTORY,
        heartbeat_label=f"dim={dim + 1}"
    )
    df_roll["dimension"] = dim + 1
    rolling_frames.append(df_roll)

    if SAVE_PARTIALS:
        pd.concat(rolling_frames, ignore_index=True).to_csv(rolling_csv_path, index=False)
        hb(f"Saved partial rolling CSV after dim {dim + 1}")

rolling_df = pd.concat(rolling_frames, ignore_index=True) if rolling_frames else pd.DataFrame()
rolling_df.to_csv(rolling_csv_path, index=False)
hb(f"Saved rolling CSV: {rolling_csv_path}")

# -------------------------------------------------------
# PLOT 5: ROLLING TE CURVES
# -------------------------------------------------------
if not rolling_df.empty:
    for dim in sorted(rolling_df["dimension"].unique()):
        hb(f"Plotting rolling TE curves | dim={dim}")

        df_dim = rolling_df[rolling_df["dimension"] == dim].copy()

        fig, ax = plt.subplots(figsize=(12, 5), facecolor=BG)
        style_ax(ax)

        ax.plot(
            df_dim["center"], df_dim["TE_LR"],
            linewidth=1.9, linestyle="-", color=C_PRIMARY, label="Left → Right"
        )
        ax.plot(
            df_dim["center"], df_dim["TE_RL"],
            linewidth=1.9, linestyle="--", color=C_SECONDARY, label="Right → Left"
        )
        ax.plot(
            df_dim["center"], df_dim["TE_diff"],
            linewidth=1.9, linestyle="-.", color=C_DIFF, label="Difference"
        )

        ax.set_title(f"Rolling Transfer Entropy | Dimension {dim}")
        ax.set_xlabel("Window Center")
        ax.set_ylabel("Transfer Entropy")

        legend = ax.legend()
        style_legend(legend)

        save_and_show(fig, plots_directory / f"rolling_te_dim{dim}.png")

# -------------------------------------------------------
# LAG SWEEP TE
# -------------------------------------------------------
lag_frames = []
lag_csv_path = output_directory / "transfer_entropy_lag_sweep.csv"

for dim in range(2):
    left_data = left_reduced[:, dim]
    right_data = right_reduced[:, dim]

    df_lag = lag_sweep_te(
        left_data,
        right_data,
        lag_values=LAG_VALUES,
        num_bins=NUM_BINS,
        k=K_HISTORY,
        heartbeat_label=f"dim={dim + 1}"
    )
    df_lag["dimension"] = dim + 1
    lag_frames.append(df_lag)

    if SAVE_PARTIALS:
        pd.concat(lag_frames, ignore_index=True).to_csv(lag_csv_path, index=False)
        hb(f"Saved partial lag CSV after dim {dim + 1}")

lag_df = pd.concat(lag_frames, ignore_index=True) if lag_frames else pd.DataFrame()
lag_df.to_csv(lag_csv_path, index=False)
hb(f"Saved lag sweep CSV: {lag_csv_path}")

# -------------------------------------------------------
# PLOT 6: LAG SWEEP CURVES
# -------------------------------------------------------
if not lag_df.empty:
    for dim in sorted(lag_df["dimension"].unique()):
        hb(f"Plotting lag sweep | dim={dim}")

        df_dim = lag_df[lag_df["dimension"] == dim].copy()

        fig, ax = plt.subplots(figsize=(12, 5), facecolor=BG)
        style_ax(ax)

        ax.plot(
            df_dim["lag"], df_dim["TE_LR"],
            linewidth=1.8, linestyle="-", marker="o",
            color=C_PRIMARY, label="Left → Right"
        )
        ax.plot(
            df_dim["lag"], df_dim["TE_RL"],
            linewidth=1.8, linestyle="--", marker="o",
            color=C_SECONDARY, label="Right → Left"
        )
        ax.plot(
            df_dim["lag"], df_dim["TE_diff"],
            linewidth=1.8, linestyle="-.", marker="s",
            color=C_DIFF, label="Difference"
        )

        ax.axhline(0.0, linewidth=1.0, linestyle="--", color=C_MUTED, alpha=0.8)
        ax.axvline(0.0, linewidth=1.0, linestyle="--", color=C_MUTED, alpha=0.8)

        ax.set_title(f"Lag-Sweep Transfer Entropy | Dimension {dim}")
        ax.set_xlabel("Lag (positive = Left leads Right)")
        ax.set_ylabel("Transfer Entropy")

        legend = ax.legend()
        style_legend(legend)

        save_and_show(fig, plots_directory / f"lag_sweep_te_dim{dim}.png")

# -------------------------------------------------------
# PRINT SUMMARY
# -------------------------------------------------------
print("\nExploratory summary:")
print(summary_df)

if not rolling_df.empty:
    print("\nRolling TE head:")
    print(rolling_df.head())

if not lag_df.empty:
    print("\nLag sweep head:")
    print(lag_df.head())

hb("Done.")
```

# Hemispherical Pairs

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — Transfer entropy</h3><p>Transfer entropy is a conditional mutual information that measures whether the past of a source $X$ improves prediction of the next target value $Y_{t+1}$ beyond the target's own past. It is directional and depends on the history lengths, lag, estimator, and any discretization used by the implementation.</p><p><b>Cell role.</b> This code applies that theory to the notebook's existing EEG-derived data and preserves the surrounding notebook's plotting/output conventions. Numerical shortcuts selected by a speed mode change sampling density, iteration counts, or visualization resolution; they do not change the defining formula above.</p></div>

$$T_{X\\to Y}=\\sum p(y_{t+1},y_t^{(k)},x_t^{(l)})\\log\\frac{p(y_{t+1}\\mid y_t^{(k)},x_t^{(l)})}{p(y_{t+1}\\mid y_t^{(k)})}.$$

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Mutual information measures how much knowing one variable reduces uncertainty about another. Unlike linear correlation, it can capture nonlinear statistical dependence. In delay-selection problems, mutual information between a signal and a delayed copy is examined as the delay changes; a local minimum is often used as a compromise between coordinates that are too redundant and coordinates that are almost unrelated. The estimate is sensitive to finite sample size and to how probability densities are estimated.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values. The visualization is a way of displaying already-computed numbers. Color, ordering, or ranking can make patterns easier to see but does not create additional statistical evidence beyond the values and uncertainty measures being plotted.</p></div>


```python
import os
import time
import math
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

try:
    from IPython.display import display
    _HAVE_IPYTHON_DISPLAY = True
except Exception:
    display = None
    _HAVE_IPYTHON_DISPLAY = False


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"

raw_eeg_path = os.path.join(base_dir, "eeg_data_with_channels.npy")
embedding_2d_dir = os.path.join(base_dir, "embedding_data", "2dembedding_data")

results_dir = os.path.join(base_dir, "results", "transfer_entropy_hemisphere_expert")
plots_dir = os.path.join(base_dir, "plots", "transfer_entropy_hemisphere_expert")

os.makedirs(results_dir, exist_ok=True)
os.makedirs(plots_dir, exist_ok=True)


FAST_OPTION = "fast"

if FAST_OPTION == "ultra":
    window_size = 1500
    step_size = 1500
    n_bins = 5
    max_pairs = 20
    max_points_per_channel = 25000
    lag_values = [1]
    permutation_count = 0
    top_n_pairs = 15

elif FAST_OPTION == "fast":
    window_size = 1500
    step_size = 1000
    n_bins = 6
    max_pairs = 35
    max_points_per_channel = 40000
    lag_values = [1]
    permutation_count = 0
    top_n_pairs = 20

elif FAST_OPTION == "balanced":
    window_size = 1500
    step_size = 750
    n_bins = 8
    max_pairs = 80
    max_points_per_channel = 80000
    lag_values = [1, 2]
    permutation_count = 10
    top_n_pairs = 30

elif FAST_OPTION == "full":
    window_size = 2000
    step_size = 500
    n_bins = 10
    max_pairs = None
    max_points_per_channel = None
    lag_values = [1, 2, 3]
    permutation_count = 30
    top_n_pairs = 50

else:
    raise ValueError("FAST_OPTION must be one of: 'ultra', 'fast', 'balanced', 'full'.")

sampling_rate = 1000
start_time = 814.571
end_time = 921.515

SHOW_PLOTS_IN_NOTEBOOK = True
SAVE_PLOTS_TO_DISK = True

USE_EXISTING_GLOBAL_EMBEDDINGS_IF_AVAILABLE = True
PREFER_2D_EMBEDDINGS = True
EMBEDDING_TO_SCALAR_MODE = "radius"
NORMALIZE_BEFORE_TE = True
USE_QUANTILE_BINS = True

ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"
WHITE = "#E8FFFF"
BLACK = "#000000"

CYAN_DIVERGING_CMAP = LinearSegmentedColormap.from_list(
    "cyan_diverging",
    [ACCENT_DIM, BLACK, ACCENT_SOFT],
    N=256,
)

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, ACCENT_DIM, ACCENT, ACCENT_SOFT],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})

rng = np.random.default_rng(42)

eeg_channels = [
    "Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8", "FC5", "FC1", "FC2", "FC6",
    "M1", "T7", "C3", "Cz", "C4", "T8", "M2", "CP5", "CP1", "CP2", "CP6", "P7",
    "P3", "Pz", "P4", "P8", "POz", "O1", "Oz", "O2"
]

left_channels = [
    "Fp1", "F7", "F3", "FC5", "FC1", "M1", "T7", "C3", "CP5", "CP1", "P7", "P3", "O1"
]

right_channels = [
    "Fp2", "F8", "F4", "FC2", "FC6", "M2", "T8", "C4", "CP2", "CP6", "P8", "P4", "O2"
]

midline_channels = ["Fpz", "Fz", "Cz", "Pz", "POz", "Oz"]


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT)

    if grid:
        ax.grid(True, alpha=0.25, color=ACCENT_DIM, linewidth=0.7)
    else:
        ax.grid(False)


def save_fig(fig, path, show=True):
    fig.tight_layout()

    if SAVE_PLOTS_TO_DISK and path is not None:
        fig.savefig(path, dpi=220, bbox_inches="tight", facecolor=BLACK)

    if SHOW_PLOTS_IN_NOTEBOOK and show and _HAVE_IPYTHON_DISPLAY:
        display(fig)

    plt.close(fig)


def short_label(x, max_len=28):
    x = str(x)
    return x if len(x) <= max_len else x[:max_len - 1] + "…"


def safe_zscore_1d(x):
    x = np.asarray(x, dtype=float).reshape(-1)
    x = np.where(np.isfinite(x), x, np.nan)

    if np.any(~np.isfinite(x)):
        med = np.nanmedian(x)
        if not np.isfinite(med):
            med = 0.0
        x = np.nan_to_num(x, nan=med, posinf=med, neginf=med)

    mu = np.mean(x)
    sd = np.std(x) + 1e-12

    return (x - mu) / sd


def embedding_to_scalar(arr, mode="radius"):
    arr = np.asarray(arr, dtype=float)

    if arr.ndim == 1:
        x = arr.reshape(-1)

    elif arr.ndim == 2:
        if mode == "x":
            x = arr[:, 0]
        elif mode == "y":
            x = arr[:, min(1, arr.shape[1] - 1)]
        elif mode == "radius":
            centered = arr - np.nanmean(arr, axis=0, keepdims=True)
            x = np.sqrt(np.nansum(centered ** 2, axis=1))
        elif mode == "speed":
            centered = arr - np.nanmean(arr, axis=0, keepdims=True)
            dx = np.diff(centered, axis=0, prepend=centered[[0], :])
            x = np.sqrt(np.nansum(dx ** 2, axis=1))
        elif mode == "pc1":
            X = arr - np.nanmean(arr, axis=0, keepdims=True)
            X = np.nan_to_num(X, nan=0.0)
            _, _, Vt = np.linalg.svd(X, full_matrices=False)
            x = X @ Vt[0]
        else:
            raise ValueError("EMBEDDING_TO_SCALAR_MODE must be one of: 'x', 'y', 'radius', 'speed', 'pc1'.")
    else:
        x = arr.reshape(arr.shape[0], -1)
        x = np.sqrt(np.nansum((x - np.nanmean(x, axis=0, keepdims=True)) ** 2, axis=1))

    x = np.asarray(x, dtype=float).reshape(-1)

    if NORMALIZE_BEFORE_TE:
        x = safe_zscore_1d(x)

    return x


def load_from_existing_global_embeddings():
    if not USE_EXISTING_GLOBAL_EMBEDDINGS_IF_AVAILABLE:
        return None, None

    g = globals()

    if "embedding_data_list_2D_trimmed" not in g:
        return None, None

    data_obj = g["embedding_data_list_2D_trimmed"]

    if "eeg_channels" in g:
        names = list(g["eeg_channels"])
    elif "eeg_channel_names" in g:
        names = list(g["eeg_channel_names"])
    else:
        names = list(eeg_channels)

    channel_series = {}

    try:
        if isinstance(data_obj, dict):
            for ch in names:
                if ch in data_obj:
                    channel_series[ch] = embedding_to_scalar(data_obj[ch], EMBEDDING_TO_SCALAR_MODE)

        else:
            arr = np.asarray(data_obj, dtype=object)

            if len(arr) == len(names):
                for idx, ch in enumerate(names):
                    channel_series[ch] = embedding_to_scalar(arr[idx], EMBEDDING_TO_SCALAR_MODE)

    except Exception:
        return None, None

    if len(channel_series) == 0:
        return None, None

    return channel_series, "existing_global_embedding_data_list_2D_trimmed"


def load_from_2d_embedding_files():
    channel_series = {}

    for ch in eeg_channels:
        candidates = [
            os.path.join(embedding_2d_dir, f"2dembedded_{ch}.npy"),
            os.path.join(embedding_2d_dir, f"2dembedded_{ch}_trimmed.npy"),
            os.path.join(embedding_2d_dir, f"{ch}.npy"),
        ]

        matches = []

        for c in candidates:
            if os.path.exists(c):
                matches.append(c)

        if len(matches) == 0:
            glob_matches = sorted(glob.glob(os.path.join(embedding_2d_dir, f"*{ch}*.npy")))
            matches.extend(glob_matches)

        if len(matches) == 0:
            continue

        try:
            arr = np.load(matches[0], allow_pickle=True)
            channel_series[ch] = embedding_to_scalar(arr, EMBEDDING_TO_SCALAR_MODE)
        except Exception:
            continue

    if len(channel_series) == 0:
        return None, None

    return channel_series, embedding_2d_dir


def load_from_raw_eeg():
    if not os.path.exists(raw_eeg_path):
        raise FileNotFoundError(
            "Could not find embeddings or raw EEG file. Expected raw file at:\n"
            f"{raw_eeg_path}"
        )

    EEG_data = np.load(raw_eeg_path, allow_pickle=True)
    EEG_data = np.asarray(EEG_data, dtype=float)

    if EEG_data.ndim != 2:
        raise ValueError(f"Expected raw EEG shape (time, channels), got {EEG_data.shape}")

    n_channels = min(EEG_data.shape[1], len(eeg_channels))
    names = eeg_channels[:n_channels]

    start_index = max(0, int(start_time * sampling_rate))
    end_index = min(EEG_data.shape[0], int(end_time * sampling_rate))

    if end_index <= start_index:
        raise ValueError(
            f"Invalid segment indices: start={start_index}, end={end_index}, length={EEG_data.shape[0]}"
        )

    EEG_seg = EEG_data[start_index:end_index, :n_channels]

    channel_series = {}

    for i, ch in enumerate(names):
        x = EEG_seg[:, i]
        if NORMALIZE_BEFORE_TE:
            x = safe_zscore_1d(x)
        channel_series[ch] = x

    return channel_series, raw_eeg_path


def load_channel_series():
    channel_series, source = load_from_existing_global_embeddings()

    if channel_series is not None:
        return channel_series, source

    if PREFER_2D_EMBEDDINGS:
        channel_series, source = load_from_2d_embedding_files()

        if channel_series is not None:
            return channel_series, source

    return load_from_raw_eeg()


def trim_channel_series_to_common_length(channel_series, max_points=None):
    available = {k: np.asarray(v, dtype=float).reshape(-1) for k, v in channel_series.items() if len(v) > 10}

    common_len = min(len(v) for v in available.values())

    if max_points is not None:
        common_len = min(common_len, int(max_points))

    out = {}

    for ch, x in available.items():
        out[ch] = x[:common_len]

    return out, common_len


def bin_data_quantile(x, n_bins=8):
    x = np.asarray(x, dtype=float).reshape(-1)

    if np.std(x) < 1e-12:
        return np.zeros(len(x), dtype=np.int16)

    if USE_QUANTILE_BINS:
        quantiles = np.linspace(0, 1, n_bins + 1)
        edges = np.quantile(x, quantiles)
        edges = np.unique(edges)

        if len(edges) <= 2:
            edges = np.linspace(np.min(x), np.max(x), n_bins + 1)
    else:
        edges = np.linspace(np.min(x), np.max(x), n_bins + 1)

    bins = np.digitize(x, edges[1:-1], right=False).astype(np.int16)
    bins = np.clip(bins, 0, n_bins - 1)

    return bins


def transfer_entropy_binned(source_bins, target_bins, lag=1, n_bins=None):
    x = np.asarray(source_bins, dtype=np.int64).reshape(-1)
    y = np.asarray(target_bins, dtype=np.int64).reshape(-1)

    n = min(len(x), len(y))

    if n <= lag + 2:
        return np.nan

    x = x[:n]
    y = y[:n]

    if n_bins is None:
        n_bins = int(max(np.max(x), np.max(y)) + 1)

    y_future = y[lag:]
    y_past = y[:-lag]
    x_past = x[:-lag]

    valid = (
        (x_past >= 0)
        & (x_past < n_bins)
        & (y_past >= 0)
        & (y_past < n_bins)
        & (y_future >= 0)
        & (y_future < n_bins)
    )

    x_past = x_past[valid]
    y_past = y_past[valid]
    y_future = y_future[valid]

    m = len(y_future)

    if m <= 3:
        return np.nan

    counts_yf_yp_xp = np.zeros((n_bins, n_bins, n_bins), dtype=np.float64)
    counts_yp_xp = np.zeros((n_bins, n_bins), dtype=np.float64)
    counts_yf_yp = np.zeros((n_bins, n_bins), dtype=np.float64)
    counts_yp = np.zeros(n_bins, dtype=np.float64)

    np.add.at(counts_yf_yp_xp, (y_future, y_past, x_past), 1)
    np.add.at(counts_yp_xp, (y_past, x_past), 1)
    np.add.at(counts_yf_yp, (y_future, y_past), 1)
    np.add.at(counts_yp, y_past, 1)

    te = 0.0

    nz = counts_yf_yp_xp > 0
    yf_idx, yp_idx, xp_idx = np.where(nz)

    for yf, yp, xp in zip(yf_idx, yp_idx, xp_idx):
        c_xyz = counts_yf_yp_xp[yf, yp, xp]
        c_yx = counts_yp_xp[yp, xp]
        c_yy = counts_yf_yp[yf, yp]
        c_y = counts_yp[yp]

        if c_xyz > 0 and c_yx > 0 and c_yy > 0 and c_y > 0:
            p_xyz = c_xyz / m
            p_yf_given_yp_xp = c_xyz / c_yx
            p_yf_given_yp = c_yy / c_y
            te += p_xyz * np.log2((p_yf_given_yp_xp + 1e-12) / (p_yf_given_yp + 1e-12))

    return float(max(te, 0.0))


def transfer_entropy_pair_continuous(source, target, window_size=1000, step_size=500, lag=1, n_bins=8):
    source = np.asarray(source, dtype=float).reshape(-1)
    target = np.asarray(target, dtype=float).reshape(-1)

    n = min(len(source), len(target))
    source = source[:n]
    target = target[:n]

    starts = np.arange(0, n - window_size + 1, step_size)

    if starts.size == 0:
        return np.array([], dtype=float), np.array([], dtype=int)

    te_vals = []

    for s in starts:
        e = s + window_size
        xb = bin_data_quantile(source[s:e], n_bins=n_bins)
        yb = bin_data_quantile(target[s:e], n_bins=n_bins)

        te = transfer_entropy_binned(xb, yb, lag=lag, n_bins=n_bins)
        te_vals.append(te)

    return np.asarray(te_vals, dtype=float), starts


def permutation_te_threshold(source, target, window_size, step_size, lag, n_bins, n_perm=20):
    if n_perm <= 0:
        return np.nan, np.nan

    source = np.asarray(source, dtype=float).reshape(-1)
    target = np.asarray(target, dtype=float).reshape(-1)

    null_means = []

    for p in range(n_perm):
        shift = int(rng.integers(window_size, max(window_size + 1, len(source) - window_size)))

        source_perm = np.roll(source, shift)
        vals, _ = transfer_entropy_pair_continuous(
            source_perm,
            target,
            window_size=window_size,
            step_size=step_size,
            lag=lag,
            n_bins=n_bins,
        )

        vals = vals[np.isfinite(vals)]

        if len(vals):
            null_means.append(float(np.mean(vals)))

    if len(null_means) == 0:
        return np.nan, np.nan

    null_means = np.asarray(null_means, dtype=float)

    return float(np.mean(null_means)), float(np.percentile(null_means, 95))


def make_channel_pairs(left_channels, right_channels, available_channels, max_pairs=None):
    pairs = []

    left = [ch for ch in left_channels if ch in available_channels]
    right = [ch for ch in right_channels if ch in available_channels]

    for l in left:
        for r in right:
            pairs.append((l, r))

    if max_pairs is not None and len(pairs) > max_pairs:
        idx = np.sort(rng.choice(len(pairs), size=max_pairs, replace=False))
        pairs = [pairs[i] for i in idx]

    return pairs


def compute_transfer_entropy_hemispheres(channel_series, left_channels, right_channels):
    available = set(channel_series.keys())
    pairs = make_channel_pairs(left_channels, right_channels, available, max_pairs=max_pairs)

    if len(pairs) == 0:
        raise ValueError("No valid left-right channel pairs found.")

    all_lr = []
    all_rl = []
    rows = []
    lag_rows = []

    starts_ref = None
    t0 = time.perf_counter()

    for pair_idx, (left, right) in enumerate(pairs, start=1):
        x_left = channel_series[left]
        x_right = channel_series[right]

        best_lag = None
        best_lr_mean = -np.inf
        best_rl_mean = -np.inf
        best_lr_vals = None
        best_rl_vals = None
        best_starts = None

        for lag in lag_values:
            lr_vals, starts = transfer_entropy_pair_continuous(
                x_left,
                x_right,
                window_size=window_size,
                step_size=step_size,
                lag=lag,
                n_bins=n_bins,
            )

            rl_vals, _ = transfer_entropy_pair_continuous(
                x_right,
                x_left,
                window_size=window_size,
                step_size=step_size,
                lag=lag,
                n_bins=n_bins,
            )

            lr_mean = float(np.nanmean(lr_vals)) if len(lr_vals) else np.nan
            rl_mean = float(np.nanmean(rl_vals)) if len(rl_vals) else np.nan

            lag_rows.append({
                "pair": f"{left}->{right}",
                "left_channel": left,
                "right_channel": right,
                "lag": lag,
                "mean_TE_L_to_R": lr_mean,
                "mean_TE_R_to_L": rl_mean,
                "net_TE_L_minus_R": lr_mean - rl_mean if np.isfinite(lr_mean) and np.isfinite(rl_mean) else np.nan,
            })

            score = max(lr_mean if np.isfinite(lr_mean) else -np.inf, rl_mean if np.isfinite(rl_mean) else -np.inf)

            if score > max(best_lr_mean, best_rl_mean):
                best_lag = lag
                best_lr_mean = lr_mean
                best_rl_mean = rl_mean
                best_lr_vals = lr_vals
                best_rl_vals = rl_vals
                best_starts = starts

        if starts_ref is None:
            starts_ref = best_starts

        if len(best_lr_vals) == 0:
            continue

        if len(best_rl_vals) == 0:
            continue

        null_mean_lr, null_95_lr = permutation_te_threshold(
            x_left,
            x_right,
            window_size=window_size,
            step_size=step_size,
            lag=best_lag,
            n_bins=n_bins,
            n_perm=permutation_count,
        )

        null_mean_rl, null_95_rl = permutation_te_threshold(
            x_right,
            x_left,
            window_size=window_size,
            step_size=step_size,
            lag=best_lag,
            n_bins=n_bins,
            n_perm=permutation_count,
        )

        lr_mean = float(np.nanmean(best_lr_vals))
        rl_mean = float(np.nanmean(best_rl_vals))
        lr_median = float(np.nanmedian(best_lr_vals))
        rl_median = float(np.nanmedian(best_rl_vals))
        lr_max = float(np.nanmax(best_lr_vals))
        rl_max = float(np.nanmax(best_rl_vals))

        net = lr_mean - rl_mean
        total = lr_mean + rl_mean
        asymmetry_index = net / (total + 1e-12)

        lr_z = (lr_mean - null_mean_lr) / (null_95_lr - null_mean_lr + 1e-12) if np.isfinite(null_mean_lr) and np.isfinite(null_95_lr) else np.nan
        rl_z = (rl_mean - null_mean_rl) / (null_95_rl - null_mean_rl + 1e-12) if np.isfinite(null_mean_rl) and np.isfinite(null_95_rl) else np.nan

        rows.append({
            "pair_index": pair_idx,
            "left_channel": left,
            "right_channel": right,
            "pair": f"{left}-{right}",
            "best_lag": best_lag,
            "mean_TE_L_to_R": lr_mean,
            "mean_TE_R_to_L": rl_mean,
            "median_TE_L_to_R": lr_median,
            "median_TE_R_to_L": rl_median,
            "max_TE_L_to_R": lr_max,
            "max_TE_R_to_L": rl_max,
            "net_TE_L_minus_R": net,
            "abs_net_TE": abs(net),
            "bidirectional_TE_sum": total,
            "asymmetry_index": asymmetry_index,
            "null_mean_L_to_R": null_mean_lr,
            "null_95_L_to_R": null_95_lr,
            "null_mean_R_to_L": null_mean_rl,
            "null_95_R_to_L": null_95_rl,
            "normalized_surrogate_score_L_to_R": lr_z,
            "normalized_surrogate_score_R_to_L": rl_z,
            "n_windows": len(best_lr_vals),
        })

        all_lr.append(best_lr_vals)
        all_rl.append(best_rl_vals)

        print(
            f"[{pair_idx}/{len(pairs)}] {left}->{right} mean={lr_mean:.5f}, "
            f"{right}->{left} mean={rl_mean:.5f}, net={net:.5f}, lag={best_lag}",
            flush=True,
        )

    elapsed = time.perf_counter() - t0

    if len(all_lr) == 0:
        raise ValueError("No TE values computed. Try smaller window_size or more available data.")

    te_lr_matrix = np.vstack(all_lr)
    te_rl_matrix = np.vstack(all_rl)

    summary_df = pd.DataFrame(rows)
    lag_df = pd.DataFrame(lag_rows)

    time_axis = starts_ref / sampling_rate if starts_ref is not None else np.arange(te_lr_matrix.shape[1])

    return te_lr_matrix, te_rl_matrix, time_axis, summary_df, lag_df, elapsed


def pair_matrix_from_summary(summary_df, value_col):
    lefts = [ch for ch in left_channels if ch in summary_df["left_channel"].unique()]
    rights = [ch for ch in right_channels if ch in summary_df["right_channel"].unique()]

    M = np.full((len(lefts), len(rights)), np.nan, dtype=float)

    for i, l in enumerate(lefts):
        for j, r in enumerate(rights):
            sub = summary_df[(summary_df["left_channel"] == l) & (summary_df["right_channel"] == r)]

            if len(sub):
                M[i, j] = float(sub[value_col].iloc[0])

    return M, lefts, rights


def plot_te_temporal_heatmap(te_matrix, time_axis, summary_df, path, direction_label="Left to Right"):
    order = np.argsort(summary_df["mean_TE_L_to_R"].to_numpy() if direction_label == "Left to Right" else summary_df["mean_TE_R_to_L"].to_numpy())[::-1]
    te_ord = te_matrix[order]
    labels = summary_df.iloc[order]["pair"].tolist()

    fig_height = min(10.0, max(6.0, 0.22 * len(labels)))
    fig, ax = plt.subplots(figsize=(13, fig_height))

    vmax = np.nanpercentile(te_ord, 98) if np.any(np.isfinite(te_ord)) else 1.0
    vmax = max(vmax, 1e-12)

    im = ax.imshow(
        te_ord,
        aspect="auto",
        interpolation="nearest",
        cmap=CYAN_SEQUENTIAL_CMAP,
        vmin=0,
        vmax=vmax,
        extent=[time_axis[0], time_axis[-1] if len(time_axis) > 1 else time_axis[0] + 1, len(labels) - 0.5, -0.5],
    )

    ax.set_yticks(np.arange(len(labels)))
    ax.set_yticklabels([short_label(x, 24) for x in labels], color=ACCENT, fontsize=7)

    style_ax(
        ax,
        title=f"Transfer Entropy Temporal Heatmap | {direction_label}",
        xlabel="Time from segment start (s)",
        ylabel="Channel pair",
        grid=False,
    )

    cbar = fig.colorbar(im, ax=ax, fraction=0.035, pad=0.02)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)
    cbar.set_label("Transfer entropy, bits", color=ACCENT)

    save_fig(fig, path)
    return path


def plot_te_asymmetry_heatmap(summary_df, path):
    M, lefts, rights = pair_matrix_from_summary(summary_df, "asymmetry_index")

    if M.size == 0:
        return None

    fig, ax = plt.subplots(figsize=(11, 8))

    im = ax.imshow(
        M,
        aspect="auto",
        interpolation="nearest",
        cmap=CYAN_DIVERGING_CMAP,
        vmin=-1,
        vmax=1,
    )

    ax.set_xticks(np.arange(len(rights)))
    ax.set_xticklabels(rights, rotation=45, ha="right", color=ACCENT)

    ax.set_yticks(np.arange(len(lefts)))
    ax.set_yticklabels(lefts, color=ACCENT)

    for i in range(M.shape[0]):
        for j in range(M.shape[1]):
            if np.isfinite(M[i, j]):
                ax.text(
                    j,
                    i,
                    f"{M[i, j]:.2f}",
                    ha="center",
                    va="center",
                    fontsize=7,
                    color=BLACK if abs(M[i, j]) > 0.45 else ACCENT_SOFT,
                )

    style_ax(
        ax,
        title="Directional TE Asymmetry Index: L→R vs R→L",
        xlabel="Right hemisphere channel",
        ylabel="Left hemisphere channel",
        grid=False,
    )

    cbar = fig.colorbar(im, ax=ax, fraction=0.035, pad=0.02)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)
    cbar.set_label("(L→R - R→L) / (L→R + R→L)", color=ACCENT)

    save_fig(fig, path)
    return path


def plot_te_strength_heatmap(summary_df, path):
    M, lefts, rights = pair_matrix_from_summary(summary_df, "bidirectional_TE_sum")

    if M.size == 0:
        return None

    fig, ax = plt.subplots(figsize=(11, 8))

    vmax = np.nanpercentile(M, 98) if np.any(np.isfinite(M)) else 1.0
    vmax = max(vmax, 1e-12)

    im = ax.imshow(
        M,
        aspect="auto",
        interpolation="nearest",
        cmap=CYAN_SEQUENTIAL_CMAP,
        vmin=0,
        vmax=vmax,
    )

    ax.set_xticks(np.arange(len(rights)))
    ax.set_xticklabels(rights, rotation=45, ha="right", color=ACCENT)

    ax.set_yticks(np.arange(len(lefts)))
    ax.set_yticklabels(lefts, color=ACCENT)

    style_ax(
        ax,
        title="Bidirectional Interhemispheric Transfer Entropy Strength",
        xlabel="Right hemisphere channel",
        ylabel="Left hemisphere channel",
        grid=False,
    )

    cbar = fig.colorbar(im, ax=ax, fraction=0.035, pad=0.02)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)
    cbar.set_label("TE L→R + TE R→L", color=ACCENT)

    save_fig(fig, path)
    return path


def plot_top_te_pairs(summary_df, path, top_n=30):
    df = summary_df.sort_values("bidirectional_TE_sum", ascending=False).head(top_n)
    df = df.sort_values("bidirectional_TE_sum", ascending=True)

    y = np.arange(len(df))

    fig, ax = plt.subplots(figsize=(11, max(6.0, 0.28 * len(df))))

    ax.barh(
        y,
        df["mean_TE_L_to_R"],
        color=ACCENT,
        edgecolor=ACCENT_SOFT,
        alpha=0.80,
        label="L→R",
    )

    ax.barh(
        y,
        -df["mean_TE_R_to_L"],
        color=ACCENT_DIM,
        edgecolor=ACCENT,
        alpha=0.85,
        label="R→L",
    )

    ax.axvline(0, color=WHITE, linewidth=1.0, alpha=0.85)

    ax.set_yticks(y)
    ax.set_yticklabels(df["pair"], color=ACCENT, fontsize=8)

    for yi, (_, row) in zip(y, df.iterrows()):
        side = "L→R" if row["net_TE_L_minus_R"] >= 0 else "R→L"
        ax.text(
            max(row["mean_TE_L_to_R"], 0) + 0.002,
            yi,
            f" net={row['net_TE_L_minus_R']:.3g} {side}",
            va="center",
            color=ACCENT_SOFT,
            fontsize=7,
        )

    style_ax(
        ax,
        title=f"Top {len(df)} Interhemispheric TE Pairs",
        xlabel="Mean TE, bits | left side shown positive, right side shown negative",
        ylabel="Channel pair",
    )

    leg = ax.legend(loc="best", frameon=True)
    for text in leg.get_texts():
        text.set_color(ACCENT)

    save_fig(fig, path)
    return path


def plot_mean_te_timeseries(te_lr, te_rl, time_axis, path):
    lr_mean = np.nanmean(te_lr, axis=0)
    rl_mean = np.nanmean(te_rl, axis=0)
    net = lr_mean - rl_mean

    fig, ax = plt.subplots(figsize=(12, 7))

    ax.plot(
        time_axis,
        lr_mean,
        color=ACCENT,
        linewidth=2.0,
        label="Mean L→R TE",
    )

    ax.plot(
        time_axis,
        rl_mean,
        color=ACCENT_SOFT,
        linewidth=1.8,
        alpha=0.85,
        label="Mean R→L TE",
    )

    ax.fill_between(
        time_axis,
        lr_mean,
        rl_mean,
        where=net >= 0,
        color=ACCENT,
        alpha=0.15,
        label="L→R dominance",
    )

    ax.fill_between(
        time_axis,
        lr_mean,
        rl_mean,
        where=net < 0,
        color=ACCENT_DIM,
        alpha=0.18,
        label="R→L dominance",
    )

    style_ax(
        ax,
        title="Mean Interhemispheric Transfer Entropy Over Time",
        xlabel="Time from segment start (s)",
        ylabel="Transfer entropy, bits",
    )

    leg = ax.legend(loc="best", frameon=True)

    for text in leg.get_texts():
        text.set_color(ACCENT)

    save_fig(fig, path)
    return path


def plot_lag_sensitivity(lag_df, path):
    if lag_df.empty or "lag" not in lag_df.columns:
        return None

    df = lag_df.copy()
    df = df[
        np.isfinite(df["mean_TE_L_to_R"])
        & np.isfinite(df["mean_TE_R_to_L"])
        & np.isfinite(df["net_TE_L_minus_R"])
    ].copy()

    if df.empty:
        return None

    unique_lags = sorted(df["lag"].dropna().unique())

    if len(unique_lags) <= 1:
        lag_value = unique_lags[0] if len(unique_lags) else "NA"

        df["bidirectional_TE_sum"] = df["mean_TE_L_to_R"] + df["mean_TE_R_to_L"]
        df["abs_net_TE"] = np.abs(df["net_TE_L_minus_R"])

        if "pair" not in df.columns:
            df["pair"] = df["left_channel"].astype(str) + "-" + df["right_channel"].astype(str)

        local_top_n = globals().get("top_n_pairs", 25)

        plot_df = (
            df.sort_values(["abs_net_TE", "bidirectional_TE_sum"], ascending=False)
            .head(local_top_n)
            .sort_values("abs_net_TE", ascending=True)
        )

        y = np.arange(len(plot_df))

        fig, ax = plt.subplots(figsize=(11, max(6.0, 0.30 * len(plot_df))))

        ax.barh(
            y,
            plot_df["mean_TE_L_to_R"],
            color=ACCENT,
            edgecolor=ACCENT_SOFT,
            alpha=0.82,
            label="L→R",
        )

        ax.barh(
            y,
            -plot_df["mean_TE_R_to_L"],
            color=ACCENT_DIM,
            edgecolor=ACCENT,
            alpha=0.88,
            label="R→L",
        )

        ax.axvline(0, color=WHITE, linewidth=1.0, alpha=0.90)

        max_val = np.nanmax(
            np.abs(
                plot_df[["mean_TE_L_to_R", "mean_TE_R_to_L"]]
                .to_numpy(dtype=float)
            )
        )

        if np.isfinite(max_val) and max_val > 0:
            ax.set_xlim(-max_val * 1.35, max_val * 1.35)

        ax.set_yticks(y)
        ax.set_yticklabels([short_label(p, 24) for p in plot_df["pair"]], color=ACCENT, fontsize=8)

        for yi, (_, row) in zip(y, plot_df.iterrows()):
            dominant = "L→R" if row["net_TE_L_minus_R"] >= 0 else "R→L"

            ax.text(
                row["mean_TE_L_to_R"] + max_val * 0.025,
                yi,
                f"net={row['net_TE_L_minus_R']:.4f} {dominant}",
                va="center",
                ha="left",
                color=ACCENT_SOFT,
                fontsize=7,
            )

        style_ax(
            ax,
            title=(
                f"Transfer Entropy Directionality at Lag {lag_value}\n"
                "single-lag fast mode: pair-level L→R vs R→L comparison"
            ),
            xlabel="Mean TE, bits | L→R positive, R→L negative",
            ylabel="Channel pair",
        )

        leg = ax.legend(loc="best", frameon=True)
        for text in leg.get_texts():
            text.set_color(ACCENT)

        save_fig(fig, path)
        return path

    grouped = (
        df.groupby("lag")
        .agg(
            mean_L_to_R=("mean_TE_L_to_R", "mean"),
            mean_R_to_L=("mean_TE_R_to_L", "mean"),
            mean_net=("net_TE_L_minus_R", "mean"),
            sem_L_to_R=("mean_TE_L_to_R", lambda x: np.nanstd(x) / np.sqrt(max(np.sum(np.isfinite(x)), 1))),
            sem_R_to_L=("mean_TE_R_to_L", lambda x: np.nanstd(x) / np.sqrt(max(np.sum(np.isfinite(x)), 1))),
            sem_net=("net_TE_L_minus_R", lambda x: np.nanstd(x) / np.sqrt(max(np.sum(np.isfinite(x)), 1))),
            n_pairs=("pair", "count") if "pair" in df.columns else ("lag", "count"),
        )
        .reset_index()
        .sort_values("lag")
    )

    fig, ax = plt.subplots(figsize=(10.5, 7.0))

    ax.plot(
        grouped["lag"],
        grouped["mean_L_to_R"],
        marker="o",
        color=ACCENT,
        linewidth=2.0,
        label="L→R mean",
    )

    ax.fill_between(
        grouped["lag"],
        grouped["mean_L_to_R"] - grouped["sem_L_to_R"],
        grouped["mean_L_to_R"] + grouped["sem_L_to_R"],
        color=ACCENT,
        alpha=0.12,
    )

    ax.plot(
        grouped["lag"],
        grouped["mean_R_to_L"],
        marker="o",
        color=ACCENT_SOFT,
        linewidth=1.8,
        label="R→L mean",
    )

    ax.fill_between(
        grouped["lag"],
        grouped["mean_R_to_L"] - grouped["sem_R_to_L"],
        grouped["mean_R_to_L"] + grouped["sem_R_to_L"],
        color=ACCENT_SOFT,
        alpha=0.10,
    )

    ax.plot(
        grouped["lag"],
        grouped["mean_net"],
        marker="s",
        color=WHITE,
        linewidth=1.6,
        label="Net L-R",
    )

    ax.fill_between(
        grouped["lag"],
        grouped["mean_net"] - grouped["sem_net"],
        grouped["mean_net"] + grouped["sem_net"],
        color=WHITE,
        alpha=0.08,
    )

    ax.axhline(0, color=WHITE, linewidth=0.9, alpha=0.65)

    ax.set_xticks(grouped["lag"].to_numpy())

    best_idx = int(np.nanargmax(np.abs(grouped["mean_net"].to_numpy(dtype=float))))
    best_lag = grouped.iloc[best_idx]["lag"]
    best_net = grouped.iloc[best_idx]["mean_net"]

    ax.annotate(
        f"strongest net lag={best_lag}\nnet={best_net:.4f}",
        xy=(best_lag, best_net),
        xytext=(8, 18),
        textcoords="offset points",
        color=ACCENT_SOFT,
        fontsize=9,
        arrowprops=dict(arrowstyle="->", color=ACCENT_SOFT, lw=1.0),
    )

    style_ax(
        ax,
        title="Transfer Entropy Lag Sensitivity with Pair-Level SEM",
        xlabel="Lag, samples",
        ylabel="Mean TE / net TE, bits",
    )

    leg = ax.legend(loc="best", frameon=True)
    for text in leg.get_texts():
        text.set_color(ACCENT)

    save_fig(fig, path)
    return path


def plot_surrogate_significance(summary_df, path):
    if "normalized_surrogate_score_L_to_R" not in summary_df.columns:
        return None

    df = summary_df.copy()
    df["max_surrogate_score"] = np.nanmax(
        df[["normalized_surrogate_score_L_to_R", "normalized_surrogate_score_R_to_L"]].to_numpy(dtype=float),
        axis=1,
    )

    df = df.sort_values("max_surrogate_score", ascending=False).head(top_n_pairs)
    df = df.sort_values("max_surrogate_score", ascending=True)

    fig, ax = plt.subplots(figsize=(11, max(6.0, 0.28 * len(df))))

    y = np.arange(len(df))

    ax.barh(
        y,
        df["normalized_surrogate_score_L_to_R"],
        color=ACCENT,
        edgecolor=ACCENT_SOFT,
        alpha=0.82,
        label="L→R normalized surrogate score",
    )

    ax.barh(
        y,
        -df["normalized_surrogate_score_R_to_L"],
        color=ACCENT_DIM,
        edgecolor=ACCENT,
        alpha=0.86,
        label="R→L normalized surrogate score",
    )

    ax.axvline(1.0, color=WHITE, linestyle="--", linewidth=1.0, alpha=0.8)
    ax.axvline(-1.0, color=WHITE, linestyle="--", linewidth=1.0, alpha=0.8)
    ax.axvline(0.0, color=WHITE, linewidth=1.0, alpha=0.8)

    ax.set_yticks(y)
    ax.set_yticklabels(df["pair"], color=ACCENT, fontsize=8)

    style_ax(
        ax,
        title="Surrogate-Normalized Transfer Entropy Scores",
        xlabel="Score relative to circular-shift null; ±1 roughly exceeds null 95% distance",
        ylabel="Pair",
    )

    leg = ax.legend(loc="best", frameon=True)

    for text in leg.get_texts():
        text.set_color(ACCENT)

    save_fig(fig, path)
    return path


def custom_force_layout_from_summary(summary_df, threshold_quantile=0.75, iterations=700):
    channels = sorted(set(summary_df["left_channel"]).union(set(summary_df["right_channel"])))
    idx = {ch: i for i, ch in enumerate(channels)}
    n = len(channels)

    W = np.zeros((n, n), dtype=float)

    cutoff = np.nanquantile(summary_df["bidirectional_TE_sum"], threshold_quantile)

    for _, row in summary_df.iterrows():
        if row["bidirectional_TE_sum"] >= cutoff:
            i = idx[row["left_channel"]]
            j = idx[row["right_channel"]]
            W[i, j] = row["bidirectional_TE_sum"]
            W[j, i] = row["bidirectional_TE_sum"]

    theta = np.linspace(0, 2 * np.pi, n, endpoint=False)
    pos = np.column_stack([np.cos(theta), np.sin(theta)])

    for i, ch in enumerate(channels):
        if ch in left_channels:
            pos[i, 0] -= 0.7
        elif ch in right_channels:
            pos[i, 0] += 0.7

    pos += 0.03 * rng.normal(size=pos.shape)

    edges = edge_list_from_matrix(W)

    if len(edges) == 0:
        return channels, pos, W

    max_w = max(w for _, _, w in edges)

    area = 5.0
    k = np.sqrt(area / max(n, 1))
    temp = 0.06

    for _ in range(iterations):
        disp = np.zeros_like(pos)

        for i in range(n):
            delta = pos[i] - pos
            dist = np.linalg.norm(delta, axis=1) + 1e-9
            repulse = (k * k / dist)[:, None] * delta / dist[:, None]
            repulse[i] = 0
            disp[i] += np.sum(repulse, axis=0)

        for i, j, w in edges:
            delta = pos[i] - pos[j]
            dist = np.linalg.norm(delta) + 1e-9
            ww = w / (max_w + 1e-12)
            attract = (dist * dist / k) * delta / dist * (0.45 + 1.8 * ww)
            disp[i] -= attract
            disp[j] += attract

        for i, ch in enumerate(channels):
            if ch in left_channels:
                disp[i, 0] += -0.015
            elif ch in right_channels:
                disp[i, 0] += 0.015

        lengths = np.linalg.norm(disp, axis=1) + 1e-9
        pos += disp / lengths[:, None] * np.minimum(lengths, temp)[:, None]
        temp *= 0.995

    pos -= pos.mean(axis=0, keepdims=True)
    scale = np.max(np.linalg.norm(pos, axis=1))

    if scale > 0:
        pos /= scale

    return channels, pos, W


def edge_list_from_matrix(W):
    edges = []
    n = W.shape[0]

    for i in range(n):
        for j in range(i + 1, n):
            if W[i, j] > 0:
                edges.append((i, j, float(W[i, j])))

    return edges


def plot_te_network(summary_df, path):
    channels, pos, W = custom_force_layout_from_summary(summary_df, threshold_quantile=0.70)
    n = len(channels)

    fig, ax = plt.subplots(figsize=(11, 8))

    edges = edge_list_from_matrix(W)

    if len(edges) == 0:
        ax.text(0.5, 0.5, "No strong TE edges", transform=ax.transAxes, ha="center", va="center", color=ACCENT)
        style_ax(ax, title="Interhemispheric TE Network", grid=False)
        save_fig(fig, path)
        return path

    max_w = max(w for _, _, w in edges)

    for i, j, w in edges:
        ax.plot(
            [pos[i, 0], pos[j, 0]],
            [pos[i, 1], pos[j, 1]],
            color=ACCENT,
            linewidth=0.5 + 5.0 * w / (max_w + 1e-12),
            alpha=0.25 + 0.60 * w / (max_w + 1e-12),
            zorder=1,
        )

    strengths = W.sum(axis=1)
    max_strength = np.max(strengths) if np.max(strengths) > 0 else 1.0

    node_colors = []
    for ch in channels:
        if ch in left_channels:
            node_colors.append(ACCENT)
        elif ch in right_channels:
            node_colors.append(ACCENT_SOFT)
        else:
            node_colors.append(WHITE)

    ax.scatter(
        pos[:, 0],
        pos[:, 1],
        s=300 + 1000 * strengths / (max_strength + 1e-12),
        c=node_colors,
        edgecolors=WHITE,
        linewidths=0.9,
        alpha=0.92,
        zorder=3,
    )

    for i, ch in enumerate(channels):
        ax.text(
            pos[i, 0],
            pos[i, 1],
            ch,
            ha="center",
            va="center",
            color=BLACK,
            fontsize=8,
            fontweight="bold",
            zorder=4,
        )

    style_ax(
        ax,
        title="Custom Interhemispheric Transfer Entropy Network\nedge width = bidirectional TE strength",
        xlabel="Custom hemisphere-constrained force layout 1",
        ylabel="Custom hemisphere-constrained force layout 2",
        grid=False,
    )

    ax.set_xticks([])
    ax.set_yticks([])

    save_fig(fig, path)
    return path


def save_outputs(te_lr, te_rl, time_axis, summary_df, lag_df, source_description, elapsed_compute):
    te_lr_path = os.path.join(results_dir, f"te_left_to_right_{FAST_OPTION}.npy")
    te_rl_path = os.path.join(results_dir, f"te_right_to_left_{FAST_OPTION}.npy")
    time_axis_path = os.path.join(results_dir, f"te_time_axis_{FAST_OPTION}.npy")
    summary_csv = os.path.join(results_dir, f"te_pair_summary_{FAST_OPTION}.csv")
    lag_csv = os.path.join(results_dir, f"te_lag_sensitivity_{FAST_OPTION}.csv")
    summary_txt = os.path.join(results_dir, f"te_expert_summary_{FAST_OPTION}.txt")

    np.save(te_lr_path, te_lr)
    np.save(te_rl_path, te_rl)
    np.save(time_axis_path, time_axis)
    summary_df.to_csv(summary_csv, index=False)
    lag_df.to_csv(lag_csv, index=False)

    with open(summary_txt, "w") as f:
        f.write("Expert Interhemispheric Transfer Entropy Analysis\n")
        f.write("================================================\n\n")
        f.write("Transfer entropy implementation: custom NumPy discrete TE over quantile-binned windows.\n")
        f.write("Direction: TE(source -> target) = I(source_past; target_future | target_past).\n\n")
        f.write(f"base_dir: {base_dir}\n")
        f.write(f"data_source: {source_description}\n")
        f.write(f"FAST_OPTION: {FAST_OPTION}\n")
        f.write(f"window_size: {window_size}\n")
        f.write(f"step_size: {step_size}\n")
        f.write(f"n_bins: {n_bins}\n")
        f.write(f"lag_values: {lag_values}\n")
        f.write(f"permutation_count: {permutation_count}\n")
        f.write(f"max_pairs: {max_pairs}\n")
        f.write(f"max_points_per_channel: {max_points_per_channel}\n")
        f.write(f"embedding_scalar_mode: {EMBEDDING_TO_SCALAR_MODE}\n")
        f.write(f"elapsed_compute_sec: {elapsed_compute:.3f}\n\n")

        f.write("Top pairs by bidirectional TE:\n")
        for _, row in summary_df.sort_values("bidirectional_TE_sum", ascending=False).head(20).iterrows():
            f.write(
                f"{row['pair']}: "
                f"L->R={row['mean_TE_L_to_R']:.6g}, "
                f"R->L={row['mean_TE_R_to_L']:.6g}, "
                f"net={row['net_TE_L_minus_R']:.6g}, "
                f"AI={row['asymmetry_index']:.6g}, "
                f"lag={row['best_lag']}\n"
            )

    return {
        "te_left_to_right_npy": te_lr_path,
        "te_right_to_left_npy": te_rl_path,
        "te_time_axis_npy": time_axis_path,
        "pair_summary_csv": summary_csv,
        "lag_sensitivity_csv": lag_csv,
        "summary_txt": summary_txt,
    }


def print_summary(summary_df, lag_df, source_description, elapsed_compute):
    print("\n" + "=" * 92)
    print("EXPERT INTERHEMISPHERIC TRANSFER ENTROPY SUMMARY")
    print("=" * 92)
    print(f"Data source: {source_description}")
    print(f"FAST_OPTION: {FAST_OPTION}")
    print(f"window_size={window_size}, step_size={step_size}, n_bins={n_bins}, lags={lag_values}")
    print(f"Pairs computed: {len(summary_df)}")
    print(f"Compute elapsed: {elapsed_compute:.2f} sec")

    cols = [
        "pair",
        "best_lag",
        "mean_TE_L_to_R",
        "mean_TE_R_to_L",
        "net_TE_L_minus_R",
        "asymmetry_index",
        "bidirectional_TE_sum",
        "normalized_surrogate_score_L_to_R",
        "normalized_surrogate_score_R_to_L",
    ]

    cols = [c for c in cols if c in summary_df.columns]

    print("\nTop 15 by bidirectional TE strength:")
    print(summary_df.sort_values("bidirectional_TE_sum", ascending=False)[cols].head(15).to_string(index=False))

    print("\nTop 15 by absolute directional asymmetry:")
    print(summary_df.sort_values("abs_net_TE", ascending=False)[cols].head(15).to_string(index=False))

    if not lag_df.empty:
        print("\nLag sensitivity summary:")
        print(
            lag_df.groupby("lag")
            .agg(
                mean_L_to_R=("mean_TE_L_to_R", "mean"),
                mean_R_to_L=("mean_TE_R_to_L", "mean"),
                mean_net=("net_TE_L_minus_R", "mean"),
            )
            .reset_index()
            .to_string(index=False)
        )


pipeline_start = time.perf_counter()

print("Running expert interhemispheric transfer entropy analysis", flush=True)
print(f"FAST_OPTION: {FAST_OPTION}", flush=True)
print(f"window_size={window_size}, step_size={step_size}, n_bins={n_bins}", flush=True)
print(f"lag_values={lag_values}, permutation_count={permutation_count}", flush=True)

channel_series_raw, source_description = load_channel_series()
channel_series, common_len = trim_channel_series_to_common_length(
    channel_series_raw,
    max_points=max_points_per_channel,
)

available_left = [ch for ch in left_channels if ch in channel_series]
available_right = [ch for ch in right_channels if ch in channel_series]

print(f"Data source: {source_description}", flush=True)
print(f"Common length: {common_len}", flush=True)
print(f"Available left channels: {available_left}", flush=True)
print(f"Available right channels: {available_right}", flush=True)

te_lr_matrix, te_rl_matrix, te_time_axis, te_summary_df, te_lag_df, te_compute_elapsed = compute_transfer_entropy_hemispheres(
    channel_series,
    available_left,
    available_right,
)

output_paths = save_outputs(
    te_lr_matrix,
    te_rl_matrix,
    te_time_axis,
    te_summary_df,
    te_lag_df,
    source_description,
    te_compute_elapsed,
)

plot_paths = []

plot_paths.append(plot_te_temporal_heatmap(
    te_lr_matrix,
    te_time_axis,
    te_summary_df,
    os.path.join(plots_dir, f"01_te_temporal_heatmap_left_to_right_{FAST_OPTION}.png"),
    direction_label="Left to Right",
))

plot_paths.append(plot_te_temporal_heatmap(
    te_rl_matrix,
    te_time_axis,
    te_summary_df,
    os.path.join(plots_dir, f"02_te_temporal_heatmap_right_to_left_{FAST_OPTION}.png"),
    direction_label="Right to Left",
))

plot_paths.append(plot_te_strength_heatmap(
    te_summary_df,
    os.path.join(plots_dir, f"03_bidirectional_te_strength_heatmap_{FAST_OPTION}.png"),
))

plot_paths.append(plot_te_asymmetry_heatmap(
    te_summary_df,
    os.path.join(plots_dir, f"04_te_directional_asymmetry_heatmap_{FAST_OPTION}.png"),
))

plot_paths.append(plot_top_te_pairs(
    te_summary_df,
    os.path.join(plots_dir, f"05_top_te_pairs_bidirectional_and_directional_{FAST_OPTION}.png"),
    top_n=top_n_pairs,
))

plot_paths.append(plot_mean_te_timeseries(
    te_lr_matrix,
    te_rl_matrix,
    te_time_axis,
    os.path.join(plots_dir, f"06_mean_te_timeseries_directionality_{FAST_OPTION}.png"),
))

plot_paths.append(plot_lag_sensitivity(
    te_lag_df,
    os.path.join(plots_dir, f"07_te_lag_sensitivity_{FAST_OPTION}.png"),
))

if permutation_count > 0:
    plot_paths.append(plot_surrogate_significance(
        te_summary_df,
        os.path.join(plots_dir, f"08_te_surrogate_significance_{FAST_OPTION}.png"),
    ))

plot_paths.append(plot_te_network(
    te_summary_df,
    os.path.join(plots_dir, f"09_interhemispheric_te_network_{FAST_OPTION}.png"),
))

plot_paths = [p for p in plot_paths if p is not None]

pipeline_elapsed = time.perf_counter() - pipeline_start

print_summary(
    te_summary_df,
    te_lag_df,
    source_description,
    te_compute_elapsed,
)

print("\nSaved outputs:")
for k, v in output_paths.items():
    print(f"{k}: {v}")

print("\nSaved plots:")
for p in plot_paths:
    print(p)

print(f"\nTotal pipeline elapsed time: {pipeline_elapsed / 60:.2f} min")
print("Done.")
```

# Regional

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — Transfer entropy</h3><p>Transfer entropy is a conditional mutual information that measures whether the past of a source $X$ improves prediction of the next target value $Y_{t+1}$ beyond the target's own past. It is directional and depends on the history lengths, lag, estimator, and any discretization used by the implementation.</p><p><b>Cell role.</b> This code applies that theory to the notebook's existing EEG-derived data and preserves the surrounding notebook's plotting/output conventions. Numerical shortcuts selected by a speed mode change sampling density, iteration counts, or visualization resolution; they do not change the defining formula above.</p></div>

$$T_{X\\to Y}=\\sum p(y_{t+1},y_t^{(k)},x_t^{(l)})\\log\\frac{p(y_{t+1}\\mid y_t^{(k)},x_t^{(l)})}{p(y_{t+1}\\mid y_t^{(k)})}.$$

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Mutual information measures how much knowing one variable reduces uncertainty about another. Unlike linear correlation, it can capture nonlinear statistical dependence. In delay-selection problems, mutual information between a signal and a delayed copy is examined as the delay changes; a local minimum is often used as a compromise between coordinates that are too redundant and coordinates that are almost unrelated. The estimate is sensitive to finite sample size and to how probability densities are estimated.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values. The visualization is a way of displaying already-computed numbers. Color, ordering, or ranking can make patterns easier to see but does not create additional statistical evidence beyond the values and uncertainty measures being plotted.</p></div>


```python
import os
import io
import re
import zipfile
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
from pyinform import transfer_entropy

# =======================================================
# Regional Transfer Entropy from 2D embeddings
# Robust archive discovery + black/cyan visualization
# =======================================================

base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
zip_file_path = os.path.join(base_dir, "2d_embedded_data.zip")

if not os.path.isfile(zip_file_path):
    raise FileNotFoundError(f"Could not find archive:\n{zip_file_path}")

eeg_channels = [
    "Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8",
    "FC5", "FC1", "FC2", "FC6", "M1", "T7", "C3", "Cz",
    "C4", "T8", "M2", "CP5", "CP1", "CP2", "CP6",
    "P7", "P3", "Pz", "P4", "P8", "POz", "O1", "Oz", "O2"
]

FAST_OPTION = "fast"  # options: "ultra", "fast", "balanced", "full"
if FAST_OPTION not in {"ultra", "fast", "balanced", "full"}:
    raise ValueError("FAST_OPTION must be one of: ultra, fast, balanced, full")

desired_length = {"ultra": 50_000, "fast": 150_000, "balanced": 500_000, "full": 4_227_688}[FAST_OPTION]
num_bins = {"ultra": 16, "fast": 32, "balanced": 64, "full": 1000}[FAST_OPTION]

ACCENT = "cyan"
BG = "black"
TE_CMAP = LinearSegmentedColormap.from_list("black_cyan", [(0.0, BG), (1.0, ACCENT)])

plt.rcParams.update({
    "figure.facecolor": BG,
    "axes.facecolor": BG,
    "savefig.facecolor": BG,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT,
})

def style_ax(ax):
    ax.set_facecolor(BG)
    ax.tick_params(colors=ACCENT)
    ax.xaxis.label.set_color(ACCENT)
    ax.yaxis.label.set_color(ACCENT)
    ax.title.set_color(ACCENT)
    for spine in ax.spines.values():
        spine.set_color(ACCENT)
    ax.grid(True, color=ACCENT, alpha=0.18)

def normalized_name(path):
    return re.sub(r"[^a-z0-9]+", "", path.lower())

def load_member(zipf, member):
    return np.load(io.BytesIO(zipf.read(member)), allow_pickle=True)

def find_members(zipf, channels):
    members = [m for m in zipf.namelist() if not m.endswith("/") and m.lower().endswith(".npy")]
    found = {}

    for channel in channels:
        ch = normalized_name(channel)
        candidates = []

        for member in members:
            base = normalized_name(os.path.splitext(os.path.basename(member))[0])

            score = None
            if base == ch:
                score = 0
            elif base in {
                f"2dembedded{ch}",
                f"2dembedded{ch}data",
                f"{ch}2dembedded",
                f"{ch}2dembedding",
                f"2dembedding{ch}",
                f"{ch}embedding",
                f"embedding{ch}",
                f"{ch}data",
            }:
                score = 1
            elif ch in base:
                score = 2

            if score is not None:
                candidates.append((score, len(member), member))

        if candidates:
            candidates.sort()
            found[channel] = candidates[0][2]

    return found, members

embedded = {}

with zipfile.ZipFile(zip_file_path, "r") as zipf:
    found, members = find_members(zipf, eeg_channels)
    print(f"ZIP contains {len(members)} .npy files.")

    missing = [ch for ch in eeg_channels if ch not in found]
    if missing:
        raise FileNotFoundError("Missing EEG embeddings: " + ", ".join(missing))

    print("\nFound individual files for all EEG channels.")

    for channel in eeg_channels:
        member = found[channel]
        arr = np.asarray(load_member(zipf, member), dtype=float)

        if arr.ndim != 2:
            raise ValueError(f"{member} has shape {arr.shape}; expected 2-D embedding.")

        if arr.shape[0] <= 4 and arr.shape[0] < arr.shape[1]:
            arr = arr.T

        if arr.shape[1] < 1:
            raise ValueError(f"{member} has no usable embedding coordinate.")

        embedded[channel] = arr
        print(f"{channel:>3s}: {member} | shape={arr.shape}")

common_length = min(len(embedded[ch]) for ch in eeg_channels)
used_length = min(desired_length, common_length)

embedding_data_list_2D_trimmed = [embedded[ch][:used_length] for ch in eeg_channels]

print(f"\nLoaded all {len(eeg_channels)} channels.")
print(f"Common available length: {common_length:,}")
print(f"Length used ({FAST_OPTION}): {used_length:,}")

region_channels = {
    "Frontal": ["Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8"],
    "Temporal": ["T7", "T8"],
    "Parietal": ["CP5", "CP1", "CP2", "CP6", "P7", "P3", "Pz", "P4", "P8"],
    "Occipital": ["O1", "Oz", "O2"],
}

region_series = {}

for region, channels in region_channels.items():
    idx = [eeg_channels.index(ch) for ch in channels]
    avg_embedding = np.mean([embedding_data_list_2D_trimmed[i] for i in idx], axis=0)
    region_series[region] = avg_embedding[:, 0]

def bin_data(data, bins):
    data = np.asarray(data, dtype=float).reshape(-1)
    finite = np.isfinite(data)

    if not np.any(finite):
        raise ValueError("Cannot bin a series with no finite values.")

    if not np.all(finite):
        data = data.copy()
        data[~finite] = np.nanmedian(data[finite])

    _, edges = np.histogram(data, bins=bins)
    return np.clip(np.digitize(data, edges[1:-1], right=False), 0, bins - 1).astype(np.int32)

region_data_binned = {
    region: bin_data(series, num_bins)
    for region, series in region_series.items()
}

k = 1
results_dir = os.path.join(base_dir, "results", "transfer_entropy_regional")
os.makedirs(results_dir, exist_ok=True)

TE_results = {}

for source_region, source_data in region_data_binned.items():
    for target_region, target_data in region_data_binned.items():
        if source_region == target_region:
            continue

        TE_value = transfer_entropy(source_data.tolist(), target_data.tolist(), k=k)
        key = f"{source_region}_to_{target_region}"
        TE_results[key] = float(TE_value)

        print(
            f"Transfer Entropy "
            f"{source_region:>9s} -> {target_region:<9s}: "
            f"{TE_value:.8g}"
        )

results_file_path = os.path.join(results_dir, "regional_transfer_entropy_results.npy")
np.save(results_file_path, TE_results, allow_pickle=True)

print("\nTransfer entropy results saved to:")
print(results_file_path)

# Plot only when TWO OR MORE finite result values exist.
regions = list(region_data_binned.keys())
TE_matrix = np.full((len(regions), len(regions)), np.nan, dtype=float)

for i, source_region in enumerate(regions):
    for j, target_region in enumerate(regions):
        if source_region != target_region:
            TE_matrix[i, j] = TE_results.get(
                f"{source_region}_to_{target_region}",
                np.nan,
            )

if np.isfinite(TE_matrix).sum() >= 2:
    fig, ax = plt.subplots(figsize=(7.5, 6.2), facecolor=BG)
    style_ax(ax)

    im = ax.imshow(
        TE_matrix,
        aspect="equal",
        cmap=TE_CMAP,
        interpolation="nearest",
    )

    ax.set_xticks(np.arange(len(regions)))
    ax.set_xticklabels(regions, rotation=30, ha="right")
    ax.set_yticks(np.arange(len(regions)))
    ax.set_yticklabels(regions)
    ax.set_xlabel("Target region")
    ax.set_ylabel("Source region")
    ax.set_title("Regional Transfer Entropy")

    for i in range(len(regions)):
        for j in range(len(regions)):
            if np.isfinite(TE_matrix[i, j]):
                ax.text(
                    j,
                    i,
                    f"{TE_matrix[i, j]:.3f}",
                    ha="center",
                    va="center",
                    color=ACCENT,
                    fontsize=9,
                )

    cbar = fig.colorbar(im, ax=ax)
    cbar.set_label("Transfer entropy")
    cbar.outline.set_edgecolor(ACCENT)
    cbar.ax.tick_params(color=ACCENT, labelcolor=ACCENT)

    fig.tight_layout()

    plot_path = os.path.join(
        results_dir,
        "regional_transfer_entropy_black_cyan.png",
    )

    fig.savefig(
        plot_path,
        dpi=180,
        bbox_inches="tight",
        facecolor=BG,
    )

    plt.show()
    plt.close(fig)

    print("Saved plot:", plot_path)
```

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — numerical support logic</h3><p>This cell implements numerical support logic for the surrounding analysis. Where it computes descriptive summaries, they are finite-sample quantities evaluated directly from the data. The cell does not introduce an independent physical law; its correctness therefore depends mainly on consistent indexing, normalization, units, and use of the estimator defined by the surrounding section.</p><p><b>Cell role.</b> The implementation uses the same dataset, channel ordering, result directories, and plotting conventions as the rest of this notebook.</p></div>

The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. ### Full Granularity

This section evaluates the entropy estimators at the native analysis granularity before any optional multiscale coarse-graining. Computationally expensive estimators later reuse the notebook-wide `full / balanced / fast / ultra` controls; changing speed mode changes sampling/iteration budgets, not the entropy definitions.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — Transfer entropy</h3><p>Transfer entropy is a conditional mutual information that measures whether the past of a source $X$ improves prediction of the next target value $Y_{t+1}$ beyond the target's own past. It is directional and depends on the history lengths, lag, estimator, and any discretization used by the implementation.</p><p><b>Cell role.</b> This code applies that theory to the notebook's existing EEG-derived data and preserves the surrounding notebook's plotting/output conventions. Numerical shortcuts selected by a speed mode change sampling density, iteration counts, or visualization resolution; they do not change the defining formula above.</p></div>

$$T_{X\\to Y}=\\sum p(y_{t+1},y_t^{(k)},x_t^{(l)})\\log\\frac{p(y_{t+1}\\mid y_t^{(k)},x_t^{(l)})}{p(y_{t+1}\\mid y_t^{(k)})}.$$

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Mutual information measures how much knowing one variable reduces uncertainty about another. Unlike linear correlation, it can capture nonlinear statistical dependence. In delay-selection problems, mutual information between a signal and a delayed copy is examined as the delay changes; a local minimum is often used as a compromise between coordinates that are too redundant and coordinates that are almost unrelated. The estimate is sensitive to finite sample size and to how probability densities are estimated.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values. The visualization is a way of displaying already-computed numbers. Color, ordering, or ranking can make patterns easier to see but does not create additional statistical evidence beyond the values and uncertainty measures being plotted.</p></div>


```python
import os
import io
import re
import zipfile
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
from pyinform import transfer_entropy

# =======================================================
# Full-granularity channel-to-channel Transfer Entropy
# Robust archive discovery + black/cyan visualization
# =======================================================

base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
zip_file_path = os.path.join(base_dir, "2d_embedded_data.zip")
results_dir = os.path.join(base_dir, "results", "transfer_entropy_full_granularity")
results_file_name = "full_granularity_transfer_entropy_results.npy"

eeg_channels = [
    "Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8",
    "FC5", "FC1", "FC2", "FC6", "M1", "T7", "C3", "Cz",
    "C4", "T8", "M2", "CP5", "CP1", "CP2", "CP6",
    "P7", "P3", "Pz", "P4", "P8", "POz", "O1", "Oz", "O2"
]

FAST_OPTION = "ultra"  # options: "ultra", "fast", "balanced", "full"
if FAST_OPTION not in {"ultra", "fast", "balanced", "full"}:
    raise ValueError("FAST_OPTION must be one of: ultra, fast, balanced, full")

desired_length = {"ultra": 30_000, "fast": 80_000, "balanced": 250_000, "full": 4_227_688}[FAST_OPTION]
num_bins = {"ultra": 12, "fast": 24, "balanced": 48, "full": 1000}[FAST_OPTION]
source_limit = {"ultra": 8, "fast": 16, "balanced": 24, "full": len(eeg_channels)}[FAST_OPTION]
k = 1

ACCENT = "cyan"
BG = "black"
TE_CMAP = LinearSegmentedColormap.from_list("black_cyan_full_te", [(0.0, BG), (1.0, ACCENT)])

plt.rcParams.update({
    "figure.facecolor": BG,
    "axes.facecolor": BG,
    "savefig.facecolor": BG,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT,
})

def style_ax(ax):
    ax.set_facecolor(BG)
    ax.tick_params(colors=ACCENT)
    ax.xaxis.label.set_color(ACCENT)
    ax.yaxis.label.set_color(ACCENT)
    ax.title.set_color(ACCENT)
    for spine in ax.spines.values():
        spine.set_color(ACCENT)

def normalized_name(path):
    return re.sub(r"[^a-z0-9]+", "", path.lower())

def load_member(zipf, member):
    return np.load(io.BytesIO(zipf.read(member)), allow_pickle=True)

def find_members(zipf, channels):
    members = [m for m in zipf.namelist() if not m.endswith("/") and m.lower().endswith(".npy")]
    found = {}

    for channel in channels:
        ch = normalized_name(channel)
        candidates = []

        for member in members:
            base = normalized_name(os.path.splitext(os.path.basename(member))[0])

            score = None
            if base == ch:
                score = 0
            elif base in {
                f"2dembedded{ch}",
                f"2dembedded{ch}data",
                f"{ch}2dembedded",
                f"{ch}2dembedding",
                f"2dembedding{ch}",
                f"{ch}embedding",
                f"embedding{ch}",
                f"{ch}data",
            }:
                score = 1
            elif ch in base:
                score = 2

            if score is not None:
                candidates.append((score, len(member), member))

        if candidates:
            candidates.sort()
            found[channel] = candidates[0][2]

    return found

def bin_data(data, bins):
    data = np.asarray(data, dtype=float).reshape(-1)
    finite = np.isfinite(data)

    if not np.any(finite):
        raise ValueError("Cannot bin a series with no finite values.")

    if not np.all(finite):
        data = data.copy()
        data[~finite] = np.nanmedian(data[finite])

    _, edges = np.histogram(data, bins=bins)
    return np.clip(np.digitize(data, edges[1:-1], right=False), 0, bins - 1).astype(np.int32)

os.makedirs(results_dir, exist_ok=True)

if not os.path.isfile(zip_file_path):
    raise FileNotFoundError(f"Could not find archive:\n{zip_file_path}")

channel_embeddings = {}

with zipfile.ZipFile(zip_file_path, "r") as zipf:
    found = find_members(zipf, eeg_channels)

    missing = [ch for ch in eeg_channels if ch not in found]
    if missing:
        raise FileNotFoundError(
            "Could not map all EEG channel embeddings. Missing: "
            + ", ".join(missing)
        )

    for channel in eeg_channels:
        arr = np.asarray(load_member(zipf, found[channel]), dtype=float)

        if arr.ndim != 2:
            raise ValueError(
                f"{found[channel]} has shape {arr.shape}; expected 2-D array."
            )

        if arr.shape[0] <= 4 and arr.shape[0] < arr.shape[1]:
            arr = arr.T

        channel_embeddings[channel] = arr

common_length = min(len(channel_embeddings[ch]) for ch in eeg_channels)
used_length = min(desired_length, common_length)

channel_binned = [
    bin_data(channel_embeddings[ch][:used_length, 0], num_bins)
    for ch in eeg_channels
]

print(
    f"FAST_OPTION={FAST_OPTION} | "
    f"used_length={used_length:,} | "
    f"bins={num_bins} | "
    f"source channels={source_limit}"
)

TE_results = {}

sources = eeg_channels[:source_limit]
targets = eeg_channels

for i, source_channel in enumerate(sources):
    for j, target_channel in enumerate(targets):
        if source_channel == target_channel:
            continue

        TE = transfer_entropy(
            channel_binned[i].tolist(),
            channel_binned[j].tolist(),
            k=k,
        )

        key = f"{source_channel}_to_{target_channel}"
        TE_results[key] = float(TE)

        print(
            f"Transfer Entropy "
            f"{source_channel:>3s} -> "
            f"{target_channel:<3s}: "
            f"{TE:.8g}"
        )

results_file_path = os.path.join(results_dir, results_file_name)
np.save(results_file_path, TE_results, allow_pickle=True)

print("\nTransfer entropy results saved to:")
print(results_file_path)

# Plot only when TWO OR MORE finite result values exist.
TE_matrix = np.full((len(sources), len(targets)), np.nan, dtype=float)

for i, source_channel in enumerate(sources):
    for j, target_channel in enumerate(targets):
        if source_channel != target_channel:
            TE_matrix[i, j] = TE_results.get(
                f"{source_channel}_to_{target_channel}",
                np.nan,
            )

if np.isfinite(TE_matrix).sum() >= 2:
    fig, ax = plt.subplots(
        figsize=(15, max(5.5, 0.42 * len(sources))),
        facecolor=BG,
    )
    style_ax(ax)

    im = ax.imshow(
        TE_matrix,
        aspect="auto",
        cmap=TE_CMAP,
        interpolation="nearest",
    )

    ax.set_xticks(np.arange(len(targets)))
    ax.set_xticklabels(targets, rotation=90, fontsize=8)
    ax.set_yticks(np.arange(len(sources)))
    ax.set_yticklabels(sources)
    ax.set_xlabel("Target channel")
    ax.set_ylabel("Source channel")
    ax.set_title(
        f"Channel-to-channel Transfer Entropy — {FAST_OPTION.upper()}"
    )

    cbar = fig.colorbar(im, ax=ax)
    cbar.set_label("Transfer entropy")
    cbar.outline.set_edgecolor(ACCENT)
    cbar.ax.tick_params(color=ACCENT, labelcolor=ACCENT)

    fig.tight_layout()

    plot_path = os.path.join(
        results_dir,
        f"transfer_entropy_heatmap_{FAST_OPTION}.png",
    )

    fig.savefig(
        plot_path,
        dpi=180,
        bbox_inches="tight",
        facecolor=BG,
    )

    plt.show()
    plt.close(fig)

    print("Saved plot:", plot_path)
```

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — Discrete Fourier analysis</h3><p>The discrete Fourier transform decomposes a finite sequence into complex sinusoids. The squared magnitude is related to spectral power, while the complex argument carries phase. Correct frequency axes require the sampling frequency used for the data actually supplied to the transform.</p><p><b>Cell role.</b> This code applies that theory to the notebook's existing EEG-derived data and preserves the surrounding notebook's plotting/output conventions. Numerical shortcuts selected by a speed mode change sampling density, iteration counts, or visualization resolution; they do not change the defining formula above.</p></div>

$$X_k=\\sum_{n=0}^{N-1}x_n e^{-i2\\pi kn/N}.$$

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The Fourier transform represents a time-domain signal as a weighted sum of sinusoidal components. Each Fourier coefficient is complex: its magnitude describes the strength of a frequency component, while its angle describes phase relative to the chosen time origin. The discrete Fourier transform evaluates this representation only at a finite set of frequencies determined by the sampling rate and record length. It describes global frequency content over the analyzed window and does not by itself show when a frequency occurs.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. An exponential describes multiplicative growth or decay. In dynamical systems, a positive exponent corresponds to growth with time and a negative exponent to decay. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values. The visualization is a way of displaying already-computed numbers. Color, ordering, or ranking can make patterns easier to see but does not create additional statistical evidence beyond the values and uncertainty measures being plotted.</p></div>


```python
import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

# -------------------------------------------------------
# CONFIG
# -------------------------------------------------------
base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
lyap_dir = os.path.join(base_dir, "results", "lyapunov_custom")
out_dir = os.path.join(base_dir, "results", "ks_entropy_from_lyapunov")
plots_dir = os.path.join(out_dir, "plots")
os.makedirs(out_dir, exist_ok=True)
os.makedirs(plots_dir, exist_ok=True)

ACCENT = "cyan"
BG = "black"
eps = 1e-12
_KS_CMAP = LinearSegmentedColormap.from_list("black_cyan_ks", [(0.0, BG), (1.0, ACCENT)])

# IMPORTANT:
# If your Lyapunov code used DT = 1.0, then KS entropy here is "per sample".
# If your Lyapunov code used DT = 1/fs, then KS entropy here is "per second".
KS_UNITS = "per sample"

plt.rcParams.update({
    "figure.facecolor": BG,
    "axes.facecolor": BG,
    "savefig.facecolor": BG,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT,
    "legend.facecolor": BG,
    "legend.edgecolor": ACCENT,
})

# -------------------------------------------------------
# STYLE
# -------------------------------------------------------
def style_ax(ax):
    ax.set_facecolor(BG)
    ax.tick_params(colors=ACCENT)
    ax.grid(True, alpha=0.20, color=ACCENT)
    for spine in ax.spines.values():
        spine.set_color(ACCENT)

# -------------------------------------------------------
# HELPERS
# -------------------------------------------------------
def find_latest_lyapunov_csv(results_dir):
    csvs = sorted(glob.glob(os.path.join(results_dir, "lyap_spectrum_*.csv")))
    if not csvs:
        raise FileNotFoundError(f"No lyap_spectrum_*.csv files found in: {results_dir}")
    return max(csvs, key=os.path.getmtime)

def get_exp_cols(df):
    exp_cols = [c for c in df.columns if str(c).startswith("Exp")]
    exp_cols = sorted(exp_cols, key=lambda c: int(str(c).replace("Exp", "")))
    return exp_cols

def infer_channel_from_stem(stem):
    stem = str(stem)
    if "dembedded_" in stem:
        return stem.split("dembedded_", 1)[1]
    return stem

def infer_embedding_dimension(row):
    if "embedding_dimension" in row.index and pd.notna(row["embedding_dimension"]):
        try:
            return int(row["embedding_dimension"])
        except Exception:
            pass

    stem = str(row["stem"])
    if "dembedded_" in stem:
        left = stem.split("dembedded_", 1)[0]
        try:
            return int(left)
        except Exception:
            pass

    if "state_dim" in row.index and pd.notna(row["state_dim"]):
        try:
            return int(row["state_dim"])
        except Exception:
            pass

    return np.nan

def row_exponents(row, exp_cols):
    vals = []
    for c in exp_cols:
        if c in row.index and pd.notna(row[c]):
            vals.append(float(row[c]))
    return np.asarray(vals, dtype=float)

def ks_entropy_estimate(lyapunov_exponents):
    """
    Pesin-type estimate:
        h_KS ≈ sum of positive Lyapunov exponents
    """
    exps = np.asarray(lyapunov_exponents, dtype=float)
    exps = exps[np.isfinite(exps)]
    if exps.size == 0:
        return np.nan
    return float(np.sum(exps[exps > 0]))

def positive_lyapunov_count(lyapunov_exponents):
    exps = np.asarray(lyapunov_exponents, dtype=float)
    exps = exps[np.isfinite(exps)]
    return int(np.sum(exps > 0))

def largest_lyapunov(lyapunov_exponents):
    exps = np.asarray(lyapunov_exponents, dtype=float)
    exps = exps[np.isfinite(exps)]
    if exps.size == 0:
        return np.nan
    return float(np.max(exps))

# -------------------------------------------------------
# LOAD LATEST LYAPUNOV CSV
# -------------------------------------------------------
lyap_csv = find_latest_lyapunov_csv(lyap_dir)
df = pd.read_csv(lyap_csv)

print("Using Lyapunov CSV:")
print(lyap_csv)

if "status" in df.columns:
    df = df[df["status"] == "ok"].copy()

if df.empty:
    raise ValueError("No successful rows found in the Lyapunov CSV.")

exp_cols = get_exp_cols(df)
if not exp_cols:
    raise ValueError("No Exp columns found in the Lyapunov CSV.")

print("\nDetected exponent columns:")
print(exp_cols)

# -------------------------------------------------------
# BUILD KS TABLE
# -------------------------------------------------------
records = []

for _, row in df.iterrows():
    stem = row["stem"] if "stem" in row.index else ""
    channel = infer_channel_from_stem(stem)
    emb_dim = infer_embedding_dimension(row)
    exps = row_exponents(row, exp_cols)

    if exps.size == 0:
        continue

    exps_sorted = np.sort(exps)[::-1]
    ks_est = ks_entropy_estimate(exps_sorted)
    n_pos = positive_lyapunov_count(exps_sorted)
    lmax = largest_lyapunov(exps_sorted)

    rec = {
        "file": row["file"] if "file" in row.index else "",
        "stem": stem,
        "channel": channel,
        "embedding_dimension": emb_dim,
        "state_dim": int(row["state_dim"]) if "state_dim" in row.index and pd.notna(row["state_dim"]) else np.nan,
        "n_samples": int(row["n_samples"]) if "n_samples" in row.index and pd.notna(row["n_samples"]) else np.nan,
        "steps_used": int(row["steps_used"]) if "steps_used" in row.index and pd.notna(row["steps_used"]) else np.nan,
        "n_exponents_used": int(exps_sorted.size),
        "largest_lyapunov_exponent": float(lmax),
        "n_positive_exponents": int(n_pos),
        "sum_positive_exponents": float(ks_est),
        "ks_entropy_estimate": float(ks_est),
        "sum_all_exponents": float(np.sum(exps_sorted)),
    }

    for i, val in enumerate(exps_sorted, start=1):
        rec[f"sorted_Exp{i}"] = float(val)

    records.append(rec)

ks_df = pd.DataFrame(records)

if ks_df.empty:
    raise ValueError("No KS entropy records were produced.")

ks_df = ks_df.sort_values(["channel", "embedding_dimension"]).reset_index(drop=True)

# -------------------------------------------------------
# SUMMARIES
# -------------------------------------------------------
all_dims = list(range(1, 11))

dim_summary_df = (
    ks_df.groupby("embedding_dimension", dropna=True)
    .agg(
        n_records=("ks_entropy_estimate", "count"),
        mean_ks_entropy=("ks_entropy_estimate", "mean"),
        std_ks_entropy=("ks_entropy_estimate", "std"),
        mean_lmax=("largest_lyapunov_exponent", "mean"),
        mean_positive_count=("n_positive_exponents", "mean"),
        mean_sum_all_exponents=("sum_all_exponents", "mean"),
    )
    .reset_index()
)

dim_summary_full = pd.DataFrame({"embedding_dimension": all_dims}).merge(
    dim_summary_df,
    on="embedding_dimension",
    how="left"
)

ks_matrix = ks_df.pivot_table(
    index="channel",
    columns="embedding_dimension",
    values="ks_entropy_estimate",
    aggfunc="mean"
).reindex(columns=all_dims)

poscount_matrix = ks_df.pivot_table(
    index="channel",
    columns="embedding_dimension",
    values="n_positive_exponents",
    aggfunc="mean"
).reindex(columns=all_dims)

# highest available dimension per channel
highest_dim_df = (
    ks_df.sort_values(["channel", "embedding_dimension"])
    .groupby("channel", as_index=False)
    .tail(1)
    .reset_index(drop=True)
)

# -------------------------------------------------------
# SAVE TABLES
# -------------------------------------------------------
ks_csv = os.path.join(out_dir, "ks_entropy_per_file.csv")
dim_csv = os.path.join(out_dir, "ks_entropy_by_dimension.csv")
matrix_csv = os.path.join(out_dir, "ks_entropy_channel_dimension_matrix.csv")
poscount_csv = os.path.join(out_dir, "positive_exponent_count_channel_dimension_matrix.csv")
highest_csv = os.path.join(out_dir, "ks_entropy_highest_available_dimension_per_channel.csv")

ks_df.to_csv(ks_csv, index=False)
dim_summary_full.to_csv(dim_csv, index=False)
ks_matrix.to_csv(matrix_csv)
poscount_matrix.to_csv(poscount_csv)
highest_dim_df.to_csv(highest_csv, index=False)

summary_txt = os.path.join(out_dir, "ks_entropy_summary.txt")
with open(summary_txt, "w") as f:
    f.write("Kolmogorov-Sinai Entropy Estimate from Saved Lyapunov Spectra\n")
    f.write("===========================================================\n\n")
    f.write(f"Source CSV: {lyap_csv}\n")
    f.write(f"Rows analyzed: {len(ks_df)}\n")
    f.write(f"Exponent columns used: {', '.join(exp_cols)}\n")
    f.write(f"KS entropy units: {KS_UNITS}\n\n")
    f.write("Estimate used:\n")
    f.write("  h_KS ≈ sum of positive Lyapunov exponents\n\n")

    f.write("Mean KS entropy by embedding dimension:\n")
    for _, row in dim_summary_full.iterrows():
        d = int(row["embedding_dimension"])
        nrec = row["n_records"]
        if pd.isna(nrec):
            f.write(f"  {d}D: no data\n")
        else:
            f.write(
                f"  {d}D: n={int(nrec)}, "
                f"mean_KS={row['mean_ks_entropy']:.6f}, "
                f"std_KS={0.0 if pd.isna(row['std_ks_entropy']) else row['std_ks_entropy']:.6f}, "
                f"mean_Lmax={row['mean_lmax']:.6f}, "
                f"mean_positive_count={row['mean_positive_count']:.3f}, "
                f"mean_sum_all_exp={row['mean_sum_all_exponents']:.6f}\n"
            )

# -------------------------------------------------------
# PRINT
# -------------------------------------------------------
print("\nKS entropy by embedding dimension:")
print(dim_summary_full)

print("\nSample per-file KS results:")
print(ks_df.head())

# -------------------------------------------------------
# PLOTS
# -------------------------------------------------------

# Plot 1: mean KS entropy vs embedding dimension
mask = dim_summary_full["mean_ks_entropy"].notna().values
x = dim_summary_full.loc[mask, "embedding_dimension"].values
y = dim_summary_full.loc[mask, "mean_ks_entropy"].values
s = dim_summary_full.loc[mask, "std_ks_entropy"].fillna(0.0).values

if np.isfinite(y).sum() >= 2:
    fig, ax = plt.subplots(figsize=(8, 5), facecolor=BG)
    style_ax(ax)

    ax.plot(x, y, marker="o", color=ACCENT, linewidth=1.5)
    ax.fill_between(x, y - s, y + s, color=ACCENT, alpha=0.15)
    ax.set_title("Mean KS Entropy Estimate vs Embedding Dimension")
    ax.set_xlabel("Embedding Dimension")
    ax.set_ylabel(f"KS Entropy Estimate ({KS_UNITS})")
    ax.set_xticks(all_dims)

    plot1_path = os.path.join(plots_dir, "mean_ks_entropy_vs_embedding_dimension.png")
    plt.tight_layout()
    plt.savefig(plot1_path, dpi=180, bbox_inches="tight", facecolor=BG)
    plt.show()
    plt.close()

# Plot 2: mean positive-exponent count vs embedding dimension
mask = dim_summary_full["mean_positive_count"].notna().values
x2 = dim_summary_full.loc[mask, "embedding_dimension"].values
y2 = dim_summary_full.loc[mask, "mean_positive_count"].values

if np.isfinite(y2).sum() >= 2:
    fig, ax = plt.subplots(figsize=(8, 5), facecolor=BG)
    style_ax(ax)

    ax.plot(
        x2,
        y2,
        marker="o",
        color=ACCENT,
        linewidth=1.5,
    )
    ax.set_title("Mean Number of Positive Lyapunov Exponents vs Embedding Dimension")
    ax.set_xlabel("Embedding Dimension")
    ax.set_ylabel("Mean Positive Exponent Count")
    ax.set_xticks(all_dims)

    plot2_path = os.path.join(plots_dir, "mean_positive_exponent_count_vs_embedding_dimension.png")
    plt.tight_layout()
    plt.savefig(plot2_path, dpi=180, bbox_inches="tight", facecolor=BG)
    plt.show()
    plt.close()

# Plot 3: channel x dimension heatmap of KS entropy
heat_vals = ks_matrix.values.astype(float)

if np.isfinite(heat_vals).sum() >= 2:
    plot_vals = np.where(np.isnan(heat_vals), 0.0, heat_vals)

    fig, ax = plt.subplots(figsize=(12, 10), facecolor=BG)
    style_ax(ax)

    im = ax.imshow(
        plot_vals,
        aspect="auto",
        cmap=_KS_CMAP,
        interpolation="nearest",
    )
    ax.set_title("KS Entropy Estimate by Channel and Embedding Dimension")
    ax.set_xlabel("Embedding Dimension")
    ax.set_ylabel("Channel")
    ax.set_xticks(np.arange(len(all_dims)))
    ax.set_xticklabels(all_dims)
    ax.set_yticks(np.arange(len(ks_matrix.index)))
    ax.set_yticklabels(ks_matrix.index)

    cbar = plt.colorbar(im, ax=ax)
    cbar.outline.set_edgecolor(ACCENT)
    cbar.ax.tick_params(color=ACCENT, labelcolor=ACCENT)
    for label in cbar.ax.get_yticklabels():
        label.set_color(ACCENT)

    plot3_path = os.path.join(plots_dir, "ks_entropy_heatmap_channel_by_dimension.png")
    plt.tight_layout()
    plt.savefig(plot3_path, dpi=180, bbox_inches="tight", facecolor=BG)
    plt.show()
    plt.close()

# Plot 4: channel trajectories across dimensions
trajectory_count = 0
for ch in ks_matrix.index:
    yy = ks_matrix.loc[ch].values.astype(float)
    if np.isfinite(yy).sum() >= 2:
        trajectory_count += 1

if trajectory_count >= 1:
    fig, ax = plt.subplots(figsize=(12, 8), facecolor=BG)
    style_ax(ax)

    for ch in ks_matrix.index:
        yy = ks_matrix.loc[ch].values.astype(float)
        xx = np.array(all_dims, dtype=float)
        mask = np.isfinite(yy)

        # Never draw a one-number trajectory.
        if np.sum(mask) >= 2:
            ax.plot(
                xx[mask],
                yy[mask],
                linewidth=1.0,
                alpha=0.7,
                color=ACCENT,
            )

    ax.set_title("KS Entropy Trajectories Across Embedding Dimensions")
    ax.set_xlabel("Embedding Dimension")
    ax.set_ylabel(f"KS Entropy Estimate ({KS_UNITS})")
    ax.set_xticks(all_dims)

    plot4_path = os.path.join(plots_dir, "ks_entropy_trajectories_across_dimensions.png")
    plt.tight_layout()
    plt.savefig(plot4_path, dpi=180, bbox_inches="tight", facecolor=BG)
    plt.show()
    plt.close()

print(f"\nSaved per-file KS CSV: {ks_csv}")
print(f"Saved dimension summary CSV: {dim_csv}")
print(f"Saved KS matrix CSV: {matrix_csv}")
print(f"Saved positive-count matrix CSV: {poscount_csv}")
print(f"Saved highest-dimension summary CSV: {highest_csv}")
print(f"Saved summary TXT: {summary_txt}")
print(f"Saved plots to: {plots_dir}")
```

<h3 style="color:#E8E8E8;margin-top:0;">Theory and role — shared entropy estimators for methods 1–64</h3>
<p>This foundation cell defines the reusable estimators used by the Sample/Fuzzy/Permutation/Dispersion and generalized-information catalogue entries. Representative definitions include Sample Entropy
\(\mathrm{SampEn}(m,r)=-\log(A/B)\), Shannon entropy \(H=-\sum_i p_i\log p_i\), Rényi entropy \(H_q=(1-q)^{-1}\log\sum_i p_i^q\), and Tsallis entropy \(S_q=(1-\sum_i p_i^q)/(q-1)\). The helper routines centralize symbolization, distances, coarse graining, and numerical safeguards so all methods use the same EEG segment and runtime budget.</p>
</div>

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Shannon entropy summarizes uncertainty in a probability distribution. Each outcome contributes according to both its probability and the information associated with that probability, so evenly distributed outcomes give larger entropy than a distribution concentrated on a few outcomes. An entropy rate extends this idea to a time process and asks how much new uncertainty is produced per time step after accounting for the past.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values. Because the calculation is performed on finite windows or segments, the reported value is a local summary of that selected data interval; a different interval can legitimately produce a different estimate if the EEG is nonstationary.</p></div>


```python
# =======================================================
# ENTROPY METHOD FOUNDATION — explicit methods 1–287
# Shared numerical primitives only; every numbered method below contains its own explicit calculation.
# =======================================================
import os, math, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import signal, stats, special
from scipy.spatial.distance import pdist, squareform
from scipy.spatial import cKDTree
from numpy.lib.stride_tricks import sliding_window_view
from collections import Counter
from IPython.display import display

ENT_EPS=1e-12
ENT_BG="#000000"; ENT_ACCENT="#00FFFF"
plt.rcParams.update({"figure.facecolor":ENT_BG,"axes.facecolor":ENT_BG,"savefig.facecolor":ENT_BG,
    "text.color":ENT_ACCENT,"axes.labelcolor":ENT_ACCENT,"axes.edgecolor":ENT_ACCENT,
    "axes.titlecolor":ENT_ACCENT,"xtick.color":ENT_ACCENT,"ytick.color":ENT_ACCENT,
    "grid.color":ENT_ACCENT})
FAST_OPTION=globals().get("FAST_OPTION","ultra")
if FAST_OPTION not in {"full","balanced","fast","ultra"}: raise ValueError("FAST_OPTION must be one of: full, balanced, fast, ultra")
_ENT_SPEED={
 "full":dict(down=1,max_samples=106944,patterns=8000,windows=36,scales=20,boot=60,sur=40),
 "balanced":dict(down=2,max_samples=70000,patterns=5000,windows=28,scales=15,boot=32,sur=24),
 "fast":dict(down=4,max_samples=40000,patterns=3000,windows=20,scales=10,boot=16,sur=12),
 "ultra":dict(down=8,max_samples=18000,patterns=1400,windows=12,scales=7,boot=6,sur=5),
}[FAST_OPTION]
_ENT_NAMES=['Fp1','Fpz','Fp2','F7','F3','Fz','F4','F8','FC5','FC1','FC2','FC6','M1','T7','C3','Cz','C4','T8','M2','CP5','CP1','CP2','CP6','P7','P3','Pz','P4','P8','POz','O1','Oz','O2']
_ENT_BANDS={'delta':(1,4),'theta':(4,8),'alpha':(8,13),'beta':(13,30),'gamma':(30,45)}

def _ent_layout(a):
    a=np.asarray(a,float)
    if a.ndim!=2: raise ValueError('EEG array must be 2-D')
    if a.shape[1]<=64 and a.shape[0]>a.shape[1]: return a
    if a.shape[0]<=64: return a.T
    return a

if 'ENTROPY_DATA_OVERRIDE' in globals():
    _ent_raw=_ent_layout(globals()['ENTROPY_DATA_OVERRIDE']); _ent_fs0=float(globals().get('ENTROPY_FS_OVERRIDE',1000.0))
else:
    _p='/home/a/projects/Complete-Neural-Signal-Analysis/eeg_data_with_channels.npy'; _ent_fs0=1000.0
    _a=_ent_layout(np.load(_p,allow_pickle=True)); a,b=int(814.571*_ent_fs0),int(921.515*_ent_fs0); _ent_raw=_a[a:min(b,len(_a))]
_ent_raw=np.asarray(_ent_raw[:_ENT_SPEED['max_samples']],float)
_ent_raw=signal.detrend(_ent_raw,axis=0,type='constant')
_d=int(_ENT_SPEED['down']); _ent_x=signal.resample_poly(_ent_raw,1,_d,axis=0) if _d>1 else _ent_raw.copy(); _ent_fs=_ent_fs0/_d
_ent_x=(_ent_x-np.nanmean(_ent_x,axis=0,keepdims=True))/(np.nanstd(_ent_x,axis=0,keepdims=True)+ENT_EPS)
_ent_names=_ENT_NAMES[:_ent_x.shape[1]]; _ent_y=_ent_x[:,0]

def _cap(y,n=None):
    y=np.asarray(y,float).ravel(); n=int(n or _ENT_SPEED['patterns'])
    if len(y)>n:
        idx=np.linspace(0,len(y)-1,n).astype(int); y=y[idx]
    return y

def _embed(y,m,tau=1,cap=True):
    y=np.asarray(y,float).ravel(); n=len(y)-(m-1)*tau
    if n<=1:return np.empty((0,m))
    X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
    if cap and len(X)>_ENT_SPEED['patterns']:
        idx=np.linspace(0,len(X)-1,_ENT_SPEED['patterns']).astype(int);X=X[idx]
    return X

def _p_from_labels(labels,weights=None,full_states=None):
    labels=list(map(tuple,labels)) if np.asarray(labels).ndim>1 else list(labels)
    if weights is None:
        C=Counter(labels); vals=np.array(list(C.values()),float)
    else:
        C={}
        for k,w in zip(labels,np.asarray(weights,float)): C[k]=C.get(k,0.0)+float(w)
        vals=np.array(list(C.values()),float)
    p=vals/(vals.sum()+ENT_EPS); return p[p>0]

def _H(p,base=np.e):
    p=np.asarray(p,float);p=p[p>0];return float(-np.sum(p*np.log(p))/np.log(base)) if len(p) else 0.0

def _hist_p(y,bins=16,quantile=False):
    y=np.asarray(y,float); y=y[np.isfinite(y)]
    if not len(y):return np.array([1.0])
    if quantile:
        e=np.unique(np.quantile(y,np.linspace(0,1,bins+1)))
        if len(e)<2:return np.array([1.0])
        h,_=np.histogram(y,bins=e)
    else:h,_=np.histogram(y,bins=bins)
    h=h[h>0].astype(float);return h/(h.sum()+ENT_EPS)

def _match_pairs(X,r):
    if len(X)<2:return 0
    return len(cKDTree(np.asarray(X,float)).query_pairs(float(r),p=np.inf))

def _sampen(y,m=2,r=.2):
    y=np.asarray(y,float);sd=np.std(y);rr=float(r*sd if r<1 else r)
    Xm=_embed(y,m);X1=_embed(y,m+1);B=_match_pairs(Xm,rr);A=_match_pairs(X1,rr)
    denB=max(1,len(Xm)*(len(Xm)-1)/2);denA=max(1,len(X1)*(len(X1)-1)/2);pB=B/denB;pA=A/denA
    if pA<=0 or pB<=0:return np.inf
    return float(-np.log(pA/pB))

def _apen(y,m=2,r=.2):
    y=_cap(y);sd=np.std(y);rr=float(r*sd if r<1 else r)
    def phi(k):
        X=_embed(y,k,cap=False)
        if not len(X):return np.nan
        tr=cKDTree(X);cnt=np.array([len(v) for v in tr.query_ball_point(X,rr,p=np.inf)],float)/len(X)
        return float(np.mean(np.log(cnt+ENT_EPS)))
    return float(phi(m)-phi(m+1))

def _fuzzyen(y,m=2,r=.2,n=2):
    y=_cap(y);sd=np.std(y);rr=max(ENT_EPS,r*sd)
    def sim(k):
        X=_embed(y,k,cap=False);X=X-X.mean(1,keepdims=True)
        if len(X)<2:return np.nan
        d=pdist(X,metric='chebyshev');return float(np.mean(np.exp(-(d**n)/(rr+ENT_EPS))))
    a,b=sim(m+1),sim(m);return float(-np.log((a+ENT_EPS)/(b+ENT_EPS)))

def _cross_sampen(x,y,m=2,r=.2):
    x=_cap(x);y=_cap(y);sd=np.std(np.r_[x,y]);rr=max(ENT_EPS,r*sd)
    def count(k):
        X=_embed(x,k,cap=False);Y=_embed(y,k,cap=False); tr=cKDTree(Y)
        return sum(len(v) for v in tr.query_ball_point(X,rr,p=np.inf))
    X0,Y0=_embed(x,m,cap=False),_embed(y,m,cap=False);X1,Y1=_embed(x,m+1,cap=False),_embed(y,m+1,cap=False)
    B,A=count(m),count(m+1);pB=B/max(1,len(X0)*len(Y0));pA=A/max(1,len(X1)*len(Y1));return float(-np.log((pA+ENT_EPS)/(pB+ENT_EPS)))

def _multivar_sampen(X,m=2,r=.2):
    X=np.asarray(X,float); X=(X-X.mean(0))/(X.std(0)+ENT_EPS); n=len(X)-(m-1)
    Z=np.hstack([X[j:j+n] for j in range(m)]); Z1=np.hstack([X[j:j+n-1] for j in range(m+1)])
    if len(Z)>_ENT_SPEED['patterns']:
        ids=np.linspace(0,len(Z)-1,_ENT_SPEED['patterns']).astype(int);Z=Z[ids]
    if len(Z1)>_ENT_SPEED['patterns']:
        ids=np.linspace(0,len(Z1)-1,_ENT_SPEED['patterns']).astype(int);Z1=Z1[ids]
    B=_match_pairs(Z,r);A=_match_pairs(Z1,r);pB=B/max(1,len(Z)*(len(Z)-1)/2);pA=A/max(1,len(Z1)*(len(Z1)-1)/2);return float(-np.log((pA+ENT_EPS)/(pB+ENT_EPS)))

def _ordinal_labels(y,m=3,tau=1):
    X=_embed(y,m,tau); return [tuple(np.argsort(v,kind='mergesort')) for v in X]

def _perm_entropy(y,m=3,tau=1,weighted=False,normalize=True):
    X=_embed(y,m,tau);labs=[tuple(np.argsort(v,kind='mergesort')) for v in X]
    w=np.var(X,axis=1) if weighted and len(X) else None;p=_p_from_labels(labs,w);h=_H(p)
    return float(h/(math.lgamma(m+1)+ENT_EPS)) if normalize else h

def _disp_classes(y,c=6):
    y=np.asarray(y,float);z=(y-np.mean(y))/(np.std(y)+ENT_EPS);u=stats.norm.cdf(z);cl=np.rint(c*u+.5).astype(int);return np.clip(cl,1,c)

def _disp_entropy(y,m=3,c=6,tau=1,fluct=False,normalize=True,reverse=False):
    z=_disp_classes(y,c);X=_embed(z,m,tau)
    P=np.diff(X,axis=1).astype(int) if fluct else X.astype(int); p=_p_from_labels(P)
    if reverse:
        M=(2*c-1)**(m-1) if fluct else c**m; val=float(np.sum((p-1/M)**2)+(M-len(p))*(1/M)**2);return val/(1-1/M+ENT_EPS)
    h=_H(p);M=(2*c-1)**(m-1) if fluct else c**m;return float(h/(np.log(M)+ENT_EPS)) if normalize else h

def _coarse(y,s,offset=0):
    y=np.asarray(y,float);n=(len(y)-offset)//s
    return y[offset:offset+n*s].reshape(n,s).mean(1) if n>0 else np.empty(0)

def _distribution_entropy(y,m=3,bins=32):
    X=_embed(y,m);d=pdist(X)
    if not len(d):return 0.0
    p=_hist_p(d,bins=bins);return _H(p)/np.log(max(2,bins))

def _bubble_entropy(y,m=3):
    # Bubble-sort swap count is inversion count of each embedded vector.
    def invdist(k):
        X=_embed(y,k); inv=[]
        for v in X:
            inv.append(sum(v[a]>v[b] for a in range(k) for b in range(a+1,k)))
        p=_p_from_labels(inv); return -np.log(np.sum(p*p)+ENT_EPS)
    return float((invdist(m+1)-invdist(m))/np.log((m+1)/(m-1)))

def _slope_entropy(y,m=4,gamma=1.0,delta=.01):
    X=_embed(y,m);D=np.diff(X,axis=1);s=np.zeros_like(D,int);s[D>delta]=1;s[D>gamma]=2;s[D<-delta]=-1;s[D<-gamma]=-2
    return _H(_p_from_labels(s))/np.log(5**(m-1))

def _cosine_similarity_entropy(y,m=3,r=.1):
    X=_embed(y,m);X=X-X.mean(1,keepdims=True);norm=np.linalg.norm(X,axis=1);X=X/(norm[:,None]+ENT_EPS)
    if len(X)<2:return 0.0
    cs=1-pdist(X,metric='cosine'); B=np.mean(cs>1-r)
    X1=_embed(y,m+1);X1=X1-X1.mean(1,keepdims=True);X1=X1/(np.linalg.norm(X1,axis=1)[:,None]+ENT_EPS);cs1=1-pdist(X1,metric='cosine');A=np.mean(cs1>1-r)
    return float(-np.log((A+ENT_EPS)/(B+ENT_EPS)))

def _increment_entropy(y,m=3,q=4):
    d=np.diff(np.asarray(y,float));mag=np.abs(d);edges=np.quantile(mag,np.linspace(0,1,q+1));cls=np.clip(np.digitize(mag,edges[1:-1]),0,q-1);sym=np.sign(d).astype(int)*(cls+1);X=_embed(sym,m)
    return _H(_p_from_labels(X))/np.log((2*q)**m)

def _grid_entropy(y,bins=12):
    y=np.asarray(y,float);a,b=y[:-1],y[1:];H,_,_=np.histogram2d(a,b,bins=bins);p=H.ravel();p=p[p>0];p=p/p.sum();return _H(p)/np.log(bins*bins)

def _attention_entropy(y):
    y=np.asarray(y,float);mx,_=signal.find_peaks(y);mn,_=signal.find_peaks(-y)
    vals=[]
    for p in (mx,mn):
        if len(p)>2: vals.append(_H(_hist_p(np.diff(p),bins=min(16,max(4,int(np.sqrt(len(p))))))))
    return float(np.mean(vals)) if vals else 0.0

def _symbolic_dynamic_entropy(y,c=6):
    z=_disp_classes(y,c);p=_hist_p(z,bins=np.arange(.5,c+1.6,1));pairs=np.column_stack([z[:-1],z[1:]]);pt=_p_from_labels(pairs)
    return float(_H(p)+_H(pt)-_H(_hist_p(z[:-1],bins=np.arange(.5,c+1.6,1))))

def _sax_entropy(y,segments=32,alphabet=6):
    y=np.asarray(y,float);n=len(y)//segments
    if n<1:return 0.0
    paa=y[:n*segments].reshape(segments,n).mean(1);z=(paa-paa.mean())/(paa.std()+ENT_EPS);cuts=stats.norm.ppf(np.arange(1,alphabet)/alphabet);sym=np.digitize(z,cuts)
    return _H(_p_from_labels(sym))/np.log(alphabet)

def _renyi(p,q):
    p=np.asarray(p,float);p=p[p>0]
    if abs(q-1)<1e-8:return _H(p)
    return float(np.log(np.sum(p**q))/(1-q))

def _tsallis(p,q):
    p=np.asarray(p,float);p=p[p>0]
    if abs(q-1)<1e-8:return _H(p)
    return float((1-np.sum(p**q))/(q-1))

def _sharma_mittal(p,q,r):
    p=np.asarray(p,float);p=p[p>0]
    if abs(r-1)<1e-8:return _renyi(p,q)
    if abs(q-1)<1e-8:
        # continuous q->1 limit gives (exp((1-r)H)-1)/(1-r)
        H=_H(p);return float((np.exp((1-r)*H)-1)/(1-r))
    return float((np.sum(p**q)**((1-r)/(1-q))-1)/(1-r))

def _arimoto_conditional(x,y,alpha=2.0,bins=8):
    # H_alpha(X|Y)=alpha/(1-alpha) log sum_y p(y)||p(X|y)||_alpha.
    ex=np.quantile(x,np.linspace(0,1,bins+1));ey=np.quantile(y,np.linspace(0,1,bins+1));ex=np.unique(ex);ey=np.unique(ey)
    ix=np.clip(np.digitize(x,ex[1:-1]),0,len(ex)-2);iy=np.clip(np.digitize(y,ey[1:-1]),0,len(ey)-2)
    H=np.zeros((max(1,len(ex)-1),max(1,len(ey)-1)))
    for a,b in zip(ix,iy):H[a,b]+=1
    py=H.sum(0); py=py/(py.sum()+ENT_EPS);val=0.
    for j,w in enumerate(py):
        if w<=0:continue
        px=H[:,j]/(H[:,j].sum()+ENT_EPS);val+=w*(np.sum(px**alpha)**(1/alpha))
    return float(alpha/(1-alpha)*np.log(val+ENT_EPS))

def _band(y,lo,hi):
    hi=min(hi,_ent_fs/2-.5)
    if hi<=lo:return np.asarray(y,float)
    sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_ent_fs,output='sos');return signal.sosfiltfilt(sos,y)


from matplotlib.colors import LinearSegmentedColormap
from cycler import cycler

ENT_DIM = "#007A7A"
ENT_CMAP = LinearSegmentedColormap.from_list("entropy_black_cyan", [ENT_BG, "#003A3A", ENT_ACCENT])

plt.rcParams.update({
    "figure.facecolor": ENT_BG,
    "axes.facecolor": ENT_BG,
    "savefig.facecolor": ENT_BG,
    "text.color": ENT_ACCENT,
    "axes.labelcolor": ENT_ACCENT,
    "axes.edgecolor": ENT_ACCENT,
    "axes.titlecolor": ENT_ACCENT,
    "xtick.color": ENT_ACCENT,
    "ytick.color": ENT_ACCENT,
    "grid.color": ENT_ACCENT,
    "legend.facecolor": ENT_BG,
    "legend.edgecolor": ENT_ACCENT,
    "legend.labelcolor": ENT_ACCENT,
    "axes.prop_cycle": cycler(color=[ENT_ACCENT]),
})

def _ent_style_ax(ax, grid=True):
    ax.set_facecolor(ENT_BG)
    ax.tick_params(colors=ENT_ACCENT)
    ax.xaxis.label.set_color(ENT_ACCENT)
    ax.yaxis.label.set_color(ENT_ACCENT)
    ax.title.set_color(ENT_ACCENT)
    for spine in ax.spines.values():
        spine.set_color(ENT_ACCENT)
    ax.grid(bool(grid), alpha=0.20, color=ENT_ACCENT)

def _ent_style_legend(legend):
    if legend is None:
        return
    frame = legend.get_frame()
    frame.set_facecolor(ENT_BG)
    frame.set_edgecolor(ENT_ACCENT)
    for txt in legend.get_texts():
        txt.set_color(ENT_ACCENT)

def _ent_style_colorbar(cbar):
    cbar.outline.set_edgecolor(ENT_ACCENT)
    cbar.ax.tick_params(color=ENT_ACCENT, labelcolor=ENT_ACCENT)
    for txt in cbar.ax.get_yticklabels():
        txt.set_color(ENT_ACCENT)

def _ent_finish_plot(fig):
    fig.tight_layout()
    plt.show()
    plt.close(fig)

def _line(x, y, title, xlabel="x", ylabel="entropy"):
    fig, ax = plt.subplots(figsize=(7.2, 3.8), facecolor=ENT_BG)
    _ent_style_ax(ax)
    ax.plot(
        np.asarray(x),
        np.asarray(y),
        color=ENT_ACCENT,
        marker="o",
        markerfacecolor=ENT_BG,
        markeredgecolor=ENT_ACCENT,
        linewidth=1.7,
        markersize=4.5,
    )
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    _ent_finish_plot(fig)

def _show_scalar(name, val):
    print(f"{name}: {val:.6g}" if np.isscalar(val) and np.isfinite(val) else f"{name}: {val}")

def _ent_is_number(v):
    return isinstance(v, (int, float, complex, np.integer, np.floating)) and not isinstance(v, (bool, np.bool_))

def _ent_plot_dataframe(df, title):
    if not isinstance(df, pd.DataFrame) or df.empty:
        return False
    num = df.select_dtypes(include=[np.number]).replace([np.inf, -np.inf], np.nan)
    if num.empty:
        return False
    num = num.dropna(axis=1, how="all")
    if num.empty:
        return False

    # Do not create a graph for a result containing fewer than two
    # finite numeric values. A single number is clearer as text.
    finite_value_count = int(np.isfinite(num.to_numpy(dtype=float)).sum())
    if finite_value_count < 2:
        return False

    nr, nc = num.shape
    if nr == 1:
        fig, ax = plt.subplots(figsize=(max(7, 0.75 * nc), 4), facecolor=ENT_BG)
        _ent_style_ax(ax)
        vals = num.iloc[0].to_numpy(float)
        ax.bar(np.arange(nc), vals, facecolor=ENT_BG, edgecolor=ENT_ACCENT, linewidth=1.4)
        ax.set_xticks(np.arange(nc))
        ax.set_xticklabels([str(c) for c in num.columns], rotation=40, ha="right", color=ENT_ACCENT)
        ax.set_ylabel("value")
        ax.set_title(title)
        _ent_finish_plot(fig)
        return True

    if nr == nc and 2 <= nr <= 64:
        fig, ax = plt.subplots(figsize=(7, 6), facecolor=ENT_BG)
        _ent_style_ax(ax, grid=False)
        arr = num.to_numpy(float)
        im = ax.imshow(arr, aspect="auto", cmap=ENT_CMAP, interpolation="nearest")
        ax.set_title(title)
        ax.set_xticks(np.arange(nc))
        ax.set_yticks(np.arange(nr))
        ax.set_xticklabels([str(c) for c in num.columns], rotation=45, ha="right", color=ENT_ACCENT, fontsize=7)
        ax.set_yticklabels([str(v) for v in num.index], color=ENT_ACCENT, fontsize=7)
        cbar = fig.colorbar(im, ax=ax)
        _ent_style_colorbar(cbar)
        _ent_finish_plot(fig)
        return True

    if nc <= 6:
        fig, ax = plt.subplots(figsize=(8.5, 4.6), facecolor=ENT_BG)
        _ent_style_ax(ax)
        styles = ["-", "--", "-.", ":", (0, (5, 2)), (0, (1, 1))]
        x = np.arange(nr)
        for k, col in enumerate(num.columns):
            vals = num[col].to_numpy(float)
            ax.plot(
                x,
                vals,
                color=ENT_ACCENT,
                linestyle=styles[k % len(styles)],
                marker="o" if nr <= 40 else None,
                markerfacecolor=ENT_BG,
                markeredgecolor=ENT_ACCENT,
                linewidth=1.5,
                alpha=max(0.45, 1.0 - 0.10 * k),
                label=str(col),
            )
        ax.set_xlabel("row")
        ax.set_ylabel("value")
        ax.set_title(title)
        legend = ax.legend()
        _ent_style_legend(legend)
        _ent_finish_plot(fig)
        return True

    arr = num.to_numpy(float)
    if nr > 80:
        take = np.linspace(0, nr - 1, 80).astype(int)
        arr = arr[take]
    if nc > 40:
        arr = arr[:, :40]
        cols = list(num.columns[:40])
    else:
        cols = list(num.columns)
    fig, ax = plt.subplots(figsize=(9, 5.5), facecolor=ENT_BG)
    _ent_style_ax(ax, grid=False)
    im = ax.imshow(arr, aspect="auto", cmap=ENT_CMAP, interpolation="nearest")
    ax.set_title(title)
    if len(cols) <= 20:
        ax.set_xticks(np.arange(len(cols)))
        ax.set_xticklabels([str(c) for c in cols], rotation=45, ha="right", color=ENT_ACCENT, fontsize=7)
    cbar = fig.colorbar(im, ax=ax)
    _ent_style_colorbar(cbar)
    _ent_finish_plot(fig)
    return True

def _ent_auto_plot(method_id, title, local_ns, candidate_names=()):
    """
    Plot a method result only when the result contains at least two
    finite numeric values that can meaningfully be compared.

    Single-number entropy results are printed only; no decorative bar
    chart or unrelated input-context plot is generated.
    """
    if local_ns is None:
        local_ns = {}

    def _finite_numeric_count(obj):
        """Count finite numeric values in a result-like object."""
        if obj is None:
            return 0

        if _ent_is_number(obj):
            try:
                return int(np.isfinite(float(np.real(obj))))
            except Exception:
                return 0

        if isinstance(obj, pd.DataFrame):
            num = obj.select_dtypes(include=[np.number]).replace(
                [np.inf, -np.inf], np.nan
            )
            if num.empty:
                return 0
            return int(np.isfinite(num.to_numpy(dtype=float)).sum())

        if isinstance(obj, pd.Series):
            if not np.issubdtype(obj.dtype, np.number):
                return 0
            arr = pd.to_numeric(obj, errors="coerce").to_numpy(dtype=float)
            return int(np.isfinite(arr).sum())

        if isinstance(obj, dict):
            return sum(_finite_numeric_count(v) for v in obj.values())

        if isinstance(obj, (list, tuple)):
            return sum(
                _finite_numeric_count(v)
                for v in obj
                if not isinstance(v, str)
            )

        try:
            arr = np.asarray(obj)
        except Exception:
            return 0

        if arr.ndim == 0:
            try:
                z = float(np.real(arr.item()))
                return int(np.isfinite(z))
            except Exception:
                return 0

        if not np.issubdtype(arr.dtype, np.number):
            return 0

        try:
            arr = np.asarray(arr, dtype=float)
        except Exception:
            return 0

        return int(np.isfinite(arr).sum())

    # --------------------------------------------------------------
    # 1) Explicit table/series result has highest priority.
    #    If it contains only one numeric result, intentionally stop:
    #    do not fall through and graph internal working arrays.
    # --------------------------------------------------------------
    obj = local_ns.get("_ent_display_result", None)

    if isinstance(obj, pd.DataFrame):
        if _finite_numeric_count(obj) < 2:
            return
        _ent_plot_dataframe(obj, f"{method_id}. {title}")
        return

    if isinstance(obj, pd.Series):
        if _finite_numeric_count(obj) < 2:
            return
        obj_df = obj.to_frame(name=obj.name or "value")
        _ent_plot_dataframe(obj_df, f"{method_id}. {title}")
        return

    # --------------------------------------------------------------
    # 2) An explicitly declared scalar result is a one-number result.
    #    Print it, but do not graph it and do not graph intermediates.
    # --------------------------------------------------------------
    scalar_result = local_ns.get("_ent_scalar_result", None)
    if _ent_is_number(scalar_result):
        return

    # --------------------------------------------------------------
    # 3) Collect named method-local outputs, excluding raw inputs and
    #    obvious loop/index variables.
    # --------------------------------------------------------------
    skip = {
        "x", "y", "X", "S", "sym", "rng", "j", "k", "a", "b", "c", "s", "_",
        "n", "m", "i", "lo", "hi", "st", "idx", "e", "aa", "bb", "cc",
        "_ent_print_args", "_ent_display_result",
    }

    objects = []
    for name in candidate_names:
        if name in skip or name not in local_ns:
            continue
        objects.append((name, local_ns[name]))

    # DataFrames first.
    for name, value in reversed(objects):
        if isinstance(value, pd.DataFrame):
            if _finite_numeric_count(value) < 2:
                continue
            if _ent_plot_dataframe(value, f"{method_id}. {title}"):
                return

    # Matrix-like outputs next.
    for name, value in reversed(objects):
        try:
            arr = np.asarray(value)
        except Exception:
            continue

        if (
            arr.ndim == 2
            and arr.size
            and 2 <= arr.shape[0] <= 200
            and 2 <= arr.shape[1] <= 200
            and np.issubdtype(arr.dtype, np.number)
        ):
            arr = arr.astype(float)
            if np.isfinite(arr).sum() >= 2:
                fig, ax = plt.subplots(figsize=(7.4, 5.8), facecolor=ENT_BG)
                _ent_style_ax(ax, grid=False)
                im = ax.imshow(
                    arr,
                    aspect="auto",
                    cmap=ENT_CMAP,
                    interpolation="nearest",
                )
                ax.set_title(f"{method_id}. {title} — {name}")
                ax.set_xlabel("column")
                ax.set_ylabel("row")
                cbar = fig.colorbar(im, ax=ax)
                _ent_style_colorbar(cbar)
                _ent_finish_plot(fig)
                return

    # Useful 1-D method outputs.
    vectors = []
    for name, value in reversed(objects):
        try:
            arr = np.asarray(value)
        except Exception:
            continue

        if (
            arr.ndim == 1
            and 2 <= arr.size <= 2000
            and np.issubdtype(arr.dtype, np.number)
        ):
            arr = arr.astype(float)
            if np.isfinite(arr).sum() >= 2:
                vectors.append((name, arr))

        if len(vectors) >= 4:
            break

    if vectors:
        fig, ax = plt.subplots(figsize=(8.2, 4.4), facecolor=ENT_BG)
        _ent_style_ax(ax)
        styles = ["-", "--", "-.", ":"]

        for q, (name, arr) in enumerate(reversed(vectors)):
            ax.plot(
                np.arange(len(arr)),
                arr,
                color=ENT_ACCENT,
                linestyle=styles[q % len(styles)],
                linewidth=1.5,
                alpha=max(0.55, 1 - 0.12 * q),
                label=name,
            )

        ax.set_title(f"{method_id}. {title}")
        ax.set_xlabel("index")
        ax.set_ylabel("value")
        legend = ax.legend()
        _ent_style_legend(legend)
        _ent_finish_plot(fig)
        return

    # --------------------------------------------------------------
    # 4) Multiple explicitly reported scalar values may be compared.
    #    ONE scalar value is intentionally left as text only.
    # --------------------------------------------------------------
    scalars = {}

    printed = local_ns.get("_ent_print_args", None)

    def collect_scalar(obj, prefix="value"):
        if _ent_is_number(obj):
            z = float(np.real(obj))
            if np.isfinite(z):
                scalars[prefix] = z

        elif isinstance(obj, dict):
            for kk, vv in obj.items():
                collect_scalar(vv, str(kk))

        elif isinstance(obj, (list, tuple)):
            for q, vv in enumerate(obj):
                if not isinstance(vv, str):
                    collect_scalar(vv, f"{prefix}_{q}")

        elif isinstance(obj, np.ndarray) and obj.ndim == 0:
            collect_scalar(obj.item(), prefix)

    collect_scalar(printed, "result")

    if not scalars:
        for name, value in objects:
            if _ent_is_number(value):
                z = float(np.real(value))
                if np.isfinite(z):
                    scalars[name] = z
            if len(scalars) >= 10:
                break

    # No plot unless at least two scalar results exist.
    if len(scalars) >= 2:
        labels = list(scalars)[:10]
        vals = [scalars[k] for k in labels]

        fig, ax = plt.subplots(
            figsize=(max(7, 0.8 * len(labels)), 4.2),
            facecolor=ENT_BG,
        )
        _ent_style_ax(ax)
        ax.bar(
            np.arange(len(vals)),
            vals,
            facecolor=ENT_BG,
            edgecolor=ENT_ACCENT,
            linewidth=1.5,
        )
        ax.set_xticks(np.arange(len(vals)))
        ax.set_xticklabels(
            labels,
            rotation=40,
            ha="right",
            color=ENT_ACCENT,
        )
        ax.set_ylabel("value")
        ax.set_title(f"{method_id}. {title}")
        _ent_finish_plot(fig)

    # If there are zero or one comparable numeric results, do nothing.
    return

print(f'Entropy explicit-method foundation ready: mode={FAST_OPTION}, samples={len(_ent_x)}, channels={_ent_x.shape[1]}, fs={_ent_fs:g} Hz')
```

### 1. Standalone Sample Entropy suite

Sample entropy estimates the negative log conditional probability that two length-m patterns that match within tolerance r also match when extended to m+1: SampEn=-ln(A_m+1/B_m). Self-matches are excluded.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Sample entropy quantifies how often similar short patterns remain similar when one additional sample is included. It compares the number of matching patterns of length m with the number of matching patterns of length m+1, using a tolerance r to define what counts as similar. Fewer continued matches produce larger entropy, interpreted as lower short-term predictability. The result depends on m, r, record length, and preprocessing, so comparisons should use consistent settings.</p><p><b>How to read the mathematics.</b> A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# =======================================================
# 1. Standalone Sample Entropy suite
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Sample entropy estimates the negative log conditional probability that two length-m
# patterns that match within tolerance r also match when extended to m+1: SampEn=-ln(A_m+1/B_m). Self-
# matches are excluded.
# =======================================================
def _entropy_method_001():
    y = _ent_y
    vals = [_sampen(_ent_x[:, j]) for j in range(min(8, _ent_x.shape[1]))]
    _ent_display_result = pd.DataFrame({'channel': _ent_names[:len(vals)], 'SampleEntropy': vals})
    display(_ent_display_result)
    return locals()
_ent_method_ns = _entropy_method_001()
_ent_auto_plot(1, 'Standalone Sample Entropy suite', _ent_method_ns, ['y', 'vals', '_ent_display_result', 'j'])
```

### 2. Sample Entropy parameter sweep

This cell sweeps the defining Sample Entropy parameters m and r. The estimator is unchanged; the grid shows how finite-data regularity depends on embedding dimension and tolerance.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Delay embedding turns one measured time series into a sequence of state-space vectors by combining the present sample with delayed copies of the same signal. ### 5. Rolling-window Sample Entropy

Rolling Sample Entropy applies the same SampEn estimator to overlapping or evenly spaced windows, producing a time-resolved complexity trajectory.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Sample entropy quantifies how often similar short patterns remain similar when one additional sample is included. It compares the number of matching patterns of length m with the number of matching patterns of length m+1, using a tolerance r to define what counts as similar. Fewer continued matches produce larger entropy, interpreted as lower short-term predictability. The result depends on m, r, record length, and preprocessing, so comparisons should use consistent settings.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> Because the calculation is performed on finite windows or segments, the reported value is a local summary of that selected data interval; a different interval can legitimately produce a different estimate if the EEG is nonstationary.</p></div>


```python
# =======================================================
# 5. Rolling-window Sample Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Rolling Sample Entropy applies the same SampEn estimator to overlapping or evenly
# spaced windows, producing a time-resolved complexity trajectory.
# =======================================================
def _entropy_method_005():
    y = _ent_y
    W = max(200, int(4 * _ent_fs))
    starts = np.linspace(0, max(0, len(y) - W), _ENT_SPEED['windows']).astype(int)
    v = [_sampen(y[s:s + W]) for s in starts]
    _line(starts / _ent_fs, v, 'Rolling-window Sample Entropy', 'time (s)', 'SampEn')
    return locals()
_ent_method_ns = _entropy_method_005()
```

### 6. Event-locked Sample Entropy

Event-locked Sample Entropy evaluates SampEn in windows centered on supplied event samples. If no external events are provided, internally detected envelope peaks are explicitly labelled demonstration events.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Sample entropy quantifies how often similar short patterns remain similar when one additional sample is included. It compares the number of matching patterns of length m with the number of matching patterns of length m+1, using a tolerance r to define what counts as similar. Fewer continued matches produce larger entropy, interpreted as lower short-term predictability. The result depends on m, r, record length, and preprocessing, so comparisons should use consistent settings.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> Because the calculation is performed on finite windows or segments, the reported value is a local summary of that selected data interval; a different interval can legitimately produce a different estimate if the EEG is nonstationary.</p></div>


```python
# =======================================================
# 6. Event-locked Sample Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Event-locked Sample Entropy evaluates SampEn in windows centered on supplied event
# samples. If no external events are provided, internally detected envelope peaks are explicitly
# labelled demonstration events.
# =======================================================
def _entropy_method_006():
    y = _ent_y
    ev = globals().get('ENTROPY_EVENT_SAMPLES', None)
    label = 'supplied events'
    if ev is None:
        e = np.abs(signal.hilbert(y))
        pk, _ = signal.find_peaks(e, distance=max(1, int(0.5 * _ent_fs)), prominence=np.std(e))
        ev = pk[np.argsort(e[pk])[-min(6, len(pk)):]] if len(pk) else []
        label = 'internal high-envelope events (demonstration)'
    half = max(100, int(0.5 * _ent_fs))
    rows = []
    for k in ev:
        a = max(0, int(k) - half)
        b = min(len(y), int(k) + half)
        rows.append([int(k), _sampen(y[a:b])])
    _ent_print_args = [label]
    print(*_ent_print_args)
    _ent_display_result = pd.DataFrame(rows, columns=['event_sample', 'SampEn'])
    display(_ent_display_result)
    return locals()
_ent_method_ns = _entropy_method_006()
_ent_auto_plot(6, 'Event-locked Sample Entropy', _ent_method_ns, ['y', 'ev', 'label', 'half', 'rows', 'k', '_ent_print_args', '_ent_display_result', 'e', 'a', 'b', 'pk', '_'])
```

### 10. Fuzzy Entropy

### 12. Refined Composite Multiscale Fuzzy Entropy

Refined-composite multiscale entropy averages entropy estimates over all s coarse-graining offsets, reducing the variance caused by choosing a single block origin.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Multiscale entropy asks how signal irregularity changes when the data are viewed at progressively coarser time scales. ### 13. Cross-Fuzzy Entropy

Cross-Fuzzy Entropy applies fuzzy similarity to delay vectors from two different signals and compares similarity at dimensions m and m+1.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# =======================================================
# 13. Cross-Fuzzy Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Cross-Fuzzy Entropy applies fuzzy similarity to delay vectors from two different
# signals and compares similarity at dimensions m and m+1.
# =======================================================

def _entropy_method_013():
    y = _ent_y

    x = _cap(_ent_x[:, 0])
    z = _cap(_ent_x[:, 1])

    rr = 0.2 * np.std(np.r_[x, z])

    def sim(k):
        X = _embed(x, k, cap=False)
        Z = _embed(z, k, cap=False)

        if X is None or Z is None or len(X) == 0 or len(Z) == 0:
            return np.nan

        # Cross-channel Chebyshev distances between delay vectors
        D = np.max(
            np.abs(X[:, None, :] - Z[None, :, :]),
            axis=2,
        )

        # Fuzzy membership:
        # mu(d) = exp(-d^2 / r)
        phi = np.mean(
            np.exp(
                -(D ** 2) / (rr + ENT_EPS)
            )
        )

        return float(phi)

    phi_m = sim(2)
    phi_m1 = sim(3)

    _ent_scalar_label = "Cross-Fuzzy Entropy"

    if (
        not np.isfinite(phi_m)
        or not np.isfinite(phi_m1)
    ):
        _ent_scalar_result = np.nan
    else:
        _ent_scalar_result = -np.log(
            (phi_m1 + ENT_EPS)
            / (phi_m + ENT_EPS)
        )

    _show_scalar(
        _ent_scalar_label,
        _ent_scalar_result,
    )

    return locals()


_ent_method_ns = _entropy_method_013()

_ent_auto_plot(
    13,
    "Cross-Fuzzy Entropy",
    _ent_method_ns,
    [
        "y",
        "rr",
        "phi_m",
        "phi_m1",
        "_ent_scalar_label",
        "_ent_scalar_result",
        "x",
        "z",
    ],
)
```

### 14. Multivariate Fuzzy Entropy

Multivariate Fuzzy Entropy measures fuzzy recurrence of joint multichannel delay vectors, extending Fuzzy Entropy from one coordinate to a synchronized vector process.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values.</p></div>


```python
# =======================================================
# 14. Multivariate Fuzzy Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Multivariate Fuzzy Entropy measures fuzzy recurrence of joint multichannel delay
# vectors, extending Fuzzy Entropy from one coordinate to a synchronized vector process.
# =======================================================
def _entropy_method_014():
    y = _ent_y
    X = _ent_x[:, :min(4, _ent_x.shape[1])]
    n = min(len(X) - 3, _ENT_SPEED['patterns'])
    ids = np.linspace(0, len(X) - 4, n).astype(int)
    Z = np.hstack([X[ids + k] for k in range(2)])
    Z1 = np.hstack([X[ids + k] for k in range(3)])
    r = 0.25
    f = lambda A: np.mean(np.exp(-pdist(A, 'chebyshev') ** 2 / (r + ENT_EPS)))
    _ent_scalar_label = 'Multivariate Fuzzy Entropy'
    _ent_scalar_result = -np.log((f(Z1) + ENT_EPS) / (f(Z) + ENT_EPS))
    _show_scalar(_ent_scalar_label, _ent_scalar_result)
    return locals()
_ent_method_ns = _entropy_method_014()
_ent_auto_plot(14, 'Multivariate Fuzzy Entropy', _ent_method_ns, ['y', 'X', 'n', 'ids', 'Z', 'Z1', 'r', 'f', '_ent_scalar_label', '_ent_scalar_result', 'k'])
```

### 15. Permutation Entropy

Permutation Entropy is the Shannon entropy of Bandt-Pompe ordinal patterns. Normalization by ln(m!) maps the maximum possible entropy to one.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Permutation entropy replaces the exact amplitudes of short windows by their rank order. For example, a three-sample window is represented by which sample is smallest, middle, and largest. The probability distribution of these ordinal patterns is then summarized by Shannon entropy. The embedding order and delay determine which time scales and pattern lengths are emphasized.</p><p><b>How to read the mathematics.</b> A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently.
```python
# =======================================================
# 15. Permutation Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Permutation Entropy is the Shannon entropy of Bandt-Pompe ordinal patterns.
# Normalization by ln(m!) maps the maximum possible entropy to one.
# =======================================================
def _entropy_method_015():
    y = _ent_y
    _ent_scalar_label = 'Permutation Entropy'
    _ent_scalar_result = _perm_entropy(y)
    _show_scalar(_ent_scalar_label, _ent_scalar_result)
    return locals()
_ent_method_ns = _entropy_method_015()
_ent_auto_plot(15, 'Permutation Entropy', _ent_method_ns, ['y', '_ent_scalar_label', '_ent_scalar_result'])
```

### 19. Ordinal Pattern Entropy

Ordinal Pattern Entropy is Shannon entropy of the empirical distribution of rank-order permutations of embedded samples.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Shannon entropy summarizes uncertainty in a probability distribution. An entropy rate extends this idea to a time process and asks how much new uncertainty is produced per time step after accounting for the past.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> The visualization is a way of displaying already-computed numbers.
```python
# =======================================================
# 19. Ordinal Pattern Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Ordinal Pattern Entropy is Shannon entropy of the empirical distribution of rank-order
# permutations of embedded samples.
# =======================================================
def _entropy_method_019():
    y = _ent_y
    labs = _ordinal_labels(y, 4)
    p = _p_from_labels(labs)
    _ent_print_args = ['ordinal patterns=', len(p), 'entropy=', _H(p), 'normalized=', _H(p) / np.log(math.factorial(4))]
    print(*_ent_print_args)
    return locals()
_ent_method_ns = _entropy_method_019()
_ent_auto_plot(19, 'Ordinal Pattern Entropy', _ent_method_ns, ['y', 'labs', 'p', '_ent_print_args'])
```

### 22. Cross-Permutation Entropy

Cross-Permutation Entropy is implemented as conditional entropy H(P_Y|P_X)=H(P_X,P_Y)-H(P_X) of synchronized ordinal patterns.

```python
# =======================================================
# 23. Permutation Entropy asymmetry analysis
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Permutation asymmetry is quantified by Jensen-Shannon divergence between forward and
# time-reversed ordinal-pattern distributions; zero indicates ordinal time-reversal symmetry.
# =======================================================
def _entropy_method_023():
    y = _ent_y
    p1 = Counter(_ordinal_labels(y, 4))
    p2 = Counter(_ordinal_labels(y[::-1], 4))
    keys = sorted(set(p1) | set(p2))
    a = np.array([p1[k] for k in keys], float)
    b = np.array([p2[k] for k in keys], float)
    a /= a.sum()
    b /= b.sum()
    m = 0.5 * (a + b)
    js = 0.5 * np.sum(a * np.log((a + ENT_EPS) / (m + ENT_EPS))) + 0.5 * np.sum(b * np.log((b + ENT_EPS) / (m + ENT_EPS)))
    _ent_scalar_label = 'Permutation time-reversal asymmetry (JSD)'
    _ent_scalar_result = js
    _show_scalar(_ent_scalar_label, _ent_scalar_result)
    return locals()
_ent_method_ns = _entropy_method_023()
_ent_auto_plot(23, 'Permutation Entropy asymmetry analysis', _ent_method_ns, ['y', 'p1', 'p2', 'keys', 'a', 'b', 'm', 'js', '_ent_scalar_label', '_ent_scalar_result', 'k'])
```

```python
# =======================================================
# 31. Cross-Dispersion Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Cross-Dispersion Entropy forms synchronized pairs of dispersion symbols from two
# channels and computes normalized joint Shannon entropy.
# =======================================================
def _entropy_method_031():
    y = _ent_y
    a, b = (_disp_classes(_ent_x[:, 0]), _disp_classes(_ent_x[:, 1]))
    n = min(len(a), len(b))
    pairs = np.column_stack([a[:n], b[:n]])
    _ent_scalar_label = 'Cross-Dispersion Entropy'
    _ent_scalar_result = _H(_p_from_labels(pairs)) / np.log(36)
    _show_scalar(_ent_scalar_label, _ent_scalar_result)
    return locals()
_ent_method_ns = _entropy_method_031()
_ent_auto_plot(31, 'Cross-Dispersion Entropy', _ent_method_ns, ['y', 'n', 'pairs', '_ent_scalar_label', '_ent_scalar_result', 'a', 'b'])
```

### 43. Differential entropy

Differential entropy for a histogram density f is approximated by -sum p_i ln f_i, including bin-width information rather than treating bins as dimensionless symbols.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# =======================================================
# 43. Differential entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Differential entropy for a histogram density f is approximated by -sum p_i ln f_i,
# including bin-width information rather than treating bins as dimensionless symbols.
# =======================================================
def _entropy_method_043():
    y = _ent_y
    p = _hist_p(y, bins=24)
    h, e = np.histogram(y, bins=32, density=True)
    w = np.diff(e)
    q = h * w
    q = q[q > 0]
    _ent_scalar_label = 'Histogram differential entropy'
    _ent_scalar_result = -np.sum(q * np.log(h[h > 0] + ENT_EPS))
    _show_scalar(_ent_scalar_label, _ent_scalar_result)
    return locals()
_ent_method_ns = _entropy_method_043()
_ent_auto_plot(43, 'Differential entropy', _ent_method_ns, ['y', 'p', 'w', 'q', '_ent_scalar_label', '_ent_scalar_result', 'h', 'e'])
```

### 44. kNN differential entropy

The k-nearest-neighbor differential-entropy estimator uses neighbor radii rather than fixed bins; in one dimension the Kozachenko-Leonenko form is psi(N)-psi(k)+ln 2 + mean ln epsilon_i.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# =======================================================
# 44. kNN differential entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: The k-nearest-neighbor differential-entropy estimator uses neighbor radii rather than
# fixed bins; in one dimension the Kozachenko-Leonenko form is psi(N)-psi(k)+ln 2 + mean ln epsilon_i.
# =======================================================
def _entropy_method_044():
    y = _ent_y
    p = _hist_p(y, bins=24)
    z = _cap(y)
    n = len(z)
    k = 3
    tree = cKDTree(z[:, None])
    d = tree.query(z[:, None], k=k + 1)[0][:, -1]
    val = special.digamma(n) - special.digamma(k) + np.log(2) + np.mean(np.log(d + ENT_EPS))
    _ent_scalar_label = 'kNN differential entropy'
    _ent_scalar_result = val
    _show_scalar(_ent_scalar_label, _ent_scalar_result)
    return locals()
_ent_method_ns = _entropy_method_044()
_ent_auto_plot(44, 'kNN differential entropy', _ent_method_ns, ['y', 'p', 'z', 'n', 'k', 'tree', 'd', 'val', '_ent_scalar_label', '_ent_scalar_result'])
```

### 49. State-occupancy entropy

State-occupancy entropy discretizes a finite state description (amplitude and local derivative) and computes Shannon entropy of the occupied states.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Shannon entropy summarizes uncertainty in a probability distribution. An entropy rate extends this idea to a time process and asks how much new uncertainty is produced per time step after accounting for the past.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# =======================================================
# 49. State-occupancy entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: State-occupancy entropy discretizes a finite state description (amplitude and local
# derivative) and computes Shannon entropy of the occupied states.
# =======================================================
def _entropy_method_049():
    y = _ent_y
    p = _hist_p(y, bins=24)
    dy = np.gradient(y)
    a = np.digitize(y, np.quantile(y, [0.25, 0.5, 0.75]))
    b = np.digitize(dy, np.quantile(dy, [0.25, 0.5, 0.75]))
    _ent_scalar_label = 'State-occupancy entropy'
    _ent_scalar_result = _H(_p_from_labels(np.column_stack([a, b]))) / np.log(16)
    _show_scalar(_ent_scalar_label, _ent_scalar_result)
    return locals()
_ent_method_ns = _entropy_method_049()
_ent_auto_plot(49, 'State-occupancy entropy', _ent_method_ns, ['y', 'p', 'dy', 'a', 'b', '_ent_scalar_label', '_ent_scalar_result'])
```

```python
# =======================================================
# 50. Regional state-occupancy entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Regional state-occupancy entropy applies the same finite-state Shannon measure to
# region-averaged EEG signals.
# =======================================================
def _entropy_method_050():
    y = _ent_y
    p = _hist_p(y, bins=24)
    regs = {'frontal': range(0, min(12, _ent_x.shape[1])), 'central': range(min(12, _ent_x.shape[1]), min(23, _ent_x.shape[1])), 'posterior': range(min(23, _ent_x.shape[1]), _ent_x.shape[1])}
    rows = []
    for name, inds in regs.items():
        inds = list(inds)
        if inds:
            s = np.mean(_ent_x[:, inds], axis=1)
            rows.append([name, _H(_hist_p(s, 16, quantile=True))])
    _ent_display_result = pd.DataFrame(rows, columns=['region', 'occupancy_entropy'])
    display(_ent_display_result)
    return locals()
_ent_method_ns = _entropy_method_050()
_ent_auto_plot(50, 'Regional state-occupancy entropy', _ent_method_ns, ['y', 'p', 'regs', 'rows', '_ent_display_result', 'name', 'inds', 's'])
```

### 61. Escort-distribution entropy

The q-escort distribution is P_i^(q)=p_i^q/sum_j p_j^q. This cell reports Shannon entropy of that reweighted distribution.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Shannon entropy summarizes uncertainty in a probability distribution. Each outcome contributes according to both its probability and the information associated with that probability, so evenly distributed outcomes give larger entropy than a distribution concentrated on a few outcomes. An entropy rate extends this idea to a time process and asks how much new uncertainty is produced per time step after accounting for the past.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# =======================================================
# 61. Escort-distribution entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: The q-escort distribution is P_i^(q)=p_i^q/sum_j p_j^q. This cell reports Shannon
# entropy of that reweighted distribution.
# =======================================================
def _entropy_method_061():
    y = _ent_y
    p = _hist_p(y, bins=24)
    q = 2.0
    e = p ** q / np.sum(p ** q)
    _ent_scalar_label = 'Escort-distribution Shannon entropy q=2'
    _ent_scalar_result = _H(e)
    _show_scalar(_ent_scalar_label, _ent_scalar_result)
    return locals()
_ent_method_ns = _entropy_method_061()
_ent_auto_plot(61, 'Escort-distribution entropy', _ent_method_ns, ['y', 'p', 'q', 'e', '_ent_scalar_label', '_ent_scalar_result'])
```

<h3 style="color:#E8E8E8;margin-top:0;">Theory and role — multiscale and decomposition helpers for methods 65–99</h3>
<p>This helper cell supplies coarse-graining, wavelet/time-frequency, and adaptive-decomposition utilities. Multiscale entropy evaluates an entropy functional after scale-dependent coarse graining; wavelet entropy normalizes scale energies \(p_s=E_s/\sum_rE_r\) and uses \(-\sum_s p_s\log p_s\). EMD/EEMD/VMD branches decompose the signal before applying the stated entropy functional. The speed mode changes scale, ensemble, or iteration budgets only.</p>
</div>

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Multiscale entropy asks how signal irregularity changes when the data are viewed at progressively coarser time scales. At scale one the original samples are analyzed; at larger scales, groups of samples are combined to create a coarse-grained series before an entropy statistic is computed. A signal can therefore look irregular at short scales but more predictable at long scales, or vice versa. Because coarse-graining shortens the series, estimates at large scales generally have greater uncertainty.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# =======================================================
# Shared primitives only; numbered cells below contain the method-specific mathematics.
# SHARED NUMERICAL PRIMITIVES — methods 65–99
# Multiscale, wavelet/time-frequency, EMD/EEMD/VMD entropy.
# =======================================================
def _sampen_stats(y,m=2,r=.2):
    y=np.asarray(y,float);rr=max(ENT_EPS,r*np.std(y));Xm=_embed(y,m);X1=_embed(y,m+1);B=_match_pairs(Xm,rr);A=_match_pairs(X1,rr)
    return A,max(1,len(X1)*(len(X1)-1)/2),B,max(1,len(Xm)*(len(Xm)-1)/2)

def _moving_coarse(y,s):
    y=np.asarray(y,float);s=int(s)
    if s<=1:return y.copy()
    return np.convolve(y,np.ones(s)/s,mode='valid')

def _variance_coarse(y,s):
    y=np.asarray(y,float);s=int(s);n=len(y)//s
    return np.var(y[:n*s].reshape(n,s),axis=1) if n else np.empty(0)

def _hier_nodes(y,levels=4):
    nodes=[np.asarray(y,float)]
    level=[np.asarray(y,float)]
    for _ in range(levels):
        nxt=[]
        for z in level:
            n=(len(z)//2)*2;z=z[:n]
            if n<4:continue
            a=(z[0::2]+z[1::2])/np.sqrt(2);d=(z[0::2]-z[1::2])/np.sqrt(2);nxt.extend([a,d])
        nodes.extend(nxt);level=nxt
        if not level:break
    return nodes

def _morlet(y,freqs=None,n_cycles=7.):
    y=np.asarray(y,float)
    if freqs is None:freqs=np.geomspace(2,min(45,.42*_ent_fs),24 if FAST_OPTION=='ultra' else 36)
    C=[]
    for f in freqs:
        sig=n_cycles/(2*np.pi*f);half=int(max(8,4*sig*_ent_fs));tt=np.arange(-half,half+1)/_ent_fs
        w=np.exp(2j*np.pi*f*tt)*np.exp(-(tt**2)/(2*sig**2));w/=np.sqrt(np.sum(np.abs(w)**2)+ENT_EPS);C.append(signal.fftconvolve(y,np.conj(w[::-1]),mode='same'))
    return np.asarray(freqs),np.asarray(C)

def _haar_packet(y,level=4):
    level_data=[np.asarray(y,float)]
    for _ in range(level):
        nxt=[]
        for z in level_data:
            n=(len(z)//2)*2;z=z[:n]
            if n<2:continue
            nxt.extend([(z[0::2]+z[1::2])/np.sqrt(2),(z[0::2]-z[1::2])/np.sqrt(2)])
        level_data=nxt
    return level_data

def _haar_details(y,levels=5):
    a=np.asarray(y,float);out=[]
    for _ in range(levels):
        n=(len(a)//2)*2;a=a[:n]
        if n<2:break
        out.append((a[0::2]-a[1::2])/np.sqrt(2));a=(a[0::2]+a[1::2])/np.sqrt(2)
    out.append(a);return out

def _energy_entropy(parts):
    E=np.array([np.sum(np.abs(z)**2) for z in parts],float);p=E/(E.sum()+ENT_EPS);return _H(p),p

def _st_entropy(y,nper=None):
    nper=int(nper or min(256,max(64,len(y)//8)));nover=nper//2;f,t,Z=signal.stft(y,fs=_ent_fs,nperseg=nper,noverlap=nover,boundary=None);P=np.abs(Z)**2;P/=P.sum()+ENT_EPS;return _H(P.ravel()),f,t,P

def _stockwell(y,freqs=None):
    y=np.asarray(y,float);freqs=np.asarray(freqs if freqs is not None else np.geomspace(2,min(45,.42*_ent_fs),20),float);S=[]
    for f in freqs:
        sig=1/(2*np.pi*f);half=int(max(8,4*sig*_ent_fs));tt=np.arange(-half,half+1)/_ent_fs
        g=(f/np.sqrt(2*np.pi))*np.exp(-.5*(f*tt)**2)*np.exp(-2j*np.pi*f*tt);S.append(signal.fftconvolve(y,g,mode='same'))
    return freqs,np.asarray(S)

def _emd_simple(y,max_imfs=None,max_sift=None):
    max_imfs=int(max_imfs or {'full':7,'balanced':6,'fast':5,'ultra':4}[FAST_OPTION]);max_sift=int(max_sift or {'full':30,'balanced':22,'fast':16,'ultra':10}[FAST_OPTION]);r=np.asarray(y,float).copy();imfs=[]
    for _ in range(max_imfs):
        h=r.copy()
        for _s in range(max_sift):
            ma=signal.find_peaks(h)[0];mi=signal.find_peaks(-h)[0]
            if len(ma)<2 or len(mi)<2:break
            x=np.arange(len(h));up=np.interp(x,ma,h[ma]);lo=np.interp(x,mi,h[mi]);mean=(up+lo)/2;hn=h-mean
            if np.linalg.norm(hn-h)/(np.linalg.norm(h)+ENT_EPS)<.05:h=hn;break
            h=hn
        if len(signal.find_peaks(h)[0])<2:break
        imfs.append(h);r=r-h
        if np.std(r)<1e-4*np.std(y):break
    return imfs,r

def _eemd_energy(y):
    rng=np.random.default_rng(94);ens={'full':20,'balanced':12,'fast':7,'ultra':4}[FAST_OPTION];allE=[]
    for _ in range(ens):
        yy=np.asarray(y,float)+.15*np.std(y)*rng.standard_normal(len(y));imfs,_=_emd_simple(yy);allE.append([np.sum(z*z) for z in imfs])
    K=max(map(len,allE),default=0);E=np.array([np.pad(v,(0,K-len(v))) for v in allE]).mean(0) if K else np.array([]);p=E/(E.sum()+ENT_EPS);return _H(p),p

def _vmd_simple(y,K=4,alpha=2000,tau=0.):
    y=np.asarray(y,float);n=min(len(y),4096 if FAST_OPTION=='ultra' else 8192);y=y[:n];it={'full':160,'balanced':120,'fast':80,'ultra':45}[FAST_OPTION]
    f=np.fft.fftfreq(n);Y=np.fft.fft(y);u=np.zeros((K,n),complex);omega=np.linspace(0,.4,K+2)[1:-1];lam=np.zeros(n,complex);pos=f>=0
    for _ in range(it):
        us=np.sum(u,axis=0)
        for k in range(K):
            rest=us-u[k];new=(Y-rest+lam/2)/(1+2*alpha*(f-omega[k])**2);us=rest+new;u[k]=new;pw=np.abs(new[pos])**2;omega[k]=np.sum(f[pos]*pw)/(np.sum(pw)+ENT_EPS)
        if tau:lam+=tau*(Y-us)
    return np.real(np.fft.ifft(u,axis=1)),omega*_ent_fs
```

A signal can therefore look irregular at short scales but more predictable at long scales, or vice versa. Because coarse-graining shortens the series, estimates at large scales generally have greater uncertainty.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless.
```python
# =======================================================
# 65. Composite Multiscale Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Composite Multiscale Entropy averages Sample Entropy over all coarse-graining offsets
# at each scale, reducing dependence on the arbitrary block origin.
# =======================================================
def _entropy_method_065():
    y = _ent_y
    sc = np.arange(1, _ENT_SPEED['scales'] + 1)
    v = []
    for s in sc:
        z = [_sampen(_coarse(y, int(s), o)) for o in range(int(s)) if len(_coarse(y, int(s), o)) > 30]
        v.append(np.nanmean(z))
    _line(sc, v, 'Composite Multiscale Entropy', 'scale', 'CMSE')
    return locals()
_ent_method_ns = _entropy_method_065()
```

### 66. Refined Composite Multiscale Entropy

Refined Composite Multiscale Entropy pools matching statistics across offsets before taking the logarithm, rather than averaging already-log-transformed SampEn values.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Multiscale entropy asks how signal irregularity changes when the data are viewed at progressively coarser time scales. Because coarse-graining shortens the series, estimates at large scales generally have greater uncertainty.</p><p><b>How to read the mathematics.</b> A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# =======================================================
# 66. Refined Composite Multiscale Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Refined Composite Multiscale Entropy pools matching statistics across offsets before
# taking the logarithm, rather than averaging already-log-transformed SampEn values.
# =======================================================
def _entropy_method_066():
    y = _ent_y
    sc = np.arange(1, _ENT_SPEED['scales'] + 1)
    v = []
    for s in sc:
        AA = BA = 0.0
        Ad = Bd = 0.0
        for o in range(int(s)):
            z = _coarse(y, int(s), o)
            if len(z) > 30:
                A, da, B, db = _sampen_stats(z)
                AA += A
                Ad += da
                BA += B
                Bd += db
        pA = AA / max(1, Ad)
        pB = BA / max(1, Bd)
        v.append(-np.log((pA + ENT_EPS) / (pB + ENT_EPS)))
    _line(sc, v, 'Refined Composite Multiscale Entropy', 'scale', 'RCMSE')
    return locals()
_ent_method_ns = _entropy_method_066()
```

```python
# =======================================================
# 70. Hierarchical Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Hierarchical Entropy recursively splits a signal into Haar-like average and difference
# branches and measures entropy on every node of the resulting multiresolution tree.
# =======================================================
def _entropy_method_070():
    y = _ent_y
    sc = np.arange(1, _ENT_SPEED['scales'] + 1)
    nodes = _hier_nodes(y, levels=3)
    vals = [_H(_hist_p(z, 16)) for z in nodes]
    _ent_display_result = pd.DataFrame({'node': range(len(nodes)), 'Shannon_entropy': vals})
    display(_ent_display_result)
    return locals()
_ent_method_ns = _entropy_method_070()
_ent_auto_plot(70, 'Hierarchical Entropy', _ent_method_ns, ['y', 'sc', 'nodes', 'vals', '_ent_display_result', 'z'])
```

### 71. Hierarchical Sample Entropy

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Sample entropy quantifies how often similar short patterns remain similar when one additional sample is included. ### 72. Multivariate Multiscale Entropy

Multivariate Multiscale Entropy coarse-grains synchronized channels jointly and computes Sample Entropy in the resulting multivariate delay space.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Multiscale entropy asks how signal irregularity changes when the data are viewed at progressively coarser time scales. At scale one the original samples are analyzed; at larger scales, groups of samples are combined to create a coarse-grained series before an entropy statistic is computed. A signal can therefore look irregular at short scales but more predictable at long scales, or vice versa. Because coarse-graining shortens the series, estimates at large scales generally have greater uncertainty.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values.</p></div>


```python
# =======================================================
# 72. Multivariate Multiscale Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Multivariate Multiscale Entropy coarse-grains synchronized channels jointly and
# computes Sample Entropy in the resulting multivariate delay space.
# =======================================================
def _entropy_method_072():
    y = _ent_y
    sc = np.arange(1, _ENT_SPEED['scales'] + 1)
    X = _ent_x[:, :min(4, _ent_x.shape[1])]
    v = []
    for s in sc:
        n = len(X) // s
        cg = X[:n * s].reshape(n, s, X.shape[1]).mean(1)
        v.append(_multivar_sampen(cg, 2, 0.25))
    _line(sc, v, 'Multivariate Multiscale Entropy', 'scale', 'MvMSE')
    return locals()
_ent_method_ns = _entropy_method_072()
```

```python
# =======================================================
# 73. Multichannel Multiscale Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Multichannel Multiscale Entropy reports the mean of channel-wise multiscale Sample
# Entropy, which is distinct from joint-state multivariate MSE.
# =======================================================
def _entropy_method_073():
    y = _ent_y
    sc = np.arange(1, _ENT_SPEED['scales'] + 1)
    ch = min(6, _ent_x.shape[1])
    v = []
    for s in sc:
        v.append(np.nanmean([_sampen(_coarse(_ent_x[:, j], int(s))) for j in range(ch)]))
    _line(sc, v, 'Multichannel Multiscale Entropy', 'scale', 'mean channel SampEn')
    return locals()
_ent_method_ns = _entropy_method_073()
```

### 74. Multiscale Cross-Entropy

### 75. Multiscale Joint Entropy

Multiscale Joint Entropy is Shannon entropy H(X_s,Y_s) of synchronized quantized coarse-grained channel states at each scale.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Shannon entropy summarizes uncertainty in a probability distribution. Each outcome contributes according to both its probability and the information associated with that probability, so evenly distributed outcomes give larger entropy than a distribution concentrated on a few outcomes. An entropy rate extends this idea to a time process and asks how much new uncertainty is produced per time step after accounting for the past.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values.</p></div>


```python
# =======================================================
# 75. Multiscale Joint Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Multiscale Joint Entropy is Shannon entropy H(X_s,Y_s) of synchronized quantized
# coarse-grained channel states at each scale.
# =======================================================
def _entropy_method_075():
    y = _ent_y
    sc = np.arange(1, _ENT_SPEED['scales'] + 1)
    vals = []
    for s in sc:
        a = _coarse(_ent_x[:, 0], int(s))
        b = _coarse(_ent_x[:, 1], int(s))
        n = min(len(a), len(b))
        qa = np.digitize(a[:n], np.quantile(a[:n], [0.25, 0.5, 0.75]))
        qb = np.digitize(b[:n], np.quantile(b[:n], [0.25, 0.5, 0.75]))
        Hxy = _H(_p_from_labels(np.column_stack([qa, qb])))
        Hx = _H(_p_from_labels(qa))
        vals.append(Hxy)
    _line(sc, vals, 'Multiscale Joint Entropy', 'scale', 'entropy')
    return locals()
_ent_method_ns = _entropy_method_075()
```

### 76. Multiscale Conditional Entropy

Multiscale Conditional Entropy uses H(Y_s|X_s)=H(X_s,Y_s)-H(X_s) on the coarse-grained symbolic states.

```python
# =======================================================
# 76. Multiscale Conditional Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Multiscale Conditional Entropy uses H(Y_s|X_s)=H(X_s,Y_s)-H(X_s) on the coarse-grained
# symbolic states.
# =======================================================
def _entropy_method_076():
    y = _ent_y
    sc = np.arange(1, _ENT_SPEED['scales'] + 1)
    vals = []
    for s in sc:
        a = _coarse(_ent_x[:, 0], int(s))
        b = _coarse(_ent_x[:, 1], int(s))
        n = min(len(a), len(b))
        qa = np.digitize(a[:n], np.quantile(a[:n], [0.25, 0.5, 0.75]))
        qb = np.digitize(b[:n], np.quantile(b[:n], [0.25, 0.5, 0.75]))
        Hxy = _H(_p_from_labels(np.column_stack([qa, qb])))
        Hx = _H(_p_from_labels(qa))
        vals.append(Hxy - Hx)
    _line(sc, vals, 'Multiscale Conditional Entropy H(Y|X)', 'scale', 'entropy')
    return locals()
_ent_method_ns = _entropy_method_076()
```

### 80. Wavelet Singular Entropy

Wavelet Singular Entropy forms a time-scale coefficient matrix, normalizes its singular values, and computes Shannon entropy of that singular-value spectrum.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Shannon entropy summarizes uncertainty in a probability distribution.
```python
# =======================================================
# 80. Wavelet Singular Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Wavelet Singular Entropy forms a time-scale coefficient matrix, normalizes its singular
# values, and computes Shannon entropy of that singular-value spectrum.
# =======================================================
def _entropy_method_080():
    y = _ent_y
    sc = np.arange(1, _ENT_SPEED['scales'] + 1)
    f, C = _morlet(_cap(y, 2500))
    s = np.linalg.svd(np.abs(C), compute_uv=False)
    p = s / (s.sum() + ENT_EPS)
    _ent_print_args = ['Wavelet Singular Entropy=', _H(p)]
    print(*_ent_print_args)
    return locals()
_ent_method_ns = _entropy_method_080()
_ent_auto_plot(80, 'Wavelet Singular Entropy', _ent_method_ns, ['y', 'sc', 's', 'p', '_ent_print_args', 'f', 'C'])
```

### 82. CWT Scalogram Entropy

CWT Scalogram Entropy normalizes the Morlet scalogram energy over time and scale and computes Shannon entropy of that 2-D distribution.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Shannon entropy summarizes uncertainty in a probability distribution. ### 83. Wavelet Coherence Entropy

Wavelet Coherence Entropy normalizes the squared wavelet-coherence field R^2=|S(Wxy)|^2/[S(|Wx|^2)S(|Wy|^2)] and computes entropy of its spatial distribution.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Coherence is a frequency-specific measure of linear association between two signals. It compares the magnitude of their cross-spectrum with the power spectra of the individual signals. Values near one mean that a stable linear relationship at that frequency explains a large fraction of the spectral activity, while values near zero indicate weak consistency. High coherence can arise from common input, volume conduction, or shared reference effects, so it should not automatically be interpreted as direct connectivity.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# =======================================================
# 83. Wavelet Coherence Entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Wavelet Coherence Entropy normalizes the squared wavelet-coherence field
# R^2=|S(Wxy)|^2/[S(|Wx|^2)S(|Wy|^2)] and computes entropy of its spatial distribution.
# =======================================================
def _entropy_method_083():
    y = _ent_y
    sc = np.arange(1, _ENT_SPEED['scales'] + 1)
    f, C1 = _morlet(_cap(_ent_x[:, 0], 2500))
    _, C2 = _morlet(_cap(_ent_x[:, 1], 2500), f)
    S12 = signal.convolve2d(C1 * np.conj(C2), np.ones((3, 9)) / 27, mode='same', boundary='symm')
    S1 = signal.convolve2d(np.abs(C1) ** 2, np.ones((3, 9)) / 27, mode='same', boundary='symm')
    S2 = signal.convolve2d(np.abs(C2) ** 2, np.ones((3, 9)) / 27, mode='same', boundary='symm')
    coh = np.abs(S12) ** 2 / (S1 * S2 + ENT_EPS)
    p = coh.ravel()
    p = p / (p.sum() + ENT_EPS)
    _ent_print_args = ['Wavelet coherence entropy=', _H(p)]
    print(*_ent_print_args)
    return locals()
_ent_method_ns = _entropy_method_083()
_ent_auto_plot(83, 'Wavelet Coherence Entropy', _ent_method_ns, ['y', 'sc', 'S12', 'S1', 'S2', 'coh', 'p', '_ent_print_args', 'f', 'C1', '_', 'C2'])
```

### 86. Spectrogram entropy trajectories

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Shannon entropy summarizes uncertainty in a probability distribution. ### 89. Rényi spectral entropy

Rényi spectral entropy applies Rényi H_q to normalized power-spectral density rather than amplitude-state probabilities.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> A power spectral density describes how signal power is distributed over frequency. The area under the PSD over a frequency band corresponds to power in that band, subject to the normalization convention used by the implementation.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless.
```python
# =======================================================
# 89. Rényi spectral entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Rényi spectral entropy applies Rényi H_q to normalized power-spectral density rather
# than amplitude-state probabilities.
# =======================================================
def _entropy_method_089():
    y = _ent_y
    sc = np.arange(1, _ENT_SPEED['scales'] + 1)
    f, P = signal.welch(y, fs=_ent_fs, nperseg=min(1024, len(y)))
    p = P / (P.sum() + ENT_EPS)
    q = 2.0
    _ent_print_args = ['Rényi', 'spectral entropy=', _renyi(p, q)]
    print(*_ent_print_args)
    return locals()
_ent_method_ns = _entropy_method_089()
_ent_auto_plot(89, 'Rényi spectral entropy', _ent_method_ns, ['y', 'sc', 'p', 'q', '_ent_print_args', 'f', 'P'])
```

### 90. Tsallis spectral entropy

### 91. S-transform entropy

```python
# =======================================================
# 91. S-transform entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: The Stockwell transform uses a frequency-dependent Gaussian window; S-transform entropy
# is Shannon entropy of its normalized time-frequency energy.
# =======================================================
def _entropy_method_091():
    y = _ent_y
    sc = np.arange(1, _ENT_SPEED['scales'] + 1)
    f, S = _stockwell(_cap(y, 3000))
    P = np.abs(S) ** 2
    P /= P.sum() + ENT_EPS
    _ent_print_args = ['S-transform energy entropy=', _H(P.ravel())]
    print(*_ent_print_args)
    return locals()
_ent_method_ns = _entropy_method_091()
_ent_auto_plot(91, 'S-transform entropy', _ent_method_ns, ['y', 'sc', 'P', '_ent_print_args', 'f', 'S'])
```

### 92. Hilbert-Huang entropy

Hilbert-Huang entropy decomposes the signal into intrinsic mode functions, obtains analytic-signal instantaneous energies, and measures entropy across modes as a function of time.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# =======================================================
# 92. Hilbert-Huang entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Hilbert-Huang entropy decomposes the signal into intrinsic mode functions, obtains
# analytic-signal instantaneous energies, and measures entropy across modes as a function of time.
# =======================================================
def _entropy_method_092():
    y = _ent_y
    sc = np.arange(1, _ENT_SPEED['scales'] + 1)
    imfs, res = _emd_simple(_cap(y, 3000))
    if not imfs:
        _ent_print_args = ['no IMFs resolved']
        print(*_ent_print_args)
        return locals()
    H = []
    for z in imfs:
        a = signal.hilbert(z)
        H.append(np.abs(a) ** 2)
    E = np.vstack(H)
    p = E / (E.sum(0, keepdims=True) + ENT_EPS)
    ht = np.array([-np.sum(p[:, k] * np.log(p[:, k] + ENT_EPS)) for k in range(p.shape[1])])
    _ent_print_args = ['Hilbert-Huang mean instantaneous mode entropy=', float(np.mean(ht))]
    print(*_ent_print_args)
    return locals()
_ent_method_ns = _entropy_method_092()
_ent_auto_plot(92, 'Hilbert-Huang entropy', _ent_method_ns, ['y', 'sc', 'H', 'z', 'E', 'p', 'ht', '_ent_print_args', 'imfs', 'res', 'a', 'k'])
```

### 97. Instantaneous-frequency entropy

<h3 style="color:#E8E8E8;margin-top:0;">Theory and role — matrix, algorithmic, and entropy-rate helpers for methods 100–130</h3>
<p>This foundation defines trajectory/SVD, covariance-spectrum, compression/complexity, and entropy-rate utilities. For normalized singular/eigenvalue weights \(p_i\), spectral entropy is \(-\sum_i p_i\log p_i\) and effective rank is \(\exp(H)\). Finite symbolic entropy-rate estimates use block entropies, e.g. \(h_L=H(S_1^L)-H(S_1^{L-1})\), rather than claiming the infinite-length limit from finite EEG.</p>
</div>

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Spectral entropy treats the normalized power spectrum as a probability distribution over frequency bins and applies Shannon entropy to that distribution. If power is concentrated in only a few frequencies, the distribution is uneven and entropy is low. If power is spread broadly across many frequencies, entropy is higher. The result therefore describes spectral dispersion, not temporal unpredictability in exactly the same sense as Sample Entropy or permutation entropy.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. An exponential describes multiplicative growth or decay. In dynamical systems, a positive exponent corresponds to growth with time and a negative exponent to decay.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values. The visualization is a way of displaying already-computed numbers. Color, ordering, or ranking can make patterns easier to see but does not create additional statistical evidence beyond the values and uncertainty measures being plotted.</p></div>


```python
# =======================================================
# Shared primitives only; numbered cells below contain the method-specific mathematics.
# SHARED NUMERICAL PRIMITIVES — methods 100–130
# Matrix, algorithmic, entropy-rate and predictive-information measures.
# =======================================================
import zlib
from scipy.special import gammaln, logsumexp

def _trajectory_matrix(y,m=20):
    X=_embed(y,m,1);return X-X.mean(0,keepdims=True)

def _eig_probs(M):
    w=np.linalg.eigvalsh((M+M.T)/2);w=np.clip(w,0,None);return w/(w.sum()+ENT_EPS)

def _svd_entropy(A):
    s=np.linalg.svd(np.asarray(A,float),compute_uv=False);p=s*s/(np.sum(s*s)+ENT_EPS);return _H(p),p

def _symbolize(y,k=4):
    y=np.asarray(y,float);e=np.unique(np.quantile(y,np.linspace(0,1,k+1)))
    if len(e)<2:return np.zeros(len(y),dtype=int)
    return np.clip(np.digitize(y,e[1:-1]),0,len(e)-2).astype(int)

def _lz76(seq):
    # Exhaustive LZ76 phrase parsing; returns phrase count.
    s=tuple(seq);n=len(s);i=0;c=0;dictionary=set()
    while i<n:
        L=1
        while i+L<=n and tuple(s[i:i+L]) in dictionary:L+=1
        phrase=tuple(s[i:min(n,i+L)]);dictionary.add(phrase);c+=1;i+=len(phrase)
    return c

def _lz_norm(seq,alphabet=None):
    seq=list(seq);n=len(seq)
    if n<2:return 0.0
    a=int(alphabet or max(2,len(set(seq))));return float(_lz76(seq)*np.log(n)/(n*np.log(a)))

def _to_bytes_symbols(seq):
    a=np.asarray(seq,int);return bytes(np.mod(a,256).astype(np.uint8).tolist())

def _comp_bits(seq):
    return 8*len(zlib.compress(_to_bytes_symbols(seq),9))

def _ncd(a,b):
    A=_to_bytes_symbols(a);B=_to_bytes_symbols(b);ca=len(zlib.compress(A,9));cb=len(zlib.compress(B,9));cab=len(zlib.compress(A+B,9));return float((cab-min(ca,cb))/max(ca,cb,1))

def _bdm_proxy(seq,block=12):
    # BDM structure: sum_j [K(block_j)+log2 multiplicity_j]; K is approximated by compressed block bits because no CTM table is bundled.
    s=list(seq);blocks=[tuple(s[i:i+block]) for i in range(0,len(s)-block+1,block)];C=Counter(blocks);tot=0.
    for b,n in C.items():tot+=_comp_bits(b)+np.log2(n)
    return float(tot)

def _kt_logprob(n0,n1):
    return float(gammaln(n0+.5)+gammaln(n1+.5)-gammaln(n0+n1+1)-2*gammaln(.5)+gammaln(1.0))

def _ctw_logprob(bits,depth=5):
    bits=np.asarray(bits,int).ravel();depth=int(depth)
    # Each node stores counts of target symbols seen under a particular context suffix.
    tree={():[0,0]}
    for t,b in enumerate(bits):
        tree.setdefault((),[0,0])[int(b)]+=1
        ctx=[]
        for d in range(1,min(depth,t)+1):
            ctx.append(int(bits[t-d]));key=tuple(ctx);tree.setdefault(key,[0,0])[int(b)]+=1
    from functools import lru_cache
    @lru_cache(None)
    def rec(key):
        n0,n1=tree.get(key,[0,0]);lk=_kt_logprob(n0,n1)
        if len(key)>=depth:return lk
        c0=key+(0,);c1=key+(1,);lc=rec(c0)+rec(c1)
        return float(logsumexp([np.log(.5)+lk,np.log(.5)+lc]))
    return rec(())

def _block_entropy(sym,L):
    sym=np.asarray(sym,int);n=len(sym)-L+1
    if n<=0:return np.nan
    X=sliding_window_view(sym,L);return _H(_p_from_labels(X))

def _cond_entropy_markov(sym,order=1):
    sym=np.asarray(sym,int);return float(_block_entropy(sym,order+1)-_block_entropy(sym,order))

def _joint_symbol_entropy(a,b):
    n=min(len(a),len(b));return _H(_p_from_labels(np.column_stack([a[:n],b[:n]])))

def _mi_symbols(a,b):
    n=min(len(a),len(b));a=np.asarray(a)[:n];b=np.asarray(b)[:n];return float(_H(_p_from_labels(a))+_H(_p_from_labels(b))-_joint_symbol_entropy(a,b))

def _causal_states(sym,L=3,decimals=2):
    sym=np.asarray(sym,int);K=int(sym.max()+1);counts={}
    for t in range(L,len(sym)):
        past=tuple(sym[t-L:t]);counts.setdefault(past,np.zeros(K))[sym[t]]+=1
    states={};past_state={}
    for past,c in counts.items():
        p=(c+.5)/(c.sum()+.5*K);key=tuple(np.round(p,decimals));
        if key not in states:states[key]=len(states)
        past_state[past]=states[key]
    seq=[]
    for t in range(L,len(sym)):
        p=tuple(sym[t-L:t]);seq.append(past_state.get(p,-1))
    return np.asarray(seq,int),past_state,states,counts
```

### 102. Covariance eigenvalue entropy

Covariance eigenvalue entropy normalizes the eigenvalues of the channel covariance matrix and measures how broadly variance is distributed over covariance modes.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> A Jacobian matrix is the local linear approximation of how small changes in each state variable affect the next state. Its eigenvalues and eigenvectors summarize locally preferred directions of growth, decay, or rotation. For a discrete-time map, eigenvalue magnitude below one indicates local contraction and magnitude above one indicates local expansion; for continuous-time systems, the sign of the real part is the corresponding stability indicator. When the Jacobian is estimated from data, these conclusions are local and depend on the quality of the regression or neighborhood used to estimate it.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values.</p></div>


```python
# =======================================================
# 102. Covariance eigenvalue entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Covariance eigenvalue entropy normalizes the eigenvalues of the channel covariance
# matrix and measures how broadly variance is distributed over covariance modes.
# =======================================================
def _entropy_method_102():
    y = _ent_y
    X = _ent_x
    C = np.cov(_ent_raw[:min(len(_ent_raw), 20000)], rowvar=False)
    p = _eig_probs(C)
    _ent_print_args = ['Covariance eigenvalue entropy=', _H(p)]
    print(*_ent_print_args)
    return locals()
_ent_method_ns = _entropy_method_102()
_ent_auto_plot(102, 'Covariance eigenvalue entropy', _ent_method_ns, ['y', 'X', 'C', 'p', '_ent_print_args'])
```

### 103. Correlation-matrix entropy

Correlation-matrix entropy applies the same eigenvalue normalization to the channel correlation matrix, removing channel scale from the geometry.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Covariance measures how two variables vary together around their means. Positive covariance means they tend to deviate in the same direction; negative covariance means opposite directions. Correlation rescales covariance by the variables' standard deviations, making it dimensionless and bounded between -1 and 1 for ordinary Pearson correlation. Both are linear summaries and can miss nonlinear dependence.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values.</p></div>


```python
# =======================================================
# 103. Correlation-matrix entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Correlation-matrix entropy applies the same eigenvalue normalization to the channel
# correlation matrix, removing channel scale from the geometry.
# =======================================================
def _entropy_method_103():
    y = _ent_y
    X = _ent_x
    R = np.corrcoef(X, rowvar=False)
    p = _eig_probs(R)
    _ent_print_args = ['Correlation-matrix eigenvalue entropy=', _H(p)]
    print(*_ent_print_args)
    return locals()
_ent_method_ns = _entropy_method_103()
_ent_auto_plot(103, 'Correlation-matrix entropy', _ent_method_ns, ['y', 'X', 'R', 'p', '_ent_print_args'])
```

### 104. Effective-rank entropy

Effective rank is exp(H(p)), where p are normalized nonnegative matrix eigenvalues. It is the entropy-equivalent number of active modes.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> A Jacobian matrix is the local linear approximation of how small changes in each state variable affect the next state. Its eigenvalues and eigenvectors summarize locally preferred directions of growth, decay, or rotation. For a discrete-time map, eigenvalue magnitude below one indicates local contraction and magnitude above one indicates local expansion; for continuous-time systems, the sign of the real part is the corresponding stability indicator. When the Jacobian is estimated from data, these conclusions are local and depend on the quality of the regression or neighborhood used to estimate it.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> The visualization is a way of displaying already-computed numbers. Color, ordering, or ranking can make patterns easier to see but does not create additional statistical evidence beyond the values and uncertainty measures being plotted.</p></div>


```python
# =======================================================
# 104. Effective-rank entropy
# Explicit method implementation — no catalogue dispatcher.
# Mathematics: Effective rank is exp(H(p)), where p are normalized nonnegative matrix eigenvalues. It
# is the entropy-equivalent number of active modes.
# =======================================================
def _entropy_method_104():
    y = _ent_y
    X = _ent_x
    p = _eig_probs(np.cov(X, rowvar=False))
    h = _H(p)
    _ent_print_args = ['effective rank exp(H)=', np.exp(h), 'H=', h]
    print(*_ent_print_args)
    return locals()
_ent_method_ns = _entropy_method_104()
_ent_auto_plot(104, 'Effective-rank entropy', _ent_method_ns, ['y', 'X', 'p', 'h', '_ent_print_args'])
```
