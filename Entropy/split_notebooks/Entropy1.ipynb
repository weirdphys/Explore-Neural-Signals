### Multiscale Entropy


<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — Multiscale entropy</h3><p>Multiscale entropy first coarse-grains the signal at a scale $\tau$ and then evaluates an entropy statistic, commonly Sample Entropy, on each coarse-grained series. For the standard non-overlapping mean coarse-graining, each new sample is the mean of $\tau$ consecutive original samples.</p><p><b>Cell role.</b> This code applies that theory to the notebook's existing EEG-derived data and preserves the surrounding notebook's plotting/output conventions. Numerical shortcuts selected by a speed mode change sampling density, iteration counts, or visualization resolution; they do not change the defining formula above.</p></div>

$$y_j^{(\\tau)}=\\frac1{\\tau}\\sum_{i=(j-1)\\tau+1}^{j\\tau}x_i.$$


```python
import sys
import time
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import nolds

# -------------------------------------------------------
# CONFIG
# -------------------------------------------------------
# Reuse the project-wide runtime modes. ``full`` preserves the original cell exactly;
# faster modes only thin the embedding trajectory before the same MSE calculation.
FAST_OPTION = globals().get("FAST_OPTION", "full")
if FAST_OPTION not in {"full", "balanced", "fast", "ultra"}:
    raise ValueError("FAST_OPTION must be one of: full, balanced, fast, ultra")
ENTROPY_LEGACY_SPEED = {
    "full":     {"embedding_stride": 1},
    "balanced": {"embedding_stride": 2},
    "fast":     {"embedding_stride": 4},
    "ultra":    {"embedding_stride": 8},
}[FAST_OPTION]

base_dir = Path("/home/a/projects/Complete-Neural-Signal-Analysis")
embedding_3d_dir = base_dir / "embedding_data" / "3dembedding_data"

output_directory = base_dir / "multiscale_entropy_3d"
plots_directory = output_directory / "plots"

output_directory.mkdir(parents=True, exist_ok=True)
plots_directory.mkdir(parents=True, exist_ok=True)

max_scale = 5
emb_dim = 2
tolerance_mode = "default"   # "default" or float
SHOW_PLOTS = True            # ALWAYS display as well as save
SAVE_PARTIALS = True         # save CSVs after each channel
HEARTBEAT = True             # print frequent progress updates

BG = "black"
ACCENT = "cyan"

# -------------------------------------------------------
# GLOBAL PLOT STYLE
# -------------------------------------------------------
plt.rcParams.update({
    "figure.facecolor": BG,
    "axes.facecolor": BG,
    "savefig.facecolor": BG,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT,
})

# -------------------------------------------------------
# HEARTBEAT HELPERS
# -------------------------------------------------------
t0_global = time.perf_counter()

def hb(msg):
    if HEARTBEAT:
        elapsed = time.perf_counter() - t0_global
        print(f"[{elapsed:8.1f}s] {msg}", flush=True)

# -------------------------------------------------------
# STYLE HELPER
# -------------------------------------------------------
def style_ax(ax):
    ax.set_facecolor(BG)
    ax.tick_params(colors=ACCENT)
    for spine in ax.spines.values():
        spine.set_color(ACCENT)
    ax.grid(True, alpha=0.20, color=ACCENT)

# -------------------------------------------------------
# HELPERS
# -------------------------------------------------------
def extract_channel_name(file_path):
    name = Path(file_path).stem
    prefixes = ["3dembedded_", "embedded_"]
    suffixes = ["_data"]

    for prefix in prefixes:
        if name.startswith(prefix):
            name = name[len(prefix):]

    for suffix in suffixes:
        if name.endswith(suffix):
            name = name[:-len(suffix)]

    return name

def coarse_grain(time_series, scale):
    """
    Fast coarse-graining by averaging non-overlapping windows.
    Keeps the final partial window if needed.
    """
    x = np.asarray(time_series, dtype=float)
    n = len(x)

    if scale <= 0:
        raise ValueError("scale must be >= 1")
    if n == 0:
        return np.array([], dtype=float)

    full_len = (n // scale) * scale
    coarse_parts = []

    if full_len > 0:
        coarse_parts.append(x[:full_len].reshape(-1, scale).mean(axis=1))

    if full_len < n:
        coarse_parts.append(np.array([x[full_len:].mean()], dtype=float))

    if not coarse_parts:
        return np.array([], dtype=float)

    return np.concatenate(coarse_parts)

def compute_tolerance(x, tolerance_mode="default"):
    x = np.asarray(x, dtype=float)
    if tolerance_mode == "default":
        return 0.1 * np.std(x)
    return float(tolerance_mode)

def safe_sampen(x, emb_dim=2, tolerance=0.2):
    x = np.asarray(x, dtype=float)

    if len(x) <= emb_dim + 1:
        return np.nan

    try:
        return float(nolds.sampen(x, emb_dim=emb_dim, tolerance=tolerance))
    except Exception as e:
        hb(f"[WARN] sampen failed: {e}")
        return np.nan

def multiscale_entropy_verbose(time_series, max_scale, emb_dim=2, tolerance_mode="default",
                               channel_name=None, dim_idx=None):
    """
    Compute multiscale entropy across scales 1..max_scale with heartbeat output.
    """
    x = np.asarray(time_series, dtype=float)
    tol = compute_tolerance(x, tolerance_mode=tolerance_mode)

    mse = np.full(max_scale, np.nan, dtype=float)

    hb(
        f"Starting MSE | channel={channel_name} | dim={dim_idx} | "
        f"len={len(x)} | tol={tol:.6f}"
    )

    for scale in range(1, max_scale + 1):
        t_scale = time.perf_counter()

        coarse_ts = coarse_grain(x, scale)
        mse[scale - 1] = safe_sampen(coarse_ts, emb_dim=emb_dim, tolerance=tol)

        dt = time.perf_counter() - t_scale
        hb(
            f"  scale={scale}/{max_scale} | channel={channel_name} | dim={dim_idx} | "
            f"coarse_len={len(coarse_ts)} | mse={mse[scale - 1]:.6f} | dt={dt:.2f}s"
        )

    return mse

def save_and_show(fig, out_path):
    plt.tight_layout()
    plt.savefig(out_path, dpi=180, bbox_inches="tight")
    hb(f"Saved plot: {out_path}")

    if SHOW_PLOTS:
        plt.show()
        hb(f"Displayed plot: {out_path.name}")

    plt.close(fig)

# -------------------------------------------------------
# LOAD FILES
# -------------------------------------------------------
file_paths = sorted(embedding_3d_dir.glob("*.npy"))

if not file_paths:
    raise FileNotFoundError(f"No .npy files found in: {embedding_3d_dir}")

hb(f"Found {len(file_paths)} 3D embedding files in: {embedding_3d_dir}")

# -------------------------------------------------------
# MAIN COMPUTATION
# -------------------------------------------------------
long_rows = []
summary_rows = []

long_csv_path = output_directory / "mse_3d_long.csv"
summary_csv_path = output_directory / "mse_3d_summary.csv"

n_files = len(file_paths)

for file_idx, file_path in enumerate(file_paths, start=1):
    t_file = time.perf_counter()

    hb(f"Loading file {file_idx}/{n_files}: {file_path.name}")
    data = np.load(file_path)
    channel_name = extract_channel_name(file_path)

    data = np.asarray(data, dtype=float)
    _speed_stride = int(ENTROPY_LEGACY_SPEED["embedding_stride"])
    if _speed_stride > 1:
        data = data[::_speed_stride]
    if data.ndim != 2:
        hb(f"[SKIP] {file_path.name}: expected 2D array, got shape {data.shape}")
        continue

    if data.shape[1] < 1:
        hb(f"[SKIP] {file_path.name}: no dimensions found")
        continue

    hb(f"Processing {file_path.name} | channel={channel_name} | shape={data.shape}")

    dim_curves = []

    for dim in range(data.shape[1]):
        t_dim = time.perf_counter()

        mse = multiscale_entropy_verbose(
            data[:, dim],
            max_scale=max_scale,
            emb_dim=emb_dim,
            tolerance_mode=tolerance_mode,
            channel_name=channel_name,
            dim_idx=dim + 1
        )
        dim_curves.append(mse)

        for scale_idx, value in enumerate(mse, start=1):
            long_rows.append({
                "channel": channel_name,
                "file": file_path.name,
                "dimension": dim + 1,
                "scale": scale_idx,
                "mse": value,
            })

        hb(
            f"Finished dimension {dim + 1}/{data.shape[1]} | channel={channel_name} | "
            f"curve={np.array2string(mse, precision=5, separator=', ')} | "
            f"dt={time.perf_counter() - t_dim:.2f}s"
        )

    dim_curves = np.asarray(dim_curves, dtype=float)
    mean_curve = np.nanmean(dim_curves, axis=0)

    for scale_idx, value in enumerate(mean_curve, start=1):
        summary_rows.append({
            "channel": channel_name,
            "scale": scale_idx,
            "mean_mse_across_dims": value,
        })

    # ---------------------------------------------------
    # PER-CHANNEL PLOT
    # ---------------------------------------------------
    hb(f"Plotting channel summary: {channel_name}")

    fig, ax = plt.subplots(figsize=(9, 6), facecolor=BG)
    style_ax(ax)

    scales = np.arange(1, max_scale + 1)

    for dim in range(dim_curves.shape[0]):
        ax.plot(
            scales,
            dim_curves[dim],
            marker="o",
            linewidth=1.6,
            label=f"Dim {dim + 1}"
        )

    ax.plot(
        scales,
        mean_curve,
        marker="s",
        linewidth=2.2,
        linestyle="--",
        label="Mean"
    )

    ax.set_title(f"Multiscale Entropy | {channel_name}")
    ax.set_xlabel("Scale")
    ax.set_ylabel("Sample Entropy")
    ax.set_xticks(scales)

    legend = ax.legend(facecolor=BG, edgecolor=ACCENT, labelcolor=ACCENT)
    for text in legend.get_texts():
        text.set_color(ACCENT)

    out_path = plots_directory / f"mse_{channel_name}.png"
    save_and_show(fig, out_path)

    # ---------------------------------------------------
    # SAVE PARTIAL RESULTS
    # ---------------------------------------------------
    if SAVE_PARTIALS:
        pd.DataFrame(long_rows).to_csv(long_csv_path, index=False)
        pd.DataFrame(summary_rows).to_csv(summary_csv_path, index=False)
        hb(f"Saved partial CSVs after channel {channel_name}")

    hb(
        f"Finished file {file_idx}/{n_files}: {file_path.name} | "
        f"dt={time.perf_counter() - t_file:.2f}s"
    )

# -------------------------------------------------------
# FINAL SAVE TABLES
# -------------------------------------------------------
long_df = pd.DataFrame(long_rows)
summary_df = pd.DataFrame(summary_rows)

long_df.to_csv(long_csv_path, index=False)
summary_df.to_csv(summary_csv_path, index=False)

hb(f"Saved long-form MSE results: {long_csv_path}")
hb(f"Saved summary MSE results:   {summary_csv_path}")

# -------------------------------------------------------
# GRAND SUMMARY PLOT
# -------------------------------------------------------
if not summary_df.empty:
    hb("Building grand summary plot")

    pivot_df = summary_df.pivot(index="scale", columns="channel", values="mean_mse_across_dims")

    fig, ax = plt.subplots(figsize=(12, 7), facecolor=BG)
    style_ax(ax)

    for channel in pivot_df.columns:
        ax.plot(
            pivot_df.index.to_numpy(),
            pivot_df[channel].to_numpy(),
            linewidth=1.6,
            marker="o",
            label=channel
        )

    ax.set_title("Multiscale Entropy Across Channels (mean across 3D embedding dimensions)")
    ax.set_xlabel("Scale")
    ax.set_ylabel("Mean Sample Entropy")
    ax.set_xticks(np.arange(1, max_scale + 1))

    legend = ax.legend(
        facecolor=BG,
        edgecolor=ACCENT,
        labelcolor=ACCENT,
        ncol=4,
        fontsize=8
    )
    for text in legend.get_texts():
        text.set_color(ACCENT)

    summary_plot_path = plots_directory / "mse_3d_all_channels_mean.png"
    save_and_show(fig, summary_plot_path)

hb("Done.")
```

### Approximate Entropy

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — Approximate entropy</h3><p>Approximate entropy compares average logarithmic match frequencies for templates of length $m$ and $m+1$. Unlike Sample Entropy, the classical ApEn definition includes self-matches, which biases small samples downward but prevents zero counts in many cases.</p><p><b>Cell role.</b> This code applies that theory to the notebook's existing EEG-derived data and preserves the surrounding notebook's plotting/output conventions. Numerical shortcuts selected by a speed mode change sampling density, iteration counts, or visualization resolution; they do not change the defining formula above.</p></div>

$$\\operatorname{ApEn}(m,r)=\\phi_m(r)-\\phi_{m+1}(r),$$

$$\\phi_m(r)=\\frac1{N-m+1}\\sum_i\\log C_i^{(m)}(r).$$


```python
import time
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from numpy.lib.stride_tricks import sliding_window_view

# -------------------------------------------------------
# CONFIG
# -------------------------------------------------------
base_dir = Path("/home/a/projects/Complete-Neural-Signal-Analysis")
eeg_path = base_dir / "eeg_data_with_channels.npy"

output_directory = base_dir / "approximate_entropy_fast"
plots_directory = output_directory / "plots"

output_directory.mkdir(parents=True, exist_ok=True)
plots_directory.mkdir(parents=True, exist_ok=True)

eeg_channels = [
    'Fp1', 'Fpz', 'Fp2', 'F7', 'F3', 'Fz', 'F4', 'F8', 'FC5', 'FC1', 'FC2', 'FC6',
    'M1', 'T7', 'C3', 'Cz', 'C4', 'T8', 'M2', 'CP5', 'CP1', 'CP2', 'CP6',
    'P7', 'P3', 'Pz', 'P4', 'P8', 'POz', 'O1', 'Oz', 'O2'
]

sampling_rate = 1000
start_time, end_time = 814.571, 921.515
start_index, end_index = int(start_time * sampling_rate), int(end_time * sampling_rate)

# Approximate Entropy parameters
m = 2
r_factor = 0.2

# SPEED / HEARTBEAT CONTROLS
FAST_OPTION = globals().get("FAST_OPTION", "full")
if FAST_OPTION not in {"full", "balanced", "fast", "ultra"}:
    raise ValueError("FAST_OPTION must be one of: full, balanced, fast, ultra")
_APEN_SPEED = {
    "full":     {"max_points": 2000, "chunk": 400},
    "balanced": {"max_points": 1500, "chunk": 350},
    "fast":     {"max_points": 1000, "chunk": 300},
    "ultra":    {"max_points": 600,  "chunk": 200},
}[FAST_OPTION]
MAX_POINTS_PER_CHANNEL = _APEN_SPEED["max_points"]
CHUNK_SIZE = _APEN_SPEED["chunk"]                # chunk size for pairwise Chebyshev comparisons
SHOW_PLOTS = True               # ALWAYS display + save
SAVE_PARTIALS = True
HEARTBEAT = True
HEARTBEAT_EVERY_CHUNK = 1       # print every chunk

BG = "black"
ACCENT = "cyan"

# -------------------------------------------------------
# GLOBAL STYLE
# -------------------------------------------------------
plt.rcParams.update({
    "figure.facecolor": BG,
    "axes.facecolor": BG,
    "savefig.facecolor": BG,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT,
})

# -------------------------------------------------------
# HEARTBEAT
# -------------------------------------------------------
t0_global = time.perf_counter()

def hb(msg):
    if HEARTBEAT:
        dt = time.perf_counter() - t0_global
        print(f"[{dt:8.1f}s] {msg}", flush=True)

# -------------------------------------------------------
# STYLE
# -------------------------------------------------------
def style_ax(ax):
    ax.set_facecolor(BG)
    ax.tick_params(colors=ACCENT)
    for spine in ax.spines.values():
        spine.set_color(ACCENT)
    ax.grid(True, alpha=0.20, color=ACCENT)

def save_and_show(fig, out_path):
    plt.tight_layout()
    plt.savefig(out_path, dpi=180, bbox_inches="tight")
    hb(f"Saved plot: {out_path.name}")
    if SHOW_PLOTS:
        plt.show()
        hb(f"Displayed plot: {out_path.name}")
    plt.close(fig)

# -------------------------------------------------------
# HELPERS
# -------------------------------------------------------
def reduce_series_length(x, max_points=2000):
    """
    Fast stride-based reduction to keep ApEn tractable.
    """
    x = np.asarray(x, dtype=float)
    n = len(x)

    if n <= max_points:
        return x, 1

    stride = int(np.ceil(n / max_points))
    return x[::stride], stride

def apen_phi_chunked(x, emb_dim, r, chunk_size=400, label="phi"):
    """
    Compute phi(m) in chunks using Chebyshev distance.
    This keeps memory bounded and gives heartbeat progress.
    """
    x = np.asarray(x, dtype=np.float32)

    if len(x) <= emb_dim:
        return np.nan

    X = sliding_window_view(x, window_shape=emb_dim).astype(np.float32, copy=False)
    n_vectors = len(X)

    hb(f"Starting {label} | emb_dim={emb_dim} | vectors={n_vectors} | r={r:.6f}")

    counts = np.empty(n_vectors, dtype=np.float32)
    n_chunks = int(np.ceil(n_vectors / chunk_size))

    for chunk_idx, start in enumerate(range(0, n_vectors, chunk_size), start=1):
        end = min(start + chunk_size, n_vectors)
        Xi = X[start:end]  # shape: (chunk, emb_dim)

        # Chebyshev distance chunk against all vectors
        # result shape: (chunk, n_vectors)
        dist = np.max(np.abs(Xi[:, None, :] - X[None, :, :]), axis=2)
        counts[start:end] = np.sum(dist <= r, axis=1)

        if HEARTBEAT and (chunk_idx % HEARTBEAT_EVERY_CHUNK == 0 or chunk_idx == n_chunks):
            hb(f"  {label} chunk {chunk_idx}/{n_chunks} complete")

    C = counts / n_vectors
    phi_val = float(np.mean(np.log(C + 1e-12)))

    hb(f"Finished {label} | emb_dim={emb_dim} | phi={phi_val:.6f}")
    return phi_val

def approximate_entropy_chunked(x, m=2, r=0.2, chunk_size=400):
    """
    ApEn(m, r) = |phi(m) - phi(m+1)|
    """
    x = np.asarray(x, dtype=float)

    if len(x) <= m + 1:
        return np.nan

    phi_m = apen_phi_chunked(x, emb_dim=m, r=r, chunk_size=chunk_size, label="phi(m)")
    phi_m1 = apen_phi_chunked(x, emb_dim=m + 1, r=r, chunk_size=chunk_size, label="phi(m+1)")

    if not np.isfinite(phi_m) or not np.isfinite(phi_m1):
        return np.nan

    apen = abs(phi_m - phi_m1)
    hb(f"Computed ApEn = {apen:.6f}")
    return apen

# -------------------------------------------------------
# LOAD EEG
# -------------------------------------------------------
hb(f"Loading EEG data from: {eeg_path}")
EEG_data = np.load(eeg_path, allow_pickle=True)

if EEG_data.ndim != 2:
    raise ValueError(f"Expected 2D EEG array, got shape {EEG_data.shape}")

# Handle either (time, channels) or (channels, time)
if EEG_data.shape[1] == len(eeg_channels):
    EEG_segment = EEG_data[start_index:end_index, :]
elif EEG_data.shape[0] == len(eeg_channels):
    EEG_segment = EEG_data[:, start_index:end_index].T
else:
    raise ValueError(f"Unexpected EEG shape: {EEG_data.shape}")

hb(f"Loaded EEG segment: shape={EEG_segment.shape}")
hb(f"Time window: {start_time:.3f}s to {end_time:.3f}s")

# -------------------------------------------------------
# MAIN COMPUTATION
# -------------------------------------------------------
rows = []
csv_path = output_directory / "approximate_entropy_results.csv"

for i, channel in enumerate(eeg_channels):
    t_channel = time.perf_counter()

    hb(f"Starting channel {i+1}/{len(eeg_channels)}: {channel}")

    x_full = EEG_segment[:, i].astype(float)
    x_used, stride = reduce_series_length(x_full, max_points=MAX_POINTS_PER_CHANNEL)

    # keep original formulation: tolerance from the channel std
    r = r_factor * np.std(x_full)

    hb(
        f"Prepared channel {channel} | raw_len={len(x_full)} | "
        f"used_len={len(x_used)} | stride={stride} | r={r:.6f}"
    )

    apen = approximate_entropy_chunked(
        x_used,
        m=m,
        r=r,
        chunk_size=CHUNK_SIZE
    )

    rows.append({
        "Channel": channel,
        "Approximate Entropy": apen,
        "Raw Length": len(x_full),
        "Used Length": len(x_used),
        "Stride Used": stride,
        "m": m,
        "r": r,
        "Chunk Size": CHUNK_SIZE,
    })

    hb(
        f"Finished channel {channel} | ApEn={apen:.6f} | "
        f"dt={time.perf_counter() - t_channel:.2f}s"
    )

    if SAVE_PARTIALS:
        pd.DataFrame(rows).to_csv(csv_path, index=False)
        hb(f"Saved partial CSV after {channel}")

# -------------------------------------------------------
# SAVE RESULTS
# -------------------------------------------------------
approx_entropy_df = pd.DataFrame(rows)
approx_entropy_df["Channel"] = pd.Categorical(
    approx_entropy_df["Channel"],
    categories=eeg_channels,
    ordered=True
)
approx_entropy_df = approx_entropy_df.sort_values("Channel").reset_index(drop=True)
approx_entropy_df.to_csv(csv_path, index=False)

hb(f"Saved final CSV: {csv_path}")
print("\nApproximate Entropy results:")
print(approx_entropy_df)

# -------------------------------------------------------
# PLOT: CHANNEL ORDER
# -------------------------------------------------------
fig, ax = plt.subplots(figsize=(14, 7), facecolor=BG)
style_ax(ax)

ax.bar(
    approx_entropy_df["Channel"],
    approx_entropy_df["Approximate Entropy"].to_numpy(dtype=float)
)

ax.set_title("Approximate Entropy of EEG Channels")
ax.set_xlabel("EEG Channel")
ax.set_ylabel("Approximate Entropy")
ax.tick_params(axis="x", rotation=90)

bar_plot_path = plots_directory / "approximate_entropy_by_channel.png"
save_and_show(fig, bar_plot_path)

# -------------------------------------------------------
# PLOT: SORTED
# -------------------------------------------------------
sorted_df = approx_entropy_df.sort_values("Approximate Entropy", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(14, 7), facecolor=BG)
style_ax(ax)

ax.bar(
    sorted_df["Channel"],
    sorted_df["Approximate Entropy"].to_numpy(dtype=float)
)

ax.set_title("Approximate Entropy of EEG Channels (Sorted)")
ax.set_xlabel("EEG Channel")
ax.set_ylabel("Approximate Entropy")
ax.tick_params(axis="x", rotation=90)

sorted_plot_path = plots_directory / "approximate_entropy_sorted.png"
save_and_show(fig, sorted_plot_path)

print("\nTop channels by Approximate Entropy:")
print(sorted_df[["Channel", "Approximate Entropy"]].head(10))

hb("Done.")
```

### Fractal Dimension of Network Connectivity

# Transfer Entropy

# Hemispheric 

<div style="font-size: 13px; font-family: 'Times New Roman', Times, serif; background-color: #181818; color: #D0D0D0; padding: 20px; border-radius: 8px; margin: 10px; display: flex; flex-wrap: nowrap; justify-content: space-between;">
    <!-- Column 1 -->
    <div style="flex: 1; margin-right: 10px;">
        <h2>Introduction</h2>
        <p>
            This script performs analysis on EEG data using the concept of Transfer Entropy (TE). Specifically, it investigates the directionality of information flow between different EEG channels based on their 2D embeddings.
        </p>
        <h2>Data Preparation</h2>
        <p>
            Initially, the code unzips a dataset of 2D embedded EEG data for multiple channels. The embedded data is then loaded into Python, trimmed to a desired length, and grouped by hemisphere (left and right).
        </p>
    </div>
    <!-- Column 2 -->
    <div style="flex: 1; margin-left: 10px;">
        <h2>Analysis</h2>
        <p>
            Transfer Entropy calculations are conducted to quantify the amount of information flowing from one EEG channel to another. This is done after binning the data and applying the PyInform library’s TE function.
        </p>
        <p>
            The TE formula is represented as:
            \[
            TE_{X \rightarrow Y} = \sum_{y_{t+1}, y_t, x_t} p(y_{t+1}, y_t, x_t) \log \frac{p(y_{t+1} | y_t, x_t)}{p(y_{t+1} | y_t)}
            \]
            Here \(X\) and \(Y\) are random variables, \(p(y_{t+1}, y_t, x_t)\) is the joint probability distribution function, and \(p(y_{t+1} | y_t, x_t)\) and \(p(y_{t+1} | y_t)\) are conditional probability distribution functions.
        </p>
        </p>
    </div>
</div>

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — Transfer entropy</h3><p>Transfer entropy is a conditional mutual information that measures whether the past of a source $X$ improves prediction of the next target value $Y_{t+1}$ beyond the target's own past. It is directional and depends on the history lengths, lag, estimator, and any discretization used by the implementation.</p><p><b>Cell role.</b> This code applies that theory to the notebook's existing EEG-derived data and preserves the surrounding notebook's plotting/output conventions. Numerical shortcuts selected by a speed mode change sampling density, iteration counts, or visualization resolution; they do not change the defining formula above.</p></div>

$$T_{X\\to Y}=\\sum p(y_{t+1},y_t^{(k)},x_t^{(l)})\\log\\frac{p(y_{t+1}\\mid y_t^{(k)},x_t^{(l)})}{p(y_{t+1}\\mid y_t^{(k)})}.$$


```python
import time
import zipfile
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pyinform import transfer_entropy

# -------------------------------------------------------
# CONFIG
# -------------------------------------------------------
base_dir = Path("/home/a/projects/Complete-Neural-Signal-Analysis")
zipped_file_path = base_dir / "2d_embedded_data.zip"

output_directory = base_dir / "transfer_entropy_2d"
plots_directory = output_directory / "plots"

output_directory.mkdir(parents=True, exist_ok=True)
plots_directory.mkdir(parents=True, exist_ok=True)

eeg_channels = [
    'Fp1', 'Fpz', 'Fp2', 'F7', 'F3', 'Fz', 'F4', 'F8', 'FC5', 'FC1', 'FC2', 'FC6',
    'M1', 'T7', 'C3', 'Cz', 'C4', 'T8', 'M2', 'CP5', 'CP1', 'CP2', 'CP6',
    'P7', 'P3', 'Pz', 'P4', 'P8', 'POz', 'O1', 'Oz', 'O2'
]

left_channels = ['Fp1', 'F7', 'F3', 'FC5', 'M1', 'T7', 'C3', 'CP5', 'P7', 'P3', 'O1']
right_channels = ['Fp2', 'F8', 'F4', 'FC6', 'M2', 'T8', 'C4', 'CP6', 'P8', 'P4', 'O2']

# central_channels existed in your original script, but was not used there either
central_channels = ['Fpz', 'Fz', 'FC1', 'FC2', 'Cz', 'CP1', 'CP2', 'Pz', 'POz', 'Oz']

# TE settings
K_HISTORY = 1
NUM_BINS = 32

FAST_OPTION = globals().get("FAST_OPTION", "full")
if FAST_OPTION not in {"full", "balanced", "fast", "ultra"}:
    raise ValueError("FAST_OPTION must be one of: full, balanced, fast, ultra")
_TE_PAIR_SPEED = {
    "full":     {"max_points": 20000, "preview": 3000},
    "balanced": {"max_points": 12000, "preview": 2500},
    "fast":     {"max_points": 7000,  "preview": 1800},
    "ultra":    {"max_points": 3500,  "preview": 1200},
}[FAST_OPTION]

# SPEED / OUTPUT CONTROLS
MAX_POINTS_PER_SERIES = _TE_PAIR_SPEED["max_points"]
PREVIEW_POINTS = _TE_PAIR_SPEED["preview"]
SHOW_PLOTS = True               # ALWAYS display + save
HEARTBEAT = True
SAVE_PARTIALS = True

BG = "black"
ACCENT = "cyan"

# -------------------------------------------------------
# GLOBAL STYLE
# -------------------------------------------------------
plt.rcParams.update({
    "figure.facecolor": BG,
    "axes.facecolor": BG,
    "savefig.facecolor": BG,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT,
})

# -------------------------------------------------------
# HEARTBEAT
# -------------------------------------------------------
t0_global = time.perf_counter()

def hb(msg):
    if HEARTBEAT:
        dt = time.perf_counter() - t0_global
        print(f"[{dt:8.1f}s] {msg}", flush=True)

# -------------------------------------------------------
# STYLE
# -------------------------------------------------------
def style_ax(ax):
    ax.set_facecolor(BG)
    ax.tick_params(colors=ACCENT)
    for spine in ax.spines.values():
        spine.set_color(ACCENT)
    ax.grid(True, alpha=0.20, color=ACCENT)

def save_and_show(fig, out_path):
    plt.tight_layout()
    plt.savefig(out_path, dpi=180, bbox_inches="tight")
    hb(f"Saved plot: {out_path.name}")
    if SHOW_PLOTS:
        plt.show()
        hb(f"Displayed plot: {out_path.name}")
    plt.close(fig)

# -------------------------------------------------------
# HELPERS
# -------------------------------------------------------
def load_embedded_channel_from_zip(zipf, channel_name):
    """
    Supports both:
      - 2dembedded_<channel>.npy
      - 2dembedded_<channel>_data.npy
    """
    candidates = [
        f"2dembedded_{channel_name}.npy",
        f"2dembedded_{channel_name}_data.npy",
    ]

    for name in candidates:
        try:
            with zipf.open(name) as f:
                arr = np.load(f)
            arr = np.asarray(arr, dtype=float)
            if arr.ndim != 2 or arr.shape[1] != 2:
                raise ValueError(f"{name} must have shape (N, 2), got {arr.shape}")
            return arr
        except KeyError:
            continue

    raise FileNotFoundError(f"No 2D embedding file found for channel {channel_name}")

def reduce_series_length(x, max_points=20000):
    x = np.asarray(x, dtype=float)
    n = len(x)

    if n <= max_points:
        return x, 1

    stride = int(np.ceil(n / max_points))
    return x[::stride], stride

def standardize_1d(x):
    x = np.asarray(x, dtype=float)
    return (x - np.mean(x)) / (np.std(x) + 1e-12)

def joint_bin_equal_width(x, y, num_bins=32):
    """
    Standardize each series, then bin both using the same edges
    derived from the combined value range.
    """
    x = standardize_1d(x)
    y = standardize_1d(y)

    joint = np.concatenate([x, y])
    xmin, xmax = np.min(joint), np.max(joint)

    if not np.isfinite(xmin) or not np.isfinite(xmax) or xmax <= xmin:
        return np.zeros_like(x, dtype=np.int32), np.zeros_like(y, dtype=np.int32)

    edges = np.linspace(xmin, xmax, num_bins + 1)

    xb = np.digitize(x, edges[1:-1], right=False).astype(np.int32)
    yb = np.digitize(y, edges[1:-1], right=False).astype(np.int32)

    xb = np.clip(xb, 0, num_bins - 1)
    yb = np.clip(yb, 0, num_bins - 1)

    return xb, yb

def compute_te_pair(source, target, k=1):
    """
    Transfer entropy from source -> target.
    """
    return float(transfer_entropy(source, target, k))

# -------------------------------------------------------
# LOAD ONLY NEEDED CHANNELS
# -------------------------------------------------------
channels_needed = sorted(set(left_channels + right_channels))
embedded = {}

hb(f"Opening zip: {zipped_file_path}")

with zipfile.ZipFile(zipped_file_path, "r") as zipf:
    for idx, ch in enumerate(channels_needed, start=1):
        hb(f"Loading channel {idx}/{len(channels_needed)}: {ch}")
        embedded[ch] = load_embedded_channel_from_zip(zipf, ch)
        hb(f"Loaded {ch}: shape={embedded[ch].shape}")

# -------------------------------------------------------
# ALIGN LENGTHS
# -------------------------------------------------------
common_length = min(len(embedded[ch]) for ch in channels_needed)
hb(f"Common length across required channels: {common_length}")

for ch in channels_needed:
    embedded[ch] = embedded[ch][:common_length]

# -------------------------------------------------------
# BUILD HEMISPHERE AVERAGES
# -------------------------------------------------------
hb("Computing left/right average embeddings")

left_stack = np.stack([embedded[ch] for ch in left_channels], axis=0)
right_stack = np.stack([embedded[ch] for ch in right_channels], axis=0)

left_embedding_avg = np.mean(left_stack, axis=0)
right_embedding_avg = np.mean(right_stack, axis=0)

hb(f"Left average shape:  {left_embedding_avg.shape}")
hb(f"Right average shape: {right_embedding_avg.shape}")

# -------------------------------------------------------
# REDUCE LENGTH FOR SPEED
# -------------------------------------------------------
left_reduced = np.zeros((0, 2), dtype=float)
right_reduced = np.zeros((0, 2), dtype=float)

left_reduced_dim0, stride0 = reduce_series_length(left_embedding_avg[:, 0], MAX_POINTS_PER_SERIES)
right_reduced_dim0, _ = reduce_series_length(right_embedding_avg[:, 0], MAX_POINTS_PER_SERIES)

left_reduced_dim1, stride1 = reduce_series_length(left_embedding_avg[:, 1], MAX_POINTS_PER_SERIES)
right_reduced_dim1, _ = reduce_series_length(right_embedding_avg[:, 1], MAX_POINTS_PER_SERIES)

left_reduced = np.column_stack([left_reduced_dim0, left_reduced_dim1])
right_reduced = np.column_stack([right_reduced_dim0, right_reduced_dim1])

hb(
    f"Reduced dimension 1 | left_len={len(left_reduced_dim0)} | right_len={len(right_reduced_dim0)} | stride={stride0}"
)
hb(
    f"Reduced dimension 2 | left_len={len(left_reduced_dim1)} | right_len={len(right_reduced_dim1)} | stride={stride1}"
)

# -------------------------------------------------------
# COMPUTE TE BY EMBEDDING DIMENSION
# -------------------------------------------------------
rows = []
csv_path = output_directory / "transfer_entropy_results.csv"

for dim in range(2):
    t_dim = time.perf_counter()

    hb(f"Starting TE for embedding dimension {dim + 1}/2")

    left_data = left_reduced[:, dim]
    right_data = right_reduced[:, dim]

    hb(f"Binning dimension {dim + 1} with {NUM_BINS} bins")
    left_binned, right_binned = joint_bin_equal_width(left_data, right_data, num_bins=NUM_BINS)

    hb(f"Computing TE Left -> Right | dim={dim + 1}")
    te_left_to_right = compute_te_pair(left_binned, right_binned, k=K_HISTORY)

    hb(f"Computing TE Right -> Left | dim={dim + 1}")
    te_right_to_left = compute_te_pair(right_binned, left_binned, k=K_HISTORY)

    rows.append({
        "dimension": dim + 1,
        "TE_left_to_right": te_left_to_right,
        "TE_right_to_left": te_right_to_left,
        "TE_asymmetry_LR_minus_RL": te_left_to_right - te_right_to_left,
        "left_length_used": len(left_data),
        "right_length_used": len(right_data),
        "stride_used": stride0 if dim == 0 else stride1,
        "num_bins": NUM_BINS,
        "k_history": K_HISTORY,
    })

    hb(
        f"Finished dimension {dim + 1} | "
        f"TE L->R={te_left_to_right:.6f} | TE R->L={te_right_to_left:.6f} | "
        f"dt={time.perf_counter() - t_dim:.2f}s"
    )

    if SAVE_PARTIALS:
        pd.DataFrame(rows).to_csv(csv_path, index=False)
        hb(f"Saved partial CSV after dimension {dim + 1}")

# -------------------------------------------------------
# SAVE RESULTS
# -------------------------------------------------------
results_df = pd.DataFrame(rows)
results_df.to_csv(csv_path, index=False)

hb(f"Saved final CSV: {csv_path}")
print("\nTransfer Entropy results:")
print(results_df)

# -------------------------------------------------------
# SUMMARY PLOT
# -------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 6), facecolor=BG)
style_ax(ax)

x = np.arange(len(results_df))
width = 0.35

ax.bar(
    x - width / 2,
    results_df["TE_left_to_right"].to_numpy(dtype=float),
    width=width,
    color=ACCENT,
    alpha=0.85,
    label="Left → Right"
)

ax.bar(
    x + width / 2,
    results_df["TE_right_to_left"].to_numpy(dtype=float),
    width=width,
    color=ACCENT,
    alpha=0.35,
    label="Right → Left"
)

ax.set_title("Transfer Entropy Between Left and Right Hemisphere Averages")
ax.set_xlabel("Embedding Dimension")
ax.set_ylabel("Transfer Entropy")
ax.set_xticks(x)
ax.set_xticklabels([f"Dim {d}" for d in results_df["dimension"]])

legend = ax.legend(facecolor=BG, edgecolor=ACCENT, labelcolor=ACCENT)
for text in legend.get_texts():
    text.set_color(ACCENT)

summary_plot_path = plots_directory / "transfer_entropy_summary.png"
save_and_show(fig, summary_plot_path)

# -------------------------------------------------------
# ASYMMETRY PLOT
# -------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 6), facecolor=BG)
style_ax(ax)

ax.bar(
    [f"Dim {d}" for d in results_df["dimension"]],
    results_df["TE_asymmetry_LR_minus_RL"].to_numpy(dtype=float),
    color=ACCENT,
    alpha=0.85
)

ax.axhline(0.0, color=ACCENT, linewidth=1.0, linestyle="--")
ax.set_title("Transfer Entropy Asymmetry (Left→Right minus Right→Left)")
ax.set_xlabel("Embedding Dimension")
ax.set_ylabel("TE Asymmetry")

asymmetry_plot_path = plots_directory / "transfer_entropy_asymmetry.png"
save_and_show(fig, asymmetry_plot_path)

# -------------------------------------------------------
# PREVIEW PLOT OF REDUCED HEMISPHERE SIGNALS
# -------------------------------------------------------
for dim in range(2):
    fig, ax = plt.subplots(figsize=(12, 5), facecolor=BG)
    style_ax(ax)

    n_preview = min(PREVIEW_POINTS, len(left_reduced), len(right_reduced))
    t = np.arange(n_preview)

    ax.plot(
        t,
        left_reduced[:n_preview, dim],
        color=ACCENT,
        linewidth=1.4,
        linestyle="-",
        label="Left hemisphere average"
    )

    ax.plot(
        t,
        right_reduced[:n_preview, dim],
        color=ACCENT,
        linewidth=1.4,
        linestyle="--",
        label="Right hemisphere average"
    )

    ax.set_title(f"Hemisphere-Averaged Embedded Signals | Dimension {dim + 1}")
    ax.set_xlabel("Sample Index")
    ax.set_ylabel("Amplitude")

    legend = ax.legend(facecolor=BG, edgecolor=ACCENT, labelcolor=ACCENT)
    for text in legend.get_texts():
        text.set_color(ACCENT)

    preview_plot_path = plots_directory / f"hemisphere_average_preview_dim{dim + 1}.png"
    save_and_show(fig, preview_plot_path)

# -------------------------------------------------------
# PRINT SUMMARY
# -------------------------------------------------------
mean_lr = results_df["TE_left_to_right"].mean()
mean_rl = results_df["TE_right_to_left"].mean()

print(f"\nMean TE Left → Right: {mean_lr:.6f}")
print(f"Mean TE Right → Left: {mean_rl:.6f}")
print(f"Mean asymmetry (L→R - R→L): {(mean_lr - mean_rl):.6f}")

hb("Done.")
```

```python
import time
import zipfile
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pyinform import transfer_entropy

# -------------------------------------------------------
# CONFIG
# -------------------------------------------------------
base_dir = Path("/home/a/projects/Complete-Neural-Signal-Analysis")
zipped_file_path = base_dir / "2d_embedded_data.zip"

output_directory = base_dir / "transfer_entropy_2d_exploratory"
plots_directory = output_directory / "plots"

output_directory.mkdir(parents=True, exist_ok=True)
plots_directory.mkdir(parents=True, exist_ok=True)

eeg_channels = [
    'Fp1', 'Fpz', 'Fp2', 'F7', 'F3', 'Fz', 'F4', 'F8', 'FC5', 'FC1', 'FC2', 'FC6',
    'M1', 'T7', 'C3', 'Cz', 'C4', 'T8', 'M2', 'CP5', 'CP1', 'CP2', 'CP6',
    'P7', 'P3', 'Pz', 'P4', 'P8', 'POz', 'O1', 'Oz', 'O2'
]

left_channels = ['Fp1', 'F7', 'F3', 'FC5', 'M1', 'T7', 'C3', 'CP5', 'P7', 'P3', 'O1']
right_channels = ['Fp2', 'F8', 'F4', 'FC6', 'M2', 'T8', 'C4', 'CP6', 'P8', 'P4', 'O2']

# TE settings
K_HISTORY = 1
NUM_BINS = 32

FAST_OPTION = globals().get("FAST_OPTION", "full")
if FAST_OPTION not in {"full", "balanced", "fast", "ultra"}:
    raise ValueError("FAST_OPTION must be one of: full, balanced, fast, ultra")
_TE_DYNAMIC_SPEED = {
    "full":     {"max_points": 12000, "preview": 3000, "step": 300, "lag_step": 2},
    "balanced": {"max_points": 9000,  "preview": 2500, "step": 450, "lag_step": 2},
    "fast":     {"max_points": 6000,  "preview": 1800, "step": 600, "lag_step": 4},
    "ultra":    {"max_points": 3500,  "preview": 1200, "step": 900, "lag_step": 5},
}[FAST_OPTION]

# SPEED / EXPLORATION CONTROLS
MAX_POINTS_PER_SERIES = _TE_DYNAMIC_SPEED["max_points"]
PREVIEW_POINTS = _TE_DYNAMIC_SPEED["preview"]
ROLLING_WINDOW = 1500
ROLLING_STEP = _TE_DYNAMIC_SPEED["step"]
LAG_VALUES = np.arange(-20, 21, _TE_DYNAMIC_SPEED["lag_step"])

# OUTPUT CONTROLS
SHOW_PLOTS = True          # ALWAYS display + save
HEARTBEAT = True
SAVE_PARTIALS = True

# -------------------------------------------------------
# COLOR PALETTE
# -------------------------------------------------------
BG = "black"
ACCENT = "#00E5FF"      # axes/ticks/text/grid cyan
C_PRIMARY = "#00E5FF"   # bright cyan
C_SECONDARY = "#14B8A6" # teal
C_DIFF = "#A78BFA"      # soft violet
C_HIGHLIGHT = "#7DD3FC" # pale aqua
C_MUTED = "#67E8F9"     # light cyan

# -------------------------------------------------------
# GLOBAL STYLE
# -------------------------------------------------------
plt.rcParams.update({
    "figure.facecolor": BG,
    "axes.facecolor": BG,
    "savefig.facecolor": BG,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT,
})

# -------------------------------------------------------
# HEARTBEAT
# -------------------------------------------------------
t0_global = time.perf_counter()

def hb(msg):
    if HEARTBEAT:
        dt = time.perf_counter() - t0_global
        print(f"[{dt:8.1f}s] {msg}", flush=True)

# -------------------------------------------------------
# STYLING HELPERS
# -------------------------------------------------------
def style_ax(ax):
    ax.set_facecolor(BG)
    ax.tick_params(colors=ACCENT)
    for spine in ax.spines.values():
        spine.set_color(ACCENT)
    ax.grid(True, alpha=0.18, color=ACCENT)

def style_legend(legend):
    if legend is None:
        return
    frame = legend.get_frame()
    frame.set_facecolor(BG)
    frame.set_edgecolor(ACCENT)
    for text in legend.get_texts():
        text.set_color(ACCENT)

def save_and_show(fig, out_path):
    plt.tight_layout()
    plt.savefig(out_path, dpi=180, bbox_inches="tight")
    hb(f"Saved plot: {out_path.name}")
    if SHOW_PLOTS:
        plt.show()
        hb(f"Displayed plot: {out_path.name}")
    plt.close(fig)

# -------------------------------------------------------
# HELPERS
# -------------------------------------------------------
def load_embedded_channel_from_zip(zipf, channel_name):
    candidates = [
        f"2dembedded_{channel_name}.npy",
        f"2dembedded_{channel_name}_data.npy",
    ]
    for name in candidates:
        try:
            with zipf.open(name) as f:
                arr = np.load(f)
            arr = np.asarray(arr, dtype=float)
            if arr.ndim != 2 or arr.shape[1] != 2:
                raise ValueError(f"{name} must have shape (N, 2), got {arr.shape}")
            return arr
        except KeyError:
            continue
    raise FileNotFoundError(f"No 2D embedding file found for channel {channel_name}")

def reduce_series_length(x, max_points=12000):
    x = np.asarray(x, dtype=float)
    n = len(x)
    if n <= max_points:
        return x, 1
    stride = int(np.ceil(n / max_points))
    return x[::stride], stride

def standardize_1d(x):
    x = np.asarray(x, dtype=float)
    return (x - np.mean(x)) / (np.std(x) + 1e-12)

def joint_bin_equal_width(x, y, num_bins=32):
    """
    Standardize each series, then bin them using shared edges.
    """
    x = standardize_1d(x)
    y = standardize_1d(y)

    joint = np.concatenate([x, y])
    xmin, xmax = np.min(joint), np.max(joint)

    if not np.isfinite(xmin) or not np.isfinite(xmax) or xmax <= xmin:
        return np.zeros_like(x, dtype=np.int32), np.zeros_like(y, dtype=np.int32)

    edges = np.linspace(xmin, xmax, num_bins + 1)

    xb = np.digitize(x, edges[1:-1], right=False).astype(np.int32)
    yb = np.digitize(y, edges[1:-1], right=False).astype(np.int32)

    xb = np.clip(xb, 0, num_bins - 1)
    yb = np.clip(yb, 0, num_bins - 1)

    return xb, yb

def compute_te_pair(source, target, k=1):
    return float(transfer_entropy(source, target, k))

def rolling_te(x, y, window=1500, step=300, num_bins=32, k=1, heartbeat_label=""):
    """
    Rolling TE for exploratory time-localized directionality.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    rows = []
    n = len(x)

    if n < window:
        return pd.DataFrame(columns=["start", "center", "TE_LR", "TE_RL", "TE_diff"])

    starts = list(range(0, n - window + 1, step))
    hb(f"Starting rolling TE {heartbeat_label} | windows={len(starts)}")

    for idx, start in enumerate(starts, start=1):
        end = start + window
        xs = x[start:end]
        ys = y[start:end]

        xb, yb = joint_bin_equal_width(xs, ys, num_bins=num_bins)
        te_lr = compute_te_pair(xb, yb, k=k)
        te_rl = compute_te_pair(yb, xb, k=k)

        rows.append({
            "start": start,
            "center": start + window // 2,
            "TE_LR": te_lr,
            "TE_RL": te_rl,
            "TE_diff": te_lr - te_rl,
        })

        if idx == 1 or idx % 5 == 0 or idx == len(starts):
            hb(f"  rolling TE {heartbeat_label} | window {idx}/{len(starts)}")

    return pd.DataFrame(rows)

def align_for_lag(x, y, lag):
    """
    Positive lag means x leads y by 'lag' samples.
    Returns aligned (x_aligned, y_aligned).
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    if lag > 0:
        return x[:-lag], y[lag:]
    if lag < 0:
        lag = abs(lag)
        return x[lag:], y[:-lag]
    return x, y

def lag_sweep_te(x, y, lag_values, num_bins=32, k=1, heartbeat_label=""):
    """
    Exploratory TE as a function of alignment lag.
    """
    rows = []
    hb(f"Starting lag sweep {heartbeat_label} | n_lags={len(lag_values)}")

    for idx, lag in enumerate(lag_values, start=1):
        xa, ya = align_for_lag(x, y, lag)

        xb, yb = joint_bin_equal_width(xa, ya, num_bins=num_bins)
        te_lr = compute_te_pair(xb, yb, k=k)
        te_rl = compute_te_pair(yb, xb, k=k)

        rows.append({
            "lag": lag,
            "TE_LR": te_lr,
            "TE_RL": te_rl,
            "TE_diff": te_lr - te_rl,
            "N_used": len(xa),
        })

        if idx == 1 or idx % 5 == 0 or idx == len(lag_values):
            hb(f"  lag sweep {heartbeat_label} | lag {idx}/{len(lag_values)}")

    return pd.DataFrame(rows)

# -------------------------------------------------------
# LOAD ONLY NEEDED CHANNELS
# -------------------------------------------------------
channels_needed = sorted(set(left_channels + right_channels))
embedded = {}

hb(f"Opening zip: {zipped_file_path}")

with zipfile.ZipFile(zipped_file_path, "r") as zipf:
    for idx, ch in enumerate(channels_needed, start=1):
        hb(f"Loading channel {idx}/{len(channels_needed)}: {ch}")
        embedded[ch] = load_embedded_channel_from_zip(zipf, ch)
        hb(f"Loaded {ch}: shape={embedded[ch].shape}")

# -------------------------------------------------------
# ALIGN LENGTHS
# -------------------------------------------------------
common_length = min(len(embedded[ch]) for ch in channels_needed)
hb(f"Common length across required channels: {common_length}")

for ch in channels_needed:
    embedded[ch] = embedded[ch][:common_length]

# -------------------------------------------------------
# BUILD HEMISPHERE AVERAGES
# -------------------------------------------------------
hb("Computing hemisphere-average embeddings")

left_stack = np.stack([embedded[ch] for ch in left_channels], axis=0)
right_stack = np.stack([embedded[ch] for ch in right_channels], axis=0)

left_embedding_avg = np.mean(left_stack, axis=0)
right_embedding_avg = np.mean(right_stack, axis=0)

hb(f"Left average shape:  {left_embedding_avg.shape}")
hb(f"Right average shape: {right_embedding_avg.shape}")

# -------------------------------------------------------
# REDUCE LENGTH FOR SPEED
# -------------------------------------------------------
left_dim0, stride0 = reduce_series_length(left_embedding_avg[:, 0], MAX_POINTS_PER_SERIES)
right_dim0, _ = reduce_series_length(right_embedding_avg[:, 0], MAX_POINTS_PER_SERIES)

left_dim1, stride1 = reduce_series_length(left_embedding_avg[:, 1], MAX_POINTS_PER_SERIES)
right_dim1, _ = reduce_series_length(right_embedding_avg[:, 1], MAX_POINTS_PER_SERIES)

left_reduced = np.column_stack([left_dim0, left_dim1])
right_reduced = np.column_stack([right_dim0, right_dim1])

hb(f"Reduced Dim 1 | len={len(left_dim0)} | stride={stride0}")
hb(f"Reduced Dim 2 | len={len(left_dim1)} | stride={stride1}")

# -------------------------------------------------------
# BASIC TE SUMMARY
# -------------------------------------------------------
summary_rows = []
summary_csv_path = output_directory / "transfer_entropy_summary.csv"

for dim in range(2):
    hb(f"Computing basic TE summary | dim={dim + 1}")

    left_data = left_reduced[:, dim]
    right_data = right_reduced[:, dim]

    xb, yb = joint_bin_equal_width(left_data, right_data, num_bins=NUM_BINS)
    te_lr = compute_te_pair(xb, yb, k=K_HISTORY)
    te_rl = compute_te_pair(yb, xb, k=K_HISTORY)

    summary_rows.append({
        "dimension": dim + 1,
        "TE_left_to_right": te_lr,
        "TE_right_to_left": te_rl,
        "TE_asymmetry_LR_minus_RL": te_lr - te_rl,
        "length_used": len(left_data),
        "stride_used": stride0 if dim == 0 else stride1,
        "num_bins": NUM_BINS,
        "k_history": K_HISTORY,
    })

    if SAVE_PARTIALS:
        pd.DataFrame(summary_rows).to_csv(summary_csv_path, index=False)
        hb(f"Saved partial summary CSV after dim {dim + 1}")

summary_df = pd.DataFrame(summary_rows)
summary_df.to_csv(summary_csv_path, index=False)
hb(f"Saved summary CSV: {summary_csv_path}")
print("\nBasic Transfer Entropy summary:")
print(summary_df)

# -------------------------------------------------------
# PLOT 1: BASIC TE SUMMARY
# -------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 6), facecolor=BG)
style_ax(ax)

x = np.arange(len(summary_df))
width = 0.35

ax.bar(
    x - width / 2,
    summary_df["TE_left_to_right"].to_numpy(dtype=float),
    width=width,
    color=C_PRIMARY,
    alpha=0.90,
    label="Left → Right"
)

ax.bar(
    x + width / 2,
    summary_df["TE_right_to_left"].to_numpy(dtype=float),
    width=width,
    color=C_SECONDARY,
    alpha=0.90,
    label="Right → Left"
)

ax.set_title("Transfer Entropy Between Left and Right Hemisphere Averages")
ax.set_xlabel("Embedding Dimension")
ax.set_ylabel("Transfer Entropy")
ax.set_xticks(x)
ax.set_xticklabels([f"Dim {d}" for d in summary_df["dimension"]])

legend = ax.legend()
style_legend(legend)

save_and_show(fig, plots_directory / "te_summary.png")

# -------------------------------------------------------
# PLOT 2: TE ASYMMETRY
# -------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 6), facecolor=BG)
style_ax(ax)

ax.bar(
    [f"Dim {d}" for d in summary_df["dimension"]],
    summary_df["TE_asymmetry_LR_minus_RL"].to_numpy(dtype=float),
    color=C_DIFF,
    alpha=0.92
)

ax.axhline(0.0, color=C_MUTED, linewidth=1.0, linestyle="--")
ax.set_title("Transfer Entropy Asymmetry (Left→Right minus Right→Left)")
ax.set_xlabel("Embedding Dimension")
ax.set_ylabel("TE Asymmetry")

save_and_show(fig, plots_directory / "te_asymmetry.png")

# -------------------------------------------------------
# PLOT 3: Z-SCORED PREVIEWS
# -------------------------------------------------------
for dim in range(2):
    hb(f"Plotting z-scored preview | dim={dim + 1}")

    left_z = standardize_1d(left_reduced[:, dim])
    right_z = standardize_1d(right_reduced[:, dim])

    n_preview = min(PREVIEW_POINTS, len(left_z), len(right_z))
    t = np.arange(n_preview)

    fig, ax = plt.subplots(figsize=(12, 5), facecolor=BG)
    style_ax(ax)

    ax.plot(
        t,
        left_z[:n_preview],
        linewidth=1.4,
        linestyle="-",
        color=C_PRIMARY,
        label="Left hemisphere z-score"
    )
    ax.plot(
        t,
        right_z[:n_preview],
        linewidth=1.4,
        linestyle="--",
        color=C_SECONDARY,
        label="Right hemisphere z-score"
    )

    ax.set_title(f"Z-Scored Hemisphere-Averaged Embedded Signals | Dimension {dim + 1}")
    ax.set_xlabel("Sample Index")
    ax.set_ylabel("Z-Score")

    legend = ax.legend()
    style_legend(legend)

    save_and_show(fig, plots_directory / f"zscore_preview_dim{dim + 1}.png")

# -------------------------------------------------------
# PLOT 4: LEFT VS RIGHT SCATTER
# -------------------------------------------------------
for dim in range(2):
    hb(f"Plotting left-vs-right scatter | dim={dim + 1}")

    left_z = standardize_1d(left_reduced[:, dim])
    right_z = standardize_1d(right_reduced[:, dim])

    n_scatter = min(PREVIEW_POINTS, len(left_z), len(right_z))
    xs = left_z[:n_scatter]
    ys = right_z[:n_scatter]

    fig, ax = plt.subplots(figsize=(7, 7), facecolor=BG)
    style_ax(ax)

    ax.scatter(xs, ys, s=6, alpha=0.55, color=C_HIGHLIGHT)

    lo = min(xs.min(), ys.min())
    hi = max(xs.max(), ys.max())
    ax.plot([lo, hi], [lo, hi], linestyle="--", linewidth=1.0, color=C_DIFF, alpha=0.85)

    ax.set_title(f"Left vs Right Hemisphere Scatter | Dimension {dim + 1}")
    ax.set_xlabel("Left hemisphere z-score")
    ax.set_ylabel("Right hemisphere z-score")

    save_and_show(fig, plots_directory / f"left_vs_right_scatter_dim{dim + 1}.png")

# -------------------------------------------------------
# ROLLING TE
# -------------------------------------------------------
rolling_frames = []
rolling_csv_path = output_directory / "transfer_entropy_rolling.csv"

for dim in range(2):
    left_data = left_reduced[:, dim]
    right_data = right_reduced[:, dim]

    df_roll = rolling_te(
        left_data,
        right_data,
        window=ROLLING_WINDOW,
        step=ROLLING_STEP,
        num_bins=NUM_BINS,
        k=K_HISTORY,
        heartbeat_label=f"dim={dim + 1}"
    )
    df_roll["dimension"] = dim + 1
    rolling_frames.append(df_roll)

    if SAVE_PARTIALS:
        pd.concat(rolling_frames, ignore_index=True).to_csv(rolling_csv_path, index=False)
        hb(f"Saved partial rolling CSV after dim {dim + 1}")

rolling_df = pd.concat(rolling_frames, ignore_index=True) if rolling_frames else pd.DataFrame()
rolling_df.to_csv(rolling_csv_path, index=False)
hb(f"Saved rolling CSV: {rolling_csv_path}")

# -------------------------------------------------------
# PLOT 5: ROLLING TE CURVES
# -------------------------------------------------------
if not rolling_df.empty:
    for dim in sorted(rolling_df["dimension"].unique()):
        hb(f"Plotting rolling TE curves | dim={dim}")

        df_dim = rolling_df[rolling_df["dimension"] == dim].copy()

        fig, ax = plt.subplots(figsize=(12, 5), facecolor=BG)
        style_ax(ax)

        ax.plot(
            df_dim["center"], df_dim["TE_LR"],
            linewidth=1.9, linestyle="-", color=C_PRIMARY, label="Left → Right"
        )
        ax.plot(
            df_dim["center"], df_dim["TE_RL"],
            linewidth=1.9, linestyle="--", color=C_SECONDARY, label="Right → Left"
        )
        ax.plot(
            df_dim["center"], df_dim["TE_diff"],
            linewidth=1.9, linestyle="-.", color=C_DIFF, label="Difference"
        )

        ax.set_title(f"Rolling Transfer Entropy | Dimension {dim}")
        ax.set_xlabel("Window Center")
        ax.set_ylabel("Transfer Entropy")

        legend = ax.legend()
        style_legend(legend)

        save_and_show(fig, plots_directory / f"rolling_te_dim{dim}.png")

# -------------------------------------------------------
# LAG SWEEP TE
# -------------------------------------------------------
lag_frames = []
lag_csv_path = output_directory / "transfer_entropy_lag_sweep.csv"

for dim in range(2):
    left_data = left_reduced[:, dim]
    right_data = right_reduced[:, dim]

    df_lag = lag_sweep_te(
        left_data,
        right_data,
        lag_values=LAG_VALUES,
        num_bins=NUM_BINS,
        k=K_HISTORY,
        heartbeat_label=f"dim={dim + 1}"
    )
    df_lag["dimension"] = dim + 1
    lag_frames.append(df_lag)

    if SAVE_PARTIALS:
        pd.concat(lag_frames, ignore_index=True).to_csv(lag_csv_path, index=False)
        hb(f"Saved partial lag CSV after dim {dim + 1}")

lag_df = pd.concat(lag_frames, ignore_index=True) if lag_frames else pd.DataFrame()
lag_df.to_csv(lag_csv_path, index=False)
hb(f"Saved lag sweep CSV: {lag_csv_path}")

# -------------------------------------------------------
# PLOT 6: LAG SWEEP CURVES
# -------------------------------------------------------
if not lag_df.empty:
    for dim in sorted(lag_df["dimension"].unique()):
        hb(f"Plotting lag sweep | dim={dim}")

        df_dim = lag_df[lag_df["dimension"] == dim].copy()

        fig, ax = plt.subplots(figsize=(12, 5), facecolor=BG)
        style_ax(ax)

        ax.plot(
            df_dim["lag"], df_dim["TE_LR"],
            linewidth=1.8, linestyle="-", marker="o",
            color=C_PRIMARY, label="Left → Right"
        )
        ax.plot(
            df_dim["lag"], df_dim["TE_RL"],
            linewidth=1.8, linestyle="--", marker="o",
            color=C_SECONDARY, label="Right → Left"
        )
        ax.plot(
            df_dim["lag"], df_dim["TE_diff"],
            linewidth=1.8, linestyle="-.", marker="s",
            color=C_DIFF, label="Difference"
        )

        ax.axhline(0.0, linewidth=1.0, linestyle="--", color=C_MUTED, alpha=0.8)
        ax.axvline(0.0, linewidth=1.0, linestyle="--", color=C_MUTED, alpha=0.8)

        ax.set_title(f"Lag-Sweep Transfer Entropy | Dimension {dim}")
        ax.set_xlabel("Lag (positive = Left leads Right)")
        ax.set_ylabel("Transfer Entropy")

        legend = ax.legend()
        style_legend(legend)

        save_and_show(fig, plots_directory / f"lag_sweep_te_dim{dim}.png")

# -------------------------------------------------------
# PRINT SUMMARY
# -------------------------------------------------------
print("\nExploratory summary:")
print(summary_df)

if not rolling_df.empty:
    print("\nRolling TE head:")
    print(rolling_df.head())

if not lag_df.empty:
    print("\nLag sweep head:")
    print(lag_df.head())

hb("Done.")
```

# Hemispherical Pairs

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — Transfer entropy</h3><p>Transfer entropy is a conditional mutual information that measures whether the past of a source $X$ improves prediction of the next target value $Y_{t+1}$ beyond the target's own past. It is directional and depends on the history lengths, lag, estimator, and any discretization used by the implementation.</p><p><b>Cell role.</b> This code applies that theory to the notebook's existing EEG-derived data and preserves the surrounding notebook's plotting/output conventions. Numerical shortcuts selected by a speed mode change sampling density, iteration counts, or visualization resolution; they do not change the defining formula above.</p></div>

$$T_{X\\to Y}=\\sum p(y_{t+1},y_t^{(k)},x_t^{(l)})\\log\\frac{p(y_{t+1}\\mid y_t^{(k)},x_t^{(l)})}{p(y_{t+1}\\mid y_t^{(k)})}.$$


```python
import os
import time
import math
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

try:
    from IPython.display import display
    _HAVE_IPYTHON_DISPLAY = True
except Exception:
    display = None
    _HAVE_IPYTHON_DISPLAY = False


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"

raw_eeg_path = os.path.join(base_dir, "eeg_data_with_channels.npy")
embedding_2d_dir = os.path.join(base_dir, "embedding_data", "2dembedding_data")

results_dir = os.path.join(base_dir, "results", "transfer_entropy_hemisphere_expert")
plots_dir = os.path.join(base_dir, "plots", "transfer_entropy_hemisphere_expert")

os.makedirs(results_dir, exist_ok=True)
os.makedirs(plots_dir, exist_ok=True)


FAST_OPTION = "fast"

if FAST_OPTION == "ultra":
    window_size = 1500
    step_size = 1500
    n_bins = 5
    max_pairs = 20
    max_points_per_channel = 25000
    lag_values = [1]
    permutation_count = 0
    top_n_pairs = 15

elif FAST_OPTION == "fast":
    window_size = 1500
    step_size = 1000
    n_bins = 6
    max_pairs = 35
    max_points_per_channel = 40000
    lag_values = [1]
    permutation_count = 0
    top_n_pairs = 20

elif FAST_OPTION == "balanced":
    window_size = 1500
    step_size = 750
    n_bins = 8
    max_pairs = 80
    max_points_per_channel = 80000
    lag_values = [1, 2]
    permutation_count = 10
    top_n_pairs = 30

elif FAST_OPTION == "full":
    window_size = 2000
    step_size = 500
    n_bins = 10
    max_pairs = None
    max_points_per_channel = None
    lag_values = [1, 2, 3]
    permutation_count = 30
    top_n_pairs = 50

else:
    raise ValueError("FAST_OPTION must be one of: 'ultra', 'fast', 'balanced', 'full'.")

sampling_rate = 1000
start_time = 814.571
end_time = 921.515

SHOW_PLOTS_IN_NOTEBOOK = True
SAVE_PLOTS_TO_DISK = True

USE_EXISTING_GLOBAL_EMBEDDINGS_IF_AVAILABLE = True
PREFER_2D_EMBEDDINGS = True
EMBEDDING_TO_SCALAR_MODE = "radius"
NORMALIZE_BEFORE_TE = True
USE_QUANTILE_BINS = True

ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"
WHITE = "#E8FFFF"
BLACK = "#000000"

CYAN_DIVERGING_CMAP = LinearSegmentedColormap.from_list(
    "cyan_diverging",
    [ACCENT_DIM, BLACK, ACCENT_SOFT],
    N=256,
)

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, ACCENT_DIM, ACCENT, ACCENT_SOFT],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})

rng = np.random.default_rng(42)

eeg_channels = [
    "Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8", "FC5", "FC1", "FC2", "FC6",
    "M1", "T7", "C3", "Cz", "C4", "T8", "M2", "CP5", "CP1", "CP2", "CP6", "P7",
    "P3", "Pz", "P4", "P8", "POz", "O1", "Oz", "O2"
]

left_channels = [
    "Fp1", "F7", "F3", "FC5", "FC1", "M1", "T7", "C3", "CP5", "CP1", "P7", "P3", "O1"
]

right_channels = [
    "Fp2", "F8", "F4", "FC2", "FC6", "M2", "T8", "C4", "CP2", "CP6", "P8", "P4", "O2"
]

midline_channels = ["Fpz", "Fz", "Cz", "Pz", "POz", "Oz"]


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT)

    if grid:
        ax.grid(True, alpha=0.25, color=ACCENT_DIM, linewidth=0.7)
    else:
        ax.grid(False)


def save_fig(fig, path, show=True):
    fig.tight_layout()

    if SAVE_PLOTS_TO_DISK and path is not None:
        fig.savefig(path, dpi=220, bbox_inches="tight", facecolor=BLACK)

    if SHOW_PLOTS_IN_NOTEBOOK and show and _HAVE_IPYTHON_DISPLAY:
        display(fig)

    plt.close(fig)


def short_label(x, max_len=28):
    x = str(x)
    return x if len(x) <= max_len else x[:max_len - 1] + "…"


def safe_zscore_1d(x):
    x = np.asarray(x, dtype=float).reshape(-1)
    x = np.where(np.isfinite(x), x, np.nan)

    if np.any(~np.isfinite(x)):
        med = np.nanmedian(x)
        if not np.isfinite(med):
            med = 0.0
        x = np.nan_to_num(x, nan=med, posinf=med, neginf=med)

    mu = np.mean(x)
    sd = np.std(x) + 1e-12

    return (x - mu) / sd


def embedding_to_scalar(arr, mode="radius"):
    arr = np.asarray(arr, dtype=float)

    if arr.ndim == 1:
        x = arr.reshape(-1)

    elif arr.ndim == 2:
        if mode == "x":
            x = arr[:, 0]
        elif mode == "y":
            x = arr[:, min(1, arr.shape[1] - 1)]
        elif mode == "radius":
            centered = arr - np.nanmean(arr, axis=0, keepdims=True)
            x = np.sqrt(np.nansum(centered ** 2, axis=1))
        elif mode == "speed":
            centered = arr - np.nanmean(arr, axis=0, keepdims=True)
            dx = np.diff(centered, axis=0, prepend=centered[[0], :])
            x = np.sqrt(np.nansum(dx ** 2, axis=1))
        elif mode == "pc1":
            X = arr - np.nanmean(arr, axis=0, keepdims=True)
            X = np.nan_to_num(X, nan=0.0)
            _, _, Vt = np.linalg.svd(X, full_matrices=False)
            x = X @ Vt[0]
        else:
            raise ValueError("EMBEDDING_TO_SCALAR_MODE must be one of: 'x', 'y', 'radius', 'speed', 'pc1'.")
    else:
        x = arr.reshape(arr.shape[0], -1)
        x = np.sqrt(np.nansum((x - np.nanmean(x, axis=0, keepdims=True)) ** 2, axis=1))

    x = np.asarray(x, dtype=float).reshape(-1)

    if NORMALIZE_BEFORE_TE:
        x = safe_zscore_1d(x)

    return x


def load_from_existing_global_embeddings():
    if not USE_EXISTING_GLOBAL_EMBEDDINGS_IF_AVAILABLE:
        return None, None

    g = globals()

    if "embedding_data_list_2D_trimmed" not in g:
        return None, None

    data_obj = g["embedding_data_list_2D_trimmed"]

    if "eeg_channels" in g:
        names = list(g["eeg_channels"])
    elif "eeg_channel_names" in g:
        names = list(g["eeg_channel_names"])
    else:
        names = list(eeg_channels)

    channel_series = {}

    try:
        if isinstance(data_obj, dict):
            for ch in names:
                if ch in data_obj:
                    channel_series[ch] = embedding_to_scalar(data_obj[ch], EMBEDDING_TO_SCALAR_MODE)

        else:
            arr = np.asarray(data_obj, dtype=object)

            if len(arr) == len(names):
                for idx, ch in enumerate(names):
                    channel_series[ch] = embedding_to_scalar(arr[idx], EMBEDDING_TO_SCALAR_MODE)

    except Exception:
        return None, None

    if len(channel_series) == 0:
        return None, None

    return channel_series, "existing_global_embedding_data_list_2D_trimmed"


def load_from_2d_embedding_files():
    channel_series = {}

    for ch in eeg_channels:
        candidates = [
            os.path.join(embedding_2d_dir, f"2dembedded_{ch}.npy"),
            os.path.join(embedding_2d_dir, f"2dembedded_{ch}_trimmed.npy"),
            os.path.join(embedding_2d_dir, f"{ch}.npy"),
        ]

        matches = []

        for c in candidates:
            if os.path.exists(c):
                matches.append(c)

        if len(matches) == 0:
            glob_matches = sorted(glob.glob(os.path.join(embedding_2d_dir, f"*{ch}*.npy")))
            matches.extend(glob_matches)

        if len(matches) == 0:
            continue

        try:
            arr = np.load(matches[0], allow_pickle=True)
            channel_series[ch] = embedding_to_scalar(arr, EMBEDDING_TO_SCALAR_MODE)
        except Exception:
            continue

    if len(channel_series) == 0:
        return None, None

    return channel_series, embedding_2d_dir


def load_from_raw_eeg():
    if not os.path.exists(raw_eeg_path):
        raise FileNotFoundError(
            "Could not find embeddings or raw EEG file. Expected raw file at:\n"
            f"{raw_eeg_path}"
        )

    EEG_data = np.load(raw_eeg_path, allow_pickle=True)
    EEG_data = np.asarray(EEG_data, dtype=float)

    if EEG_data.ndim != 2:
        raise ValueError(f"Expected raw EEG shape (time, channels), got {EEG_data.shape}")

    n_channels = min(EEG_data.shape[1], len(eeg_channels))
    names = eeg_channels[:n_channels]

    start_index = max(0, int(start_time * sampling_rate))
    end_index = min(EEG_data.shape[0], int(end_time * sampling_rate))

    if end_index <= start_index:
        raise ValueError(
            f"Invalid segment indices: start={start_index}, end={end_index}, length={EEG_data.shape[0]}"
        )

    EEG_seg = EEG_data[start_index:end_index, :n_channels]

    channel_series = {}

    for i, ch in enumerate(names):
        x = EEG_seg[:, i]
        if NORMALIZE_BEFORE_TE:
            x = safe_zscore_1d(x)
        channel_series[ch] = x

    return channel_series, raw_eeg_path


def load_channel_series():
    channel_series, source = load_from_existing_global_embeddings()

    if channel_series is not None:
        return channel_series, source

    if PREFER_2D_EMBEDDINGS:
        channel_series, source = load_from_2d_embedding_files()

        if channel_series is not None:
            return channel_series, source

    return load_from_raw_eeg()


def trim_channel_series_to_common_length(channel_series, max_points=None):
    available = {k: np.asarray(v, dtype=float).reshape(-1) for k, v in channel_series.items() if len(v) > 10}

    common_len = min(len(v) for v in available.values())

    if max_points is not None:
        common_len = min(common_len, int(max_points))

    out = {}

    for ch, x in available.items():
        out[ch] = x[:common_len]

    return out, common_len


def bin_data_quantile(x, n_bins=8):
    x = np.asarray(x, dtype=float).reshape(-1)

    if np.std(x) < 1e-12:
        return np.zeros(len(x), dtype=np.int16)

    if USE_QUANTILE_BINS:
        quantiles = np.linspace(0, 1, n_bins + 1)
        edges = np.quantile(x, quantiles)
        edges = np.unique(edges)

        if len(edges) <= 2:
            edges = np.linspace(np.min(x), np.max(x), n_bins + 1)
    else:
        edges = np.linspace(np.min(x), np.max(x), n_bins + 1)

    bins = np.digitize(x, edges[1:-1], right=False).astype(np.int16)
    bins = np.clip(bins, 0, n_bins - 1)

    return bins


def transfer_entropy_binned(source_bins, target_bins, lag=1, n_bins=None):
    x = np.asarray(source_bins, dtype=np.int64).reshape(-1)
    y = np.asarray(target_bins, dtype=np.int64).reshape(-1)

    n = min(len(x), len(y))

    if n <= lag + 2:
        return np.nan

    x = x[:n]
    y = y[:n]

    if n_bins is None:
        n_bins = int(max(np.max(x), np.max(y)) + 1)

    y_future = y[lag:]
    y_past = y[:-lag]
    x_past = x[:-lag]

    valid = (
        (x_past >= 0)
        & (x_past < n_bins)
        & (y_past >= 0)
        & (y_past < n_bins)
        & (y_future >= 0)
        & (y_future < n_bins)
    )

    x_past = x_past[valid]
    y_past = y_past[valid]
    y_future = y_future[valid]

    m = len(y_future)

    if m <= 3:
        return np.nan

    counts_yf_yp_xp = np.zeros((n_bins, n_bins, n_bins), dtype=np.float64)
    counts_yp_xp = np.zeros((n_bins, n_bins), dtype=np.float64)
    counts_yf_yp = np.zeros((n_bins, n_bins), dtype=np.float64)
    counts_yp = np.zeros(n_bins, dtype=np.float64)

    np.add.at(counts_yf_yp_xp, (y_future, y_past, x_past), 1)
    np.add.at(counts_yp_xp, (y_past, x_past), 1)
    np.add.at(counts_yf_yp, (y_future, y_past), 1)
    np.add.at(counts_yp, y_past, 1)

    te = 0.0

    nz = counts_yf_yp_xp > 0
    yf_idx, yp_idx, xp_idx = np.where(nz)

    for yf, yp, xp in zip(yf_idx, yp_idx, xp_idx):
        c_xyz = counts_yf_yp_xp[yf, yp, xp]
        c_yx = counts_yp_xp[yp, xp]
        c_yy = counts_yf_yp[yf, yp]
        c_y = counts_yp[yp]

        if c_xyz > 0 and c_yx > 0 and c_yy > 0 and c_y > 0:
            p_xyz = c_xyz / m
            p_yf_given_yp_xp = c_xyz / c_yx
            p_yf_given_yp = c_yy / c_y
            te += p_xyz * np.log2((p_yf_given_yp_xp + 1e-12) / (p_yf_given_yp + 1e-12))

    return float(max(te, 0.0))


def transfer_entropy_pair_continuous(source, target, window_size=1000, step_size=500, lag=1, n_bins=8):
    source = np.asarray(source, dtype=float).reshape(-1)
    target = np.asarray(target, dtype=float).reshape(-1)

    n = min(len(source), len(target))
    source = source[:n]
    target = target[:n]

    starts = np.arange(0, n - window_size + 1, step_size)

    if starts.size == 0:
        return np.array([], dtype=float), np.array([], dtype=int)

    te_vals = []

    for s in starts:
        e = s + window_size
        xb = bin_data_quantile(source[s:e], n_bins=n_bins)
        yb = bin_data_quantile(target[s:e], n_bins=n_bins)

        te = transfer_entropy_binned(xb, yb, lag=lag, n_bins=n_bins)
        te_vals.append(te)

    return np.asarray(te_vals, dtype=float), starts


def permutation_te_threshold(source, target, window_size, step_size, lag, n_bins, n_perm=20):
    if n_perm <= 0:
        return np.nan, np.nan

    source = np.asarray(source, dtype=float).reshape(-1)
    target = np.asarray(target, dtype=float).reshape(-1)

    null_means = []

    for p in range(n_perm):
        shift = int(rng.integers(window_size, max(window_size + 1, len(source) - window_size)))

        source_perm = np.roll(source, shift)
        vals, _ = transfer_entropy_pair_continuous(
            source_perm,
            target,
            window_size=window_size,
            step_size=step_size,
            lag=lag,
            n_bins=n_bins,
        )

        vals = vals[np.isfinite(vals)]

        if len(vals):
            null_means.append(float(np.mean(vals)))

    if len(null_means) == 0:
        return np.nan, np.nan

    null_means = np.asarray(null_means, dtype=float)

    return float(np.mean(null_means)), float(np.percentile(null_means, 95))


def make_channel_pairs(left_channels, right_channels, available_channels, max_pairs=None):
    pairs = []

    left = [ch for ch in left_channels if ch in available_channels]
    right = [ch for ch in right_channels if ch in available_channels]

    for l in left:
        for r in right:
            pairs.append((l, r))

    if max_pairs is not None and len(pairs) > max_pairs:
        idx = np.sort(rng.choice(len(pairs), size=max_pairs, replace=False))
        pairs = [pairs[i] for i in idx]

    return pairs


def compute_transfer_entropy_hemispheres(channel_series, left_channels, right_channels):
    available = set(channel_series.keys())
    pairs = make_channel_pairs(left_channels, right_channels, available, max_pairs=max_pairs)

    if len(pairs) == 0:
        raise ValueError("No valid left-right channel pairs found.")

    all_lr = []
    all_rl = []
    rows = []
    lag_rows = []

    starts_ref = None
    t0 = time.perf_counter()

    for pair_idx, (left, right) in enumerate(pairs, start=1):
        x_left = channel_series[left]
        x_right = channel_series[right]

        best_lag = None
        best_lr_mean = -np.inf
        best_rl_mean = -np.inf
        best_lr_vals = None
        best_rl_vals = None
        best_starts = None

        for lag in lag_values:
            lr_vals, starts = transfer_entropy_pair_continuous(
                x_left,
                x_right,
                window_size=window_size,
                step_size=step_size,
                lag=lag,
                n_bins=n_bins,
            )

            rl_vals, _ = transfer_entropy_pair_continuous(
                x_right,
                x_left,
                window_size=window_size,
                step_size=step_size,
                lag=lag,
                n_bins=n_bins,
            )

            lr_mean = float(np.nanmean(lr_vals)) if len(lr_vals) else np.nan
            rl_mean = float(np.nanmean(rl_vals)) if len(rl_vals) else np.nan

            lag_rows.append({
                "pair": f"{left}->{right}",
                "left_channel": left,
                "right_channel": right,
                "lag": lag,
                "mean_TE_L_to_R": lr_mean,
                "mean_TE_R_to_L": rl_mean,
                "net_TE_L_minus_R": lr_mean - rl_mean if np.isfinite(lr_mean) and np.isfinite(rl_mean) else np.nan,
            })

            score = max(lr_mean if np.isfinite(lr_mean) else -np.inf, rl_mean if np.isfinite(rl_mean) else -np.inf)

            if score > max(best_lr_mean, best_rl_mean):
                best_lag = lag
                best_lr_mean = lr_mean
                best_rl_mean = rl_mean
                best_lr_vals = lr_vals
                best_rl_vals = rl_vals
                best_starts = starts

        if starts_ref is None:
            starts_ref = best_starts

        if len(best_lr_vals) == 0:
            continue

        if len(best_rl_vals) == 0:
            continue

        null_mean_lr, null_95_lr = permutation_te_threshold(
            x_left,
            x_right,
            window_size=window_size,
            step_size=step_size,
            lag=best_lag,
            n_bins=n_bins,
            n_perm=permutation_count,
        )

        null_mean_rl, null_95_rl = permutation_te_threshold(
            x_right,
            x_left,
            window_size=window_size,
            step_size=step_size,
            lag=best_lag,
            n_bins=n_bins,
            n_perm=permutation_count,
        )

        lr_mean = float(np.nanmean(best_lr_vals))
        rl_mean = float(np.nanmean(best_rl_vals))
        lr_median = float(np.nanmedian(best_lr_vals))
        rl_median = float(np.nanmedian(best_rl_vals))
        lr_max = float(np.nanmax(best_lr_vals))
        rl_max = float(np.nanmax(best_rl_vals))

        net = lr_mean - rl_mean
        total = lr_mean + rl_mean
        asymmetry_index = net / (total + 1e-12)

        lr_z = (lr_mean - null_mean_lr) / (null_95_lr - null_mean_lr + 1e-12) if np.isfinite(null_mean_lr) and np.isfinite(null_95_lr) else np.nan
        rl_z = (rl_mean - null_mean_rl) / (null_95_rl - null_mean_rl + 1e-12) if np.isfinite(null_mean_rl) and np.isfinite(null_95_rl) else np.nan

        rows.append({
            "pair_index": pair_idx,
            "left_channel": left,
            "right_channel": right,
            "pair": f"{left}-{right}",
            "best_lag": best_lag,
            "mean_TE_L_to_R": lr_mean,
            "mean_TE_R_to_L": rl_mean,
            "median_TE_L_to_R": lr_median,
            "median_TE_R_to_L": rl_median,
            "max_TE_L_to_R": lr_max,
            "max_TE_R_to_L": rl_max,
            "net_TE_L_minus_R": net,
            "abs_net_TE": abs(net),
            "bidirectional_TE_sum": total,
            "asymmetry_index": asymmetry_index,
            "null_mean_L_to_R": null_mean_lr,
            "null_95_L_to_R": null_95_lr,
            "null_mean_R_to_L": null_mean_rl,
            "null_95_R_to_L": null_95_rl,
            "normalized_surrogate_score_L_to_R": lr_z,
            "normalized_surrogate_score_R_to_L": rl_z,
            "n_windows": len(best_lr_vals),
        })

        all_lr.append(best_lr_vals)
        all_rl.append(best_rl_vals)

        print(
            f"[{pair_idx}/{len(pairs)}] {left}->{right} mean={lr_mean:.5f}, "
            f"{right}->{left} mean={rl_mean:.5f}, net={net:.5f}, lag={best_lag}",
            flush=True,
        )

    elapsed = time.perf_counter() - t0

    if len(all_lr) == 0:
        raise ValueError("No TE values computed. Try smaller window_size or more available data.")

    te_lr_matrix = np.vstack(all_lr)
    te_rl_matrix = np.vstack(all_rl)

    summary_df = pd.DataFrame(rows)
    lag_df = pd.DataFrame(lag_rows)

    time_axis = starts_ref / sampling_rate if starts_ref is not None else np.arange(te_lr_matrix.shape[1])

    return te_lr_matrix, te_rl_matrix, time_axis, summary_df, lag_df, elapsed


def pair_matrix_from_summary(summary_df, value_col):
    lefts = [ch for ch in left_channels if ch in summary_df["left_channel"].unique()]
    rights = [ch for ch in right_channels if ch in summary_df["right_channel"].unique()]

    M = np.full((len(lefts), len(rights)), np.nan, dtype=float)

    for i, l in enumerate(lefts):
        for j, r in enumerate(rights):
            sub = summary_df[(summary_df["left_channel"] == l) & (summary_df["right_channel"] == r)]

            if len(sub):
                M[i, j] = float(sub[value_col].iloc[0])

    return M, lefts, rights


def plot_te_temporal_heatmap(te_matrix, time_axis, summary_df, path, direction_label="Left to Right"):
    order = np.argsort(summary_df["mean_TE_L_to_R"].to_numpy() if direction_label == "Left to Right" else summary_df["mean_TE_R_to_L"].to_numpy())[::-1]
    te_ord = te_matrix[order]
    labels = summary_df.iloc[order]["pair"].tolist()

    fig_height = min(10.0, max(6.0, 0.22 * len(labels)))
    fig, ax = plt.subplots(figsize=(13, fig_height))

    vmax = np.nanpercentile(te_ord, 98) if np.any(np.isfinite(te_ord)) else 1.0
    vmax = max(vmax, 1e-12)

    im = ax.imshow(
        te_ord,
        aspect="auto",
        interpolation="nearest",
        cmap=CYAN_SEQUENTIAL_CMAP,
        vmin=0,
        vmax=vmax,
        extent=[time_axis[0], time_axis[-1] if len(time_axis) > 1 else time_axis[0] + 1, len(labels) - 0.5, -0.5],
    )

    ax.set_yticks(np.arange(len(labels)))
    ax.set_yticklabels([short_label(x, 24) for x in labels], color=ACCENT, fontsize=7)

    style_ax(
        ax,
        title=f"Transfer Entropy Temporal Heatmap | {direction_label}",
        xlabel="Time from segment start (s)",
        ylabel="Channel pair",
        grid=False,
    )

    cbar = fig.colorbar(im, ax=ax, fraction=0.035, pad=0.02)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)
    cbar.set_label("Transfer entropy, bits", color=ACCENT)

    save_fig(fig, path)
    return path


def plot_te_asymmetry_heatmap(summary_df, path):
    M, lefts, rights = pair_matrix_from_summary(summary_df, "asymmetry_index")

    if M.size == 0:
        return None

    fig, ax = plt.subplots(figsize=(11, 8))

    im = ax.imshow(
        M,
        aspect="auto",
        interpolation="nearest",
        cmap=CYAN_DIVERGING_CMAP,
        vmin=-1,
        vmax=1,
    )

    ax.set_xticks(np.arange(len(rights)))
    ax.set_xticklabels(rights, rotation=45, ha="right", color=ACCENT)

    ax.set_yticks(np.arange(len(lefts)))
    ax.set_yticklabels(lefts, color=ACCENT)

    for i in range(M.shape[0]):
        for j in range(M.shape[1]):
            if np.isfinite(M[i, j]):
                ax.text(
                    j,
                    i,
                    f"{M[i, j]:.2f}",
                    ha="center",
                    va="center",
                    fontsize=7,
                    color=BLACK if abs(M[i, j]) > 0.45 else ACCENT_SOFT,
                )

    style_ax(
        ax,
        title="Directional TE Asymmetry Index: L→R vs R→L",
        xlabel="Right hemisphere channel",
        ylabel="Left hemisphere channel",
        grid=False,
    )

    cbar = fig.colorbar(im, ax=ax, fraction=0.035, pad=0.02)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)
    cbar.set_label("(L→R - R→L) / (L→R + R→L)", color=ACCENT)

    save_fig(fig, path)
    return path


def plot_te_strength_heatmap(summary_df, path):
    M, lefts, rights = pair_matrix_from_summary(summary_df, "bidirectional_TE_sum")

    if M.size == 0:
        return None

    fig, ax = plt.subplots(figsize=(11, 8))

    vmax = np.nanpercentile(M, 98) if np.any(np.isfinite(M)) else 1.0
    vmax = max(vmax, 1e-12)

    im = ax.imshow(
        M,
        aspect="auto",
        interpolation="nearest",
        cmap=CYAN_SEQUENTIAL_CMAP,
        vmin=0,
        vmax=vmax,
    )

    ax.set_xticks(np.arange(len(rights)))
    ax.set_xticklabels(rights, rotation=45, ha="right", color=ACCENT)

    ax.set_yticks(np.arange(len(lefts)))
    ax.set_yticklabels(lefts, color=ACCENT)

    style_ax(
        ax,
        title="Bidirectional Interhemispheric Transfer Entropy Strength",
        xlabel="Right hemisphere channel",
        ylabel="Left hemisphere channel",
        grid=False,
    )

    cbar = fig.colorbar(im, ax=ax, fraction=0.035, pad=0.02)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)
    cbar.set_label("TE L→R + TE R→L", color=ACCENT)

    save_fig(fig, path)
    return path


def plot_top_te_pairs(summary_df, path, top_n=30):
    df = summary_df.sort_values("bidirectional_TE_sum", ascending=False).head(top_n)
    df = df.sort_values("bidirectional_TE_sum", ascending=True)

    y = np.arange(len(df))

    fig, ax = plt.subplots(figsize=(11, max(6.0, 0.28 * len(df))))

    ax.barh(
        y,
        df["mean_TE_L_to_R"],
        color=ACCENT,
        edgecolor=ACCENT_SOFT,
        alpha=0.80,
        label="L→R",
    )

    ax.barh(
        y,
        -df["mean_TE_R_to_L"],
        color=ACCENT_DIM,
        edgecolor=ACCENT,
        alpha=0.85,
        label="R→L",
    )

    ax.axvline(0, color=WHITE, linewidth=1.0, alpha=0.85)

    ax.set_yticks(y)
    ax.set_yticklabels(df["pair"], color=ACCENT, fontsize=8)

    for yi, (_, row) in zip(y, df.iterrows()):
        side = "L→R" if row["net_TE_L_minus_R"] >= 0 else "R→L"
        ax.text(
            max(row["mean_TE_L_to_R"], 0) + 0.002,
            yi,
            f" net={row['net_TE_L_minus_R']:.3g} {side}",
            va="center",
            color=ACCENT_SOFT,
            fontsize=7,
        )

    style_ax(
        ax,
        title=f"Top {len(df)} Interhemispheric TE Pairs",
        xlabel="Mean TE, bits | left side shown positive, right side shown negative",
        ylabel="Channel pair",
    )

    leg = ax.legend(loc="best", frameon=True)
    for text in leg.get_texts():
        text.set_color(ACCENT)

    save_fig(fig, path)
    return path


def plot_mean_te_timeseries(te_lr, te_rl, time_axis, path):
    lr_mean = np.nanmean(te_lr, axis=0)
    rl_mean = np.nanmean(te_rl, axis=0)
    net = lr_mean - rl_mean

    fig, ax = plt.subplots(figsize=(12, 7))

    ax.plot(
        time_axis,
        lr_mean,
        color=ACCENT,
        linewidth=2.0,
        label="Mean L→R TE",
    )

    ax.plot(
        time_axis,
        rl_mean,
        color=ACCENT_SOFT,
        linewidth=1.8,
        alpha=0.85,
        label="Mean R→L TE",
    )

    ax.fill_between(
        time_axis,
        lr_mean,
        rl_mean,
        where=net >= 0,
        color=ACCENT,
        alpha=0.15,
        label="L→R dominance",
    )

    ax.fill_between(
        time_axis,
        lr_mean,
        rl_mean,
        where=net < 0,
        color=ACCENT_DIM,
        alpha=0.18,
        label="R→L dominance",
    )

    style_ax(
        ax,
        title="Mean Interhemispheric Transfer Entropy Over Time",
        xlabel="Time from segment start (s)",
        ylabel="Transfer entropy, bits",
    )

    leg = ax.legend(loc="best", frameon=True)

    for text in leg.get_texts():
        text.set_color(ACCENT)

    save_fig(fig, path)
    return path


def plot_lag_sensitivity(lag_df, path):
    if lag_df.empty or "lag" not in lag_df.columns:
        return None

    df = lag_df.copy()
    df = df[
        np.isfinite(df["mean_TE_L_to_R"])
        & np.isfinite(df["mean_TE_R_to_L"])
        & np.isfinite(df["net_TE_L_minus_R"])
    ].copy()

    if df.empty:
        return None

    unique_lags = sorted(df["lag"].dropna().unique())

    if len(unique_lags) <= 1:
        lag_value = unique_lags[0] if len(unique_lags) else "NA"

        df["bidirectional_TE_sum"] = df["mean_TE_L_to_R"] + df["mean_TE_R_to_L"]
        df["abs_net_TE"] = np.abs(df["net_TE_L_minus_R"])

        if "pair" not in df.columns:
            df["pair"] = df["left_channel"].astype(str) + "-" + df["right_channel"].astype(str)

        local_top_n = globals().get("top_n_pairs", 25)

        plot_df = (
            df.sort_values(["abs_net_TE", "bidirectional_TE_sum"], ascending=False)
            .head(local_top_n)
            .sort_values("abs_net_TE", ascending=True)
        )

        y = np.arange(len(plot_df))

        fig, ax = plt.subplots(figsize=(11, max(6.0, 0.30 * len(plot_df))))

        ax.barh(
            y,
            plot_df["mean_TE_L_to_R"],
            color=ACCENT,
            edgecolor=ACCENT_SOFT,
            alpha=0.82,
            label="L→R",
        )

        ax.barh(
            y,
            -plot_df["mean_TE_R_to_L"],
            color=ACCENT_DIM,
            edgecolor=ACCENT,
            alpha=0.88,
            label="R→L",
        )

        ax.axvline(0, color=WHITE, linewidth=1.0, alpha=0.90)

        max_val = np.nanmax(
            np.abs(
                plot_df[["mean_TE_L_to_R", "mean_TE_R_to_L"]]
                .to_numpy(dtype=float)
            )
        )

        if np.isfinite(max_val) and max_val > 0:
            ax.set_xlim(-max_val * 1.35, max_val * 1.35)

        ax.set_yticks(y)
        ax.set_yticklabels([short_label(p, 24) for p in plot_df["pair"]], color=ACCENT, fontsize=8)

        for yi, (_, row) in zip(y, plot_df.iterrows()):
            dominant = "L→R" if row["net_TE_L_minus_R"] >= 0 else "R→L"

            ax.text(
                row["mean_TE_L_to_R"] + max_val * 0.025,
                yi,
                f"net={row['net_TE_L_minus_R']:.4f} {dominant}",
                va="center",
                ha="left",
                color=ACCENT_SOFT,
                fontsize=7,
            )

        style_ax(
            ax,
            title=(
                f"Transfer Entropy Directionality at Lag {lag_value}\n"
                "single-lag fast mode: pair-level L→R vs R→L comparison"
            ),
            xlabel="Mean TE, bits | L→R positive, R→L negative",
            ylabel="Channel pair",
        )

        leg = ax.legend(loc="best", frameon=True)
        for text in leg.get_texts():
            text.set_color(ACCENT)

        save_fig(fig, path)
        return path

    grouped = (
        df.groupby("lag")
        .agg(
            mean_L_to_R=("mean_TE_L_to_R", "mean"),
            mean_R_to_L=("mean_TE_R_to_L", "mean"),
            mean_net=("net_TE_L_minus_R", "mean"),
            sem_L_to_R=("mean_TE_L_to_R", lambda x: np.nanstd(x) / np.sqrt(max(np.sum(np.isfinite(x)), 1))),
            sem_R_to_L=("mean_TE_R_to_L", lambda x: np.nanstd(x) / np.sqrt(max(np.sum(np.isfinite(x)), 1))),
            sem_net=("net_TE_L_minus_R", lambda x: np.nanstd(x) / np.sqrt(max(np.sum(np.isfinite(x)), 1))),
            n_pairs=("pair", "count") if "pair" in df.columns else ("lag", "count"),
        )
        .reset_index()
        .sort_values("lag")
    )

    fig, ax = plt.subplots(figsize=(10.5, 7.0))

    ax.plot(
        grouped["lag"],
        grouped["mean_L_to_R"],
        marker="o",
        color=ACCENT,
        linewidth=2.0,
        label="L→R mean",
    )

    ax.fill_between(
        grouped["lag"],
        grouped["mean_L_to_R"] - grouped["sem_L_to_R"],
        grouped["mean_L_to_R"] + grouped["sem_L_to_R"],
        color=ACCENT,
        alpha=0.12,
    )

    ax.plot(
        grouped["lag"],
        grouped["mean_R_to_L"],
        marker="o",
        color=ACCENT_SOFT,
        linewidth=1.8,
        label="R→L mean",
    )

    ax.fill_between(
        grouped["lag"],
        grouped["mean_R_to_L"] - grouped["sem_R_to_L"],
        grouped["mean_R_to_L"] + grouped["sem_R_to_L"],
        color=ACCENT_SOFT,
        alpha=0.10,
    )

    ax.plot(
        grouped["lag"],
        grouped["mean_net"],
        marker="s",
        color=WHITE,
        linewidth=1.6,
        label="Net L-R",
    )

    ax.fill_between(
        grouped["lag"],
        grouped["mean_net"] - grouped["sem_net"],
        grouped["mean_net"] + grouped["sem_net"],
        color=WHITE,
        alpha=0.08,
    )

    ax.axhline(0, color=WHITE, linewidth=0.9, alpha=0.65)

    ax.set_xticks(grouped["lag"].to_numpy())

    best_idx = int(np.nanargmax(np.abs(grouped["mean_net"].to_numpy(dtype=float))))
    best_lag = grouped.iloc[best_idx]["lag"]
    best_net = grouped.iloc[best_idx]["mean_net"]

    ax.annotate(
        f"strongest net lag={best_lag}\nnet={best_net:.4f}",
        xy=(best_lag, best_net),
        xytext=(8, 18),
        textcoords="offset points",
        color=ACCENT_SOFT,
        fontsize=9,
        arrowprops=dict(arrowstyle="->", color=ACCENT_SOFT, lw=1.0),
    )

    style_ax(
        ax,
        title="Transfer Entropy Lag Sensitivity with Pair-Level SEM",
        xlabel="Lag, samples",
        ylabel="Mean TE / net TE, bits",
    )

    leg = ax.legend(loc="best", frameon=True)
    for text in leg.get_texts():
        text.set_color(ACCENT)

    save_fig(fig, path)
    return path


def plot_surrogate_significance(summary_df, path):
    if "normalized_surrogate_score_L_to_R" not in summary_df.columns:
        return None

    df = summary_df.copy()
    df["max_surrogate_score"] = np.nanmax(
        df[["normalized_surrogate_score_L_to_R", "normalized_surrogate_score_R_to_L"]].to_numpy(dtype=float),
        axis=1,
    )

    df = df.sort_values("max_surrogate_score", ascending=False).head(top_n_pairs)
    df = df.sort_values("max_surrogate_score", ascending=True)

    fig, ax = plt.subplots(figsize=(11, max(6.0, 0.28 * len(df))))

    y = np.arange(len(df))

    ax.barh(
        y,
        df["normalized_surrogate_score_L_to_R"],
        color=ACCENT,
        edgecolor=ACCENT_SOFT,
        alpha=0.82,
        label="L→R normalized surrogate score",
    )

    ax.barh(
        y,
        -df["normalized_surrogate_score_R_to_L"],
        color=ACCENT_DIM,
        edgecolor=ACCENT,
        alpha=0.86,
        label="R→L normalized surrogate score",
    )

    ax.axvline(1.0, color=WHITE, linestyle="--", linewidth=1.0, alpha=0.8)
    ax.axvline(-1.0, color=WHITE, linestyle="--", linewidth=1.0, alpha=0.8)
    ax.axvline(0.0, color=WHITE, linewidth=1.0, alpha=0.8)

    ax.set_yticks(y)
    ax.set_yticklabels(df["pair"], color=ACCENT, fontsize=8)

    style_ax(
        ax,
        title="Surrogate-Normalized Transfer Entropy Scores",
        xlabel="Score relative to circular-shift null; ±1 roughly exceeds null 95% distance",
        ylabel="Pair",
    )

    leg = ax.legend(loc="best", frameon=True)

    for text in leg.get_texts():
        text.set_color(ACCENT)

    save_fig(fig, path)
    return path


def custom_force_layout_from_summary(summary_df, threshold_quantile=0.75, iterations=700):
    channels = sorted(set(summary_df["left_channel"]).union(set(summary_df["right_channel"])))
    idx = {ch: i for i, ch in enumerate(channels)}
    n = len(channels)

    W = np.zeros((n, n), dtype=float)

    cutoff = np.nanquantile(summary_df["bidirectional_TE_sum"], threshold_quantile)

    for _, row in summary_df.iterrows():
        if row["bidirectional_TE_sum"] >= cutoff:
            i = idx[row["left_channel"]]
            j = idx[row["right_channel"]]
            W[i, j] = row["bidirectional_TE_sum"]
            W[j, i] = row["bidirectional_TE_sum"]

    theta = np.linspace(0, 2 * np.pi, n, endpoint=False)
    pos = np.column_stack([np.cos(theta), np.sin(theta)])

    for i, ch in enumerate(channels):
        if ch in left_channels:
            pos[i, 0] -= 0.7
        elif ch in right_channels:
            pos[i, 0] += 0.7

    pos += 0.03 * rng.normal(size=pos.shape)

    edges = edge_list_from_matrix(W)

    if len(edges) == 0:
        return channels, pos, W

    max_w = max(w for _, _, w in edges)

    area = 5.0
    k = np.sqrt(area / max(n, 1))
    temp = 0.06

    for _ in range(iterations):
        disp = np.zeros_like(pos)

        for i in range(n):
            delta = pos[i] - pos
            dist = np.linalg.norm(delta, axis=1) + 1e-9
            repulse = (k * k / dist)[:, None] * delta / dist[:, None]
            repulse[i] = 0
            disp[i] += np.sum(repulse, axis=0)

        for i, j, w in edges:
            delta = pos[i] - pos[j]
            dist = np.linalg.norm(delta) + 1e-9
            ww = w / (max_w + 1e-12)
            attract = (dist * dist / k) * delta / dist * (0.45 + 1.8 * ww)
            disp[i] -= attract
            disp[j] += attract

        for i, ch in enumerate(channels):
            if ch in left_channels:
                disp[i, 0] += -0.015
            elif ch in right_channels:
                disp[i, 0] += 0.015

        lengths = np.linalg.norm(disp, axis=1) + 1e-9
        pos += disp / lengths[:, None] * np.minimum(lengths, temp)[:, None]
        temp *= 0.995

    pos -= pos.mean(axis=0, keepdims=True)
    scale = np.max(np.linalg.norm(pos, axis=1))

    if scale > 0:
        pos /= scale

    return channels, pos, W


def edge_list_from_matrix(W):
    edges = []
    n = W.shape[0]

    for i in range(n):
        for j in range(i + 1, n):
            if W[i, j] > 0:
                edges.append((i, j, float(W[i, j])))

    return edges


def plot_te_network(summary_df, path):
    channels, pos, W = custom_force_layout_from_summary(summary_df, threshold_quantile=0.70)
    n = len(channels)

    fig, ax = plt.subplots(figsize=(11, 8))

    edges = edge_list_from_matrix(W)

    if len(edges) == 0:
        ax.text(0.5, 0.5, "No strong TE edges", transform=ax.transAxes, ha="center", va="center", color=ACCENT)
        style_ax(ax, title="Interhemispheric TE Network", grid=False)
        save_fig(fig, path)
        return path

    max_w = max(w for _, _, w in edges)

    for i, j, w in edges:
        ax.plot(
            [pos[i, 0], pos[j, 0]],
            [pos[i, 1], pos[j, 1]],
            color=ACCENT,
            linewidth=0.5 + 5.0 * w / (max_w + 1e-12),
            alpha=0.25 + 0.60 * w / (max_w + 1e-12),
            zorder=1,
        )

    strengths = W.sum(axis=1)
    max_strength = np.max(strengths) if np.max(strengths) > 0 else 1.0

    node_colors = []
    for ch in channels:
        if ch in left_channels:
            node_colors.append(ACCENT)
        elif ch in right_channels:
            node_colors.append(ACCENT_SOFT)
        else:
            node_colors.append(WHITE)

    ax.scatter(
        pos[:, 0],
        pos[:, 1],
        s=300 + 1000 * strengths / (max_strength + 1e-12),
        c=node_colors,
        edgecolors=WHITE,
        linewidths=0.9,
        alpha=0.92,
        zorder=3,
    )

    for i, ch in enumerate(channels):
        ax.text(
            pos[i, 0],
            pos[i, 1],
            ch,
            ha="center",
            va="center",
            color=BLACK,
            fontsize=8,
            fontweight="bold",
            zorder=4,
        )

    style_ax(
        ax,
        title="Custom Interhemispheric Transfer Entropy Network\nedge width = bidirectional TE strength",
        xlabel="Custom hemisphere-constrained force layout 1",
        ylabel="Custom hemisphere-constrained force layout 2",
        grid=False,
    )

    ax.set_xticks([])
    ax.set_yticks([])

    save_fig(fig, path)
    return path


def save_outputs(te_lr, te_rl, time_axis, summary_df, lag_df, source_description, elapsed_compute):
    te_lr_path = os.path.join(results_dir, f"te_left_to_right_{FAST_OPTION}.npy")
    te_rl_path = os.path.join(results_dir, f"te_right_to_left_{FAST_OPTION}.npy")
    time_axis_path = os.path.join(results_dir, f"te_time_axis_{FAST_OPTION}.npy")
    summary_csv = os.path.join(results_dir, f"te_pair_summary_{FAST_OPTION}.csv")
    lag_csv = os.path.join(results_dir, f"te_lag_sensitivity_{FAST_OPTION}.csv")
    summary_txt = os.path.join(results_dir, f"te_expert_summary_{FAST_OPTION}.txt")

    np.save(te_lr_path, te_lr)
    np.save(te_rl_path, te_rl)
    np.save(time_axis_path, time_axis)
    summary_df.to_csv(summary_csv, index=False)
    lag_df.to_csv(lag_csv, index=False)

    with open(summary_txt, "w") as f:
        f.write("Expert Interhemispheric Transfer Entropy Analysis\n")
        f.write("================================================\n\n")
        f.write("Transfer entropy implementation: custom NumPy discrete TE over quantile-binned windows.\n")
        f.write("Direction: TE(source -> target) = I(source_past; target_future | target_past).\n\n")
        f.write(f"base_dir: {base_dir}\n")
        f.write(f"data_source: {source_description}\n")
        f.write(f"FAST_OPTION: {FAST_OPTION}\n")
        f.write(f"window_size: {window_size}\n")
        f.write(f"step_size: {step_size}\n")
        f.write(f"n_bins: {n_bins}\n")
        f.write(f"lag_values: {lag_values}\n")
        f.write(f"permutation_count: {permutation_count}\n")
        f.write(f"max_pairs: {max_pairs}\n")
        f.write(f"max_points_per_channel: {max_points_per_channel}\n")
        f.write(f"embedding_scalar_mode: {EMBEDDING_TO_SCALAR_MODE}\n")
        f.write(f"elapsed_compute_sec: {elapsed_compute:.3f}\n\n")

        f.write("Top pairs by bidirectional TE:\n")
        for _, row in summary_df.sort_values("bidirectional_TE_sum", ascending=False).head(20).iterrows():
            f.write(
                f"{row['pair']}: "
                f"L->R={row['mean_TE_L_to_R']:.6g}, "
                f"R->L={row['mean_TE_R_to_L']:.6g}, "
                f"net={row['net_TE_L_minus_R']:.6g}, "
                f"AI={row['asymmetry_index']:.6g}, "
                f"lag={row['best_lag']}\n"
            )

    return {
        "te_left_to_right_npy": te_lr_path,
        "te_right_to_left_npy": te_rl_path,
        "te_time_axis_npy": time_axis_path,
        "pair_summary_csv": summary_csv,
        "lag_sensitivity_csv": lag_csv,
        "summary_txt": summary_txt,
    }


def print_summary(summary_df, lag_df, source_description, elapsed_compute):
    print("\n" + "=" * 92)
    print("EXPERT INTERHEMISPHERIC TRANSFER ENTROPY SUMMARY")
    print("=" * 92)
    print(f"Data source: {source_description}")
    print(f"FAST_OPTION: {FAST_OPTION}")
    print(f"window_size={window_size}, step_size={step_size}, n_bins={n_bins}, lags={lag_values}")
    print(f"Pairs computed: {len(summary_df)}")
    print(f"Compute elapsed: {elapsed_compute:.2f} sec")

    cols = [
        "pair",
        "best_lag",
        "mean_TE_L_to_R",
        "mean_TE_R_to_L",
        "net_TE_L_minus_R",
        "asymmetry_index",
        "bidirectional_TE_sum",
        "normalized_surrogate_score_L_to_R",
        "normalized_surrogate_score_R_to_L",
    ]

    cols = [c for c in cols if c in summary_df.columns]

    print("\nTop 15 by bidirectional TE strength:")
    print(summary_df.sort_values("bidirectional_TE_sum", ascending=False)[cols].head(15).to_string(index=False))

    print("\nTop 15 by absolute directional asymmetry:")
    print(summary_df.sort_values("abs_net_TE", ascending=False)[cols].head(15).to_string(index=False))

    if not lag_df.empty:
        print("\nLag sensitivity summary:")
        print(
            lag_df.groupby("lag")
            .agg(
                mean_L_to_R=("mean_TE_L_to_R", "mean"),
                mean_R_to_L=("mean_TE_R_to_L", "mean"),
                mean_net=("net_TE_L_minus_R", "mean"),
            )
            .reset_index()
            .to_string(index=False)
        )


pipeline_start = time.perf_counter()

print("Running expert interhemispheric transfer entropy analysis", flush=True)
print(f"FAST_OPTION: {FAST_OPTION}", flush=True)
print(f"window_size={window_size}, step_size={step_size}, n_bins={n_bins}", flush=True)
print(f"lag_values={lag_values}, permutation_count={permutation_count}", flush=True)

channel_series_raw, source_description = load_channel_series()
channel_series, common_len = trim_channel_series_to_common_length(
    channel_series_raw,
    max_points=max_points_per_channel,
)

available_left = [ch for ch in left_channels if ch in channel_series]
available_right = [ch for ch in right_channels if ch in channel_series]

print(f"Data source: {source_description}", flush=True)
print(f"Common length: {common_len}", flush=True)
print(f"Available left channels: {available_left}", flush=True)
print(f"Available right channels: {available_right}", flush=True)

te_lr_matrix, te_rl_matrix, te_time_axis, te_summary_df, te_lag_df, te_compute_elapsed = compute_transfer_entropy_hemispheres(
    channel_series,
    available_left,
    available_right,
)

output_paths = save_outputs(
    te_lr_matrix,
    te_rl_matrix,
    te_time_axis,
    te_summary_df,
    te_lag_df,
    source_description,
    te_compute_elapsed,
)

plot_paths = []

plot_paths.append(plot_te_temporal_heatmap(
    te_lr_matrix,
    te_time_axis,
    te_summary_df,
    os.path.join(plots_dir, f"01_te_temporal_heatmap_left_to_right_{FAST_OPTION}.png"),
    direction_label="Left to Right",
))

plot_paths.append(plot_te_temporal_heatmap(
    te_rl_matrix,
    te_time_axis,
    te_summary_df,
    os.path.join(plots_dir, f"02_te_temporal_heatmap_right_to_left_{FAST_OPTION}.png"),
    direction_label="Right to Left",
))

plot_paths.append(plot_te_strength_heatmap(
    te_summary_df,
    os.path.join(plots_dir, f"03_bidirectional_te_strength_heatmap_{FAST_OPTION}.png"),
))

plot_paths.append(plot_te_asymmetry_heatmap(
    te_summary_df,
    os.path.join(plots_dir, f"04_te_directional_asymmetry_heatmap_{FAST_OPTION}.png"),
))

plot_paths.append(plot_top_te_pairs(
    te_summary_df,
    os.path.join(plots_dir, f"05_top_te_pairs_bidirectional_and_directional_{FAST_OPTION}.png"),
    top_n=top_n_pairs,
))

plot_paths.append(plot_mean_te_timeseries(
    te_lr_matrix,
    te_rl_matrix,
    te_time_axis,
    os.path.join(plots_dir, f"06_mean_te_timeseries_directionality_{FAST_OPTION}.png"),
))

plot_paths.append(plot_lag_sensitivity(
    te_lag_df,
    os.path.join(plots_dir, f"07_te_lag_sensitivity_{FAST_OPTION}.png"),
))

if permutation_count > 0:
    plot_paths.append(plot_surrogate_significance(
        te_summary_df,
        os.path.join(plots_dir, f"08_te_surrogate_significance_{FAST_OPTION}.png"),
    ))

plot_paths.append(plot_te_network(
    te_summary_df,
    os.path.join(plots_dir, f"09_interhemispheric_te_network_{FAST_OPTION}.png"),
))

plot_paths = [p for p in plot_paths if p is not None]

pipeline_elapsed = time.perf_counter() - pipeline_start

print_summary(
    te_summary_df,
    te_lag_df,
    source_description,
    te_compute_elapsed,
)

print("\nSaved outputs:")
for k, v in output_paths.items():
    print(f"{k}: {v}")

print("\nSaved plots:")
for p in plot_paths:
    print(p)

print(f"\nTotal pipeline elapsed time: {pipeline_elapsed / 60:.2f} min")
print("Done.")
```

# Regional

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — Transfer entropy</h3><p>Transfer entropy is a conditional mutual information that measures whether the past of a source $X$ improves prediction of the next target value $Y_{t+1}$ beyond the target's own past. It is directional and depends on the history lengths, lag, estimator, and any discretization used by the implementation.</p><p><b>Cell role.</b> This code applies that theory to the notebook's existing EEG-derived data and preserves the surrounding notebook's plotting/output conventions. Numerical shortcuts selected by a speed mode change sampling density, iteration counts, or visualization resolution; they do not change the defining formula above.</p></div>

$$T_{X\\to Y}=\\sum p(y_{t+1},y_t^{(k)},x_t^{(l)})\\log\\frac{p(y_{t+1}\\mid y_t^{(k)},x_t^{(l)})}{p(y_{t+1}\\mid y_t^{(k)})}.$$


```python
import zipfile
import numpy as np
import os
import shutil
from pyinform import transfer_entropy

# Path to the zipped 2D embedded data
base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
zip_file_path = os.path.join(base_dir, "2d_embedded_data.zip")
extraction_dir = os.path.join(base_dir, "_tmp_2d_embedded_transfer_entropy_regional")

embedding_data_list_2D = []

# Extract zipped data to a temporary directory
with zipfile.ZipFile(zip_file_path, 'r') as zipf:
    zipf.extractall(extraction_dir)

eeg_channels = ['Fp1', 'Fpz', 'Fp2', 'F7', 'F3', 'Fz', 'F4', 'F8', 'FC5', 'FC1', 'FC2', 'FC6',
                'M1', 'T7', 'C3', 'Cz', 'C4', 'T8', 'M2', 'CP5', 'CP1', 'CP2', 'CP6',
                'P7', 'P3', 'Pz', 'P4', 'P8', 'POz', 'O1', 'Oz', 'O2']

for channel in eeg_channels:
    file_path = os.path.join(extraction_dir, f'2dembedded_{channel}_data.npy')
    embedding_data = np.load(file_path)
    embedding_data_list_2D.append(embedding_data)

FAST_OPTION = "fast"  # options: "ultra", "fast", "balanced", "full"
_speed_max_points = {"ultra": 50000, "fast": 150000, "balanced": 500000, "full": 4227688}[FAST_OPTION]
desired_length = _speed_max_points
embedding_data_list_2D_trimmed = [data[:desired_length] for data in embedding_data_list_2D]

# Brain regions
frontal_channels = ['Fp1', 'Fpz', 'Fp2', 'F7', 'F3', 'Fz', 'F4', 'F8']
temporal_channels = ['T7', 'T8']
parietal_channels = ['CP5', 'CP1', 'CP2', 'CP6', 'P7', 'P3', 'Pz', 'P4', 'P8']
occipital_channels = ['O1', 'Oz', 'O2']

frontal_indices = [eeg_channels.index(ch) for ch in frontal_channels]
temporal_indices = [eeg_channels.index(ch) for ch in temporal_channels]
parietal_indices = [eeg_channels.index(ch) for ch in parietal_channels]
occipital_indices = [eeg_channels.index(ch) for ch in occipital_channels]

# Getting average for each region
frontal_data_avg = np.mean([embedding_data_list_2D_trimmed[i] for i in frontal_indices], axis=0)
temporal_data_avg = np.mean([embedding_data_list_2D_trimmed[i] for i in temporal_indices], axis=0)
parietal_data_avg = np.mean([embedding_data_list_2D_trimmed[i] for i in parietal_indices], axis=0)
occipital_data_avg = np.mean([embedding_data_list_2D_trimmed[i] for i in occipital_indices], axis=0)

# Use one of the dimensions for Transfer Entropy calculation
frontal_data = frontal_data_avg[:, 0]
temporal_data = temporal_data_avg[:, 0]
parietal_data = parietal_data_avg[:, 0]
occipital_data = occipital_data_avg[:, 0]

def bin_data(data, num_bins):
    hist, bins = np.histogram(data, bins=num_bins)
    binned_data = np.digitize(data, bins[:-1]) - 1
    return binned_data

frontal_data_binned = bin_data(frontal_data, {"ultra": 16, "fast": 32, "balanced": 64, "full": 1000}[FAST_OPTION])
temporal_data_binned = bin_data(temporal_data, {"ultra": 16, "fast": 32, "balanced": 64, "full": 1000}[FAST_OPTION])
parietal_data_binned = bin_data(parietal_data, {"ultra": 16, "fast": 32, "balanced": 64, "full": 1000}[FAST_OPTION])
occipital_data_binned = bin_data(occipital_data, {"ultra": 16, "fast": 32, "balanced": 64, "full": 1000}[FAST_OPTION])

# Store the regional data in a dictionary for easier processing
region_data_binned = {
    "Frontal": frontal_data_binned,
    "Temporal": temporal_data_binned,
    "Parietal": parietal_data_binned,
    "Occipital": occipital_data_binned
}

k = 1


# Directory to save the transfer entropy results
results_dir = os.path.join(base_dir, "results", "transfer_entropy_regional")
if not os.path.exists(results_dir):
    os.makedirs(results_dir)  # Create the directory if it doesn't exist

# Dictionary to store results
TE_results = {}

# Compute and print the transfer entropy from each region to every other region
for source_region, source_data in region_data_binned.items():
    for target_region, target_data in region_data_binned.items():
        if source_region != target_region:
            try:
                TE_value = transfer_entropy(source_data, target_data, k)
                key = f"{source_region}_to_{target_region}"
                TE_results[key] = TE_value
                print(f"Transfer Entropy from {source_region} to {target_region} (2D embeddings): {TE_value}")
            except Exception as e:
                print(f"Error computing Transfer Entropy from {source_region} to {target_region}: {e}")

# Save the computed Transfer Entropy results
results_file_path = os.path.join(results_dir, 'regional_transfer_entropy_results.npy')
np.save(results_file_path, TE_results)

print(f"Transfer entropy results saved to: {results_file_path}")
```

The cell does not introduce an independent physical law; its correctness therefore depends mainly on consistent indexing, normalization, units, and use of the estimator defined by the surrounding section.</p><p><b>Cell role.</b> The implementation uses the same dataset, channel ordering, result directories, and plotting conventions as the rest of this notebook.</p></div>


### Full Granularity

This section evaluates the entropy estimators at the native analysis granularity before any optional multiscale coarse-graining. Computationally expensive estimators later reuse the notebook-wide `full / balanced / fast / ultra` controls; changing speed mode changes sampling/iteration budgets, not the entropy definitions.


<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — Transfer entropy</h3><p>Transfer entropy is a conditional mutual information that measures whether the past of a source $X$ improves prediction of the next target value $Y_{t+1}$ beyond the target's own past. It is directional and depends on the history lengths, lag, estimator, and any discretization used by the implementation.</p><p><b>Cell role.</b> This code applies that theory to the notebook's existing EEG-derived data and preserves the surrounding notebook's plotting/output conventions. Numerical shortcuts selected by a speed mode change sampling density, iteration counts, or visualization resolution; they do not change the defining formula above.</p></div>

$$T_{X\\to Y}=\\sum p(y_{t+1},y_t^{(k)},x_t^{(l)})\\log\\frac{p(y_{t+1}\\mid y_t^{(k)},x_t^{(l)})}{p(y_{t+1}\\mid y_t^{(k)})}.$$


```python
import zipfile
import numpy as np
import os
from pyinform import transfer_entropy
import shutil

# Paths and constants
base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
zip_file_path = os.path.join(base_dir, "2d_embedded_data.zip")
extraction_dir = os.path.join(base_dir, "_tmp_2d_embedded_transfer_entropy_full")
results_dir = os.path.join(base_dir, "results", "transfer_entropy_full_granularity")
results_file_name = 'full_granularity_transfer_entropy_results.npy'

eeg_channels = ['Fp1', 'Fpz', 'Fp2', 'F7', 'F3', 'Fz', 'F4', 'F8', 'FC5', 'FC1', 'FC2', 'FC6',
                'M1', 'T7', 'C3', 'Cz', 'C4', 'T8', 'M2', 'CP5', 'CP1', 'CP2', 'CP6',
                'P7', 'P3', 'Pz', 'P4', 'P8', 'POz', 'O1', 'Oz', 'O2']

FAST_OPTION = "ultra"  # options: "ultra", "fast", "balanced", "full"
if FAST_OPTION not in {"ultra", "fast", "balanced", "full"}:
    raise ValueError("FAST_OPTION must be one of: ultra, fast, balanced, full")
desired_length = {"ultra": 30000, "fast": 80000, "balanced": 250000, "full": 4227688}[FAST_OPTION]
num_bins = {"ultra": 12, "fast": 24, "balanced": 48, "full": 1000}[FAST_OPTION]
k = 1

# Helper functions
def extract_data(zip_path, extraction_path):
    with zipfile.ZipFile(zip_path, 'r') as zipf:
        zipf.extractall(extraction_path)
    return extraction_path

def load_embedded_data(channels, extraction_path):
    data = []
    for channel in channels:
        file_path = os.path.join(extraction_path, f'2dembedded_{channel}_data.npy')
        embedding_data = np.load(file_path)
        data.append(embedding_data)
    return data

def bin_data(data, bins):
    hist, bin_edges = np.histogram(data, bins=bins)
    binned_data = np.digitize(data, bin_edges[:-1]) - 1  # subtract 1 to start binning from 0
    return binned_data

# Create results directory if it doesn't exist
if not os.path.exists(results_dir):
    os.makedirs(results_dir)

# Main execution
extract_data(zip_file_path, extraction_dir)
embedding_data_list_2D = load_embedded_data(eeg_channels, extraction_dir)

# Trimming and binning the data
embedding_data_list_2D_trimmed = [data[:desired_length, 0] for data in embedding_data_list_2D] 
embedding_data_list_2D_binned = [bin_data(data, num_bins) for data in embedding_data_list_2D_trimmed]

# Dictionary to store transfer entropy results
TE_results = {}

# Compute Transfer Entropy for channel pairs. Faster modes limit the number of source channels; full preserves all pairs.
_source_limit = {"ultra": 8, "fast": 16, "balanced": 24, "full": len(eeg_channels)}[FAST_OPTION]
for i, source_channel in enumerate(eeg_channels[:_source_limit]):
    for j, target_channel in enumerate(eeg_channels):
        if i != j:  # To avoid computing Transfer Entropy for the same channel
            try:
                TE = transfer_entropy(embedding_data_list_2D_binned[i], embedding_data_list_2D_binned[j], k)
                key = f"{source_channel}_to_{target_channel}"
                TE_results[key] = TE
                print(f"Transfer Entropy from {source_channel} to {target_channel}: {TE}")
            except Exception as e:
                print(f"Error computing Transfer Entropy from {source_channel} to {target_channel}: {e}")

# Save the computed Transfer Entropy results to the specified directory
results_file_path = os.path.join(results_dir, results_file_name)
np.save(results_file_path, TE_results)

print(f"Transfer entropy results saved to: {results_file_path}")

# Optionally remove the extraction directory after loading the data
shutil.rmtree(extraction_dir)
```

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;"><h3 style="color:#E8E8E8;margin-top:0;">Theory and method — Discrete Fourier analysis</h3><p>The discrete Fourier transform decomposes a finite sequence into complex sinusoids. The squared magnitude is related to spectral power, while the complex argument carries phase. Correct frequency axes require the sampling frequency used for the data actually supplied to the transform.</p><p><b>Cell role.</b> This code applies that theory to the notebook's existing EEG-derived data and preserves the surrounding notebook's plotting/output conventions. Numerical shortcuts selected by a speed mode change sampling density, iteration counts, or visualization resolution; they do not change the defining formula above.</p></div>

$$X_k=\\sum_{n=0}^{N-1}x_n e^{-i2\\pi kn/N}.$$


```python
import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -------------------------------------------------------
# CONFIG
# -------------------------------------------------------
base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
lyap_dir = os.path.join(base_dir, "results", "lyapunov_custom")
out_dir = os.path.join(base_dir, "results", "ks_entropy_from_lyapunov")
plots_dir = os.path.join(out_dir, "plots")
os.makedirs(out_dir, exist_ok=True)
os.makedirs(plots_dir, exist_ok=True)

ACCENT = "cyan"
BG = "black"
eps = 1e-12

# IMPORTANT:
# If your Lyapunov code used DT = 1.0, then KS entropy here is "per sample".
# If your Lyapunov code used DT = 1/fs, then KS entropy here is "per second".
KS_UNITS = "per sample"

plt.rcParams.update({
    "figure.facecolor": BG,
    "axes.facecolor": BG,
    "savefig.facecolor": BG,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT,
    "legend.facecolor": BG,
    "legend.edgecolor": ACCENT,
})

# -------------------------------------------------------
# STYLE
# -------------------------------------------------------
def style_ax(ax):
    ax.set_facecolor(BG)
    ax.tick_params(colors=ACCENT)
    ax.grid(True, alpha=0.20, color=ACCENT)
    for spine in ax.spines.values():
        spine.set_color(ACCENT)

# -------------------------------------------------------
# HELPERS
# -------------------------------------------------------
def find_latest_lyapunov_csv(results_dir):
    csvs = sorted(glob.glob(os.path.join(results_dir, "lyap_spectrum_*.csv")))
    if not csvs:
        raise FileNotFoundError(f"No lyap_spectrum_*.csv files found in: {results_dir}")
    return max(csvs, key=os.path.getmtime)

def get_exp_cols(df):
    exp_cols = [c for c in df.columns if str(c).startswith("Exp")]
    exp_cols = sorted(exp_cols, key=lambda c: int(str(c).replace("Exp", "")))
    return exp_cols

def infer_channel_from_stem(stem):
    stem = str(stem)
    if "dembedded_" in stem:
        return stem.split("dembedded_", 1)[1]
    return stem

def infer_embedding_dimension(row):
    if "embedding_dimension" in row.index and pd.notna(row["embedding_dimension"]):
        try:
            return int(row["embedding_dimension"])
        except Exception:
            pass

    stem = str(row["stem"])
    if "dembedded_" in stem:
        left = stem.split("dembedded_", 1)[0]
        try:
            return int(left)
        except Exception:
            pass

    if "state_dim" in row.index and pd.notna(row["state_dim"]):
        try:
            return int(row["state_dim"])
        except Exception:
            pass

    return np.nan

def row_exponents(row, exp_cols):
    vals = []
    for c in exp_cols:
        if c in row.index and pd.notna(row[c]):
            vals.append(float(row[c]))
    return np.asarray(vals, dtype=float)

def ks_entropy_estimate(lyapunov_exponents):
    """
    Pesin-type estimate:
        h_KS ≈ sum of positive Lyapunov exponents
    """
    exps = np.asarray(lyapunov_exponents, dtype=float)
    exps = exps[np.isfinite(exps)]
    if exps.size == 0:
        return np.nan
    return float(np.sum(exps[exps > 0]))

def positive_lyapunov_count(lyapunov_exponents):
    exps = np.asarray(lyapunov_exponents, dtype=float)
    exps = exps[np.isfinite(exps)]
    return int(np.sum(exps > 0))

def largest_lyapunov(lyapunov_exponents):
    exps = np.asarray(lyapunov_exponents, dtype=float)
    exps = exps[np.isfinite(exps)]
    if exps.size == 0:
        return np.nan
    return float(np.max(exps))

# -------------------------------------------------------
# LOAD LATEST LYAPUNOV CSV
# -------------------------------------------------------
lyap_csv = find_latest_lyapunov_csv(lyap_dir)
df = pd.read_csv(lyap_csv)

print("Using Lyapunov CSV:")
print(lyap_csv)

if "status" in df.columns:
    df = df[df["status"] == "ok"].copy()

if df.empty:
    raise ValueError("No successful rows found in the Lyapunov CSV.")

exp_cols = get_exp_cols(df)
if not exp_cols:
    raise ValueError("No Exp columns found in the Lyapunov CSV.")

print("\nDetected exponent columns:")
print(exp_cols)

# -------------------------------------------------------
# BUILD KS TABLE
# -------------------------------------------------------
records = []

for _, row in df.iterrows():
    stem = row["stem"] if "stem" in row.index else ""
    channel = infer_channel_from_stem(stem)
    emb_dim = infer_embedding_dimension(row)
    exps = row_exponents(row, exp_cols)

    if exps.size == 0:
        continue

    exps_sorted = np.sort(exps)[::-1]
    ks_est = ks_entropy_estimate(exps_sorted)
    n_pos = positive_lyapunov_count(exps_sorted)
    lmax = largest_lyapunov(exps_sorted)

    rec = {
        "file": row["file"] if "file" in row.index else "",
        "stem": stem,
        "channel": channel,
        "embedding_dimension": emb_dim,
        "state_dim": int(row["state_dim"]) if "state_dim" in row.index and pd.notna(row["state_dim"]) else np.nan,
        "n_samples": int(row["n_samples"]) if "n_samples" in row.index and pd.notna(row["n_samples"]) else np.nan,
        "steps_used": int(row["steps_used"]) if "steps_used" in row.index and pd.notna(row["steps_used"]) else np.nan,
        "n_exponents_used": int(exps_sorted.size),
        "largest_lyapunov_exponent": float(lmax),
        "n_positive_exponents": int(n_pos),
        "sum_positive_exponents": float(ks_est),
        "ks_entropy_estimate": float(ks_est),
        "sum_all_exponents": float(np.sum(exps_sorted)),
    }

    for i, val in enumerate(exps_sorted, start=1):
        rec[f"sorted_Exp{i}"] = float(val)

    records.append(rec)

ks_df = pd.DataFrame(records)

if ks_df.empty:
    raise ValueError("No KS entropy records were produced.")

ks_df = ks_df.sort_values(["channel", "embedding_dimension"]).reset_index(drop=True)

# -------------------------------------------------------
# SUMMARIES
# -------------------------------------------------------
all_dims = list(range(1, 11))

dim_summary_df = (
    ks_df.groupby("embedding_dimension", dropna=True)
    .agg(
        n_records=("ks_entropy_estimate", "count"),
        mean_ks_entropy=("ks_entropy_estimate", "mean"),
        std_ks_entropy=("ks_entropy_estimate", "std"),
        mean_lmax=("largest_lyapunov_exponent", "mean"),
        mean_positive_count=("n_positive_exponents", "mean"),
        mean_sum_all_exponents=("sum_all_exponents", "mean"),
    )
    .reset_index()
)

dim_summary_full = pd.DataFrame({"embedding_dimension": all_dims}).merge(
    dim_summary_df,
    on="embedding_dimension",
    how="left"
)

ks_matrix = ks_df.pivot_table(
    index="channel",
    columns="embedding_dimension",
    values="ks_entropy_estimate",
    aggfunc="mean"
).reindex(columns=all_dims)

poscount_matrix = ks_df.pivot_table(
    index="channel",
    columns="embedding_dimension",
    values="n_positive_exponents",
    aggfunc="mean"
).reindex(columns=all_dims)

# highest available dimension per channel
highest_dim_df = (
    ks_df.sort_values(["channel", "embedding_dimension"])
    .groupby("channel", as_index=False)
    .tail(1)
    .reset_index(drop=True)
)

# -------------------------------------------------------
# SAVE TABLES
# -------------------------------------------------------
ks_csv = os.path.join(out_dir, "ks_entropy_per_file.csv")
dim_csv = os.path.join(out_dir, "ks_entropy_by_dimension.csv")
matrix_csv = os.path.join(out_dir, "ks_entropy_channel_dimension_matrix.csv")
poscount_csv = os.path.join(out_dir, "positive_exponent_count_channel_dimension_matrix.csv")
highest_csv = os.path.join(out_dir, "ks_entropy_highest_available_dimension_per_channel.csv")

ks_df.to_csv(ks_csv, index=False)
dim_summary_full.to_csv(dim_csv, index=False)
ks_matrix.to_csv(matrix_csv)
poscount_matrix.to_csv(poscount_csv)
highest_dim_df.to_csv(highest_csv, index=False)

summary_txt = os.path.join(out_dir, "ks_entropy_summary.txt")
with open(summary_txt, "w") as f:
    f.write("Kolmogorov-Sinai Entropy Estimate from Saved Lyapunov Spectra\n")
    f.write("===========================================================\n\n")
    f.write(f"Source CSV: {lyap_csv}\n")
    f.write(f"Rows analyzed: {len(ks_df)}\n")
    f.write(f"Exponent columns used: {', '.join(exp_cols)}\n")
    f.write(f"KS entropy units: {KS_UNITS}\n\n")
    f.write("Estimate used:\n")
    f.write("  h_KS ≈ sum of positive Lyapunov exponents\n\n")

    f.write("Mean KS entropy by embedding dimension:\n")
    for _, row in dim_summary_full.iterrows():
        d = int(row["embedding_dimension"])
        nrec = row["n_records"]
        if pd.isna(nrec):
            f.write(f"  {d}D: no data\n")
        else:
            f.write(
                f"  {d}D: n={int(nrec)}, "
                f"mean_KS={row['mean_ks_entropy']:.6f}, "
                f"std_KS={0.0 if pd.isna(row['std_ks_entropy']) else row['std_ks_entropy']:.6f}, "
                f"mean_Lmax={row['mean_lmax']:.6f}, "
                f"mean_positive_count={row['mean_positive_count']:.3f}, "
                f"mean_sum_all_exp={row['mean_sum_all_exponents']:.6f}\n"
            )

# -------------------------------------------------------
# PRINT
# -------------------------------------------------------
print("\nKS entropy by embedding dimension:")
print(dim_summary_full)

print("\nSample per-file KS results:")
print(ks_df.head())

# -------------------------------------------------------
# PLOTS
# -------------------------------------------------------
# Plot 1: mean KS entropy vs embedding dimension
fig, ax = plt.subplots(figsize=(8, 5), facecolor=BG)
style_ax(ax)

mask = dim_summary_full["mean_ks_entropy"].notna().values
x = dim_summary_full.loc[mask, "embedding_dimension"].values
y = dim_summary_full.loc[mask, "mean_ks_entropy"].values
s = dim_summary_full.loc[mask, "std_ks_entropy"].fillna(0.0).values

ax.plot(x, y, marker="o", color=ACCENT, linewidth=1.5)
ax.fill_between(x, y - s, y + s, color=ACCENT, alpha=0.15)
ax.set_title("Mean KS Entropy Estimate vs Embedding Dimension")
ax.set_xlabel("Embedding Dimension")
ax.set_ylabel(f"KS Entropy Estimate ({KS_UNITS})")
ax.set_xticks(all_dims)

plot1_path = os.path.join(plots_dir, "mean_ks_entropy_vs_embedding_dimension.png")
plt.tight_layout()
plt.savefig(plot1_path, dpi=180, bbox_inches="tight")
plt.show()
plt.close()

# Plot 2: mean number of positive exponents vs embedding dimension
fig, ax = plt.subplots(figsize=(8, 5), facecolor=BG)
style_ax(ax)

mask = dim_summary_full["mean_positive_count"].notna().values
ax.plot(
    dim_summary_full.loc[mask, "embedding_dimension"].values,
    dim_summary_full.loc[mask, "mean_positive_count"].values,
    marker="o",
    color=ACCENT,
    linewidth=1.5
)
ax.set_title("Mean Number of Positive Lyapunov Exponents vs Embedding Dimension")
ax.set_xlabel("Embedding Dimension")
ax.set_ylabel("Mean Positive Exponent Count")
ax.set_xticks(all_dims)

plot2_path = os.path.join(plots_dir, "mean_positive_exponent_count_vs_embedding_dimension.png")
plt.tight_layout()
plt.savefig(plot2_path, dpi=180, bbox_inches="tight")
plt.show()
plt.close()

# Plot 3: channel x dimension heatmap of KS entropy
fig, ax = plt.subplots(figsize=(12, 10), facecolor=BG)
style_ax(ax)

heat_vals = ks_matrix.values.astype(float)
plot_vals = np.where(np.isnan(heat_vals), 0.0, heat_vals)

im = ax.imshow(plot_vals, aspect="auto")
ax.set_title("KS Entropy Estimate by Channel and Embedding Dimension")
ax.set_xlabel("Embedding Dimension")
ax.set_ylabel("Channel")
ax.set_xticks(np.arange(len(all_dims)))
ax.set_xticklabels(all_dims)
ax.set_yticks(np.arange(len(ks_matrix.index)))
ax.set_yticklabels(ks_matrix.index)

cbar = plt.colorbar(im, ax=ax)
cbar.outline.set_edgecolor(ACCENT)
cbar.ax.tick_params(color=ACCENT)
for label in cbar.ax.get_yticklabels():
    label.set_color(ACCENT)

plot3_path = os.path.join(plots_dir, "ks_entropy_heatmap_channel_by_dimension.png")
plt.tight_layout()
plt.savefig(plot3_path, dpi=180, bbox_inches="tight")
plt.show()
plt.close()

# Plot 4: channel trajectories across dimensions
fig, ax = plt.subplots(figsize=(12, 8), facecolor=BG)
style_ax(ax)

for ch in ks_matrix.index:
    y = ks_matrix.loc[ch].values.astype(float)
    x = np.array(all_dims, dtype=float)
    mask = np.isfinite(y)
    if np.sum(mask) >= 2:
        ax.plot(x[mask], y[mask], linewidth=1.0, alpha=0.7, color=ACCENT)

ax.set_title("KS Entropy Trajectories Across Embedding Dimensions")
ax.set_xlabel("Embedding Dimension")
ax.set_ylabel(f"KS Entropy Estimate ({KS_UNITS})")
ax.set_xticks(all_dims)

plot4_path = os.path.join(plots_dir, "ks_entropy_trajectories_across_dimensions.png")
plt.tight_layout()
plt.savefig(plot4_path, dpi=180, bbox_inches="tight")
plt.show()
plt.close()

print(f"\nSaved per-file KS CSV: {ks_csv}")
print(f"Saved dimension summary CSV: {dim_csv}")
print(f"Saved KS matrix CSV: {matrix_csv}")
print(f"Saved positive-count matrix CSV: {poscount_csv}")
print(f"Saved highest-dimension summary CSV: {highest_csv}")
print(f"Saved summary TXT: {summary_txt}")
print(f"Saved plots to: {plots_dir}")
```

<h3 style="color:#E8E8E8;margin-top:0;">Theory and role — shared entropy estimators for methods 1–64</h3>
<p>This foundation cell defines the reusable estimators used by the Sample/Fuzzy/Permutation/Dispersion and generalized-information catalogue entries. Representative definitions include Sample Entropy
\(\mathrm{SampEn}(m,r)=-\log(A/B)\), Shannon entropy \(H=-\sum_i p_i\log p_i\), Rényi entropy \(H_q=(1-q)^{-1}\log\sum_i p_i^q\), and Tsallis entropy \(S_q=(1-\sum_i p_i^q)/(q-1)\). The helper routines centralize symbolization, distances, coarse graining, and numerical safeguards so all methods use the same EEG segment and runtime budget.</p>
</div>


```python
# =======================================================
# ENTROPY CATALOGUE FOUNDATION — methods 1–64
# Same EEG segment/channel convention as the existing Entropy notebook.
# =======================================================
import os, math, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import signal, stats, special
from scipy.spatial.distance import pdist, squareform
from scipy.spatial import cKDTree
from numpy.lib.stride_tricks import sliding_window_view
from collections import Counter
from IPython.display import display

ENT_EPS=1e-12
ENT_BG="#000000"; ENT_ACCENT="#00FFFF"
plt.rcParams.update({"figure.facecolor":ENT_BG,"axes.facecolor":ENT_BG,"savefig.facecolor":ENT_BG,
    "text.color":ENT_ACCENT,"axes.labelcolor":ENT_ACCENT,"axes.edgecolor":ENT_ACCENT,
    "axes.titlecolor":ENT_ACCENT,"xtick.color":ENT_ACCENT,"ytick.color":ENT_ACCENT,
    "grid.color":ENT_ACCENT})
FAST_OPTION=globals().get("FAST_OPTION","ultra")
if FAST_OPTION not in {"full","balanced","fast","ultra"}: raise ValueError("FAST_OPTION must be one of: full, balanced, fast, ultra")
_ENT_SPEED={
 "full":dict(down=1,max_samples=106944,patterns=8000,windows=36,scales=20,boot=60,sur=40),
 "balanced":dict(down=2,max_samples=70000,patterns=5000,windows=28,scales=15,boot=32,sur=24),
 "fast":dict(down=4,max_samples=40000,patterns=3000,windows=20,scales=10,boot=16,sur=12),
 "ultra":dict(down=8,max_samples=18000,patterns=1400,windows=12,scales=7,boot=6,sur=5),
}[FAST_OPTION]
_ENT_NAMES=['Fp1','Fpz','Fp2','F7','F3','Fz','F4','F8','FC5','FC1','FC2','FC6','M1','T7','C3','Cz','C4','T8','M2','CP5','CP1','CP2','CP6','P7','P3','Pz','P4','P8','POz','O1','Oz','O2']
_ENT_BANDS={'delta':(1,4),'theta':(4,8),'alpha':(8,13),'beta':(13,30),'gamma':(30,45)}

def _ent_layout(a):
    a=np.asarray(a,float)
    if a.ndim!=2: raise ValueError('EEG array must be 2-D')
    if a.shape[1]<=64 and a.shape[0]>a.shape[1]: return a
    if a.shape[0]<=64: return a.T
    return a

if 'ENTROPY_DATA_OVERRIDE' in globals():
    _ent_raw=_ent_layout(globals()['ENTROPY_DATA_OVERRIDE']); _ent_fs0=float(globals().get('ENTROPY_FS_OVERRIDE',1000.0))
else:
    _p='/home/a/projects/Complete-Neural-Signal-Analysis/eeg_data_with_channels.npy'; _ent_fs0=1000.0
    _a=_ent_layout(np.load(_p,allow_pickle=True)); a,b=int(814.571*_ent_fs0),int(921.515*_ent_fs0); _ent_raw=_a[a:min(b,len(_a))]
_ent_raw=np.asarray(_ent_raw[:_ENT_SPEED['max_samples']],float)
_ent_raw=signal.detrend(_ent_raw,axis=0,type='constant')
_d=int(_ENT_SPEED['down']); _ent_x=signal.resample_poly(_ent_raw,1,_d,axis=0) if _d>1 else _ent_raw.copy(); _ent_fs=_ent_fs0/_d
_ent_x=(_ent_x-np.nanmean(_ent_x,axis=0,keepdims=True))/(np.nanstd(_ent_x,axis=0,keepdims=True)+ENT_EPS)
_ent_names=_ENT_NAMES[:_ent_x.shape[1]]; _ent_y=_ent_x[:,0]

def _cap(y,n=None):
    y=np.asarray(y,float).ravel(); n=int(n or _ENT_SPEED['patterns'])
    if len(y)>n:
        idx=np.linspace(0,len(y)-1,n).astype(int); y=y[idx]
    return y

def _embed(y,m,tau=1,cap=True):
    y=np.asarray(y,float).ravel(); n=len(y)-(m-1)*tau
    if n<=1:return np.empty((0,m))
    X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
    if cap and len(X)>_ENT_SPEED['patterns']:
        idx=np.linspace(0,len(X)-1,_ENT_SPEED['patterns']).astype(int);X=X[idx]
    return X

def _p_from_labels(labels,weights=None,full_states=None):
    labels=list(map(tuple,labels)) if np.asarray(labels).ndim>1 else list(labels)
    if weights is None:
        C=Counter(labels); vals=np.array(list(C.values()),float)
    else:
        C={}
        for k,w in zip(labels,np.asarray(weights,float)): C[k]=C.get(k,0.0)+float(w)
        vals=np.array(list(C.values()),float)
    p=vals/(vals.sum()+ENT_EPS); return p[p>0]

def _H(p,base=np.e):
    p=np.asarray(p,float);p=p[p>0];return float(-np.sum(p*np.log(p))/np.log(base)) if len(p) else 0.0

def _hist_p(y,bins=16,quantile=False):
    y=np.asarray(y,float); y=y[np.isfinite(y)]
    if not len(y):return np.array([1.0])
    if quantile:
        e=np.unique(np.quantile(y,np.linspace(0,1,bins+1)))
        if len(e)<2:return np.array([1.0])
        h,_=np.histogram(y,bins=e)
    else:h,_=np.histogram(y,bins=bins)
    h=h[h>0].astype(float);return h/(h.sum()+ENT_EPS)

def _match_pairs(X,r):
    if len(X)<2:return 0
    return len(cKDTree(np.asarray(X,float)).query_pairs(float(r),p=np.inf))

def _sampen(y,m=2,r=.2):
    y=np.asarray(y,float);sd=np.std(y);rr=float(r*sd if r<1 else r)
    Xm=_embed(y,m);X1=_embed(y,m+1);B=_match_pairs(Xm,rr);A=_match_pairs(X1,rr)
    denB=max(1,len(Xm)*(len(Xm)-1)/2);denA=max(1,len(X1)*(len(X1)-1)/2);pB=B/denB;pA=A/denA
    if pA<=0 or pB<=0:return np.inf
    return float(-np.log(pA/pB))

def _apen(y,m=2,r=.2):
    y=_cap(y);sd=np.std(y);rr=float(r*sd if r<1 else r)
    def phi(k):
        X=_embed(y,k,cap=False)
        if not len(X):return np.nan
        tr=cKDTree(X);cnt=np.array([len(v) for v in tr.query_ball_point(X,rr,p=np.inf)],float)/len(X)
        return float(np.mean(np.log(cnt+ENT_EPS)))
    return float(phi(m)-phi(m+1))

def _fuzzyen(y,m=2,r=.2,n=2):
    y=_cap(y);sd=np.std(y);rr=max(ENT_EPS,r*sd)
    def sim(k):
        X=_embed(y,k,cap=False);X=X-X.mean(1,keepdims=True)
        if len(X)<2:return np.nan
        d=pdist(X,metric='chebyshev');return float(np.mean(np.exp(-(d**n)/(rr+ENT_EPS))))
    a,b=sim(m+1),sim(m);return float(-np.log((a+ENT_EPS)/(b+ENT_EPS)))

def _cross_sampen(x,y,m=2,r=.2):
    x=_cap(x);y=_cap(y);sd=np.std(np.r_[x,y]);rr=max(ENT_EPS,r*sd)
    def count(k):
        X=_embed(x,k,cap=False);Y=_embed(y,k,cap=False); tr=cKDTree(Y)
        return sum(len(v) for v in tr.query_ball_point(X,rr,p=np.inf))
    X0,Y0=_embed(x,m,cap=False),_embed(y,m,cap=False);X1,Y1=_embed(x,m+1,cap=False),_embed(y,m+1,cap=False)
    B,A=count(m),count(m+1);pB=B/max(1,len(X0)*len(Y0));pA=A/max(1,len(X1)*len(Y1));return float(-np.log((pA+ENT_EPS)/(pB+ENT_EPS)))

def _multivar_sampen(X,m=2,r=.2):
    X=np.asarray(X,float); X=(X-X.mean(0))/(X.std(0)+ENT_EPS); n=len(X)-(m-1)
    Z=np.hstack([X[j:j+n] for j in range(m)]); Z1=np.hstack([X[j:j+n-1] for j in range(m+1)])
    if len(Z)>_ENT_SPEED['patterns']:
        ids=np.linspace(0,len(Z)-1,_ENT_SPEED['patterns']).astype(int);Z=Z[ids]
    if len(Z1)>_ENT_SPEED['patterns']:
        ids=np.linspace(0,len(Z1)-1,_ENT_SPEED['patterns']).astype(int);Z1=Z1[ids]
    B=_match_pairs(Z,r);A=_match_pairs(Z1,r);pB=B/max(1,len(Z)*(len(Z)-1)/2);pA=A/max(1,len(Z1)*(len(Z1)-1)/2);return float(-np.log((pA+ENT_EPS)/(pB+ENT_EPS)))

def _ordinal_labels(y,m=3,tau=1):
    X=_embed(y,m,tau); return [tuple(np.argsort(v,kind='mergesort')) for v in X]

def _perm_entropy(y,m=3,tau=1,weighted=False,normalize=True):
    X=_embed(y,m,tau);labs=[tuple(np.argsort(v,kind='mergesort')) for v in X]
    w=np.var(X,axis=1) if weighted and len(X) else None;p=_p_from_labels(labs,w);h=_H(p)
    return float(h/(math.lgamma(m+1)+ENT_EPS)) if normalize else h

def _disp_classes(y,c=6):
    y=np.asarray(y,float);z=(y-np.mean(y))/(np.std(y)+ENT_EPS);u=stats.norm.cdf(z);cl=np.rint(c*u+.5).astype(int);return np.clip(cl,1,c)

def _disp_entropy(y,m=3,c=6,tau=1,fluct=False,normalize=True,reverse=False):
    z=_disp_classes(y,c);X=_embed(z,m,tau)
    P=np.diff(X,axis=1).astype(int) if fluct else X.astype(int); p=_p_from_labels(P)
    if reverse:
        M=(2*c-1)**(m-1) if fluct else c**m; val=float(np.sum((p-1/M)**2)+(M-len(p))*(1/M)**2);return val/(1-1/M+ENT_EPS)
    h=_H(p);M=(2*c-1)**(m-1) if fluct else c**m;return float(h/(np.log(M)+ENT_EPS)) if normalize else h

def _coarse(y,s,offset=0):
    y=np.asarray(y,float);n=(len(y)-offset)//s
    return y[offset:offset+n*s].reshape(n,s).mean(1) if n>0 else np.empty(0)

def _distribution_entropy(y,m=3,bins=32):
    X=_embed(y,m);d=pdist(X)
    if not len(d):return 0.0
    p=_hist_p(d,bins=bins);return _H(p)/np.log(max(2,bins))

def _bubble_entropy(y,m=3):
    # Bubble-sort swap count is inversion count of each embedded vector.
    def invdist(k):
        X=_embed(y,k); inv=[]
        for v in X:
            inv.append(sum(v[a]>v[b] for a in range(k) for b in range(a+1,k)))
        p=_p_from_labels(inv); return -np.log(np.sum(p*p)+ENT_EPS)
    return float((invdist(m+1)-invdist(m))/np.log((m+1)/(m-1)))

def _slope_entropy(y,m=4,gamma=1.0,delta=.01):
    X=_embed(y,m);D=np.diff(X,axis=1);s=np.zeros_like(D,int);s[D>delta]=1;s[D>gamma]=2;s[D<-delta]=-1;s[D<-gamma]=-2
    return _H(_p_from_labels(s))/np.log(5**(m-1))

def _cosine_similarity_entropy(y,m=3,r=.1):
    X=_embed(y,m);X=X-X.mean(1,keepdims=True);norm=np.linalg.norm(X,axis=1);X=X/(norm[:,None]+ENT_EPS)
    if len(X)<2:return 0.0
    cs=1-pdist(X,metric='cosine'); B=np.mean(cs>1-r)
    X1=_embed(y,m+1);X1=X1-X1.mean(1,keepdims=True);X1=X1/(np.linalg.norm(X1,axis=1)[:,None]+ENT_EPS);cs1=1-pdist(X1,metric='cosine');A=np.mean(cs1>1-r)
    return float(-np.log((A+ENT_EPS)/(B+ENT_EPS)))

def _increment_entropy(y,m=3,q=4):
    d=np.diff(np.asarray(y,float));mag=np.abs(d);edges=np.quantile(mag,np.linspace(0,1,q+1));cls=np.clip(np.digitize(mag,edges[1:-1]),0,q-1);sym=np.sign(d).astype(int)*(cls+1);X=_embed(sym,m)
    return _H(_p_from_labels(X))/np.log((2*q)**m)

def _grid_entropy(y,bins=12):
    y=np.asarray(y,float);a,b=y[:-1],y[1:];H,_,_=np.histogram2d(a,b,bins=bins);p=H.ravel();p=p[p>0];p=p/p.sum();return _H(p)/np.log(bins*bins)

def _attention_entropy(y):
    y=np.asarray(y,float);mx,_=signal.find_peaks(y);mn,_=signal.find_peaks(-y)
    vals=[]
    for p in (mx,mn):
        if len(p)>2: vals.append(_H(_hist_p(np.diff(p),bins=min(16,max(4,int(np.sqrt(len(p))))))))
    return float(np.mean(vals)) if vals else 0.0

def _symbolic_dynamic_entropy(y,c=6):
    z=_disp_classes(y,c);p=_hist_p(z,bins=np.arange(.5,c+1.6,1));pairs=np.column_stack([z[:-1],z[1:]]);pt=_p_from_labels(pairs)
    return float(_H(p)+_H(pt)-_H(_hist_p(z[:-1],bins=np.arange(.5,c+1.6,1))))

def _sax_entropy(y,segments=32,alphabet=6):
    y=np.asarray(y,float);n=len(y)//segments
    if n<1:return 0.0
    paa=y[:n*segments].reshape(segments,n).mean(1);z=(paa-paa.mean())/(paa.std()+ENT_EPS);cuts=stats.norm.ppf(np.arange(1,alphabet)/alphabet);sym=np.digitize(z,cuts)
    return _H(_p_from_labels(sym))/np.log(alphabet)

def _renyi(p,q):
    p=np.asarray(p,float);p=p[p>0]
    if abs(q-1)<1e-8:return _H(p)
    return float(np.log(np.sum(p**q))/(1-q))

def _tsallis(p,q):
    p=np.asarray(p,float);p=p[p>0]
    if abs(q-1)<1e-8:return _H(p)
    return float((1-np.sum(p**q))/(q-1))

def _sharma_mittal(p,q,r):
    p=np.asarray(p,float);p=p[p>0]
    if abs(r-1)<1e-8:return _renyi(p,q)
    if abs(q-1)<1e-8:
        # continuous q->1 limit gives (exp((1-r)H)-1)/(1-r)
        H=_H(p);return float((np.exp((1-r)*H)-1)/(1-r))
    return float((np.sum(p**q)**((1-r)/(1-q))-1)/(1-r))

def _arimoto_conditional(x,y,alpha=2.0,bins=8):
    # H_alpha(X|Y)=alpha/(1-alpha) log sum_y p(y)||p(X|y)||_alpha.
    ex=np.quantile(x,np.linspace(0,1,bins+1));ey=np.quantile(y,np.linspace(0,1,bins+1));ex=np.unique(ex);ey=np.unique(ey)
    ix=np.clip(np.digitize(x,ex[1:-1]),0,len(ex)-2);iy=np.clip(np.digitize(y,ey[1:-1]),0,len(ey)-2)
    H=np.zeros((max(1,len(ex)-1),max(1,len(ey)-1)))
    for a,b in zip(ix,iy):H[a,b]+=1
    py=H.sum(0); py=py/(py.sum()+ENT_EPS);val=0.
    for j,w in enumerate(py):
        if w<=0:continue
        px=H[:,j]/(H[:,j].sum()+ENT_EPS);val+=w*(np.sum(px**alpha)**(1/alpha))
    return float(alpha/(1-alpha)*np.log(val+ENT_EPS))

def _band(y,lo,hi):
    hi=min(hi,_ent_fs/2-.5)
    if hi<=lo:return np.asarray(y,float)
    sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_ent_fs,output='sos');return signal.sosfiltfilt(sos,y)

def _line(x,y,title,xlabel='x',ylabel='entropy'):
    fig,ax=plt.subplots(figsize=(7.2,3.6));ax.plot(x,y,color=ENT_ACCENT,lw=1.6);ax.set(title=title,xlabel=xlabel,ylabel=ylabel);ax.grid(alpha=.2);plt.show()

def _show_scalar(name,val): print(f'{name}: {val:.6g}' if np.isscalar(val) and np.isfinite(val) else f'{name}: {val}')
print(f'Entropy catalogue foundation ready: mode={FAST_OPTION}, samples={len(_ent_x)}, channels={_ent_x.shape[1]}, fs={_ent_fs:g} Hz')
```

Keeping the dispatch layer separate avoids duplicating shared numerical code while preserving the method-specific theory supplied immediately before each catalogue execution cell.</p>

```python
def _ent_run_1_64(i):
    y=_ent_y
    if i==1:
        vals=[_sampen(_ent_x[:,j]) for j in range(min(8,_ent_x.shape[1]))];display(pd.DataFrame({'channel':_ent_names[:len(vals)],'SampleEntropy':vals}));return
    if i==2:
        rs=np.linspace(.1,.35,6);ms=[1,2,3];rows=[[m,r,_sampen(y,m,r)] for m in ms for r in rs];display(pd.DataFrame(rows,columns=['m','r_over_SD','SampEn']));return
    if i==3:
        rs=np.linspace(.08,.4,9);v=[_sampen(y,2,r) for r in rs];_line(rs,v,'Sample Entropy tolerance sensitivity','r / SD','SampEn');return
    if i==4:
        ms=np.arange(1,5);v=[_sampen(y,int(m),.2) for m in ms];_line(ms,v,'Sample Entropy embedding-dimension sensitivity','m','SampEn');return
    if i==5:
        W=max(200,int(4*_ent_fs));starts=np.linspace(0,max(0,len(y)-W),_ENT_SPEED['windows']).astype(int);v=[_sampen(y[s:s+W]) for s in starts];_line(starts/_ent_fs,v,'Rolling-window Sample Entropy','time (s)','SampEn');return
    if i==6:
        ev=globals().get('ENTROPY_EVENT_SAMPLES',None);label='supplied events'
        if ev is None:
            e=np.abs(signal.hilbert(y));pk,_=signal.find_peaks(e,distance=max(1,int(.5*_ent_fs)),prominence=np.std(e));ev=pk[np.argsort(e[pk])[-min(6,len(pk)):]] if len(pk) else [];label='internal high-envelope events (demonstration)'
        half=max(100,int(.5*_ent_fs));rows=[]
        for k in ev:
            a=max(0,int(k)-half);b=min(len(y),int(k)+half);rows.append([int(k),_sampen(y[a:b])])
        print(label);display(pd.DataFrame(rows,columns=['event_sample','SampEn']));return
    if i==7:
        rows=[]
        for name,(lo,hi) in _ENT_BANDS.items():
            if lo<_ent_fs/2-.5:rows.append([name,_sampen(_band(y,lo,hi))])
        display(pd.DataFrame(rows,columns=['band','SampEn']));return
    if i==8:_show_scalar('Cross-Sample Entropy ch0|ch1',_cross_sampen(_ent_x[:,0],_ent_x[:,1]));return
    if i==9:_show_scalar('Multivariate Sample Entropy (first 4 channels)',_multivar_sampen(_ent_x[:,:min(4,_ent_x.shape[1])],2,.25));return
    if i==10:_show_scalar('Fuzzy Entropy',_fuzzyen(y));return
    if i==11:
        sc=np.arange(1,_ENT_SPEED['scales']+1);v=[_fuzzyen(_coarse(y,int(s))) for s in sc];_line(sc,v,'Multiscale Fuzzy Entropy','scale','FuzzyEn');return
    if i==12:
        sc=np.arange(1,_ENT_SPEED['scales']+1);v=[]
        for s in sc:
            vals=[_fuzzyen(_coarse(y,int(s),o)) for o in range(int(s)) if len(_coarse(y,int(s),o))>20];v.append(np.mean(vals) if vals else np.nan)
        _line(sc,v,'Refined Composite Multiscale Fuzzy Entropy','scale','RCMFE');return
    if i==13:
        # Cross-fuzzy analogue uses cross-embedding Chebyshev distances and fuzzy membership.
        x,z=_cap(_ent_x[:,0]),_cap(_ent_x[:,1]);rr=.2*np.std(np.r_[x,z])
        def sim(k):
            X=_embed(x,k,cap=False);Z=_embed(z,k,cap=False);D=np.max(np.abs(X[:,None,:]-Z[None,:,:]),axis=2);return np.mean(np.exp(-(D**2)/(rr+ENT_EPS)))
        _show_scalar('Cross-Fuzzy Entropy',-np.log((sim(3)+ENT_EPS)/(sim(2)+ENT_EPS)));return
    if i==14:
        # Multivariate fuzzy entropy on joint delay vectors.
        X=_ent_x[:,:min(4,_ent_x.shape[1])];n=min(len(X)-3,_ENT_SPEED['patterns']);ids=np.linspace(0,len(X)-4,n).astype(int);Z=np.hstack([X[ids+k] for k in range(2)]);Z1=np.hstack([X[ids+k] for k in range(3)]);r=.25
        f=lambda A:np.mean(np.exp(-(pdist(A,'chebyshev')**2)/(r+ENT_EPS)));_show_scalar('Multivariate Fuzzy Entropy',-np.log((f(Z1)+ENT_EPS)/(f(Z)+ENT_EPS)));return
    if i==15:_show_scalar('Permutation Entropy',_perm_entropy(y));return
    if i==16:_show_scalar('Weighted Permutation Entropy',_perm_entropy(y,weighted=True));return
    if i==17:
        sc=np.arange(1,_ENT_SPEED['scales']+1);_line(sc,[_perm_entropy(_coarse(y,int(s))) for s in sc],'Multiscale Permutation Entropy','scale','PE');return
    if i==18:
        sc=np.arange(1,_ENT_SPEED['scales']+1);v=[]
        for s in sc:v.append(np.mean([_perm_entropy(_coarse(y,int(s),o)) for o in range(int(s)) if len(_coarse(y,int(s),o))>20]))
        _line(sc,v,'Refined Composite Multiscale Permutation Entropy','scale','RCMPE');return
    if i==19:
        labs=_ordinal_labels(y,4);p=_p_from_labels(labs);print('ordinal patterns=',len(p),'entropy=',_H(p),'normalized=',_H(p)/np.log(math.factorial(4)));return
    if i==20:
        labs=_ordinal_labels(y,3);pairs=list(zip(labs[:-1],labs[1:]));_show_scalar('Ordinal transition entropy',_H(_p_from_labels(pairs)));return
    if i==21:
        h3=_H(_p_from_labels(_ordinal_labels(y,3)));h4=_H(_p_from_labels(_ordinal_labels(y,4)));_show_scalar('Conditional permutation entropy H4-H3',h4-h3);return
    if i==22:
        a=_ordinal_labels(_ent_x[:,0],3);b=_ordinal_labels(_ent_x[:,1],3);n=min(len(a),len(b));joint=_H(_p_from_labels(list(zip(a[:n],b[:n]))));hx=_H(_p_from_labels(a[:n]));_show_scalar('Cross-Permutation conditional entropy H(P2|P1)',joint-hx);return
    if i==23:
        p1=Counter(_ordinal_labels(y,4));p2=Counter(_ordinal_labels(y[::-1],4));keys=sorted(set(p1)|set(p2));a=np.array([p1[k] for k in keys],float);b=np.array([p2[k] for k in keys],float);a/=a.sum();b/=b.sum();m=.5*(a+b);js=.5*np.sum(a*np.log((a+ENT_EPS)/(m+ENT_EPS)))+.5*np.sum(b*np.log((b+ENT_EPS)/(m+ENT_EPS)));_show_scalar('Permutation time-reversal asymmetry (JSD)',js);return
    if i==24:
        rows=[[name,_perm_entropy(_band(y,lo,hi))] for name,(lo,hi) in _ENT_BANDS.items() if lo<_ent_fs/2-.5];display(pd.DataFrame(rows,columns=['band','PermutationEntropy']));return
    if i==25:
        W=max(200,int(4*_ent_fs));starts=np.linspace(0,max(0,len(y)-W),_ENT_SPEED['windows']).astype(int);_line(starts/_ent_fs,[_perm_entropy(y[s:s+W]) for s in starts],'Rolling-window Permutation Entropy','time (s)','PE');return
    if i==26:_show_scalar('Dispersion Entropy',_disp_entropy(y));return
    if i==27:_show_scalar('Fluctuation-Based Dispersion Entropy',_disp_entropy(y,fluct=True));return
    if i in (28,29):
        sc=np.arange(1,_ENT_SPEED['scales']+1);v=[]
        for s in sc:
            if i==28:v.append(_disp_entropy(_coarse(y,int(s))))
            else:v.append(np.mean([_disp_entropy(_coarse(y,int(s),o)) for o in range(int(s)) if len(_coarse(y,int(s),o))>20]))
        _line(sc,v,'Multiscale Dispersion Entropy' if i==28 else 'Refined Composite Multiscale Dispersion Entropy','scale','DispEn');return
    if i==30:_show_scalar('Reverse Dispersion Entropy',_disp_entropy(y,reverse=True));return
    if i==31:
        a,b=_disp_classes(_ent_x[:,0]),_disp_classes(_ent_x[:,1]);n=min(len(a),len(b));pairs=np.column_stack([a[:n],b[:n]]);_show_scalar('Cross-Dispersion Entropy',_H(_p_from_labels(pairs))/np.log(36));return
    if i==32:
        Z=np.column_stack([_disp_classes(_ent_x[:,j],4) for j in range(min(4,_ent_x.shape[1]))]);_show_scalar('Multivariate Dispersion Entropy',_H(_p_from_labels(Z))/np.log(4**Z.shape[1]));return
    if i==33:_show_scalar('Distribution Entropy',_distribution_entropy(y));return
    if i==34:_show_scalar('Bubble Entropy',_bubble_entropy(y));return
    if i==35:_show_scalar('Slope Entropy',_slope_entropy(y));return
    if i==36:_show_scalar('Cosine Similarity Entropy',_cosine_similarity_entropy(y));return
    if i==37:_show_scalar('Increment Entropy',_increment_entropy(y));return
    if i==38:_show_scalar('Grid Distribution Entropy',_grid_entropy(y));return
    if i==39:_show_scalar('Attention Entropy',_attention_entropy(y));return
    if i==40:_show_scalar('Symbolic Dynamic Entropy',_symbolic_dynamic_entropy(y));return
    if i==41:_show_scalar('Symbolic Aggregate Entropy (SAX)',_sax_entropy(y));return
    p=_hist_p(y,bins=24)
    if i==42:_show_scalar('Shannon amplitude entropy',_H(p));return
    if i==43:
        # Plug-in histogram differential entropy includes bin widths.
        h,e=np.histogram(y,bins=32,density=True);w=np.diff(e);q=h*w;q=q[q>0];_show_scalar('Histogram differential entropy',-np.sum(q*np.log(h[h>0]+ENT_EPS)));return
    if i in (44,45):
        # Kozachenko-Leonenko 1-D estimator; method 44 and 45 intentionally show the same canonical kNN estimator with k sensitivity vs a fixed k.
        z=_cap(y);n=len(z);k=5 if i==45 else 3;tree=cKDTree(z[:,None]);d=tree.query(z[:,None],k=k+1)[0][:,-1];val=special.digamma(n)-special.digamma(k)+np.log(2)+np.mean(np.log(d+ENT_EPS));_show_scalar('Kozachenko-Leonenko differential entropy' if i==45 else 'kNN differential entropy',val);return
    if i==46:_show_scalar('Gaussian differential entropy',.5*np.log(2*np.pi*np.e*np.var(y)));return
    if i==47:
        bs=np.array([8,12,16,24,32,48]);_line(bs,[_H(_hist_p(y,int(b))) for b in bs],'Histogram entropy bin sensitivity','bins','Shannon H');return
    if i==48:_show_scalar('Quantile-symbol entropy',_H(_hist_p(y,12,quantile=True)));return
    if i==49:
        # state occupancy from amplitude x local derivative quantile states
        dy=np.gradient(y);a=np.digitize(y,np.quantile(y,[.25,.5,.75]));b=np.digitize(dy,np.quantile(dy,[.25,.5,.75]));_show_scalar('State-occupancy entropy',_H(_p_from_labels(np.column_stack([a,b])))/np.log(16));return
    if i==50:
        regs={'frontal':range(0,min(12,_ent_x.shape[1])),'central':range(min(12,_ent_x.shape[1]),min(23,_ent_x.shape[1])),'posterior':range(min(23,_ent_x.shape[1]),_ent_x.shape[1])};rows=[]
        for name,inds in regs.items():
            inds=list(inds)
            if inds:
                s=np.mean(_ent_x[:,inds],axis=1);rows.append([name,_H(_hist_p(s,16,quantile=True))])
        display(pd.DataFrame(rows,columns=['region','occupancy_entropy']));return
    if i==51:
        vals=[_H(_hist_p(_ent_x[:,j],12,quantile=True)) for j in range(_ent_x.shape[1])];display(pd.DataFrame({'channel':_ent_names,'symbolic_entropy':vals}));return
    if i==52:_show_scalar('Rényi entropy q=2',_renyi(p,2));return
    if i==53:_show_scalar('Tsallis entropy q=2',_tsallis(p,2));return
    if i==54:_show_scalar('Collision entropy H2',-np.log(np.sum(p*p)));return
    if i==55:_show_scalar('Min-entropy H∞',-np.log(np.max(p)));return
    if i==56:_show_scalar('Max-entropy H1/2',2*np.log(np.sum(np.sqrt(p))));return
    if i==57:_show_scalar('Hartley entropy H0',np.log(np.count_nonzero(p)));return
    if i==58:_show_scalar('Arimoto conditional entropy alpha=2, H(X_t+1|X_t)',_arimoto_conditional(y[1:],y[:-1],2));return
    if i==59:
        q=2.;_show_scalar('Havrda-Charvat entropy q=2',(np.sum(p**q)-1)/(2**(1-q)-1));return
    if i==60:_show_scalar('Sharma-Mittal entropy q=2,r=0.5',_sharma_mittal(p,2,.5));return
    if i==61:
        q=2.;e=p**q/np.sum(p**q);_show_scalar('Escort-distribution Shannon entropy q=2',_H(e));return
    if i in (62,63,64):
        qs=np.linspace(.25,4,30);v=[(_renyi(p,q) if i in (62,63) else _tsallis(p,q)) for q in qs];title={62:'q-entropy sweep (Rényi family)',63:'Rényi entropy spectrum across q',64:'Tsallis entropy spectrum across q'}[i];_line(qs,v,title,'q','entropy');return
    raise ValueError(i)
```

### 1. Standalone Sample Entropy suite

Sample entropy estimates the negative log conditional probability that two length-m patterns that match within tolerance r also match when extended to m+1: SampEn=-ln(A_m+1/B_m). Self-matches are excluded.

```python
_ent_run_1_64(1)
```

### 2. Sample Entropy parameter sweep

This cell sweeps the defining Sample Entropy parameters m and r. The estimator is unchanged; the grid shows how finite-data regularity depends on embedding dimension and tolerance.

```python
_ent_run_1_64(2)
```

### 3. Sample Entropy tolerance-sensitivity analysis

Tolerance sensitivity varies r while holding m fixed. Because match counts A and B depend monotonically on the Chebyshev-radius threshold, SampEn can be unstable when r is too small.

```python
_ent_run_1_64(3)
```

### 4. Sample Entropy embedding-dimension sensitivity analysis

Embedding-dimension sensitivity varies m in SampEn=-ln(A/B). Larger m demands recurrence of longer patterns and therefore needs more data for stable counting.

```python
_ent_run_1_64(4)
```

### 5. Rolling-window Sample Entropy

Rolling Sample Entropy applies the same SampEn estimator to overlapping or evenly spaced windows, producing a time-resolved complexity trajectory.

```python
_ent_run_1_64(5)
```

### 6. Event-locked Sample Entropy

Event-locked Sample Entropy evaluates SampEn in windows centered on supplied event samples. If no external events are provided, internally detected envelope peaks are explicitly labelled demonstration events.

```python
_ent_run_1_64(6)
```

### 7. Band-limited Sample Entropy

Band-limited Sample Entropy first band-pass filters the signal and then applies the standard SampEn estimator, so differences reflect temporal irregularity within each frequency band.

```python
_ent_run_1_64(7)
```

### 8. Cross-Sample Entropy

Cross-Sample Entropy replaces self-pattern matching with matches between patterns from two signals. The conditional ratio of m+1 to m cross-matches quantifies shared pattern regularity.

```python
_ent_run_1_64(8)
```

### 9. Multivariate Sample Entropy

Multivariate Sample Entropy forms joint delay vectors from several synchronized channels and applies the same match-extension principle in the joint state space.

```python
_ent_run_1_64(9)
```

### 10. Fuzzy Entropy

Fuzzy Entropy replaces hard r-neighborhood matches with a smooth membership exp(-d^n/r), then uses -ln(phi_m+1/phi_m).

```python
_ent_run_1_64(10)
```

### 11. Multiscale Fuzzy Entropy

Multiscale Fuzzy Entropy coarse-grains the signal at scale s by averaging non-overlapping blocks before computing Fuzzy Entropy at each scale.

```python
_ent_run_1_64(11)
```

### 12. Refined Composite Multiscale Fuzzy Entropy

Refined-composite multiscale entropy averages entropy estimates over all s coarse-graining offsets, reducing the variance caused by choosing a single block origin.

```python
_ent_run_1_64(12)
```

### 13. Cross-Fuzzy Entropy

Cross-Fuzzy Entropy applies fuzzy similarity to delay vectors from two different signals and compares similarity at dimensions m and m+1.

```python
_ent_run_1_64(13)
```

### 14. Multivariate Fuzzy Entropy

Multivariate Fuzzy Entropy measures fuzzy recurrence of joint multichannel delay vectors, extending Fuzzy Entropy from one coordinate to a synchronized vector process.

```python
_ent_run_1_64(14)
```

### 15. Permutation Entropy

Permutation Entropy is the Shannon entropy of Bandt-Pompe ordinal patterns. Normalization by ln(m!) maps the maximum possible entropy to one.

```python
_ent_run_1_64(15)
```

### 16. Weighted Permutation Entropy

Weighted Permutation Entropy weights each ordinal-pattern occurrence by local variance before normalizing pattern probabilities, retaining amplitude information ignored by ordinary PE.

```python
_ent_run_1_64(16)
```

### 17. Multiscale Permutation Entropy

Multiscale Permutation Entropy computes PE after coarse-graining at increasing temporal scales.

```python
_ent_run_1_64(17)
```

### 18. Refined Composite Multiscale Permutation Entropy

Refined-composite Multiscale Permutation Entropy averages PE across all coarse-graining offsets at each scale.

```python
_ent_run_1_64(18)
```

### 19. Ordinal Pattern Entropy

Ordinal Pattern Entropy is Shannon entropy of the empirical distribution of rank-order permutations of embedded samples.

```python
_ent_run_1_64(19)
```

### 20. Ordinal Transition Entropy

Ordinal Transition Entropy is the Shannon entropy of consecutive ordinal-pattern pairs, so it captures uncertainty in symbolic transitions as well as occupancy.

```python
_ent_run_1_64(20)
```

### 21. Conditional Permutation Entropy

Conditional Permutation Entropy is implemented as the entropy increment H(P_m+1)-H(P_m), the uncertainty added when an ordinal pattern is extended by one sample.

```python
_ent_run_1_64(21)
```

### 22. Cross-Permutation Entropy

Cross-Permutation Entropy is implemented as conditional entropy H(P_Y|P_X)=H(P_X,P_Y)-H(P_X) of synchronized ordinal patterns.

```python
_ent_run_1_64(22)
```

### 23. Permutation Entropy asymmetry analysis

Permutation asymmetry is quantified by Jensen-Shannon divergence between forward and time-reversed ordinal-pattern distributions; zero indicates ordinal time-reversal symmetry.

```python
_ent_run_1_64(23)
```

### 24. Band-limited Permutation Entropy

Band-limited Permutation Entropy applies ordinal-pattern entropy separately after band-pass filtering.

```python
_ent_run_1_64(24)
```

### 25. Rolling-window Permutation Entropy

Rolling-window Permutation Entropy tracks the Bandt-Pompe ordinal entropy through time.

```python
_ent_run_1_64(25)
```

### 26. Dispersion Entropy

Dispersion Entropy maps standardized amplitudes through the normal CDF into c classes, forms m-symbol dispersion patterns, and computes normalized Shannon entropy H/ln(c^m).

```python
_ent_run_1_64(26)
```

### 27. Fluctuation-Based Dispersion Entropy

Fluctuation-Based Dispersion Entropy uses differences between adjacent dispersion classes; there are (2c-1)^(m-1) potential fluctuation patterns.

```python
_ent_run_1_64(27)
```

### 28. Multiscale Dispersion Entropy

Multiscale Dispersion Entropy coarse-grains first and computes DispEn at each temporal scale.

```python
_ent_run_1_64(28)
```

### 29. Refined Composite Multiscale Dispersion Entropy

Refined-composite Multiscale Dispersion Entropy averages DispEn over all coarse-graining offsets at each scale.

```python
_ent_run_1_64(29)
```

### 30. Reverse Dispersion Entropy

Reverse Dispersion Entropy measures squared Euclidean distance of dispersion-pattern probabilities from the uniform distribution and normalizes by 1-1/c^m.

```python
_ent_run_1_64(30)
```

### 31. Cross-Dispersion Entropy

Cross-Dispersion Entropy forms synchronized pairs of dispersion symbols from two channels and computes normalized joint Shannon entropy.

```python
_ent_run_1_64(31)
```

### 32. Multivariate Dispersion Entropy

Multivariate Dispersion Entropy forms synchronized class tuples across channels and computes the entropy of their joint occupancy.

```python
_ent_run_1_64(32)
```

### 33. Distribution Entropy

Distribution Entropy computes Shannon entropy of the empirical distribution of pairwise distances between embedded state vectors.

```python
_ent_run_1_64(33)
```

### 34. Bubble Entropy

Bubble Entropy summarizes the distribution of inversion counts (bubble-sort swaps) in embedded vectors using Rényi order-2 entropy and contrasts dimensions m and m+1.

```python
_ent_run_1_64(34)
```

### 35. Slope Entropy

Slope Entropy discretizes consecutive slopes into signed magnitude symbols and computes Shannon entropy of length-(m-1) slope-symbol words.

```python
_ent_run_1_64(35)
```

### 36. Cosine Similarity Entropy

Cosine Similarity Entropy compares centered embedding vectors by angular similarity and uses the Sample-Entropy extension ratio across m and m+1.

```python
_ent_run_1_64(36)
```

### 37. Increment Entropy

Increment Entropy symbolizes increments by sign and quantized magnitude, then computes normalized Shannon entropy of increment words.

```python
_ent_run_1_64(37)
```

### 38. Grid Distribution Entropy

Grid Distribution Entropy bins the return map (x_t,x_t+1) on a 2-D grid and computes normalized occupancy entropy.

```python
_ent_run_1_64(38)
```

### 39. Attention Entropy

Attention Entropy quantifies the entropy of intervals between successive local maxima and minima, averaging the two interval-distribution entropies.

```python
_ent_run_1_64(39)
```

### 40. Symbolic Dynamic Entropy

Symbolic Dynamic Entropy combines uncertainty of discrete symbol occupancy with uncertainty of symbol-to-symbol transitions.

```python
_ent_run_1_64(40)
```

### 41. Symbolic Aggregate Entropy

Symbolic Aggregate Entropy uses SAX: piecewise aggregate approximation followed by Gaussian breakpoint symbols, then computes Shannon entropy of symbol occupancy.

```python
_ent_run_1_64(41)
```

### 42. Shannon amplitude entropy

Shannon amplitude entropy is H=-sum p_i ln p_i for an empirical amplitude histogram.

```python
_ent_run_1_64(42)
```

### 43. Differential entropy

Differential entropy for a histogram density f is approximated by -sum p_i ln f_i, including bin-width information rather than treating bins as dimensionless symbols.

```python
_ent_run_1_64(43)
```

### 44. kNN differential entropy

The k-nearest-neighbor differential-entropy estimator uses neighbor radii rather than fixed bins; in one dimension the Kozachenko-Leonenko form is psi(N)-psi(k)+ln 2 + mean ln epsilon_i.

```python
_ent_run_1_64(44)
```

### 45. Kozachenko-Leonenko entropy

Kozachenko-Leonenko entropy is the canonical kNN differential-entropy estimator; the implementation uses the one-dimensional max-norm volume factor 2.

```python
_ent_run_1_64(45)
```

### 46. Gaussian differential entropy

For a Gaussian variable with variance sigma^2, differential entropy is h=0.5 ln(2 pi e sigma^2).

```python
_ent_run_1_64(46)
```

### 47. Histogram entropy with bin-sensitivity analysis

Histogram entropy depends on partition resolution. This cell sweeps bin count to expose that estimator dependence instead of reporting one arbitrary binning.

```python
_ent_run_1_64(47)
```

### 48. Quantile-symbol entropy

Quantile-symbol entropy uses equiprobable amplitude breakpoints. It measures symbol occupancy after rank-based discretization and is invariant to monotone amplitude transformations up to ties.

```python
_ent_run_1_64(48)
```

### 49. State-occupancy entropy

State-occupancy entropy discretizes a finite state description (amplitude and local derivative) and computes Shannon entropy of the occupied states.

```python
_ent_run_1_64(49)
```

### 50. Regional state-occupancy entropy

Regional state-occupancy entropy applies the same finite-state Shannon measure to region-averaged EEG signals.

```python
_ent_run_1_64(50)
```

### 51. Channel-wise symbolic entropy

Channel-wise symbolic entropy computes the entropy of quantile-coded amplitude symbols separately for every channel.

```python
_ent_run_1_64(51)
```

### 52. Rényi entropy

Rényi entropy of order q is H_q=(1/(1-q)) ln sum p_i^q, with the Shannon limit as q approaches 1.

```python
_ent_run_1_64(52)
```

### 53. Tsallis entropy

Tsallis entropy is S_q=(1-sum p_i^q)/(q-1), another q-generalization with Shannon limit at q=1.

```python
_ent_run_1_64(53)
```

### 54. Collision entropy

Collision entropy is Rényi entropy of order 2: H_2=-ln sum p_i^2, the negative log collision probability.

```python
_ent_run_1_64(54)
```

### 55. Min-entropy

Min-entropy is the Rényi q->infinity limit H_inf=-ln max_i p_i.

```python
_ent_run_1_64(55)
```

### 56. Max-entropy

Max-entropy is the Rényi order-1/2 quantity H_max=2 ln sum_i sqrt(p_i), distinct from Hartley entropy H_0.

```python
_ent_run_1_64(56)
```

### 57. Hartley entropy

Hartley entropy is Rényi order zero: H_0=ln |supp(p)|, depending only on the number of occupied states.

```python
_ent_run_1_64(57)
```

### 58. Arimoto entropy

Arimoto conditional Rényi entropy is H_alpha(X|Y)=alpha/(1-alpha) ln sum_y p(y)||p(X|y)||_alpha. The cell applies it to successive amplitude symbols.

```python
_ent_run_1_64(58)
```

### 59. Havrda-Charvát entropy

Havrda-Charvát entropy is a normalized power-sum entropy, here HC_q=(sum p_i^q-1)/(2^(1-q)-1), with a Shannon-proportional limit as q approaches 1.

```python
_ent_run_1_64(59)
```

### 60. Sharma-Mittal entropy

Sharma-Mittal entropy is S_q,r=[(sum p_i^q)^((1-r)/(1-q))-1]/(1-r), containing Rényi and Tsallis-type limits.

```python
_ent_run_1_64(60)
```

### 61. Escort-distribution entropy

The q-escort distribution is P_i^(q)=p_i^q/sum_j p_j^q. This cell reports Shannon entropy of that reweighted distribution.

```python
_ent_run_1_64(61)
```

### 62. q-entropy sweep

A q-entropy sweep evaluates one generalized entropy continuously across q, showing how emphasis shifts from rare states at low q to dominant states at high q.

```python
_ent_run_1_64(62)
```

### 63. Rényi entropy spectrum across q

The Rényi spectrum H_q across q summarizes distribution heterogeneity across probability orders.

```python
_ent_run_1_64(63)
```

### 64. Tsallis entropy spectrum across q

The Tsallis spectrum S_q across q provides the corresponding nonadditive q-entropy family.

```python
_ent_run_1_64(64)
```

EMD/EEMD/VMD branches decompose the signal before applying the stated entropy functional. The speed mode changes scale, ensemble, or iteration budgets only.</p>
</div>


```python
# =======================================================
# ENTROPY CATALOGUE HELPERS — methods 65–99
# Multiscale, wavelet/time-frequency, EMD/EEMD/VMD entropy.
# =======================================================
def _sampen_stats(y,m=2,r=.2):
    y=np.asarray(y,float);rr=max(ENT_EPS,r*np.std(y));Xm=_embed(y,m);X1=_embed(y,m+1);B=_match_pairs(Xm,rr);A=_match_pairs(X1,rr)
    return A,max(1,len(X1)*(len(X1)-1)/2),B,max(1,len(Xm)*(len(Xm)-1)/2)

def _moving_coarse(y,s):
    y=np.asarray(y,float);s=int(s)
    if s<=1:return y.copy()
    return np.convolve(y,np.ones(s)/s,mode='valid')

def _variance_coarse(y,s):
    y=np.asarray(y,float);s=int(s);n=len(y)//s
    return np.var(y[:n*s].reshape(n,s),axis=1) if n else np.empty(0)

def _hier_nodes(y,levels=4):
    nodes=[np.asarray(y,float)]
    level=[np.asarray(y,float)]
    for _ in range(levels):
        nxt=[]
        for z in level:
            n=(len(z)//2)*2;z=z[:n]
            if n<4:continue
            a=(z[0::2]+z[1::2])/np.sqrt(2);d=(z[0::2]-z[1::2])/np.sqrt(2);nxt.extend([a,d])
        nodes.extend(nxt);level=nxt
        if not level:break
    return nodes

def _morlet(y,freqs=None,n_cycles=7.):
    y=np.asarray(y,float)
    if freqs is None:freqs=np.geomspace(2,min(45,.42*_ent_fs),24 if FAST_OPTION=='ultra' else 36)
    C=[]
    for f in freqs:
        sig=n_cycles/(2*np.pi*f);half=int(max(8,4*sig*_ent_fs));tt=np.arange(-half,half+1)/_ent_fs
        w=np.exp(2j*np.pi*f*tt)*np.exp(-(tt**2)/(2*sig**2));w/=np.sqrt(np.sum(np.abs(w)**2)+ENT_EPS);C.append(signal.fftconvolve(y,np.conj(w[::-1]),mode='same'))
    return np.asarray(freqs),np.asarray(C)

def _haar_packet(y,level=4):
    level_data=[np.asarray(y,float)]
    for _ in range(level):
        nxt=[]
        for z in level_data:
            n=(len(z)//2)*2;z=z[:n]
            if n<2:continue
            nxt.extend([(z[0::2]+z[1::2])/np.sqrt(2),(z[0::2]-z[1::2])/np.sqrt(2)])
        level_data=nxt
    return level_data

def _haar_details(y,levels=5):
    a=np.asarray(y,float);out=[]
    for _ in range(levels):
        n=(len(a)//2)*2;a=a[:n]
        if n<2:break
        out.append((a[0::2]-a[1::2])/np.sqrt(2));a=(a[0::2]+a[1::2])/np.sqrt(2)
    out.append(a);return out

def _energy_entropy(parts):
    E=np.array([np.sum(np.abs(z)**2) for z in parts],float);p=E/(E.sum()+ENT_EPS);return _H(p),p

def _st_entropy(y,nper=None):
    nper=int(nper or min(256,max(64,len(y)//8)));nover=nper//2;f,t,Z=signal.stft(y,fs=_ent_fs,nperseg=nper,noverlap=nover,boundary=None);P=np.abs(Z)**2;P/=P.sum()+ENT_EPS;return _H(P.ravel()),f,t,P

def _stockwell(y,freqs=None):
    y=np.asarray(y,float);freqs=np.asarray(freqs if freqs is not None else np.geomspace(2,min(45,.42*_ent_fs),20),float);S=[]
    for f in freqs:
        sig=1/(2*np.pi*f);half=int(max(8,4*sig*_ent_fs));tt=np.arange(-half,half+1)/_ent_fs
        g=(f/np.sqrt(2*np.pi))*np.exp(-.5*(f*tt)**2)*np.exp(-2j*np.pi*f*tt);S.append(signal.fftconvolve(y,g,mode='same'))
    return freqs,np.asarray(S)

def _emd_simple(y,max_imfs=None,max_sift=None):
    max_imfs=int(max_imfs or {'full':7,'balanced':6,'fast':5,'ultra':4}[FAST_OPTION]);max_sift=int(max_sift or {'full':30,'balanced':22,'fast':16,'ultra':10}[FAST_OPTION]);r=np.asarray(y,float).copy();imfs=[]
    for _ in range(max_imfs):
        h=r.copy()
        for _s in range(max_sift):
            ma=signal.find_peaks(h)[0];mi=signal.find_peaks(-h)[0]
            if len(ma)<2 or len(mi)<2:break
            x=np.arange(len(h));up=np.interp(x,ma,h[ma]);lo=np.interp(x,mi,h[mi]);mean=(up+lo)/2;hn=h-mean
            if np.linalg.norm(hn-h)/(np.linalg.norm(h)+ENT_EPS)<.05:h=hn;break
            h=hn
        if len(signal.find_peaks(h)[0])<2:break
        imfs.append(h);r=r-h
        if np.std(r)<1e-4*np.std(y):break
    return imfs,r

def _eemd_energy(y):
    rng=np.random.default_rng(94);ens={'full':20,'balanced':12,'fast':7,'ultra':4}[FAST_OPTION];allE=[]
    for _ in range(ens):
        yy=np.asarray(y,float)+.15*np.std(y)*rng.standard_normal(len(y));imfs,_=_emd_simple(yy);allE.append([np.sum(z*z) for z in imfs])
    K=max(map(len,allE),default=0);E=np.array([np.pad(v,(0,K-len(v))) for v in allE]).mean(0) if K else np.array([]);p=E/(E.sum()+ENT_EPS);return _H(p),p

def _vmd_simple(y,K=4,alpha=2000,tau=0.):
    y=np.asarray(y,float);n=min(len(y),4096 if FAST_OPTION=='ultra' else 8192);y=y[:n];it={'full':160,'balanced':120,'fast':80,'ultra':45}[FAST_OPTION]
    f=np.fft.fftfreq(n);Y=np.fft.fft(y);u=np.zeros((K,n),complex);omega=np.linspace(0,.4,K+2)[1:-1];lam=np.zeros(n,complex);pos=f>=0
    for _ in range(it):
        us=np.sum(u,axis=0)
        for k in range(K):
            rest=us-u[k];new=(Y-rest+lam/2)/(1+2*alpha*(f-omega[k])**2);us=rest+new;u[k]=new;pw=np.abs(new[pos])**2;omega[k]=np.sum(f[pos]*pw)/(np.sum(pw)+ENT_EPS)
        if tau:lam+=tau*(Y-us)
    return np.real(np.fft.ifft(u,axis=1)),omega*_ent_fs
```

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#181818;color:#D0D0D0;padding:18px;border-radius:8px;margin:10px;line-height:1.55;">
<h3 style="color:#E8E8E8;margin-top:0;">Execution router — entropy methods 65–99</h3>
<p>This dispatcher maps each catalogue index to its multiscale, wavelet, time-frequency, or adaptive-decomposition implementation. It introduces no additional estimator; its purpose is to apply the method-specific branch to the common EEG data and shared speed configuration.</p>
</div>


```python
def _ent_run_65_99(i):
    y=_ent_y;sc=np.arange(1,_ENT_SPEED['scales']+1)
    if i==65:
        v=[]
        for s in sc:
            z=[_sampen(_coarse(y,int(s),o)) for o in range(int(s)) if len(_coarse(y,int(s),o))>30];v.append(np.nanmean(z))
        _line(sc,v,'Composite Multiscale Entropy','scale','CMSE');return
    if i==66:
        v=[]
        for s in sc:
            AA=BA=0.;Ad=Bd=0.
            for o in range(int(s)):
                z=_coarse(y,int(s),o)
                if len(z)>30:
                    A,da,B,db=_sampen_stats(z);AA+=A;Ad+=da;BA+=B;Bd+=db
            pA=AA/max(1,Ad);pB=BA/max(1,Bd);v.append(-np.log((pA+ENT_EPS)/(pB+ENT_EPS)))
        _line(sc,v,'Refined Composite Multiscale Entropy','scale','RCMSE');return
    if i==67:_line(sc,[_sampen(_moving_coarse(y,int(s))) for s in sc],'Modified Multiscale Entropy (moving-average coarse grain)','scale','MMSE');return
    if i==68:
        W=max(300,int(3*_ent_fs));starts=np.linspace(0,max(0,len(y)-W),min(8,_ENT_SPEED['windows'])).astype(int);M=np.array([[_sampen(_coarse(y[s:s+W],int(q))) for q in sc] for s in starts]);display(pd.DataFrame(M,index=np.round(starts/_ent_fs,2),columns=sc));return
    if i==69:_line(sc,[_sampen(_variance_coarse(y,int(s))) for s in sc],'Generalized Multiscale Entropy (variance coarse grain)','scale','GMSE');return
    if i==70:
        nodes=_hier_nodes(y,levels=3);vals=[_H(_hist_p(z,16)) for z in nodes];display(pd.DataFrame({'node':range(len(nodes)),'Shannon_entropy':vals}));return
    if i==71:
        nodes=_hier_nodes(y,levels=3);vals=[_sampen(z) if len(z)>30 else np.nan for z in nodes];display(pd.DataFrame({'node':range(len(nodes)),'SampleEntropy':vals}));return
    if i==72:
        X=_ent_x[:,:min(4,_ent_x.shape[1])];v=[]
        for s in sc:
            n=len(X)//s;cg=X[:n*s].reshape(n,s,X.shape[1]).mean(1);v.append(_multivar_sampen(cg,2,.25))
        _line(sc,v,'Multivariate Multiscale Entropy','scale','MvMSE');return
    if i==73:
        ch=min(6,_ent_x.shape[1]);v=[]
        for s in sc:v.append(np.nanmean([_sampen(_coarse(_ent_x[:,j],int(s))) for j in range(ch)]))
        _line(sc,v,'Multichannel Multiscale Entropy','scale','mean channel SampEn');return
    if i==74:_line(sc,[_cross_sampen(_coarse(_ent_x[:,0],int(s)),_coarse(_ent_x[:,1],int(s))) for s in sc],'Multiscale Cross-Entropy','scale','Cross-SampEn');return
    if i in (75,76):
        vals=[]
        for s in sc:
            a=_coarse(_ent_x[:,0],int(s));b=_coarse(_ent_x[:,1],int(s));n=min(len(a),len(b));qa=np.digitize(a[:n],np.quantile(a[:n],[.25,.5,.75]));qb=np.digitize(b[:n],np.quantile(b[:n],[.25,.5,.75]));Hxy=_H(_p_from_labels(np.column_stack([qa,qb])));Hx=_H(_p_from_labels(qa));vals.append(Hxy if i==75 else Hxy-Hx)
        _line(sc,vals,'Multiscale Joint Entropy' if i==75 else 'Multiscale Conditional Entropy H(Y|X)','scale','entropy');return
    if i==77:
        parts=_haar_details(y,5);h,p=_energy_entropy(parts);print('wavelet energy entropy=',h,'level probabilities=',p);return
    if i==78:
        parts=_haar_packet(y,4);h,p=_energy_entropy(parts);print('Haar wavelet-packet entropy=',h,'packets=',len(parts));return
    if i==79:
        parts=_haar_details(y,5);h,p=_energy_entropy(parts);display(pd.DataFrame({'component':range(len(p)),'relative_wavelet_energy':p}));print('RWE Shannon entropy=',h);return
    if i==80:
        f,C=_morlet(_cap(y,2500));s=np.linalg.svd(np.abs(C),compute_uv=False);p=s/(s.sum()+ENT_EPS);print('Wavelet Singular Entropy=',_H(p));return
    if i==81:
        levels=np.arange(2,7);vals=[]
        for l in levels:vals.append(_energy_entropy(_haar_details(y,int(l)))[0])
        _line(levels,vals,'Multiscale Wavelet Entropy','decomposition levels','entropy');return
    if i==82:
        f,C=_morlet(_cap(y,3000));P=np.abs(C)**2;P/=P.sum()+ENT_EPS;print('CWT scalogram entropy=',_H(P.ravel()));return
    if i==83:
        f,C1=_morlet(_cap(_ent_x[:,0],2500));_,C2=_morlet(_cap(_ent_x[:,1],2500),f);S12=signal.convolve2d(C1*np.conj(C2),np.ones((3,9))/27,mode='same',boundary='symm');S1=signal.convolve2d(np.abs(C1)**2,np.ones((3,9))/27,mode='same',boundary='symm');S2=signal.convolve2d(np.abs(C2)**2,np.ones((3,9))/27,mode='same',boundary='symm');coh=np.abs(S12)**2/(S1*S2+ENT_EPS);p=coh.ravel();p=p/(p.sum()+ENT_EPS);print('Wavelet coherence entropy=',_H(p));return
    if i==84:
        f,C=_morlet(_cap(y,3000));phase=np.angle(C).ravel();print('Wavelet phase entropy=',_H(_hist_p(phase,24)));return
    if i==85:
        h,_,_,_= _st_entropy(y);print('Time-frequency STFT energy entropy=',h);return
    if i==86:
        nper=min(256,max(64,len(y)//8));f,t,Z=signal.stft(y,fs=_ent_fs,nperseg=nper,noverlap=nper//2,boundary=None);P=np.abs(Z)**2;vals=[]
        for k in range(P.shape[1]):q=P[:,k]/(P[:,k].sum()+ENT_EPS);vals.append(_H(q))
        _line(t,vals,'Spectrogram entropy trajectory','time (s)','spectral entropy');return
    if i==87:
        ch=min(8,_ent_x.shape[1]);rows=[]
        for j in range(ch):h,f,t,P=_st_entropy(_ent_x[:,j]);rows.append([_ent_names[j],h])
        display(pd.DataFrame(rows,columns=['channel','STFT_entropy']));return
    if i==88:
        f,t,Z=signal.stft(y,fs=_ent_fs,nperseg=min(256,max(64,len(y)//8)),boundary=None);P=np.abs(Z)**2;rows=[]
        for name,(lo,hi) in _ENT_BANDS.items():
            mask=(f>=lo)&(f<min(hi,_ent_fs/2));q=P[mask].ravel();q=q/(q.sum()+ENT_EPS);rows.append([name,_H(q)]) if len(q) else None
        display(pd.DataFrame(rows,columns=['band','spectrogram_entropy']));return
    if i in (89,90):
        f,P=signal.welch(y,fs=_ent_fs,nperseg=min(1024,len(y)));p=P/(P.sum()+ENT_EPS);q=2.;print(('Rényi' if i==89 else 'Tsallis'),'spectral entropy=',_renyi(p,q) if i==89 else _tsallis(p,q));return
    if i==91:
        f,S=_stockwell(_cap(y,3000));P=np.abs(S)**2;P/=P.sum()+ENT_EPS;print('S-transform energy entropy=',_H(P.ravel()));return
    if i in (92,93):
        imfs,res=_emd_simple(_cap(y,3000));
        if not imfs:print('no IMFs resolved');return
        if i==93:
            h,p=_energy_entropy(imfs);print('EMD IMF energy entropy=',h,'probabilities=',p);return
        H=[]
        for z in imfs:
            a=signal.hilbert(z);H.append(np.abs(a)**2)
        E=np.vstack(H);p=E/(E.sum(0,keepdims=True)+ENT_EPS);ht=np.array([-np.sum(p[:,k]*np.log(p[:,k]+ENT_EPS)) for k in range(p.shape[1])]);print('Hilbert-Huang mean instantaneous mode entropy=',float(np.mean(ht)));return
    if i==94:
        h,p=_eemd_energy(_cap(y,2500));print('Ensemble EMD mean IMF-energy entropy=',h,'energy shares=',p);return
    if i==95:
        modes,cf=_vmd_simple(_cap(y,3500));h,p=_energy_entropy(modes);print('VMD mode-energy entropy=',h,'center frequencies Hz=',cf);return
    if i==96:
        imfs,_=_emd_simple(_cap(y,3000));h,p=_energy_entropy(imfs);print('Mode-energy entropy=',h,'mode shares=',p);return
    if i==97:
        a=signal.hilbert(y);ph=np.unwrap(np.angle(a));fi=np.diff(ph)*_ent_fs/(2*np.pi);fi=fi[np.isfinite(fi)];print('Instantaneous-frequency entropy=',_H(_hist_p(fi,32)));return
    if i==98:
        env=np.abs(signal.hilbert(y));print('Amplitude-envelope entropy=',_H(_hist_p(env,32)));return
    if i==99:
        ph=np.angle(signal.hilbert(y));print('Hilbert phase entropy=',_H(_hist_p(ph,bins=24)));return
    raise ValueError(i)
```

### 65. Composite Multiscale Entropy

Composite Multiscale Entropy averages Sample Entropy over all coarse-graining offsets at each scale, reducing dependence on the arbitrary block origin.

```python
_ent_run_65_99(65)
```

### 66. Refined Composite Multiscale Entropy

Refined Composite Multiscale Entropy pools matching statistics across offsets before taking the logarithm, rather than averaging already-log-transformed SampEn values.

```python
_ent_run_65_99(66)
```

### 67. Modified Multiscale Entropy

Modified Multiscale Entropy uses an overlapping moving-average coarse-grained series, retaining more samples than non-overlapping coarse graining.

```python
_ent_run_65_99(67)
```

### 68. Short-Time Multiscale Entropy

Short-Time Multiscale Entropy estimates a scale-by-time matrix by applying multiscale Sample Entropy inside local windows.

```python
_ent_run_65_99(68)
```

### 69. Generalized Multiscale Entropy

Generalized Multiscale Entropy can use alternative coarse-grained moments; this cell uses block variance, a common GMSE construction, before Sample Entropy.

```python
_ent_run_65_99(69)
```

### 70. Hierarchical Entropy

Hierarchical Entropy recursively splits a signal into Haar-like average and difference branches and measures entropy on every node of the resulting multiresolution tree.

```python
_ent_run_65_99(70)
```

### 71. Hierarchical Sample Entropy

Hierarchical Sample Entropy uses the same average/difference hierarchy but evaluates Sample Entropy on each node.

```python
_ent_run_65_99(71)
```

### 72. Multivariate Multiscale Entropy

Multivariate Multiscale Entropy coarse-grains synchronized channels jointly and computes Sample Entropy in the resulting multivariate delay space.

```python
_ent_run_65_99(72)
```

### 73. Multichannel Multiscale Entropy

Multichannel Multiscale Entropy reports the mean of channel-wise multiscale Sample Entropy, which is distinct from joint-state multivariate MSE.

### 74. Multiscale Cross-Entropy

Multiscale Cross-Entropy computes cross-Sample Entropy between two channels after coarse-graining at each scale.

```python
_ent_run_65_99(74)
```

### 75. Multiscale Joint Entropy

Multiscale Joint Entropy is Shannon entropy H(X_s,Y_s) of synchronized quantized coarse-grained channel states at each scale.

### 76. Multiscale Conditional Entropy

Multiscale Conditional Entropy uses H(Y_s|X_s)=H(X_s,Y_s)-H(X_s) on the coarse-grained symbolic states.

### 77. Wavelet Entropy

Wavelet Entropy normalizes discrete wavelet-component energies p_j=E_j/sum E and computes -sum p_j ln p_j.

```python
_ent_run_65_99(77)
```

### 78. Wavelet Packet Entropy

Wavelet Packet Entropy applies the same energy-probability entropy to all terminal nodes of a full Haar packet decomposition.

```python
_ent_run_65_99(78)
```

### 79. Relative Wavelet Energy Entropy

Relative Wavelet Energy is p_j=E_j/sum E; its Shannon entropy measures how broadly energy is distributed over wavelet scales.

```python
_ent_run_65_99(79)
```

### 80. Wavelet Singular Entropy

Wavelet Singular Entropy forms a time-scale coefficient matrix, normalizes its singular values, and computes Shannon entropy of that singular-value spectrum.

```python
_ent_run_65_99(80)
```

### 81. Multiscale Wavelet Entropy

Multiscale Wavelet Entropy repeats wavelet energy entropy across increasing decomposition depths.

```python
_ent_run_65_99(81)
```

### 82. CWT Scalogram Entropy

CWT Scalogram Entropy normalizes the Morlet scalogram energy over time and scale and computes Shannon entropy of that 2-D distribution.

```python
_ent_run_65_99(82)
```

### 83. Wavelet Coherence Entropy

Wavelet Coherence Entropy normalizes the squared wavelet-coherence field R^2=|S(Wxy)|^2/[S(|Wx|^2)S(|Wy|^2)] and computes entropy of its spatial distribution.

```python
_ent_run_65_99(83)
```

### 84. Wavelet Phase Entropy

Wavelet Phase Entropy is Shannon entropy of the phase-angle distribution of complex wavelet coefficients.

```python
_ent_run_65_99(84)
```

### 85. Time-frequency entropy

Time-frequency entropy treats normalized STFT energy as a joint probability distribution over time and frequency.

```python
_ent_run_65_99(85)
```

### 86. Spectrogram entropy trajectories

Spectrogram entropy trajectories normalize spectral power within each STFT time slice and track its Shannon entropy through time.

```python
_ent_run_65_99(86)
```

### 87. STFT entropy maps

STFT entropy maps compare joint time-frequency energy entropy across EEG channels.

### 88. Band-specific spectrogram entropy

Band-specific spectrogram entropy restricts the STFT energy distribution to each canonical frequency band before normalization and Shannon entropy.

### 89. Rényi spectral entropy

Rényi spectral entropy applies Rényi H_q to normalized power-spectral density rather than amplitude-state probabilities.

### 90. Tsallis spectral entropy

Tsallis spectral entropy applies Tsallis S_q to normalized spectral power.

```python
_ent_run_65_99(90)
```

### 91. S-transform entropy

The Stockwell transform uses a frequency-dependent Gaussian window; S-transform entropy is Shannon entropy of its normalized time-frequency energy.

```python
_ent_run_65_99(91)
```

### 92. Hilbert-Huang entropy

Hilbert-Huang entropy decomposes the signal into intrinsic mode functions, obtains analytic-signal instantaneous energies, and measures entropy across modes as a function of time.

### 93. EMD IMF entropy

EMD IMF entropy normalizes intrinsic-mode energies and computes Shannon entropy over the IMF energy shares.

### 94. Ensemble EMD entropy

Ensemble EMD entropy adds independent noise across an ensemble, performs EMD repeatedly, averages aligned IMF energies, and computes entropy of the mean energy shares.

### 95. Variational Mode Decomposition entropy

VMD entropy uses variational mode decomposition to obtain narrow-band modes and computes Shannon entropy of their normalized energies.

### 96. Mode-energy entropy

Mode-energy entropy is the generic Shannon entropy -sum p_k ln p_k of normalized decomposition-mode energies.

### 97. Instantaneous-frequency entropy

Instantaneous-frequency entropy uses the derivative of unwrapped Hilbert phase, f_i=(fs/2pi) Delta phi, and computes entropy of its empirical distribution.

### 99. Hilbert phase entropy

<h3 style="color:#E8E8E8;margin-top:0;">Theory and role — matrix, algorithmic, and entropy-rate helpers for methods 100–130</h3>
<p>This foundation defines trajectory/SVD, covariance-spectrum, compression/complexity, and entropy-rate utilities. For normalized singular/eigenvalue weights \(p_i\), spectral entropy is \(-\sum_i p_i\log p_i\) and effective rank is \(\exp(H)\). Finite symbolic entropy-rate estimates use block entropies, e.g. \(h_L=H(S_1^L)-H(S_1^{L-1})\), rather than claiming the infinite-length limit from finite EEG.</p>
</div>


```python
# =======================================================
# ENTROPY CATALOGUE HELPERS — methods 100–130
# Matrix, algorithmic, entropy-rate and predictive-information measures.
# =======================================================
import zlib
from scipy.special import gammaln, logsumexp

def _trajectory_matrix(y,m=20):
    X=_embed(y,m,1);return X-X.mean(0,keepdims=True)

def _eig_probs(M):
    w=np.linalg.eigvalsh((M+M.T)/2);w=np.clip(w,0,None);return w/(w.sum()+ENT_EPS)

def _svd_entropy(A):
    s=np.linalg.svd(np.asarray(A,float),compute_uv=False);p=s*s/(np.sum(s*s)+ENT_EPS);return _H(p),p

def _symbolize(y,k=4):
    y=np.asarray(y,float);e=np.unique(np.quantile(y,np.linspace(0,1,k+1)))
    if len(e)<2:return np.zeros(len(y),dtype=int)
    return np.clip(np.digitize(y,e[1:-1]),0,len(e)-2).astype(int)

def _lz76(seq):
    # Exhaustive LZ76 phrase parsing; returns phrase count.
    s=tuple(seq);n=len(s);i=0;c=0;dictionary=set()
    while i<n:
        L=1
        while i+L<=n and tuple(s[i:i+L]) in dictionary:L+=1
        phrase=tuple(s[i:min(n,i+L)]);dictionary.add(phrase);c+=1;i+=len(phrase)
    return c

def _lz_norm(seq,alphabet=None):
    seq=list(seq);n=len(seq)
    if n<2:return 0.0
    a=int(alphabet or max(2,len(set(seq))));return float(_lz76(seq)*np.log(n)/(n*np.log(a)))

def _to_bytes_symbols(seq):
    a=np.asarray(seq,int);return bytes(np.mod(a,256).astype(np.uint8).tolist())

def _comp_bits(seq):
    return 8*len(zlib.compress(_to_bytes_symbols(seq),9))

def _ncd(a,b):
    A=_to_bytes_symbols(a);B=_to_bytes_symbols(b);ca=len(zlib.compress(A,9));cb=len(zlib.compress(B,9));cab=len(zlib.compress(A+B,9));return float((cab-min(ca,cb))/max(ca,cb,1))

def _bdm_proxy(seq,block=12):
    # BDM structure: sum_j [K(block_j)+log2 multiplicity_j]; K is approximated by compressed block bits because no CTM table is bundled.
    s=list(seq);blocks=[tuple(s[i:i+block]) for i in range(0,len(s)-block+1,block)];C=Counter(blocks);tot=0.
    for b,n in C.items():tot+=_comp_bits(b)+np.log2(n)
    return float(tot)

def _kt_logprob(n0,n1):
    return float(gammaln(n0+.5)+gammaln(n1+.5)-gammaln(n0+n1+1)-2*gammaln(.5)+gammaln(1.0))

def _ctw_logprob(bits,depth=5):
    bits=np.asarray(bits,int).ravel();depth=int(depth)
    # Each node stores counts of target symbols seen under a particular context suffix.
    tree={():[0,0]}
    for t,b in enumerate(bits):
        tree.setdefault((),[0,0])[int(b)]+=1
        ctx=[]
        for d in range(1,min(depth,t)+1):
            ctx.append(int(bits[t-d]));key=tuple(ctx);tree.setdefault(key,[0,0])[int(b)]+=1
    from functools import lru_cache
    @lru_cache(None)
    def rec(key):
        n0,n1=tree.get(key,[0,0]);lk=_kt_logprob(n0,n1)
        if len(key)>=depth:return lk
        c0=key+(0,);c1=key+(1,);lc=rec(c0)+rec(c1)
        return float(logsumexp([np.log(.5)+lk,np.log(.5)+lc]))
    return rec(())

def _block_entropy(sym,L):
    sym=np.asarray(sym,int);n=len(sym)-L+1
    if n<=0:return np.nan
    X=sliding_window_view(sym,L);return _H(_p_from_labels(X))

def _cond_entropy_markov(sym,order=1):
    sym=np.asarray(sym,int);return float(_block_entropy(sym,order+1)-_block_entropy(sym,order))

def _joint_symbol_entropy(a,b):
    n=min(len(a),len(b));return _H(_p_from_labels(np.column_stack([a[:n],b[:n]])))

def _mi_symbols(a,b):
    n=min(len(a),len(b));a=np.asarray(a)[:n];b=np.asarray(b)[:n];return float(_H(_p_from_labels(a))+_H(_p_from_labels(b))-_joint_symbol_entropy(a,b))

def _causal_states(sym,L=3,decimals=2):
    sym=np.asarray(sym,int);K=int(sym.max()+1);counts={}
    for t in range(L,len(sym)):
        past=tuple(sym[t-L:t]);counts.setdefault(past,np.zeros(K))[sym[t]]+=1
    states={};past_state={}
    for past,c in counts.items():
        p=(c+.5)/(c.sum()+.5*K);key=tuple(np.round(p,decimals));
        if key not in states:states[key]=len(states)
        past_state[past]=states[key]
    seq=[]
    for t in range(L,len(sym)):
        p=tuple(sym[t-L:t]);seq.append(past_state.get(p,-1))
    return np.asarray(seq,int),past_state,states,counts
```

```python
def _ent_run_100_130(i):
    y=_ent_y;X=_ent_x
    if i==100:
        h,p=_svd_entropy(_trajectory_matrix(y,24));print('SVD entropy=',h,'effective singular modes=',np.exp(h));return
    if i==101:
        C=np.cov(X,rowvar=False);p=_eig_probs(C);print('PCA eigenvalue entropy=',_H(p));return
    if i==102:
        C=np.cov(_ent_raw[:min(len(_ent_raw),20000)],rowvar=False);p=_eig_probs(C);print('Covariance eigenvalue entropy=',_H(p));return
    if i==103:
        R=np.corrcoef(X,rowvar=False);p=_eig_probs(R);print('Correlation-matrix eigenvalue entropy=',_H(p));return
    if i==104:
        p=_eig_probs(np.cov(X,rowvar=False));h=_H(p);print('effective rank exp(H)=',np.exp(h),'H=',h);return
    if i==105:
        p=_eig_probs(np.cov(X,rowvar=False));H2=-np.log(np.sum(p*p)+ENT_EPS);print('participation entropy H2=',H2,'participation ratio=',np.exp(H2));return
    if i==106:
        sc=np.arange(1,_ENT_SPEED['scales']+1);v=[]
        for s in sc:
            n=len(X)//s;cg=X[:n*s].reshape(n,s,X.shape[1]).mean(1);v.append(_svd_entropy(cg)[0])
        _line(sc,v,'Multiscale SVD Entropy','scale','SVD entropy');return
    if i==107:
        W=max(200,int(3*_ent_fs));starts=np.linspace(0,max(0,len(X)-W),_ENT_SPEED['windows']).astype(int);v=[_svd_entropy(X[s:s+W])[0] for s in starts];_line(starts/_ent_fs,v,'Time-resolved SVD Entropy','time (s)','SVD entropy');return
    sym=_symbolize(y,4)
    if i==108:print('Lempel-Ziv normalized complexity entropy proxy=',_lz_norm(sym,4));return
    if i==109:
        labs=_ordinal_labels(y,3);ids={k:j for j,k in enumerate(sorted(set(labs)))};seq=[ids[k] for k in labs];print('Permutation Lempel-Ziv complexity=',_lz_norm(seq,max(2,len(ids))));return
    if i==110:
        sc=np.arange(1,_ENT_SPEED['scales']+1);v=[_lz_norm(_symbolize(_coarse(y,int(s)),4),4) for s in sc];_line(sc,v,'Multiscale Lempel-Ziv Complexity','scale','normalized LZ76');return
    if i==111:
        raw=max(1,len(_to_bytes_symbols(sym)));print('compression entropy proxy bits/symbol=',_comp_bits(sym)/len(sym),'compression ratio=',len(zlib.compress(_to_bytes_symbols(sym),9))/raw);return
    if i==112:
        a=_symbolize(_ent_x[:,0],4);b=_symbolize(_ent_x[:,1],4);print('normalized compression distance ch0-ch1=',_ncd(a,b));return
    if i==113:print('BDM entropy proxy bits=',_bdm_proxy(sym),'(compressed-block K proxy; no CTM table claimed)');return
    if i==114:print('algorithmic complexity proxy compressed bits/symbol=',_comp_bits(sym)/len(sym));return
    if i==115:
        bits=(y>np.median(y)).astype(int);lp=_ctw_logprob(bits,depth=5);print('CTW entropy-rate estimate bits/sample=',-lp/(len(bits)*np.log(2)));return
    if i==116:print('approximate Kolmogorov complexity proxy bits=',_comp_bits(sym),'for',len(sym),'symbols');return
    if i==117:print('Markov-partition entropy rate H(S_t|S_t-1)=',_cond_entropy_markov(sym,1));return
    if i==118:
        hs=[_block_entropy(sym,L) for L in range(1,6)];print('symbolic entropy-rate block increment=',hs[-1]-hs[-2],'block entropies=',hs);return
    if i==119:
        labs=_ordinal_labels(y,3);ids={k:j for j,k in enumerate(sorted(set(labs)))};s=np.array([ids[k] for k in labs]);print('ordinal entropy rate H(P_t|P_t-1)=',_cond_entropy_markov(s,1));return
    if i==120:
        L=np.arange(1,7);H=[_block_entropy(sym,int(k)) for k in L];_line(L,H,'Block Entropy Growth Curves','block length L','H(L)');return
    if i==121:
        L=np.arange(1,7);H=np.array([_block_entropy(sym,int(k)) for k in L]);h=np.mean(np.diff(H)[-2:]);E=float(np.mean(H[-3:]-h*L[-3:]));print('finite-block excess entropy intercept≈',E,'entropy-rate tail≈',h);return
    if i==122:
        L=3;past=np.array([sum(int(sym[t-k-1])*(4**k) for k in range(L)) for t in range(L,len(sym)-L)]);future=np.array([sum(int(sym[t+k])*(4**k) for k in range(L)) for t in range(L,len(sym)-L)]);print('predictive information I(past;future)=',_mi_symbols(past,future));return
    if i in (123,124,125):
        ss,mp,states,counts=_causal_states(sym,L=3);h=_H(_p_from_labels(ss[ss>=0]));
        if i==123:print('statistical complexity C_mu≈H(causal state)=',h,'states=',len(states));return
        if i==124:print('causal-state entropy=',h);return
        # epsilon-machine summary: state transition probabilities under observed symbols
        rows=[]
        for a,b in zip(ss[:-1],ss[1:]):
            if a>=0 and b>=0:rows.append((int(a),int(b)))
        print('epsilon-machine finite-history state entropy=',h,'state transitions=',len(set(rows)));return
    if i in (126,127,128):
        L=3;n=len(sym)-L;past=np.array([sum(int(sym[t-k-1])*(4**k) for k in range(L)) for t in range(L,len(sym))]);cur=sym[L:]
        Hcur=_H(_p_from_labels(cur));Hjoint=_joint_symbol_entropy(past,cur);Hpast=_H(_p_from_labels(past));AIS=Hpast+Hcur-Hjoint;Hpred=Hjoint-Hpast
        if i==126:print('Active Information Storage I(past;current)=',AIS);return
        if i==128:print('Predictive entropy H(current|past)=',Hpred);return
        # local AIS log p(x|past)/p(x)
        pc=Counter(cur);pp=Counter(past);pj=Counter(zip(past,cur));N=len(cur);vals=[np.log((pj[(a,b)]*N+ENT_EPS)/(pp[a]*pc[b]+ENT_EPS)) for a,b in zip(past,cur)];print('Local AIS mean=',float(np.mean(vals)),'median=',float(np.median(vals)));return
    if i==129:
        L=3;past=np.array([sum(int(sym[t-k-1])*(4**k) for k in range(L)) for t in range(L,len(sym)-L)]);future=np.array([sum(int(sym[t+k])*(4**k) for k in range(L)) for t in range(L,len(sym)-L)]);print('past-future mutual information=',_mi_symbols(past,future));return
    if i==130:
        hf=_cond_entropy_markov(sym,3);hr=_cond_entropy_markov(sym[::-1],3);print('finite-order predictive entropy asymmetry H_fwd-H_rev=',hf-hr,'(asymptotic Shannon entropy rate itself is time-reversal invariant)');return
    raise ValueError(i)
```

### 100. Singular Value Decomposition entropy

SVD entropy normalizes squared singular values of a trajectory matrix, p_k=s_k^2/sum s_j^2, and computes -sum p_k ln p_k.

```python
_ent_run_100_130(100)
```

### 101. PCA eigenvalue entropy

PCA eigenvalue entropy is Shannon entropy of normalized covariance eigenvalues, equivalent to entropy of explained-variance fractions.

```python
_ent_run_100_130(101)
```

### 102. Covariance eigenvalue entropy

Covariance eigenvalue entropy normalizes the eigenvalues of the channel covariance matrix and measures how broadly variance is distributed over covariance modes.

```python
_ent_run_100_130(102)
```

### 103. Correlation-matrix entropy

Correlation-matrix entropy applies the same eigenvalue normalization to the channel correlation matrix, removing channel scale from the geometry.

```python
_ent_run_100_130(103)
```

### 104. Effective-rank entropy

Effective rank is exp(H(p)), where p are normalized nonnegative matrix eigenvalues. It is the entropy-equivalent number of active modes.

```python
_ent_run_100_130(104)
```

### 105. Participation entropy of channel covariance

Participation entropy is the Rényi order-2 entropy H2=-ln sum p_i^2; exp(H2)=1/sum p_i^2 is the participation ratio.

```python
_ent_run_100_130(105)
```

### 106. Multiscale SVD entropy

Multiscale SVD entropy coarse-grains the multichannel time series and recomputes singular-value energy entropy at each scale.

```python
_ent_run_100_130(106)
```

### 107. Time-resolved SVD entropy

Time-resolved SVD entropy applies singular-value energy entropy to successive multichannel windows.

```python
_ent_run_100_130(107)
```

### 108. Lempel-Ziv complexity entropy proxy

Lempel-Ziv complexity counts new phrases in a symbolic sequence. The normalized LZ76 quantity c(n) ln n/[n ln alphabet] is used as an entropy-rate/algorithmic-complexity proxy.

```python
_ent_run_100_130(108)
```

### 109. Permutation Lempel-Ziv complexity

Permutation Lempel-Ziv complexity first converts the signal to ordinal-pattern symbols and then applies normalized LZ76 parsing.

```python
_ent_run_100_130(109)
```

### 110. Multiscale Lempel-Ziv complexity

Multiscale Lempel-Ziv complexity coarse-grains the signal, symbolizes each scale, and evaluates normalized LZ76 complexity.

```python
_ent_run_100_130(110)
```

### 111. Compression entropy

Compression entropy uses lossless compressed length per symbol as a practical entropy-rate proxy; finite headers and compressor modeling make it estimator-dependent.

```python
_ent_run_100_130(111)
```

### 112. Normalized compression distance entropy

Normalized compression distance is NCD(x,y)=[C(xy)-min(C(x),C(y))]/max(C(x),C(y)), a compressor-based similarity proxy.

```python
_ent_run_100_130(112)
```

### 113. Block Decomposition Method entropy proxy

Block Decomposition Method has structure BDM=sum_j[K(block_j)+log2 n_j]. Because no certified CTM table is bundled, K(block) is explicitly approximated by lossless compressed block length.

```python
_ent_run_100_130(113)
```

### 114. Algorithmic complexity entropy proxy

Approximate algorithmic complexity is estimated by the length of a losslessly compressed symbolic representation; it is a computable proxy, not exact Kolmogorov complexity.

```python
_ent_run_100_130(114)
```

### 115. Context-tree weighting entropy rate

Context-Tree Weighting recursively mixes a Krichevsky-Trofimov estimator at each binary context node with the product of its child models: P_w=0.5 P_KT+0.5 P_w0 P_w1. Entropy rate is -ln P_w/n.

```python
_ent_run_100_130(115)
```

### 116. Approximate Kolmogorov complexity proxy

Kolmogorov complexity is uncomputable in general. This cell therefore reports only a lossless-compression upper-bound proxy and labels it accordingly.

```python
_ent_run_100_130(116)
```

### 117. Entropy rate from Markov partitions

A Markov-partition entropy-rate estimate is H(S_t|S_t-1)=H(S_t,S_t-1)-H(S_t-1) after amplitude symbolization.

```python
_ent_run_100_130(117)
```

### 118. Symbolic entropy rate

Symbolic entropy rate is estimated from block-entropy increments h_L=H(L)-H(L-1), which converge toward the entropy rate as L grows under suitable conditions.

```python
_ent_run_100_130(118)
```

### 119. Ordinal entropy rate

Ordinal entropy rate applies conditional entropy to the sequence of ordinal patterns rather than amplitude symbols.

```python
_ent_run_100_130(119)
```

### 120. Block entropy growth curves

Block entropy H(L) is Shannon entropy of length-L symbolic words. Its growth curve separates extensive entropy-rate growth from subextensive structure.

```python
_ent_run_100_130(120)
```

### 121. Excess entropy

Excess entropy is the subextensive intercept E=lim[H(L)-h_mu L]. The finite-data implementation estimates h_mu from the tail block-entropy slope and then estimates the intercept.

```python
_ent_run_100_130(121)
```

### 122. Predictive information

Predictive information is mutual information between past and future blocks, I(P;F)=H(P)+H(F)-H(P,F).

```python
_ent_run_100_130(122)
```

### 123. Statistical complexity

Statistical complexity C_mu is Shannon entropy of causal-state occupancy. Causal states are approximated here by finite histories with equal rounded predictive next-symbol distributions.

```python
_ent_run_100_130(123)
```

### 124. Causal-state entropy

Causal-state entropy is H(S) over inferred predictive-equivalence states; it is the same information quantity underlying statistical complexity for an epsilon-machine.

```python
_ent_run_100_130(124)
```

### 125. ε-machine entropy

An epsilon-machine is the unifilar state model formed by predictive-equivalence causal states. The cell reports the finite-history approximation, state entropy, and observed state-transition structure.

```python
_ent_run_100_130(125)
```

### 126. Active Information Storage

Active Information Storage is I(X_past;X_t), the information in a process past that predicts its current value.

```python
_ent_run_100_130(126)
```

### 127. Local Active Information Storage

Local Active Information Storage is the pointwise log ratio log[p(x_t|past)/p(x_t)], whose expectation is AIS.

```python
_ent_run_100_130(127)
```

### 128. Predictive entropy

Predictive entropy is conditional uncertainty H(X_t|X_past). It complements AIS through H(X_t)=AIS+H(X_t|past).

```python
_ent_run_100_130(128)
```

### 129. Past-future mutual information

Past-future mutual information is I(P;F) for finite symbolic past and future blocks and is a finite-block predictive-information estimate.

```python
_ent_run_100_130(129)
```

### 130. Entropy-rate asymmetry

For a stationary process the asymptotic Shannon entropy rate is time-reversal invariant. This cell therefore reports finite-history predictive conditional-entropy asymmetry, not a fictitious difference between two exact entropy rates.

```python
_ent_run_100_130(130)
```

<h3 style="color:#E8E8E8;margin-top:0;">Theory and role — multivariate information-flow helpers for methods 131–177</h3>
<p>This block defines joint/conditional entropy, mutual information, PID, transfer entropy, directed information, and predictive-information utilities. Core identities include \(I(X;Y)=H(X)+H(Y)-H(X,Y)\) and transfer entropy \(T_{X\to Y}=I(X_{t^-};Y_t\mid Y_{t^-})\). PID branches explicitly state their chosen redundancy convention; directed quantities are interpreted under their estimator/model assumptions rather than as unconditional physical causation.</p>
</div>


```python
# =======================================================
# ENTROPY CATALOGUE HELPERS — methods 131–177
# Multivariate information, PID, transfer entropy and directed information.
# =======================================================
def _sym_ch(j,k=4): return _symbolize(_ent_x[:,j],k)

def _Hcols(*cols):
    n=min(map(len,cols));A=np.column_stack([np.asarray(c)[:n] for c in cols]);return _H(_p_from_labels(A))

def _MI(a,b): return float(_H(_p_from_labels(a))+_H(_p_from_labels(b))-_Hcols(a,b))

def _CMI(a,b,c):
    return float(_Hcols(a,c)+_Hcols(b,c)-_H(_p_from_labels(c))-_Hcols(a,b,c))

def _TC(cols):
    return float(sum(_H(_p_from_labels(c)) for c in cols)-_Hcols(*cols))

def _DTC(cols):
    Hjoint=_Hcols(*cols);return float(sum(_Hcols(*[cols[j] for j in range(len(cols)) if j!=i]) for i in range(len(cols)))-(len(cols)-1)*Hjoint)

def _pid_mmi(s1,s2,t):
    i1=_MI(s1,t);i2=_MI(s2,t);ij=_MI(np.array([a*16+b for a,b in zip(s1,s2)]),t);r=min(i1,i2);u1=i1-r;u2=i2-r;s=ij-r-u1-u2
    return dict(redundancy=float(r),unique_1=float(u1),unique_2=float(u2),synergy=float(max(0,s)),joint_MI=float(ij))

def _te_discrete(x,y,lag=1,cond=None):
    x=np.asarray(x,int);y=np.asarray(y,int);lag=int(lag);n=min(len(x),len(y))
    if n<=lag:return 0.0
    yt=y[lag:n];yp=y[:n-lag];xp=x[:n-lag]
    if cond is None:return _CMI(xp,yt,yp)
    cond=np.asarray(cond,int)[:n-lag];z=np.array([a*64+b for a,b in zip(yp,cond)]);return _CMI(xp,yt,z)

def _te_multi(srcs,y,lag=1):
    n=min([len(y)]+[len(s) for s in srcs]);yt=np.asarray(y)[lag:n];yp=np.asarray(y)[:n-lag];base=1;code=np.zeros(n-lag,dtype=np.int64)
    for s in srcs:
        ss=np.asarray(s)[:n-lag];code+=base*ss;base*=max(2,int(ss.max()+1))
    return _CMI(code,yt,yp)

def _local_te(x,y,lag=1):
    x=np.asarray(x,int);y=np.asarray(y,int);n=min(len(x),len(y));yt=y[lag:n];yp=y[:n-lag];xp=x[:n-lag];N=len(yt)
    cxyz=Counter(zip(xp,yp,yt));cxy=Counter(zip(xp,yp));cyy=Counter(zip(yp,yt));cy=Counter(yp)
    vals=[]
    for a,b,c in zip(xp,yp,yt):vals.append(np.log((cxyz[(a,b,c)]*cy[b]+ENT_EPS)/(cxy[(a,b)]*cyy[(b,c)]+ENT_EPS)))
    return np.asarray(vals,float)

def _te_matrix(ch=None,lag=1,cond_global=False):
    ch=int(ch or min(8,_ent_x.shape[1]));S=[_sym_ch(j) for j in range(ch)];M=np.zeros((ch,ch));g=_symbolize(np.mean(_ent_x[:,:ch],axis=1),4)
    for a in range(ch):
        for b in range(ch):
            if a!=b:M[a,b]=_te_discrete(S[a],S[b],lag, g if cond_global else None)
    return M

def _ordinal_seq(y,m=3):
    labs=_ordinal_labels(y,m);keys={k:i for i,k in enumerate(sorted(set(labs)))};return np.array([keys[k] for k in labs],int)

def _gc_info(x,y,lag=1):
    x=np.asarray(x,float);y=np.asarray(y,float);n=min(len(x),len(y));Y=y[lag:n];yr=y[:n-lag,None];xf=np.column_stack([y[:n-lag],x[:n-lag]]);br=np.linalg.lstsq(np.c_[np.ones(len(yr)),yr],Y,rcond=None)[0];bf=np.linalg.lstsq(np.c_[np.ones(len(xf)),xf],Y,rcond=None)[0];er=Y-np.c_[np.ones(len(yr)),yr]@br;ef=Y-np.c_[np.ones(len(xf)),xf]@bf;return float(.5*np.log((np.var(er)+ENT_EPS)/(np.var(ef)+ENT_EPS)))
```

### 131. Joint entropy

Joint entropy is H(X,Y)=-sum p(x,y) ln p(x,y) for the empirical joint symbol distribution.

```python
_ent_run_131_177(131)
```

### 132. Conditional entropy

Conditional entropy is H(Y|X)=H(X,Y)-H(X).

```python
_ent_run_131_177(132)
```

### 133. Channel-pair mutual information matrix

Pairwise mutual information is I(X;Y)=H(X)+H(Y)-H(X,Y), computed for every channel pair.

```python
_ent_run_131_177(133)
```

### 134. Normalized mutual information matrix

Normalized mutual information uses I(X;Y)/sqrt(H(X)H(Y)), yielding a dimensionless symmetric dependence score.

```python
_ent_run_131_177(134)
```

### 135. Conditional mutual information

Conditional mutual information is I(X;Y|Z)=H(X,Z)+H(Y,Z)-H(Z)-H(X,Y,Z).

```python
_ent_run_131_177(135)
```

### 136. Time-delayed mutual information matrix

Time-delayed mutual information evaluates I(X_t;Y_t+tau) across channel pairs at a fixed positive lag.

```python
_ent_run_131_177(136)
```

### 137. Lagged mutual information sweep

A lagged MI sweep evaluates I(X_t;Y_t+tau) across a range of lags to reveal delayed statistical dependence.

```python
_ent_run_131_177(137)
```

### 138. Partial mutual information

Partial mutual information is implemented as conditional mutual information I(X;Y|Z), removing dependence statistically explained by the conditioning variable.

```python
_ent_run_131_177(138)
```

### 139. Interaction information

Interaction information for three variables is I(X;Y;Z)=I(X;Y)-I(X;Y|Z); its sign depends on the redundancy/synergy convention.

```python
_ent_run_131_177(139)
```

### 140. Total correlation

Total correlation is TC=sum_i H(X_i)-H(X_1,...,X_n), a multivariate generalization of mutual information.

```python
_ent_run_131_177(140)
```

### 141. Dual total correlation

Dual total correlation is DTC=sum_i H(X_-i)-(n-1)H(X_1,...,X_n), also called binding information.

```python
_ent_run_131_177(141)
```

### 142. O-information

O-information is Omega=TC-DTC: positive values indicate redundancy-dominated dependence and negative values synergy-dominated dependence.

```python
_ent_run_131_177(142)
```

### 143. S-information

S-information is TC+DTC, a symmetric measure of total interdependence complementary to O-information.

```python
_ent_run_131_177(143)
```

### 144. Multi-information

Multi-information is another name for total correlation, sum marginal entropies minus joint entropy.

```python
_ent_run_131_177(144)
```

### 145. Redundancy-synergy entropy decomposition

The redundancy-synergy dominance summary uses O-information: positive Omega is reported as redundancy dominance and negative Omega as synergy dominance; it is not a full PID.

```python
_ent_run_131_177(145)
```

### 146. Partial Information Decomposition

This notebook uses the bivariate minimum-mutual-information PID: R=min[I(S1;T),I(S2;T)], U1=I(S1;T)-R, U2=I(S2;T)-R, and synergy=I(S1,S2;T)-R-U1-U2.

```python
_ent_run_131_177(146)
```

### 147. Redundant information

Redundant information is the MMI PID atom R=min[I(S1;T),I(S2;T)].

```python
_ent_run_131_177(147)
```

### 148. Unique information

Unique information atoms are U1=I(S1;T)-R and U2=I(S2;T)-R under the stated MMI PID convention.

```python
_ent_run_131_177(148)
```

### 149. Synergistic information

Synergistic information is the residual joint source information not assigned to redundancy or unique atoms under MMI PID.

```python
_ent_run_131_177(149)
```

### 150. PID-based channel triplet analysis

PID-based triplet analysis evaluates the stated MMI PID for channel-source pairs and a distinct target.

```python
_ent_run_131_177(150)
```

### 151. PID source-target decomposition

PID source-target decomposition ranks source-pair/target triplets by their redundancy, unique and synergy atoms.

```python
_ent_run_131_177(151)
```

### 152. Synergy-dominant EEG channel motifs

Synergy-dominant motifs are triplets whose MMI synergy exceeds their redundancy.

```python
_ent_run_131_177(152)
```

### 153. Redundancy-dominant EEG channel motifs

Redundancy-dominant motifs are triplets whose MMI redundancy exceeds their synergy.

```python
_ent_run_131_177(153)
```

### 154. Conditional Transfer Entropy

Conditional Transfer Entropy is T_X→Y|Z=I(X_t-past;Y_t|Y_past,Z_past).

```python
_ent_run_131_177(154)
```

### 155. Partial Transfer Entropy

Partial Transfer Entropy conditions source-to-target information transfer on additional observed processes to remove shared predictive drivers.

```python
_ent_run_131_177(155)
```

### 156. Multivariate Transfer Entropy

Multivariate Transfer Entropy uses a vector of source histories: I(X1_past,...,Xk_past;Y_t|Y_past).

```python
_ent_run_131_177(156)
```

### 157. Local Transfer Entropy

Local Transfer Entropy is the pointwise log ratio log[p(y_t|y_past,x_past)/p(y_t|y_past)]; its expectation is TE.

```python
_ent_run_131_177(157)
```

### 158. Symbolic Transfer Entropy

Symbolic Transfer Entropy estimates TE after quantizing amplitudes into discrete symbols.

```python
_ent_run_131_177(158)
```

### 159. Permutation Transfer Entropy

Permutation Transfer Entropy estimates directed predictive information using ordinal-pattern symbols instead of amplitude bins.

```python
_ent_run_131_177(159)
```

### 160. Effective Transfer Entropy

Effective Transfer Entropy subtracts the mean TE from shuffled-source surrogates to reduce finite-sample bias.

```python
_ent_run_131_177(160)
```

### 161. Normalized Transfer Entropy

Normalized Transfer Entropy divides TE by target innovation entropy H(Y_t|Y_past), expressing transferred information as a fraction of unpredictable target entropy.

```python
_ent_run_131_177(161)
```

### 162. Delayed Transfer Entropy sweep

Delayed Transfer Entropy sweeps source-target lag in T_X→Y(tau).

```python
_ent_run_131_177(162)
```

### 163. Frequency-band Transfer Entropy

Frequency-band Transfer Entropy applies the TE estimator after band-pass filtering and symbolization.

```python
_ent_run_131_177(163)
```

### 164. Time-frequency Transfer Entropy

Time-frequency Transfer Entropy estimates TE from band-envelope states inside local time windows, producing a band-by-time directed-information table.

```python
_ent_run_131_177(164)
```

### 165. Conditional TE network analysis

Conditional TE network analysis computes pairwise TE while conditioning on a global multichannel state symbol.

```python
_ent_run_131_177(165)
```

### 166. TE source-sink entropy

TE source-sink entropy normalizes outgoing and incoming TE weights per node and computes Shannon entropy of each directional weight distribution.

```python
_ent_run_131_177(166)
```

### 167. TE hub entropy

TE hub entropy normalizes total in+out TE strength across nodes and measures how concentrated network information flow is.

```python
_ent_run_131_177(167)
```

### 168. TE motif entropy

TE motif entropy thresholds strong directed TE edges and measures entropy of triad edge-count motif classes.

```python
_ent_run_131_177(168)
```

### 169. TE cascade entropy

TE cascade entropy measures the distributional entropy of two-step cascade weights T_i→j T_j→k.

```python
_ent_run_131_177(169)
```

### 170. Directed Information

Massey directed information is sum_t I(X^t;Y_t|Y^(t-1)). The cell reports a first-order finite-history rate I((X_t-1,X_t);Y_t|Y_t-1).

```python
_ent_run_131_177(170)
```

### 171. Causation Entropy

Causation Entropy is conditional mutual information from a candidate source past to a target future given an existing conditioning set; with the target past as condition it coincides with first-order TE.

```python
_ent_run_131_177(171)
```

### 172. Conditional Causation Entropy

Conditional Causation Entropy enlarges the conditioning set to include another observed process.

```python
_ent_run_131_177(172)
```

### 173. Information Flow Decomposition

The information-flow decomposition uses the exact chain rule I((Y_past,X_past);Y_t)=I(Y_past;Y_t)+I(X_past;Y_t|Y_past), separating self-predictive information from TE.

```python
_ent_run_131_177(173)
```

### 174. Granger-causality information entropy

For linear Gaussian variables, Granger/transfer information is 0.5 ln(var restricted / var full), the entropy reduction produced by adding the source predictor.

```python
_ent_run_131_177(174)
```

### 175. Directed information rate

Directed information rate is reported using the same finite-history Massey conditional-information approximation as the directed-information cell.

```python
_ent_run_131_177(175)
```

### 176. Feedback information decomposition

Feedback information diagnostics report TE in both directions, their sum as symmetric exchange and their difference as net directional balance; these summaries are not claimed to be orthogonal PID atoms.

```python
_ent_run_131_177(176)
```

### 177. Causal conditioning entropy

Causal conditioning entropy H(Y_t|Y_past,X_past) is the remaining target uncertainty after conditioning on both target and source histories.

```python
_ent_run_131_177(177)
```

<h3 style="color:#E8E8E8;margin-top:0;">Theory and role — graph, microstate, recurrence, phase, and divergence helpers for methods 178–233</h3>
<p>This foundation builds the probability objects used by graph/network entropy, topographic and microstate entropy, recurrence-line statistics, phase/CFC entropy, and distribution divergences. Examples include the trace-one graph Laplacian density \(\rho=L/\mathrm{Tr}(L)\) with von Neumann entropy \(-\mathrm{Tr}(\rho\log\rho)\), and Jensen–Shannon divergence \(\tfrac12D_{KL}(P\|M)+\tfrac12D_{KL}(Q\|M)\), \(M=(P+Q)/2\).</p>

```python
# =======================================================
# ENTROPY CATALOGUE HELPERS — methods 178–233
# Graph/topographic/microstate/recurrence/phase/divergence entropy.
# =======================================================
import itertools
import networkx as nx
from scipy.spatial import Delaunay

_ENT_POS = {
'Fp1':(-0.50,0.98),'Fpz':(0.00,1.05),'Fp2':(0.50,0.98),'F7':(-0.92,0.62),'F3':(-0.48,0.62),'Fz':(0.00,0.67),'F4':(0.48,0.62),'F8':(0.92,0.62),
'FC5':(-0.70,0.32),'FC1':(-0.25,0.30),'FC2':(0.25,0.30),'FC6':(0.70,0.32),'M1':(-1.05,0.00),'T7':(-1.00,0.00),'C3':(-0.50,0.00),'Cz':(0.00,0.00),'C4':(0.50,0.00),'T8':(1.00,0.00),'M2':(1.05,0.00),
'CP5':(-0.70,-0.32),'CP1':(-0.25,-0.30),'CP2':(0.25,-0.30),'CP6':(0.70,-0.32),'P7':(-0.92,-0.62),'P3':(-0.48,-0.62),'Pz':(0.00,-0.67),'P4':(0.48,-0.62),'P8':(0.92,-0.62),'POz':(0.00,-0.84),'O1':(-0.45,-0.98),'Oz':(0.00,-1.05),'O2':(0.45,-0.98)}

def _safe_prob(v):
    v=np.asarray(v,float).ravel();v=np.clip(v,0,None);s=v.sum();return v/s if s>ENT_EPS else np.ones(max(1,len(v)))/max(1,len(v))

def _corr_graph(X=None, threshold=.35, absolute=True):
    X=np.asarray(_ent_x if X is None else X,float);R=np.corrcoef(X,rowvar=False);W=np.abs(R) if absolute else R.copy();np.fill_diagonal(W,0.)
    A=np.where(W>=float(threshold),W,0.);G=nx.from_numpy_array(A)
    return G,W,A

def _edge_weight_entropy(A):
    w=np.asarray(A,float)[np.triu_indices_from(A,1)];w=w[w>0];return _H(_safe_prob(w)) if len(w) else 0.0

def _degree_distribution_entropy(G):
    deg=np.array([d for _,d in G.degree()],int);return _H(_p_from_labels(deg)) if len(deg) else 0.0

def _laplacian_probs(A):
    L=np.diag(np.sum(A,axis=1))-A;lam=np.linalg.eigvalsh((L+L.T)/2);lam=np.clip(lam,0,None);return _safe_prob(lam),L

def _heat_density_entropy(A,beta=1.0):
    _,L=_laplacian_probs(A);lam=np.linalg.eigvalsh((L+L.T)/2);z=np.exp(-float(beta)*np.clip(lam,0,None));return _H(_safe_prob(z))

def _regions_indices():
    groups={'frontal':['Fp1','Fpz','Fp2','F7','F3','Fz','F4','F8','FC5','FC1','FC2','FC6'],
            'central':['T7','C3','Cz','C4','T8','M1','M2'],
            'posterior':['CP5','CP1','CP2','CP6','P7','P3','Pz','P4','P8','POz','O1','Oz','O2']}
    out={}
    for k,names in groups.items():out[k]=[i for i,n in enumerate(_ent_names) if n in names]
    return out

def _scalp_xy():
    xy=[]
    for k,n in enumerate(_ent_names):xy.append(_ENT_POS.get(n,(math.cos(2*math.pi*k/len(_ent_names)),math.sin(2*math.pi*k/len(_ent_names)))))
    return np.asarray(xy,float)

def _spatial_edges():
    xy=_scalp_xy()
    if len(xy)<3:return [(i,i+1) for i in range(len(xy)-1)]
    tri=Delaunay(xy);E=set()
    for s in tri.simplices:
        for a,b in itertools.combinations(s,2):E.add(tuple(sorted((int(a),int(b)))))
    return sorted(E)

def _pair_asymmetry(values):
    vals=np.asarray(values,float);pairs=[]
    lr=[('Fp1','Fp2'),('F7','F8'),('F3','F4'),('FC5','FC6'),('FC1','FC2'),('T7','T8'),('C3','C4'),('CP5','CP6'),('CP1','CP2'),('P7','P8'),('P3','P4'),('O1','O2'),('M1','M2')]
    idx={n:i for i,n in enumerate(_ent_names)}
    for l,r in lr:
        if l in idx and r in idx:pairs.append(abs(vals[idx[l]]-vals[idx[r]])/(abs(vals[idx[l]])+abs(vals[idx[r]])+ENT_EPS))
    return np.asarray(pairs,float)

def _microstates(K=4):
    X=np.asarray(_ent_x,float);step=max(1,len(X)//max(300,_ENT_SPEED['patterns']));Z=X[::step]
    gfp=np.std(Z,axis=1);keep=gfp>np.quantile(gfp,.25);Z=Z[keep];Z=Z/(np.linalg.norm(Z,axis=1,keepdims=True)+ENT_EPS)
    if len(Z)<K:return np.zeros(len(Z),int),np.eye(K,Z.shape[1])[:K]
    ids=np.linspace(0,len(Z)-1,K).astype(int);C=Z[ids].copy()
    for _ in range({'full':40,'balanced':30,'fast':20,'ultra':12}[FAST_OPTION]):
        R=Z@C.T;lab=np.argmax(np.abs(R),axis=1);C2=[]
        for k in range(K):
            zz=Z[lab==k]
            if not len(zz):C2.append(C[k]);continue
            sg=np.sign(zz@C[k]);sg[sg==0]=1;v=np.mean(zz*sg[:,None],axis=0);C2.append(v/(np.linalg.norm(v)+ENT_EPS))
        C2=np.asarray(C2)
        if np.linalg.norm(np.abs(C2@C.T)-np.eye(K))<1e-5: C=C2;break
        C=C2
    lab=np.argmax(np.abs(Z@C.T),axis=1)
    return lab,C

def _run_lengths(seq):
    seq=np.asarray(seq); 
    if not len(seq):return np.array([],int),np.array([],int)
    cuts=np.r_[0,1+np.flatnonzero(seq[1:]!=seq[:-1]),len(seq)];return seq[cuts[:-1]],np.diff(cuts)

def _recurrence(y=None,m=3,tau=1,rr=.05):
    y=_ent_y if y is None else np.asarray(y,float);X=_embed(_cap(y, {'full':1200,'balanced':900,'fast':650,'ultra':420}[FAST_OPTION]),m,tau,cap=False)
    if len(X)<3:return np.zeros((len(X),len(X)),bool)
    D=squareform(pdist(X,'euclidean'));d=D[np.triu_indices(len(D),1)];eps=np.quantile(d,float(rr));R=D<=eps;np.fill_diagonal(R,False);return R

def _line_lengths_bool(v,val=True,minlen=1):
    v=np.asarray(v,bool)==bool(val);z=np.r_[False,v,False];d=np.diff(z.astype(int));starts=np.flatnonzero(d==1);ends=np.flatnonzero(d==-1);L=ends-starts;return L[L>=minlen]

def _diag_lengths(R):
    out=[];n=len(R)
    for k in range(-n+1,n):out.extend(_line_lengths_bool(np.diag(R,k),True,2).tolist())
    return np.asarray(out,int)

def _vert_lengths(R,val=True):
    out=[]
    for j in range(R.shape[1]):out.extend(_line_lengths_bool(R[:,j],val,2).tolist())
    return np.asarray(out,int)

def _phase(y):return np.angle(signal.hilbert(np.asarray(y,float)))

def _plv(a,b):return float(abs(np.mean(np.exp(1j*(np.asarray(a)-np.asarray(b))))))

def _pac_distribution(y,plo=4,phi=8,alo=30,ahi=45,bins=18):
    if ahi>=.48*_ent_fs: ahi=.45*_ent_fs
    if phi>=alo:return np.ones(bins)/bins
    ph=_phase(_band(y,plo,phi));amp=np.abs(signal.hilbert(_band(y,alo,ahi)));edges=np.linspace(-np.pi,np.pi,bins+1);idx=np.clip(np.digitize(ph,edges)-1,0,bins-1);v=np.array([np.mean(amp[idx==k]) if np.any(idx==k) else 0 for k in range(bins)]);return _safe_prob(v)

def _common_hist_probs(x,y,bins=32):
    x=np.asarray(x,float);y=np.asarray(y,float);lo=min(np.min(x),np.min(y));hi=max(np.max(x),np.max(y));e=np.linspace(lo,hi,bins+1);p=np.histogram(x,e)[0].astype(float)+.5;q=np.histogram(y,e)[0].astype(float)+.5;return _safe_prob(p),_safe_prob(q),e

def _jsd(p,q):
    p=_safe_prob(p);q=_safe_prob(q);m=.5*(p+q);return float(.5*np.sum(p*np.log((p+ENT_EPS)/(m+ENT_EPS)))+.5*np.sum(q*np.log((q+ENT_EPS)/(m+ENT_EPS))))

def _kl(p,q):
    p=_safe_prob(p);q=_safe_prob(q);return float(np.sum(p*np.log((p+ENT_EPS)/(q+ENT_EPS))))

def _renyi_div(p,q,alpha=2.):
    p=_safe_prob(p);q=_safe_prob(q);a=float(alpha);return float(np.log(np.sum((p**a)*(q**(1-a)))+ENT_EPS)/(a-1))

def _tsallis_div(p,q,qord=2.):
    p=_safe_prob(p);q=_safe_prob(q);qq=float(qord);return float((np.sum((p**qq)*(q**(1-qq)))-1)/(qq-1))
```

### 178. Graph entropy

Let $w_e\ge0$ be thresholded functional-connectivity edge weights and $p_e=w_e/\sum_f w_f$. This cell defines graph entropy as $H_E=-\sum_e p_e\ln p_e$. The definition is explicit because “graph entropy” is not unique.

**Implementation.** The existing EEG segment, channel naming, black/cyan plotting style, and the notebook's `full / balanced / fast / ultra` runtime policy are retained.

### 179. Degree-distribution entropy

If $p_k$ is the fraction of vertices with degree $k$, degree-distribution entropy is $H_D=-\sum_k p_k\ln p_k$. It measures heterogeneity of graph degree classes, not edge-weight uncertainty.

**Implementation.** The existing EEG segment, channel naming, black/cyan plotting style, and the notebook's `full / balanced / fast / ultra` runtime policy are retained.

<h3 style="color:#E8E8E8;margin-top:0;">Theory and role — surrogate inference and entropy-feature helpers for methods 234–255</h3>
<p>This cell defines null-model, bootstrap, multiple-testing, and entropy-feature utilities. Surrogate p-values use \(p=(1+\#\{T_b\ge T_{obs}\})/(B+1)\) for the stated one-sided statistic, while BH-FDR orders p-values and controls the selected false-discovery rate. Shuffle, Fourier-phase, IAAFT, and colored-noise nulls are kept separate because they preserve different signal properties.</p>
</div>


```python
# =======================================================
# ENTROPY CATALOGUE HELPERS — methods 234–255
# Surrogate inference and entropy feature/topography/state analyses.
# =======================================================
from scipy.cluster.hierarchy import linkage, fcluster
from scipy.spatial.distance import cdist

def _phase_surrogate(y,rng=None):
    rng=np.random.default_rng() if rng is None else rng;y=np.asarray(y,float);Y=np.fft.rfft(y);ph=rng.uniform(0,2*np.pi,len(Y));ph[0]=0
    if len(y)%2==0:ph[-1]=0
    Z=np.abs(Y)*np.exp(1j*ph);return np.fft.irfft(Z,n=len(y))

def _iaaft(y,rng=None,iters=None):
    rng=np.random.default_rng() if rng is None else rng;y=np.asarray(y,float);target=np.abs(np.fft.rfft(y));sorted_y=np.sort(y);z=rng.permutation(y);iters=int(iters or {'full':80,'balanced':50,'fast':30,'ultra':15}[FAST_OPTION])
    for _ in range(iters):
        Z=np.fft.rfft(z);z=np.fft.irfft(target*np.exp(1j*np.angle(Z)),n=len(y));r=np.argsort(z);zz=np.empty_like(z);zz[r]=sorted_y;z=zz
    return z

def _psd_slope(y):
    f,P=signal.welch(y,fs=_ent_fs,nperseg=min(len(y),max(64,int(2*_ent_fs))));m=(f>max(.5,1/len(y)))&(f<min(.4*_ent_fs,45))&(P>0)
    if m.sum()<4:return 0.0
    return float(-np.polyfit(np.log(f[m]),np.log(P[m]),1)[0])

def _colored_noise_like(y,rng=None):
    rng=np.random.default_rng() if rng is None else rng;y=np.asarray(y,float);n=len(y);beta=np.clip(_psd_slope(y),-2.5,3.5);f=np.fft.rfftfreq(n,d=1/_ent_fs);amp=np.ones_like(f);amp[1:]=np.maximum(f[1:],1/n)**(-beta/2);ph=rng.uniform(0,2*np.pi,len(f));ph[0]=0
    if n%2==0:ph[-1]=0
    z=np.fft.irfft(amp*np.exp(1j*ph),n=n);z=(z-z.mean())/(z.std()+ENT_EPS)*np.std(y)+np.mean(y);return z,beta

def _emp_p(obs,null,two=False):
    null=np.asarray(null,float);null=null[np.isfinite(null)]
    if not len(null):return np.nan
    if two:
        c=np.mean(null);return float((1+np.sum(np.abs(null-c)>=abs(obs-c)))/(len(null)+1))
    return float((1+np.sum(null>=obs))/(len(null)+1))

def _bh(p):
    p=np.asarray(p,float);n=len(p);order=np.argsort(p);q=np.empty(n);prev=1.
    for rank,idx in reversed(list(enumerate(order,1))):prev=min(prev,p[idx]*n/rank);q[idx]=prev
    return np.clip(q,0,1)

def _spectral_H(y):
    f,P=signal.welch(np.asarray(y,float),fs=_ent_fs,nperseg=min(len(y),max(64,int(2*_ent_fs))));p=_safe_prob(P);return _H(p)/np.log(max(2,len(p)))

def _channel_entropy_features(maxch=None):
    n=int(maxch or _ent_x.shape[1]);rows=[]
    for j in range(n):
        z=_ent_x[:,j];rows.append([_H(_hist_p(z,24))/np.log(24),_sampen(z),_perm_entropy(z),_disp_entropy(z),_spectral_H(z),_lz_norm(_symbolize(z,4),4)])
    F=np.asarray(rows,float);F[~np.isfinite(F)]=np.nan
    med=np.nanmedian(F,axis=0);inds=np.where(~np.isfinite(F));F[inds]=np.take(med,inds[1]);return F

def _zcols(F):
    F=np.asarray(F,float);return (F-F.mean(0,keepdims=True))/(F.std(0,keepdims=True)+ENT_EPS)

def _window_entropy_features():
    W=max(100,int(3*_ent_fs));starts=np.linspace(0,max(0,len(_ent_y)-W),_ENT_SPEED['windows']).astype(int);rows=[]
    for s in starts:
        z=_ent_y[s:s+W];top=_H(_safe_prob(np.var(_ent_x[s:s+W],axis=0)));rows.append([_perm_entropy(z),_disp_entropy(z),_spectral_H(z),top])
    return starts,np.asarray(rows,float)
```

### 234. Entropy surrogate testing

A surrogate test compares an observed entropy statistic $T(x)$ with statistics from null series that preserve specified properties. This cell reports empirical two-sided p-values against shuffle, Fourier-phase, and IAAFT nulls; the null family is part of the hypothesis and is never interchangeable.

### 236. Phase-randomized surrogate entropy testing

### 237. IAAFT-surrogate entropy testing

<h3 style="color:#E8E8E8;margin-top:0;">Theory and role — dynamical and maximum-entropy helpers for methods 256–275</h3>
<p>This block supports finite partition entropy rates, topological/metric-entropy estimates, correlation entropy \(K_2\), irreversibility and entropy production, and maximum-entropy models. For a stationary Markov chain, the implemented entropy-production rate uses \(\tfrac12\sum_{ij}(\pi_iP_{ij}-\pi_jP_{ji})\log[(\pi_iP_{ij})/(\pi_jP_{ji})]\). Finite-data KS/Pesin-related cells are explicitly diagnostics or bounds, not theorem-level identification.</p>

```python
# =======================================================
# ENTROPY CATALOGUE HELPERS — methods 256–275
# Dynamical entropy, irreversibility and maximum-entropy models.
# =======================================================
from scipy.optimize import minimize, brentq

def _entropy_rate_blocks(sym,maxL=6):
    H=np.array([_block_entropy(sym,L) for L in range(1,maxL+1)],float);d=np.diff(np.r_[0,H]);return H,d

def _corrsum(y,m,r):
    X=_embed(_cap(y, {'full':4500,'balanced':3000,'fast':1800,'ultra':900}[FAST_OPTION]),m,1,cap=False)
    if len(X)<3:return np.nan
    pairs=len(cKDTree(X).query_pairs(float(r),p=np.inf));return pairs/max(1,len(X)*(len(X)-1)/2)

def _k2_curve(y,m=3):
    yy=_cap(y, {'full':4500,'balanced':3000,'fast':1800,'ultra':900}[FAST_OPTION]);sd=np.std(yy);rs=np.geomspace(.08*sd,.5*sd,8);vals=[]
    for r in rs:
        c1=_corrsum(yy,m,r);c2=_corrsum(yy,m+1,r);vals.append(np.log((c1+ENT_EPS)/(c2+ENT_EPS))*_ent_fs)
    return rs,np.asarray(vals)

def _symbol_transition_matrix(sym,K=None):
    s=np.asarray(sym,int);K=int(K or (s.max()+1 if len(s) else 1));C=np.zeros((K,K),float)
    for a,b in zip(s[:-1],s[1:]):C[a,b]+=1
    P=C/(C.sum(1,keepdims=True)+ENT_EPS);return C,P

def _markov_epr(sym):
    C,P=_symbol_transition_matrix(sym);pi=C.sum(1);pi=_safe_prob(pi);e=0.
    for a in range(len(pi)):
        for b in range(len(pi)):
            f=pi[a]*P[a,b];r=pi[b]*P[b,a]
            if f>0 and r>0:e+=f*np.log(f/r)
    return float(e)

def _path_reverse_div(sym,L=3,js=False):
    s=np.asarray(sym,int);X=sliding_window_view(s,L);f=Counter(map(tuple,X));r=Counter(tuple(x[::-1]) for x in X);keys=sorted(set(f)|set(r));p=np.array([f[k]+.5 for k in keys],float);q=np.array([r[k]+.5 for k in keys],float);p=_safe_prob(p);q=_safe_prob(q);return _jsd(p,q) if js else _kl(p,q)

def _local_expansion_proxy(y,horizon=1):
    yy=_cap(y, {'full':7000,'balanced':4500,'fast':2500,'ultra':1200}[FAST_OPTION]);m=3;E=_embed(yy,m,1,cap=False);h=int(horizon);n=len(E)-h
    if n<20:return np.nan
    E0=E[:n];E1=E[h:h+n];k=min(20,max(6,int(np.sqrt(n))));tree=cKDTree(E0);ids=np.linspace(0,n-1,min(n,120 if FAST_OPTION=='full' else 80 if FAST_OPTION=='balanced' else 50 if FAST_OPTION=='fast' else 30)).astype(int);G=[]
    for ix in ids:
        nn=tree.query(E0[ix],k=k)[1];A=E0[nn]-E0[ix];B=E1[nn]-E1[ix];J=np.linalg.lstsq(A,B,rcond=1e-6)[0].T;s=np.linalg.svd(J,compute_uv=False);G.append(np.prod(s[s>1]) if np.any(s>1) else 1.)
    return float(np.log(np.mean(G)+ENT_EPS)/(h/_ent_fs))

def _linear_lyap_sum(X=None):
    X=np.asarray(_ent_x[:,:min(8,_ent_x.shape[1])] if X is None else X,float);A=X[:-1];B=X[1:];ridge=1e-3;M=np.linalg.solve(A.T@A+ridge*np.eye(A.shape[1]),A.T@B).T;e=np.linalg.eigvals(M);le=np.log(np.maximum(np.abs(e),ENT_EPS))*_ent_fs;return float(np.sum(le[le>0])),le

def _binary_states(X=None,nch=6):
    X=np.asarray(_ent_x if X is None else X,float)[:,:min(nch,_ent_x.shape[1])];return (X>np.median(X,axis=0,keepdims=True)).astype(int)

def _independent_maxent(B):
    p=np.clip(np.mean(B,axis=0),1e-6,1-1e-6);h=np.log(p/(1-p));return p,h

def _ising_fit(B):
    B=np.asarray(B,int);S=2*B-1;n=S.shape[1];states=np.array(list(__import__('itertools').product([-1.,1.],repeat=n)));pairs=list(__import__('itertools').combinations(range(n),2));emp1=S.mean(0);emp2=np.array([np.mean(S[:,a]*S[:,b]) for a,b in pairs])
    def feats(st):return np.c_[st,np.column_stack([st[:,a]*st[:,b] for a,b in pairs])]
    F=feats(states);target=np.r_[emp1,emp2]
    def obj(th):
        z=F@th;logZ=special.logsumexp(z);p=np.exp(z-logZ);return float(logZ-th@target), F.T@p-target
    res=minimize(lambda th:obj(th)[0],np.zeros(F.shape[1]),jac=lambda th:obj(th)[1],method='L-BFGS-B',options={'maxiter':{'full':300,'balanced':200,'fast':120,'ultra':70}[FAST_OPTION]});z=F@res.x;p=np.exp(z-special.logsumexp(z));return states,p,res.x,float(res.fun),bool(res.success)

def _soft_state_model(F,K=4,tau=1.0):
    Z=_zcols(F);K=min(K,len(Z));C=Z[np.linspace(0,len(Z)-1,K).astype(int)]
    for _ in range(25):
        D=cdist(Z,C,'sqeuclidean');lab=D.argmin(1);C=np.array([Z[lab==k].mean(0) if np.any(lab==k) else C[k] for k in range(K)])
    E=cdist(Z,C,'sqeuclidean');logp=-E/max(tau,ENT_EPS);logp-=special.logsumexp(logp,axis=1,keepdims=True);P=np.exp(logp);return Z,C,E,P
```

<h3 style="color:#E8E8E8;margin-top:0;">Theory and role — density-operator information helpers for methods 276–287</h3>
<p>This foundation constructs finite-dimensional positive trace-one matrices from classical EEG covariance or analytic feature states and applies quantum-information mathematics such as \(S(\rho)=-\mathrm{Tr}(\rho\log\rho)\), quantum relative entropy \(D(\rho\|\sigma)=\mathrm{Tr}[\rho(\log\rho-\log\sigma)]\), and reduced-state/Schmidt entropies. These are mathematical analogues applied to EEG-derived operators, not evidence of physical quantum states or entanglement in brain tissue.</p>

```python
# =======================================================
# ENTROPY CATALOGUE HELPERS — methods 276–287
# Quantum-information mathematics applied to explicitly defined EEG-derived
# density-operator analogues; no claim of physical quantum brain states.
# =======================================================
def _density_cov(X):
    X=np.asarray(X,float);C=np.cov(X,rowvar=False);C=(C+C.T)/2;w,V=np.linalg.eigh(C);w=np.clip(w,0,None);C=(V*w)@V.T;tr=np.trace(C);return C/(tr+ENT_EPS)

def _qentropy(rho):
    rho=np.asarray(rho,complex);w=np.linalg.eigvalsh((rho+rho.conj().T)/2).real;w=np.clip(w,0,None);w=w/(w.sum()+ENT_EPS);return _H(w),w

def _qrel(rho,sigma,floor=1e-10):
    rho=np.asarray(rho,complex);sigma=np.asarray(sigma,complex);wr,Ur=np.linalg.eigh((rho+rho.conj().T)/2);ws,Us=np.linalg.eigh((sigma+sigma.conj().T)/2);wr=np.clip(wr,0,None);wr=wr/(wr.sum()+ENT_EPS);ws=np.clip(ws,floor,None);ws=ws/ws.sum();lr=Ur@np.diag(np.log(np.clip(wr,floor,None)))@Ur.conj().T;ls=Us@np.diag(np.log(ws))@Us.conj().T;return float(np.real(np.trace(rho@(lr-ls))))

def _analytic_state(nch):
    nch=min(int(nch),_ent_x.shape[1]);Z=signal.hilbert(_ent_x[:,:nch],axis=0);idx=int(np.argmax(np.sum(np.abs(Z)**2,axis=1)));psi=Z[idx].astype(complex);return psi/(np.linalg.norm(psi)+ENT_EPS),idx

def _pure_reduced(psi,dA,dB):
    psi=np.asarray(psi,complex).ravel();assert len(psi)==dA*dB;M=psi.reshape(dA,dB);rhoA=M@M.conj().T;rhoB=M.conj().T@M;rhoA/=np.trace(rhoA)+ENT_EPS;rhoB/=np.trace(rhoB)+ENT_EPS;return rhoA,rhoB

def _partial_traces(rho,dA,dB):
    R=np.asarray(rho,complex).reshape(dA,dB,dA,dB);rhoA=np.einsum('abcb->ac',R);rhoB=np.einsum('abac->bc',R);return rhoA,rhoB

def _bitcode(B):
    B=np.asarray(B,int);return np.sum(B*(1<<np.arange(B.shape[1])),axis=1).astype(int)

def _rho_windows(nch=8):
    nch=min(nch,_ent_x.shape[1]);W=max(80,int(3*_ent_fs));starts=np.linspace(0,max(0,len(_ent_x)-W),_ENT_SPEED['windows']).astype(int);return starts,[_density_cov(_ent_x[s:s+W,:nch]) for s in starts]

def _osee(rho,dA=2,dB=2):
    R=np.asarray(rho,complex).reshape(dA,dB,dA,dB).transpose(0,2,1,3).reshape(dA*dA,dB*dB);s=np.linalg.svd(R,compute_uv=False);p=_safe_prob(s*s);return _H(p),p

def _discord_Z_proxy(rho):
    rho=np.asarray(rho,complex);rhoA,rhoB=_partial_traces(rho,2,2);SA=_qentropy(rhoA)[0];SB=_qentropy(rhoB)[0];SAB=_qentropy(rho)[0];I=SA+SB-SAB;R=rho.reshape(2,2,2,2);cond=0.
    for b in range(2):
        A=R[:,b,:,b];pb=float(np.real(np.trace(A)))
        if pb>ENT_EPS:cond+=pb*_qentropy(A/pb)[0]
    Jz=SA-cond;return float(I),float(Jz),float(max(0,I-Jz))
```

### 278. Entanglement entropy
