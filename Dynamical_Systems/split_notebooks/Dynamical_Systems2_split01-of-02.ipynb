Recurrence analysis then marks pairs of standardized states that return within a common radius $\varepsilon$.</p><p>Using one globally calibrated $\varepsilon$ makes recurrence rate comparable across channels after standardization. Recurrence rate (RR) is the fraction of recurrent pairs; determinism (DET) is the fraction of recurrence points belonging to sufficiently long diagonal lines; laminarity (LAM) is the analogous fraction in vertical lines. The diagonal-line entropy summarizes the empirical distribution of diagonal line lengths.</p><p>The embedding step speed $\|X_{t+1}-X_t\|$ is an additional geometric diagnostic and is not itself an RQA invariant.</p><p><b>Cell role.</b> The cell uses a pooled standardized-distance quantile to define the global recurrence threshold and applies the same threshold to every channel.</p></div>

$$X_t=(x_t,x_{t-\tau},x_{t-2\tau}),\qquad R_{ij}=\mathbf 1\{\|Z_i-Z_j\|_2\le\varepsilon\},$$

$$RR=\frac{\sum_{i\ne j}R_{ij}}{N(N-1)}.$$

```python
import os
import re
import time
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.spatial.distance import pdist, squareform
from matplotlib.colors import LinearSegmentedColormap
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

# -------------------------------------------------------
# SPEED MODE — shared four-level convention
# -------------------------------------------------------
# full     = original/highest-fidelity count settings
# balanced = moderate reduction for routine analysis
# fast     = stronger reduction for exploration
# ultra    = smallest defensible preview counts
FAST_OPTION = globals().get("FAST_OPTION", "fast")
if FAST_OPTION not in {"full", "balanced", "fast", "ultra"}:
    raise ValueError("FAST_OPTION must be one of: 'full', 'balanced', 'fast', 'ultra'.")
_SPEED_COUNT_FACTOR = {"full": 1.0, "balanced": 0.60, "fast": 0.35, "ultra": 0.18}[FAST_OPTION]
def _speed_count(original, minimum=1):
    return int(original) if FAST_OPTION == "full" else max(int(minimum), int(round(float(original) * _SPEED_COUNT_FACTOR)))

# -------------------------------------------------------
# CONFIG
# -------------------------------------------------------
base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"

eeg_path = os.path.join(base_dir, "eeg_data_with_channels.npy")
embedding_3d_dir = os.path.join(base_dir, "embedding_data", "3dembedding_data")

results_dir = os.path.join(base_dir, "results", "embedding_3d_recurrence_analysis_fixed")
plots_dir = os.path.join(results_dir, "plots")

os.makedirs(results_dir, exist_ok=True)
os.makedirs(plots_dir, exist_ok=True)

sampling_rate = 1000
start_time, end_time = 814.571, 921.515
start_index = int(start_time * sampling_rate)
end_index = int(end_time * sampling_rate)

eeg_channel_names = [
    "Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8",
    "FC5", "FC1", "FC2", "FC6", "M1", "T7", "C3", "Cz",
    "C4", "T8", "M2", "CP5", "CP1", "CP2", "CP6", "P7",
    "P3", "Pz", "P4", "P8", "POz", "O1", "Oz", "O2"
]

# Plot controls
MAX_CHANNELS_TO_ANALYZE = 32
MAX_EEG_PLOT_POINTS = _speed_count(20000, 5000)
MAX_3D_PLOT_POINTS = _speed_count(6000, 1500)
MAX_RECURRENCE_POINTS = _speed_count(1800, 600)

# Individual channel plots can create many files.
MAKE_INDIVIDUAL_CHANNEL_PLOTS = True

# Meaningful global thresholds
GLOBAL_RECURRENCE_TARGET_RATE = 0.03
GLOBAL_LOW_SPEED_PERCENTILE = 2.0

# RQA controls
MIN_DIAGONAL_LINE = 2
MIN_VERTICAL_LINE = 2

# Heartbeat
HEARTBEAT_EVERY_SEC = 10

# -------------------------------------------------------
# STYLE: black background + cyan
# -------------------------------------------------------
ACCENT = "cyan"
BG = "black"

CYAN_CMAP = LinearSegmentedColormap.from_list(
    "black_to_cyan",
    [(0.0, "black"), (1.0, "cyan")]
)

plt.rcParams.update({
    "figure.facecolor": BG,
    "axes.facecolor": BG,
    "savefig.facecolor": BG,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT,
})

def style_ax(ax):
    ax.set_facecolor(BG)
    ax.tick_params(axis="both", colors=ACCENT)

    ax.xaxis.label.set_color(ACCENT)
    ax.yaxis.label.set_color(ACCENT)
    ax.title.set_color(ACCENT)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)

    ax.grid(True, color=ACCENT, alpha=0.18)


def style_3d_ax(ax):
    ax.set_facecolor(BG)
    ax.tick_params(axis="both", colors=ACCENT)

    ax.xaxis.label.set_color(ACCENT)
    ax.yaxis.label.set_color(ACCENT)
    ax.zaxis.label.set_color(ACCENT)
    ax.title.set_color(ACCENT)

    ax.xaxis.pane.set_facecolor(BG)
    ax.yaxis.pane.set_facecolor(BG)
    ax.zaxis.pane.set_facecolor(BG)

    ax.xaxis.pane.set_edgecolor(ACCENT)
    ax.yaxis.pane.set_edgecolor(ACCENT)
    ax.zaxis.pane.set_edgecolor(ACCENT)


def style_colorbar(cbar):
    cbar.outline.set_edgecolor(ACCENT)
    cbar.ax.tick_params(color=ACCENT, labelcolor=ACCENT)
    cbar.ax.yaxis.label.set_color(ACCENT)

    for label in cbar.ax.get_yticklabels():
        label.set_color(ACCENT)


def save_show_close(fig, out_path, dpi=220):
    fig.tight_layout()
    fig.savefig(out_path, dpi=dpi, bbox_inches="tight", facecolor=BG)
    plt.show()
    plt.close(fig)
    print("Saved:", out_path, flush=True)


# -------------------------------------------------------
# HEARTBEAT
# -------------------------------------------------------
last_heartbeat = time.time()

def heartbeat(message, every_sec=10, force=False):
    global last_heartbeat
    now = time.time()

    if force or (now - last_heartbeat >= every_sec):
        print(message, flush=True)
        last_heartbeat = now


# -------------------------------------------------------
# HELPERS
# -------------------------------------------------------
def zscore(x, axis=0):
    x = np.asarray(x, dtype=float)
    return (x - np.nanmean(x, axis=axis, keepdims=True)) / (
        np.nanstd(x, axis=axis, keepdims=True) + 1e-12
    )


def uniform_subsample(x, max_points):
    x = np.asarray(x)

    if max_points is None or len(x) <= max_points:
        return x

    idx = np.linspace(0, len(x) - 1, max_points).astype(int)
    return x[idx]


def infer_channel_from_filename(file_path):
    stem = os.path.splitext(os.path.basename(file_path))[0]

    if "dembedded_" in stem:
        return stem.split("dembedded_", 1)[1]

    if "_" in stem:
        return stem.split("_")[-1]

    return stem


def sort_embedding_files_by_channel(file_paths, channel_order):
    channel_rank = {ch: i for i, ch in enumerate(channel_order)}

    def key_func(fp):
        ch = infer_channel_from_filename(fp)
        return channel_rank.get(ch, 10_000), os.path.basename(fp)

    return sorted(file_paths, key=key_func)


def load_embedding_safely(file_path):
    arr = np.load(file_path, allow_pickle=True)

    if isinstance(arr, np.ndarray) and arr.dtype == object:
        try:
            arr = np.asarray(arr.tolist(), dtype=float)
        except Exception:
            arr = np.asarray(arr, dtype=float)

    arr = np.asarray(arr, dtype=float)
    arr = np.squeeze(arr)

    if arr.ndim != 2:
        raise ValueError(f"Expected 2D embedding array, got shape {arr.shape}")

    if arr.shape[1] < 3:
        raise ValueError(f"Expected at least 3 columns for 3D embedding, got shape {arr.shape}")

    arr = arr[:, :3]

    mask = np.all(np.isfinite(arr), axis=1)
    arr = arr[mask]

    if arr.shape[0] < 10:
        raise ValueError("Embedding has too few valid points.")

    return arr


def embedding_speed(X):
    Xz = zscore(X, axis=0)
    steps = np.diff(Xz, axis=0)
    return np.linalg.norm(steps, axis=1)


def recurrence_matrix_with_global_eps(X, eps, max_points=1800):
    """
    Uses one global epsilon for all channels.
    This makes recurrence rate comparable and meaningful across channels.
    """
    X = uniform_subsample(X, max_points)
    Xz = zscore(X, axis=0)

    distances = squareform(pdist(Xz, metric="euclidean"))
    R = (distances <= eps).astype(np.uint8)

    # Remove line of identity so it does not inflate recurrence metrics.
    np.fill_diagonal(R, 0)

    n = R.shape[0]
    denom = max(n * n - n, 1)
    recurrence_rate = float(R.sum() / denom)

    return R, recurrence_rate, Xz


def line_lengths_1d(binary_array):
    """
    Return lengths of contiguous runs of 1s in a 1D binary array.
    """
    lengths = []
    count = 0

    for val in binary_array:
        if val:
            count += 1
        else:
            if count > 0:
                lengths.append(count)
                count = 0

    if count > 0:
        lengths.append(count)

    return lengths


def rqa_metrics(R, min_diag=2, min_vert=2):
    """
    Basic recurrence quantification analysis metrics.

    RR  = recurrence rate
    DET = fraction of recurrence points forming diagonal lines
    LAM = fraction of recurrence points forming vertical lines
    Lmax = longest diagonal line
    TT  = trapping time, mean vertical line length
    diag_entropy = Shannon entropy of diagonal line lengths
    """
    R = np.asarray(R, dtype=np.uint8)
    n = R.shape[0]
    total_rec = int(R.sum())

    if total_rec == 0:
        return {
            "determinism": 0.0,
            "laminarity": 0.0,
            "lmax": 0,
            "trapping_time": 0.0,
            "diag_entropy": 0.0,
        }

    # Diagonal line lengths, excluding main diagonal already zeroed
    diag_lengths = []

    for offset in range(-(n - 1), n):
        if offset == 0:
            continue

        diag = np.diagonal(R, offset=offset)
        diag_lengths.extend(line_lengths_1d(diag))

    diag_valid = [l for l in diag_lengths if l >= min_diag]
    diag_points = sum(diag_valid)

    determinism = diag_points / total_rec if total_rec > 0 else 0.0
    lmax = max(diag_valid) if len(diag_valid) > 0 else 0

    if len(diag_valid) > 0:
        vals, counts = np.unique(diag_valid, return_counts=True)
        p = counts.astype(float) / counts.sum()
        diag_entropy = float(-np.sum(p * np.log2(p + 1e-12)))
    else:
        diag_entropy = 0.0

    # Vertical line lengths
    vert_lengths = []

    for col in range(n):
        vert = R[:, col]
        vert_lengths.extend(line_lengths_1d(vert))

    vert_valid = [l for l in vert_lengths if l >= min_vert]
    vert_points = sum(vert_valid)

    laminarity = vert_points / total_rec if total_rec > 0 else 0.0
    trapping_time = float(np.mean(vert_valid)) if len(vert_valid) > 0 else 0.0

    return {
        "determinism": float(determinism),
        "laminarity": float(laminarity),
        "lmax": int(lmax),
        "trapping_time": float(trapping_time),
        "diag_entropy": float(diag_entropy),
    }


# -------------------------------------------------------
# LOAD EEG DATA
# -------------------------------------------------------
heartbeat("Loading EEG data...", force=True)

EEG_data = np.load(eeg_path, allow_pickle=True)
EEG_data = np.asarray(EEG_data, dtype=float)

if EEG_data.ndim != 2:
    raise ValueError(f"Expected 2D EEG data, got shape {EEG_data.shape}")

if EEG_data.shape[0] == len(eeg_channel_names) and EEG_data.shape[1] != len(eeg_channel_names):
    EEG_data = EEG_data.T

if end_index > EEG_data.shape[0]:
    raise ValueError(
        f"Requested end_index={end_index}, but EEG only has {EEG_data.shape[0]} samples."
    )

EEG_segment = EEG_data[start_index:end_index, :]

print("Loaded EEG from:", eeg_path, flush=True)
print("Full EEG shape:", EEG_data.shape, flush=True)
print("Segment shape:", EEG_segment.shape, flush=True)


# -------------------------------------------------------
# FIND 3D EMBEDDING FILES
# -------------------------------------------------------
file_paths = sorted(glob.glob(os.path.join(embedding_3d_dir, "*.npy")))

if len(file_paths) == 0:
    raise FileNotFoundError(f"No .npy files found in: {embedding_3d_dir}")

file_paths = sort_embedding_files_by_channel(file_paths, eeg_channel_names)

if MAX_CHANNELS_TO_ANALYZE is not None:
    file_paths = file_paths[:MAX_CHANNELS_TO_ANALYZE]

print(f"Analyzing {len(file_paths)} 3D embedding files.", flush=True)


# -------------------------------------------------------
# FIRST PASS: GLOBAL THRESHOLDS
# -------------------------------------------------------
heartbeat("Computing global recurrence and low-speed thresholds...", force=True)

pooled_distances = []
pooled_speeds = []

threshold_records = []

for idx, file_path in enumerate(file_paths, start=1):
    channel = infer_channel_from_filename(file_path)

    heartbeat(
        f"Threshold pass {idx}/{len(file_paths)}: {channel}",
        every_sec=HEARTBEAT_EVERY_SEC,
        force=(idx == 1)
    )

    X = load_embedding_safely(file_path)

    # Pooled distances for global recurrence epsilon
    X_rec = uniform_subsample(X, MAX_RECURRENCE_POINTS)
    X_rec_z = zscore(X_rec, axis=0)
    d = pdist(X_rec_z, metric="euclidean")

    pooled_distances.append(d)

    # Pooled speed for global low-speed threshold
    spd = embedding_speed(X)
    pooled_speeds.append(spd)

    threshold_records.append({
        "channel": channel,
        "file": os.path.basename(file_path),
        "n_points": int(X.shape[0]),
        "distance_count": int(len(d)),
        "speed_count": int(len(spd)),
    })

pooled_distances = np.concatenate(pooled_distances)
pooled_speeds = np.concatenate(pooled_speeds)

GLOBAL_EPS = float(np.percentile(
    pooled_distances,
    100 * GLOBAL_RECURRENCE_TARGET_RATE
))

GLOBAL_LOW_SPEED_THRESHOLD = float(np.percentile(
    pooled_speeds,
    GLOBAL_LOW_SPEED_PERCENTILE
))

print(f"Global recurrence epsilon: {GLOBAL_EPS:.6f}", flush=True)
print(f"Global target recurrence percentile: {GLOBAL_RECURRENCE_TARGET_RATE:.2%}", flush=True)
print(f"Global low-speed threshold: {GLOBAL_LOW_SPEED_THRESHOLD:.6f}", flush=True)
print(f"Global low-speed percentile: {GLOBAL_LOW_SPEED_PERCENTILE:.2f}", flush=True)


# -------------------------------------------------------
# PLOT 1: STACKED EEG SEGMENT
# -------------------------------------------------------
heartbeat("Plotting stacked EEG segment...", force=True)

EEG_plot = uniform_subsample(EEG_segment, MAX_EEG_PLOT_POINTS)
EEG_plot_z = zscore(EEG_plot, axis=0)

offset = 6.0
time_axis = np.linspace(start_time, end_time, EEG_plot_z.shape[0])

fig, ax = plt.subplots(figsize=(14, 9), facecolor=BG)
style_ax(ax)

n_channels = min(EEG_plot_z.shape[1], len(eeg_channel_names))

for ch in range(n_channels):
    ax.plot(
        time_axis,
        EEG_plot_z[:, ch] + ch * offset,
        color=ACCENT,
        linewidth=0.8,
        alpha=0.75
    )

ax.set_yticks(np.arange(n_channels) * offset)
ax.set_yticklabels(eeg_channel_names[:n_channels])
ax.set_xlabel("Time (s)")
ax.set_ylabel("Channel, z-scored with vertical offset")
ax.set_title("Stacked EEG Segment Used for 3D Embedding Analysis")

plot_path = os.path.join(plots_dir, "01_stacked_eeg_segment.png")
save_show_close(fig, plot_path)


# -------------------------------------------------------
# SECOND PASS: ANALYSIS
# -------------------------------------------------------
summary_records = []

for idx, file_path in enumerate(file_paths, start=1):
    channel = infer_channel_from_filename(file_path)

    heartbeat(
        f"Analysis pass {idx}/{len(file_paths)}: {channel}",
        every_sec=HEARTBEAT_EVERY_SEC,
        force=(idx == 1)
    )

    rec = {
        "file": os.path.basename(file_path),
        "path": file_path,
        "channel": channel,
        "status": "ok",
        "error": "",
        "n_points": np.nan,

        "global_recurrence_eps": GLOBAL_EPS,
        "recurrence_rate": np.nan,
        "determinism": np.nan,
        "laminarity": np.nan,
        "lmax": np.nan,
        "trapping_time": np.nan,
        "diag_entropy": np.nan,

        "mean_speed": np.nan,
        "std_speed": np.nan,
        "median_speed": np.nan,
        "low_speed_fraction_global": np.nan,
        "low_speed_threshold_global": GLOBAL_LOW_SPEED_THRESHOLD,
    }

    try:
        X = load_embedding_safely(file_path)
        rec["n_points"] = int(X.shape[0])

        spd = embedding_speed(X)

        rec["mean_speed"] = float(np.mean(spd))
        rec["std_speed"] = float(np.std(spd))
        rec["median_speed"] = float(np.median(spd))
        rec["low_speed_fraction_global"] = float(np.mean(spd <= GLOBAL_LOW_SPEED_THRESHOLD))

        R, recurrence_rate, X_rec_z = recurrence_matrix_with_global_eps(
            X,
            eps=GLOBAL_EPS,
            max_points=MAX_RECURRENCE_POINTS
        )

        rec["recurrence_rate"] = float(recurrence_rate)

        rqa = rqa_metrics(
            R,
            min_diag=MIN_DIAGONAL_LINE,
            min_vert=MIN_VERTICAL_LINE
        )

        rec.update(rqa)

        # ---------------------------------------------------
        # OPTIONAL INDIVIDUAL CHANNEL PLOTS
        # ---------------------------------------------------
        if MAKE_INDIVIDUAL_CHANNEL_PLOTS:
            # 3D trajectory
            X_plot = uniform_subsample(X, MAX_3D_PLOT_POINTS)
            X_plot_z = zscore(X_plot, axis=0)

            fig = plt.figure(figsize=(8, 7), facecolor=BG)
            ax = fig.add_subplot(111, projection="3d")
            style_3d_ax(ax)

            ax.plot(
                X_plot_z[:, 0],
                X_plot_z[:, 1],
                X_plot_z[:, 2],
                color=ACCENT,
                linewidth=0.8,
                alpha=0.8
            )

            ax.set_title(f"3D Delay Embedding Trajectory | {channel}")
            ax.set_xlabel("X1, z-scored")
            ax.set_ylabel("X2, z-scored")
            ax.set_zlabel("X3, z-scored")

            plot_path = os.path.join(plots_dir, f"02_3d_embedding_{channel}.png")
            save_show_close(fig, plot_path)

            # Recurrence plot
            fig, ax = plt.subplots(figsize=(7, 7), facecolor=BG)
            style_ax(ax)

            im = ax.imshow(
                R,
                origin="lower",
                cmap=CYAN_CMAP,
                interpolation="nearest",
                vmin=0,
                vmax=1
            )

            ax.set_title(
                f"Global-Epsilon Recurrence Plot | {channel}\n"
                f"RR={recurrence_rate:.3%}, DET={rec['determinism']:.3f}"
            )
            ax.set_xlabel("Time index")
            ax.set_ylabel("Time index")

            plot_path = os.path.join(plots_dir, f"03_recurrence_plot_{channel}.png")
            save_show_close(fig, plot_path)

            # Embedding coordinates
            X_small = uniform_subsample(X, MAX_EEG_PLOT_POINTS)
            X_small_z = zscore(X_small, axis=0)

            fig, ax = plt.subplots(figsize=(12, 6), facecolor=BG)
            style_ax(ax)

            ax.plot(X_small_z[:, 0], color=ACCENT, linewidth=1.0, alpha=0.85, label="X1")
            ax.plot(X_small_z[:, 1] + 5, color=ACCENT, linewidth=1.0, alpha=0.55, label="X2 + offset")
            ax.plot(X_small_z[:, 2] + 10, color=ACCENT, linewidth=1.0, alpha=0.35, label="X3 + offset")

            ax.set_title(f"3D Embedding Coordinates | {channel}")
            ax.set_xlabel("Time index")
            ax.set_ylabel("Z-scored coordinate with offsets")

            leg = ax.legend(facecolor=BG, edgecolor=ACCENT)
            for text in leg.get_texts():
                text.set_color(ACCENT)

            plot_path = os.path.join(plots_dir, f"04_embedding_coordinates_{channel}.png")
            save_show_close(fig, plot_path)

            # Speed with global threshold
            speed_small = embedding_speed(X_small)

            fig, ax = plt.subplots(figsize=(12, 5), facecolor=BG)
            style_ax(ax)

            ax.plot(
                speed_small,
                color=ACCENT,
                linewidth=1.0,
                alpha=0.85
            )

            ax.axhline(
                GLOBAL_LOW_SPEED_THRESHOLD,
                color=ACCENT,
                linestyle="--",
                alpha=0.7,
                label="Global low-speed threshold"
            )

            ax.set_title(f"Embedding Step Speed with Global Low-Speed Threshold | {channel}")
            ax.set_xlabel("Time index")
            ax.set_ylabel("Step speed in standardized embedding space")

            leg = ax.legend(facecolor=BG, edgecolor=ACCENT)
            for text in leg.get_texts():
                text.set_color(ACCENT)

            plot_path = os.path.join(plots_dir, f"05_embedding_speed_{channel}.png")
            save_show_close(fig, plot_path)

    except Exception as e:
        rec["status"] = "error"
        rec["error"] = f"{type(e).__name__}: {e}"
        print(f"ERROR in {os.path.basename(file_path)}: {rec['error']}", flush=True)

    summary_records.append(rec)


# -------------------------------------------------------
# SAVE SUMMARY
# -------------------------------------------------------
summary_df = pd.DataFrame(summary_records)

summary_csv = os.path.join(results_dir, "embedding_3d_rqa_summary_fixed.csv")
summary_df.to_csv(summary_csv, index=False)

threshold_df = pd.DataFrame(threshold_records)
threshold_csv = os.path.join(results_dir, "embedding_3d_threshold_pass_summary.csv")
threshold_df.to_csv(threshold_csv, index=False)

print("\nSaved summary CSV:", summary_csv, flush=True)
print("Saved threshold summary CSV:", threshold_csv, flush=True)

n_errors = int((summary_df["status"] == "error").sum())
print(f"Errors: {n_errors} of {len(summary_df)}", flush=True)


# -------------------------------------------------------
# SUMMARY PLOTS
# -------------------------------------------------------
df_ok = summary_df[summary_df["status"] == "ok"].copy()

if len(df_ok) > 0:
    # ---------------------------------------------------
    # PLOT 6: RECURRENCE RATE BY CHANNEL
    # Now meaningful because epsilon is global.
    # ---------------------------------------------------
    plot_df = df_ok.sort_values("recurrence_rate")

    fig, ax = plt.subplots(figsize=(10, 8), facecolor=BG)
    style_ax(ax)

    ax.barh(
        plot_df["channel"],
        plot_df["recurrence_rate"],
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.85
    )

    ax.axvline(
        GLOBAL_RECURRENCE_TARGET_RATE,
        color=ACCENT,
        linestyle="--",
        alpha=0.7,
        label="Global pooled target rate"
    )

    ax.set_title("Global-Epsilon Recurrence Rate by Channel")
    ax.set_xlabel("Recurrence rate")
    ax.set_ylabel("Channel")

    leg = ax.legend(facecolor=BG, edgecolor=ACCENT)
    for text in leg.get_texts():
        text.set_color(ACCENT)

    plot_path = os.path.join(plots_dir, "06_global_epsilon_recurrence_rate_by_channel.png")
    save_show_close(fig, plot_path)

    # ---------------------------------------------------
    # PLOT 7: DETERMINISM BY CHANNEL
    # ---------------------------------------------------
    plot_df = df_ok.sort_values("determinism")

    fig, ax = plt.subplots(figsize=(10, 8), facecolor=BG)
    style_ax(ax)

    ax.barh(
        plot_df["channel"],
        plot_df["determinism"],
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.85
    )

    ax.set_title("RQA Determinism by Channel")
    ax.set_xlabel("Determinism")
    ax.set_ylabel("Channel")

    plot_path = os.path.join(plots_dir, "07_rqa_determinism_by_channel.png")
    save_show_close(fig, plot_path)

    # ---------------------------------------------------
    # PLOT 8: LAMINARITY BY CHANNEL
    # ---------------------------------------------------
    plot_df = df_ok.sort_values("laminarity")

    fig, ax = plt.subplots(figsize=(10, 8), facecolor=BG)
    style_ax(ax)

    ax.barh(
        plot_df["channel"],
        plot_df["laminarity"],
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.85
    )

    ax.set_title("RQA Laminarity by Channel")
    ax.set_xlabel("Laminarity")
    ax.set_ylabel("Channel")

    plot_path = os.path.join(plots_dir, "08_rqa_laminarity_by_channel.png")
    save_show_close(fig, plot_path)

    # ---------------------------------------------------
    # PLOT 9: MEAN EMBEDDING SPEED BY CHANNEL
    # ---------------------------------------------------
    plot_df = df_ok.sort_values("mean_speed")

    fig, ax = plt.subplots(figsize=(10, 8), facecolor=BG)
    style_ax(ax)

    ax.barh(
        plot_df["channel"],
        plot_df["mean_speed"],
        xerr=plot_df["std_speed"],
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.85
    )

    ax.set_title("Mean Step Speed in 3D Embedding by Channel")
    ax.set_xlabel("Mean step speed, standardized embedding")
    ax.set_ylabel("Channel")

    plot_path = os.path.join(plots_dir, "09_mean_embedding_speed_by_channel.png")
    save_show_close(fig, plot_path)

    # ---------------------------------------------------
    # PLOT 10: GLOBAL LOW-SPEED FRACTION BY CHANNEL
    # Now meaningful because threshold is global.
    # ---------------------------------------------------
    plot_df = df_ok.sort_values("low_speed_fraction_global")

    fig, ax = plt.subplots(figsize=(10, 8), facecolor=BG)
    style_ax(ax)

    ax.barh(
        plot_df["channel"],
        plot_df["low_speed_fraction_global"],
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.85
    )

    ax.axvline(
        GLOBAL_LOW_SPEED_PERCENTILE / 100.0,
        color=ACCENT,
        linestyle="--",
        alpha=0.7,
        label="Global pooled threshold fraction"
    )

    ax.set_title("Global Low-Speed Fraction by Channel")
    ax.set_xlabel("Fraction of steps below global low-speed threshold")
    ax.set_ylabel("Channel")

    leg = ax.legend(facecolor=BG, edgecolor=ACCENT)
    for text in leg.get_texts():
        text.set_color(ACCENT)

    plot_path = os.path.join(plots_dir, "10_global_low_speed_fraction_by_channel.png")
    save_show_close(fig, plot_path)

    # ---------------------------------------------------
    # PLOT 11: RECURRENCE RATE VS MEAN SPEED
    # ---------------------------------------------------
    fig, ax = plt.subplots(figsize=(9, 6), facecolor=BG)
    style_ax(ax)

    ax.scatter(
        df_ok["mean_speed"],
        df_ok["recurrence_rate"],
        color=ACCENT,
        edgecolor=ACCENT,
        s=65,
        alpha=0.85
    )

    for _, row in df_ok.iterrows():
        ax.text(
            row["mean_speed"],
            row["recurrence_rate"],
            row["channel"],
            color=ACCENT,
            fontsize=8,
            alpha=0.75
        )

    ax.set_title("Global Recurrence Rate vs Mean Embedding Speed")
    ax.set_xlabel("Mean step speed")
    ax.set_ylabel("Global-epsilon recurrence rate")

    plot_path = os.path.join(plots_dir, "11_recurrence_rate_vs_embedding_speed.png")
    save_show_close(fig, plot_path)

    # ---------------------------------------------------
    # PLOT 12: DET vs LAM
    # ---------------------------------------------------
    fig, ax = plt.subplots(figsize=(9, 6), facecolor=BG)
    style_ax(ax)

    ax.scatter(
        df_ok["determinism"],
        df_ok["laminarity"],
        color=ACCENT,
        edgecolor=ACCENT,
        s=65,
        alpha=0.85
    )

    for _, row in df_ok.iterrows():
        ax.text(
            row["determinism"],
            row["laminarity"],
            row["channel"],
            color=ACCENT,
            fontsize=8,
            alpha=0.75
        )

    ax.set_title("RQA Determinism vs Laminarity")
    ax.set_xlabel("Determinism")
    ax.set_ylabel("Laminarity")

    plot_path = os.path.join(plots_dir, "12_determinism_vs_laminarity.png")
    save_show_close(fig, plot_path)


print("\nAll done.", flush=True)
print("Plots saved to:", plots_dir, flush=True)
```
