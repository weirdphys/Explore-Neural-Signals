```python
import os
import glob
import time
import numpy as np

if not hasattr(np, "trapz"):
    np.trapz = np.trapezoid

import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.gridspec import GridSpec
from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list
from scipy.stats import chi2

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"

FAST_OPTION = "ultra"  # options: "ultra", "fast", "balanced", "full"
if FAST_OPTION not in {"ultra", "fast", "balanced", "full"}:
    raise ValueError("FAST_OPTION must be one of: ultra, fast, balanced, full")

SAVE_PLOTS_TO_DISK = True
RENDER_ALL_IMAGES_INLINE = True
INLINE_IMAGE_WIDTH = 1100
PLOT_DPI = 150
VERBOSE_PROGRESS = True

out_dir = os.path.join(base_dir, "topology_persistence_significance_testing")
mode_out_dir = os.path.join(
    out_dir,
    f"h1_persistence_significance_testing_{FAST_OPTION}_multi_surrogate_consensus",
)
os.makedirs(mode_out_dir, exist_ok=True)

BLACK = "#000000"
ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"
CYAN_FILL = "#00D5FF"
CYAN_FILL_DARK = "#005A63"

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, CYAN_FILL_DARK, CYAN_FILL, ACCENT_SOFT],
    N=256,
)

CYAN_DIVERGING_CMAP = LinearSegmentedColormap.from_list(
    "cyan_diverging",
    [ACCENT_SOFT, BLACK, CYAN_FILL],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    if VERBOSE_PROGRESS:
        print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def force_cyan_text(ax):
    ax.title.set_color(ACCENT)
    ax.xaxis.label.set_color(ACCENT)
    ax.yaxis.label.set_color(ACCENT)

    for tick_label in ax.get_xticklabels():
        tick_label.set_color(ACCENT)

    for tick_label in ax.get_yticklabels():
        tick_label.set_color(ACCENT)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def style_legend(legend):
    if legend is not None:
        legend.get_frame().set_facecolor(BLACK)
        legend.get_frame().set_edgecolor(ACCENT)

        for text in legend.get_texts():
            text.set_color(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def bh_fdr(pvals):
    pvals = np.asarray(pvals, dtype=float)
    qvals = np.full_like(pvals, np.nan, dtype=float)

    valid = np.isfinite(pvals)
    p = pvals[valid]

    if len(p) == 0:
        return qvals

    order = np.argsort(p)
    ranked = p[order]
    m = len(ranked)

    raw_q = ranked * m / np.arange(1, m + 1)
    monotone_q = np.minimum.accumulate(raw_q[::-1])[::-1]
    monotone_q = np.clip(monotone_q, 0.0, 1.0)

    q_valid = np.empty_like(monotone_q)
    q_valid[order] = monotone_q

    qvals[valid] = q_valid

    return qvals


def percentile_score(values, higher_is_more=True):
    s = pd.Series(values, dtype=float)

    if not higher_is_more:
        s = -s

    if s.nunique(dropna=True) <= 1:
        return np.full(len(s), 50.0, dtype=float)

    return s.rank(method="average", pct=True).to_numpy(dtype=float) * 100.0


def offdiag_values(M):
    M = np.asarray(M, dtype=float)

    if M.shape[0] <= 1:
        return np.asarray([], dtype=float)

    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    vals = vals[np.isfinite(vals)]

    return vals


def dynamic_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=None, upper_bound=None):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        vmin, vmax = 0.0, 1.0
    else:
        vmin = float(np.nanpercentile(values, lower_pct))
        vmax = float(np.nanpercentile(values, upper_pct))

        if not np.isfinite(vmin):
            vmin = float(np.nanmin(values))

        if not np.isfinite(vmax):
            vmax = float(np.nanmax(values))

        if vmax <= vmin:
            center = float(np.nanmean(values))
            eps = max(1e-6, abs(center) * 0.01, 1e-5)
            vmin = center - eps
            vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    vmin -= pad
    vmax += pad

    if lower_bound is not None:
        vmin = max(lower_bound, vmin)

    if upper_bound is not None:
        vmax = min(upper_bound, vmax)

    if vmax <= vmin:
        vmax = vmin + 1e-6

    return vmin, vmax


def robust_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, include_zero=True):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        return 0.0, 1.0

    vmin = float(np.nanpercentile(values, lower_pct))
    vmax = float(np.nanpercentile(values, upper_pct))

    if include_zero:
        vmin = min(vmin, 0.0)
        vmax = max(vmax, 0.0)

    if vmax <= vmin:
        center = float(np.nanmean(values))
        eps = max(abs(center) * 0.05, 1e-6)
        vmin = center - eps
        vmax = center + eps

    pad = (vmax - vmin) * pad_frac

    return vmin - pad, vmax + pad


def robust_metric_contrast_matrix(raw_matrix):
    raw_matrix = np.asarray(raw_matrix, dtype=float)
    out = np.zeros_like(raw_matrix, dtype=float)
    info = []

    for j in range(raw_matrix.shape[1]):
        col = raw_matrix[:, j]
        finite = col[np.isfinite(col)]

        if len(finite) == 0:
            out[:, j] = 0.0
            info.append({"metric_index": j, "center": np.nan, "scale": np.nan})
            continue

        center = float(np.nanmedian(finite))
        mad = float(np.nanmedian(np.abs(finite - center)))
        scale = 1.4826 * mad

        if not np.isfinite(scale) or scale < 1e-9:
            p10 = float(np.nanpercentile(finite, 10))
            p90 = float(np.nanpercentile(finite, 90))
            scale = (p90 - p10) / 2.563

        if not np.isfinite(scale) or scale < 1e-9:
            scale = float(np.nanstd(finite))

        if not np.isfinite(scale) or scale < 1e-9:
            scale = 1.0

        out[:, j] = (col - center) / scale

        info.append({
            "metric_index": j,
            "center": center,
            "scale": scale,
        })

    return out, pd.DataFrame(info)


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


def gaussian_similarity_from_distance(D, sigma=None):
    D = np.asarray(D, dtype=float)
    off = offdiag_values(D)

    if sigma is None:
        sigma = float(np.nanmedian(off)) if len(off) else 1.0

    sigma = max(float(sigma), 1e-8)

    S = np.exp(-(D ** 2) / (2.0 * sigma ** 2))
    np.fill_diagonal(S, 1.0)

    return S, sigma


def fisher_combine_pvalues(pvals):
    pvals = np.asarray(pvals, dtype=float)
    pvals = pvals[np.isfinite(pvals)]
    pvals = np.clip(pvals, 1e-300, 1.0)

    if len(pvals) == 0:
        return np.nan, np.nan, 0

    if len(pvals) == 1:
        return float(pvals[0]), float(-2.0 * np.log(pvals[0])), 1

    stat = float(-2.0 * np.sum(np.log(pvals)))
    p_combined = float(chi2.sf(stat, 2 * len(pvals)))

    return p_combined, stat, len(pvals)


def newest_file(patterns, exclude_fragments=()):
    candidates = []

    for pattern in patterns:
        candidates.extend(glob.glob(pattern, recursive=True))

    clean = []

    for path in candidates:
        name = os.path.basename(path).lower()

        if any(fragment.lower() in name for fragment in exclude_fragments):
            continue

        if os.path.isfile(path):
            clean.append(path)

    if len(clean) == 0:
        return None

    clean = sorted(clean, key=lambda p: os.path.getmtime(p), reverse=True)

    return clean[0]


channel_regions = {
    "Fp1": "Frontal", "Fpz": "Frontal", "Fp2": "Frontal",
    "F7": "Frontal", "F3": "Frontal", "Fz": "Frontal", "F4": "Frontal", "F8": "Frontal",
    "FC5": "Frontocentral", "FC1": "Frontocentral", "FC2": "Frontocentral", "FC6": "Frontocentral",
    "M1": "Mastoid", "M2": "Mastoid",
    "T7": "Central", "C3": "Central", "Cz": "Central", "C4": "Central", "T8": "Central",
    "CP5": "Centroparietal", "CP1": "Centroparietal", "CP2": "Centroparietal", "CP6": "Centroparietal",
    "P7": "Parietal", "P3": "Parietal", "Pz": "Parietal", "P4": "Parietal", "P8": "Parietal",
    "POz": "Occipital", "O1": "Occipital", "Oz": "Occipital", "O2": "Occipital",
}

scalp_xy = {
    "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
    "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
    "F4": (0.45, 0.55), "F8": (0.95, 0.60),
    "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
    "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
    "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
    "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
    "M2": (1.15, 0.05),
    "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
    "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
    "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
    "P4": (0.45, -0.60), "P8": (0.90, -0.60),
    "POz": (0.00, -0.88),
    "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
}

primary_metric_names = [
    "total_lifetime",
    "max_lifetime",
    "q90_lifetime",
    "betti_auc",
    "betti_l2",
    "landscape_l2",
    "landscape_max",
]

primary_metric_labels = [
    "Total lifetime",
    "Max lifetime",
    "Q90 lifetime",
    "Betti AUC",
    "Betti L2",
    "Landscape L2",
    "Landscape max",
]

surrogate_specs = [
    {
        "family": "shuffle",
        "label": "Shuffle",
        "prefix_candidates": ["shuffle"],
        "root": os.path.join(base_dir, "topology_shuffle_surrogate_persistent_homology"),
        "summary_patterns": [
            os.path.join(base_dir, "topology_shuffle_surrogate_persistent_homology", "**", "h1_shuffle_surrogate_ph_summary.csv"),
            os.path.join(base_dir, "topology_shuffle_surrogate_persistent_homology", "**", "*shuffle*ph_summary.csv"),
            os.path.join(base_dir, "topology_shuffle_surrogate_persistent_homology", "**", "*summary*.csv"),
        ],
        "long_patterns": [
            os.path.join(base_dir, "topology_shuffle_surrogate_persistent_homology", "**", "*shuffle*long_table.csv"),
            os.path.join(base_dir, "topology_shuffle_surrogate_persistent_homology", "**", "*long_table.csv"),
        ],
    },
    {
        "family": "phase",
        "label": "Phase randomized",
        "prefix_candidates": ["phase"],
        "root": os.path.join(base_dir, "topology_phase_randomized_surrogate_persistent_homology"),
        "summary_patterns": [
            os.path.join(base_dir, "topology_phase_randomized_surrogate_persistent_homology", "**", "h1_phase_randomized_surrogate_ph_summary.csv"),
            os.path.join(base_dir, "topology_phase_randomized_surrogate_persistent_homology", "**", "*phase*ph_summary.csv"),
            os.path.join(base_dir, "topology_phase_randomized_surrogate_persistent_homology", "**", "*summary*.csv"),
        ],
        "long_patterns": [
            os.path.join(base_dir, "topology_phase_randomized_surrogate_persistent_homology", "**", "*phase*long_table.csv"),
            os.path.join(base_dir, "topology_phase_randomized_surrogate_persistent_homology", "**", "*long_table.csv"),
        ],
    },
    {
        "family": "iaaft",
        "label": "IAAFT",
        "prefix_candidates": ["iaaft"],
        "root": os.path.join(base_dir, "topology_iaaft_surrogate_persistent_homology"),
        "summary_patterns": [
            os.path.join(base_dir, "topology_iaaft_surrogate_persistent_homology", "**", "h1_iaaft_surrogate_ph_summary.csv"),
            os.path.join(base_dir, "topology_iaaft_surrogate_persistent_homology", "**", "*iaaft*ph_summary.csv"),
            os.path.join(base_dir, "topology_iaaft_surrogate_persistent_homology", "**", "*summary*.csv"),
        ],
        "long_patterns": [
            os.path.join(base_dir, "topology_iaaft_surrogate_persistent_homology", "**", "*iaaft*long_table.csv"),
            os.path.join(base_dir, "topology_iaaft_surrogate_persistent_homology", "**", "*long_table.csv"),
        ],
    },
    {
        "family": "colored_noise",
        "label": "Colored noise",
        "prefix_candidates": ["colored_noise"],
        "root": os.path.join(base_dir, "topology_colored_noise_surrogate_persistent_homology"),
        "summary_patterns": [
            os.path.join(base_dir, "topology_colored_noise_surrogate_persistent_homology", "**", "h1_colored_noise_surrogate_ph_summary.csv"),
            os.path.join(base_dir, "topology_colored_noise_surrogate_persistent_homology", "**", "*colored_noise*ph_summary.csv"),
            os.path.join(base_dir, "topology_colored_noise_surrogate_persistent_homology", "**", "*summary*.csv"),
        ],
        "long_patterns": [
            os.path.join(base_dir, "topology_colored_noise_surrogate_persistent_homology", "**", "*colored_noise*long_table.csv"),
            os.path.join(base_dir, "topology_colored_noise_surrogate_persistent_homology", "**", "*long_table.csv"),
        ],
    },
]

exclude_summary_fragments = [
    "region",
    "regional",
    "runtime",
    "diagnostic",
    "diagnostics",
    "signature",
    "distance",
    "similarity",
    "mds",
    "contrast",
    "fixed",
    "regularized",
    "matrix",
    "long",
    "plot",
]

loaded_families = []

log("\n" + "=" * 72)
log("Loading surrogate PH outputs for persistence significance testing")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)

for spec in surrogate_specs:
    summary_path = newest_file(
        spec["summary_patterns"],
        exclude_fragments=exclude_summary_fragments,
    )

    long_path = newest_file(
        spec["long_patterns"],
        exclude_fragments=["summary", "region", "runtime", "diagnostic", "signature", "distance", "similarity", "mds", "matrix", "fixed", "contrast"],
    )

    if summary_path is None:
        log(f"  skipped {spec['label']}: no summary CSV found")
        continue

    try:
        summary_df = pd.read_csv(summary_path)
    except Exception as exc:
        log(f"  skipped {spec['label']}: could not read {summary_path} | {exc}")
        continue

    if "channel" not in summary_df.columns:
        log(f"  skipped {spec['label']}: summary has no channel column")
        continue

    summary_df["channel"] = summary_df["channel"].astype(str)

    long_df = pd.DataFrame()

    if long_path is not None:
        try:
            long_df = pd.read_csv(long_path)

            if "channel" in long_df.columns:
                long_df["channel"] = long_df["channel"].astype(str)
            else:
                long_df = pd.DataFrame()

        except Exception as exc:
            log(f"  warning {spec['label']}: could not read long table {long_path} | {exc}")
            long_df = pd.DataFrame()

    actual_prefix = None

    for prefix_candidate in spec["prefix_candidates"]:
        good = False

        for metric in primary_metric_names:
            if f"{prefix_candidate}_mean_{metric}" in summary_df.columns or f"{prefix_candidate}_p_greater_{metric}" in summary_df.columns:
                good = True
                break

        if good:
            actual_prefix = prefix_candidate
            break

    if actual_prefix is None:
        log(f"  skipped {spec['label']}: could not infer metric prefix")
        continue

    loaded_families.append({
        "family": spec["family"],
        "label": spec["label"],
        "prefix": actual_prefix,
        "summary_path": summary_path,
        "long_path": long_path,
        "summary_df": summary_df,
        "long_df": long_df,
    })

    log(f"  loaded {spec['label']}")
    log(f"    summary: {summary_path}")
    log(f"    long:    {long_path if long_path is not None else 'not found / not used'}")
    log(f"    rows:    summary={len(summary_df)}, long={len(long_df)}")

if len(loaded_families) == 0:
    raise RuntimeError(
        "No surrogate PH outputs were found. Run at least one surrogate PH notebook cell first "
        "(shuffle, phase-randomized, IAAFT, or colored-noise)."
    )


def get_long_values(long_df, channel, metric):
    if long_df is None or len(long_df) == 0:
        return np.asarray([], dtype=float)

    if "channel" not in long_df.columns or metric not in long_df.columns:
        return np.asarray([], dtype=float)

    vals = long_df.loc[long_df["channel"].astype(str) == str(channel), metric].to_numpy(dtype=float)
    vals = vals[np.isfinite(vals)]

    return vals


def compute_regularized_family_z(summary_df, long_df, prefix, metric):
    obs_col = f"observed_{metric}"
    mean_col = f"{prefix}_mean_{metric}"
    std_col = f"{prefix}_std_{metric}"

    obs = summary_df[obs_col].to_numpy(dtype=float)
    null_mean = summary_df[mean_col].to_numpy(dtype=float)

    if std_col in summary_df.columns:
        null_std = summary_df[std_col].to_numpy(dtype=float)
    else:
        null_std = np.zeros(len(summary_df), dtype=float)

    delta = obs - null_mean

    finite_std = null_std[np.isfinite(null_std) & (null_std > 0)]
    finite_delta = np.abs(delta[np.isfinite(delta)])

    null_vals_all = np.asarray([], dtype=float)

    if long_df is not None and len(long_df) > 0 and metric in long_df.columns:
        null_vals_all = long_df[metric].to_numpy(dtype=float)
        null_vals_all = null_vals_all[np.isfinite(null_vals_all)]

    floors = []

    if len(finite_std) > 0:
        floors.append(float(np.nanmedian(finite_std)) * 0.10)
        floors.append(float(np.nanpercentile(finite_std, 25)) * 0.25)

    if len(finite_delta) > 0:
        floors.append(float(np.nanmedian(finite_delta)) * 0.05)
        floors.append(float(np.nanpercentile(finite_delta, 75)) * 0.025)

    if len(null_vals_all) > 1:
        floors.append(float(np.nanstd(null_vals_all)) * 0.03)
        floors.append(float(np.nanmedian(np.abs(null_vals_all - np.nanmedian(null_vals_all)))) * 0.10)

    if np.any(np.isfinite(obs)):
        floors.append(float(np.nanmedian(np.abs(obs[np.isfinite(obs)]))) * 1e-6)

    if np.any(np.isfinite(null_mean)):
        floors.append(float(np.nanmedian(np.abs(null_mean[np.isfinite(null_mean)]))) * 1e-6)

    floors.append(1e-6)
    floors = [v for v in floors if np.isfinite(v) and v > 0]

    std_floor = max(floors) if len(floors) else 1e-6
    z = delta / np.maximum(null_std, std_floor)

    return z, std_floor


test_rows = []
family_metric_z_info = []

for family_data in loaded_families:
    family = family_data["family"]
    label = family_data["label"]
    prefix = family_data["prefix"]
    summary_df = family_data["summary_df"].copy()
    long_df = family_data["long_df"]

    for metric in primary_metric_names:
        obs_col = f"observed_{metric}"
        mean_col = f"{prefix}_mean_{metric}"
        std_col = f"{prefix}_std_{metric}"
        p_greater_col = f"{prefix}_p_greater_{metric}"
        p_lower_col = f"{prefix}_p_lower_{metric}"

        if obs_col not in summary_df.columns or mean_col not in summary_df.columns:
            log(f"  warning: {label} missing {metric}; skipping metric")
            continue

        z_regularized, std_floor = compute_regularized_family_z(
            summary_df,
            long_df,
            prefix,
            metric,
        )

        family_metric_z_info.append({
            "family": family,
            "family_label": label,
            "metric": metric,
            "std_floor": std_floor,
        })

        for row_idx, row in summary_df.iterrows():
            ch = str(row["channel"])
            region = row["region"] if "region" in summary_df.columns else channel_regions.get(ch, "Unknown")

            obs = float(row[obs_col])
            null_mean = float(row[mean_col])
            null_std = float(row[std_col]) if std_col in summary_df.columns and np.isfinite(row[std_col]) else np.nan
            delta = obs - null_mean

            null_vals = get_long_values(long_df, ch, metric)

            if len(null_vals) > 0:
                p_greater = float((1.0 + np.sum(null_vals >= obs)) / (len(null_vals) + 1.0))
                p_lower = float((1.0 + np.sum(null_vals <= obs)) / (len(null_vals) + 1.0))
                null_n = int(len(null_vals))
                null_source = "long_table_empirical"
            else:
                p_greater = float(row[p_greater_col]) if p_greater_col in summary_df.columns and np.isfinite(row[p_greater_col]) else np.nan
                p_lower = float(row[p_lower_col]) if p_lower_col in summary_df.columns and np.isfinite(row[p_lower_col]) else np.nan
                null_n = int(row[f"n_{prefix}_surrogates"]) if f"n_{prefix}_surrogates" in summary_df.columns and np.isfinite(row[f"n_{prefix}_surrogates"]) else np.nan
                null_source = "summary_existing"

            p_two = np.nan

            if np.isfinite(p_greater) and np.isfinite(p_lower):
                p_two = float(min(1.0, 2.0 * min(p_greater, p_lower)))

            z_raw_col = f"{prefix}_z_{metric}"

            if z_raw_col in summary_df.columns and np.isfinite(row[z_raw_col]):
                z_raw = float(row[z_raw_col])
            else:
                z_raw = float(delta / (null_std + 1e-12)) if np.isfinite(null_std) else np.nan

            test_rows.append({
                "family": family,
                "family_label": label,
                "prefix": prefix,
                "channel": ch,
                "region": region,
                "metric": metric,
                "metric_label": primary_metric_labels[primary_metric_names.index(metric)],
                "observed": obs,
                "null_mean": null_mean,
                "null_std": null_std,
                "delta": delta,
                "ratio": float(obs / (null_mean + 1e-12)),
                "z_raw": z_raw,
                "z_regularized": float(z_regularized[row_idx]),
                "std_floor_used": std_floor,
                "p_greater": p_greater,
                "p_lower": p_lower,
                "p_two_sided": p_two,
                "null_n": null_n,
                "null_source": null_source,
                "summary_path": family_data["summary_path"],
                "long_path": family_data["long_path"] if family_data["long_path"] is not None else "",
            })

results_df = pd.DataFrame(test_rows)

if len(results_df) == 0:
    raise RuntimeError("No channel-metric-family tests could be computed.")

results_df["q_greater"] = np.nan
results_df["q_two_sided"] = np.nan

for family in results_df["family"].unique():
    mask = results_df["family"] == family
    results_df.loc[mask, "q_greater"] = bh_fdr(results_df.loc[mask, "p_greater"].to_numpy(dtype=float))
    results_df.loc[mask, "q_two_sided"] = bh_fdr(results_df.loc[mask, "p_two_sided"].to_numpy(dtype=float))

results_df["neglog10_p_greater"] = -np.log10(results_df["p_greater"].to_numpy(dtype=float) + 1e-12)
results_df["neglog10_q_greater"] = -np.log10(results_df["q_greater"].to_numpy(dtype=float) + 1e-12)
results_df["is_q05_greater"] = results_df["q_greater"] < 0.05
results_df["is_q10_greater"] = results_df["q_greater"] < 0.10

family_metric_z_info_df = pd.DataFrame(family_metric_z_info)

consensus_rows = []

channels = sorted(results_df["channel"].unique().tolist())

for ch in channels:
    for metric, metric_label in zip(primary_metric_names, primary_metric_labels):
        block = results_df[(results_df["channel"] == ch) & (results_df["metric"] == metric)].copy()

        if len(block) == 0:
            continue

        pvals_greater = block["p_greater"].to_numpy(dtype=float)
        pvals_two = block["p_two_sided"].to_numpy(dtype=float)

        fisher_p_greater, fisher_stat_greater, fisher_k_greater = fisher_combine_pvalues(pvals_greater)
        fisher_p_two, fisher_stat_two, fisher_k_two = fisher_combine_pvalues(pvals_two)

        n_available = int(len(block))
        n_q05 = int(np.sum(block["q_greater"].to_numpy(dtype=float) < 0.05))
        n_q10 = int(np.sum(block["q_greater"].to_numpy(dtype=float) < 0.10))

        region = block["region"].iloc[0]

        consensus_rows.append({
            "channel": ch,
            "region": region,
            "metric": metric,
            "metric_label": metric_label,
            "n_surrogate_families_available": n_available,
            "n_family_q_greater_lt_0_05": n_q05,
            "n_family_q_greater_lt_0_10": n_q10,
            "family_agreement_fraction_q05": n_q05 / max(n_available, 1),
            "family_agreement_fraction_q10": n_q10 / max(n_available, 1),
            "mean_observed": float(np.nanmean(block["observed"])),
            "mean_null_mean": float(np.nanmean(block["null_mean"])),
            "mean_delta": float(np.nanmean(block["delta"])),
            "median_delta": float(np.nanmedian(block["delta"])),
            "mean_ratio": float(np.nanmean(block["ratio"])),
            "mean_z_regularized": float(np.nanmean(block["z_regularized"])),
            "median_z_regularized": float(np.nanmedian(block["z_regularized"])),
            "max_z_regularized": float(np.nanmax(block["z_regularized"])),
            "min_p_greater": float(np.nanmin(block["p_greater"])),
            "min_q_greater": float(np.nanmin(block["q_greater"])),
            "mean_neglog10_q_greater": float(np.nanmean(block["neglog10_q_greater"])),
            "fisher_p_greater": fisher_p_greater,
            "fisher_stat_greater": fisher_stat_greater,
            "fisher_k_greater": fisher_k_greater,
            "fisher_p_two_sided": fisher_p_two,
            "fisher_stat_two_sided": fisher_stat_two,
            "fisher_k_two_sided": fisher_k_two,
            "families_used": ",".join(block["family"].tolist()),
        })

consensus_df = pd.DataFrame(consensus_rows)

consensus_df["fisher_q_greater"] = bh_fdr(consensus_df["fisher_p_greater"].to_numpy(dtype=float))
consensus_df["fisher_q_two_sided"] = bh_fdr(consensus_df["fisher_p_two_sided"].to_numpy(dtype=float))
consensus_df["neglog10_fisher_q_greater"] = -np.log10(consensus_df["fisher_q_greater"].to_numpy(dtype=float) + 1e-12)
consensus_df["neglog10_fisher_p_greater"] = -np.log10(consensus_df["fisher_p_greater"].to_numpy(dtype=float) + 1e-12)

channel_summary_df = (
    consensus_df
    .groupby(["channel", "region"])
    .agg(
        n_metrics=("metric", "count"),
        n_metrics_fisher_q_lt_0_05=("fisher_q_greater", lambda x: int(np.sum(np.asarray(x, dtype=float) < 0.05))),
        n_metrics_fisher_q_lt_0_10=("fisher_q_greater", lambda x: int(np.sum(np.asarray(x, dtype=float) < 0.10))),
        mean_neglog10_fisher_q=("neglog10_fisher_q_greater", "mean"),
        max_neglog10_fisher_q=("neglog10_fisher_q_greater", "max"),
        mean_z_regularized=("mean_z_regularized", "mean"),
        median_z_regularized=("median_z_regularized", "median"),
        mean_family_agreement_q05=("family_agreement_fraction_q05", "mean"),
        max_family_agreement_q05=("family_agreement_fraction_q05", "max"),
        mean_delta=("mean_delta", "mean"),
        mean_ratio=("mean_ratio", "mean"),
        min_fisher_q=("fisher_q_greater", "min"),
    )
    .reset_index()
)

channel_summary_df["significance_component_q"] = percentile_score(channel_summary_df["mean_neglog10_fisher_q"], higher_is_more=True)
channel_summary_df["significance_component_agreement"] = percentile_score(channel_summary_df["mean_family_agreement_q05"], higher_is_more=True)
channel_summary_df["significance_component_z"] = percentile_score(channel_summary_df["mean_z_regularized"], higher_is_more=True)
channel_summary_df["significance_component_metric_count"] = percentile_score(channel_summary_df["n_metrics_fisher_q_lt_0_05"], higher_is_more=True)

channel_summary_df["persistence_significance_score"] = (
    0.42 * channel_summary_df["significance_component_q"]
    + 0.26 * channel_summary_df["significance_component_agreement"]
    + 0.22 * channel_summary_df["significance_component_z"]
    + 0.10 * channel_summary_df["significance_component_metric_count"]
)

channel_summary_df = channel_summary_df.sort_values("persistence_significance_score", ascending=False).reset_index(drop=True)

region_summary_df = (
    channel_summary_df
    .groupby("region")
    .agg(
        n_channels=("channel", "count"),
        mean_persistence_significance_score=("persistence_significance_score", "mean"),
        sem_persistence_significance_score=("persistence_significance_score", lambda x: float(np.std(x, ddof=1) / np.sqrt(len(x))) if len(x) > 1 else 0.0),
        mean_neglog10_fisher_q=("mean_neglog10_fisher_q", "mean"),
        mean_z_regularized=("mean_z_regularized", "mean"),
        mean_family_agreement_q05=("mean_family_agreement_q05", "mean"),
        mean_n_metrics_fisher_q_lt_0_05=("n_metrics_fisher_q_lt_0_05", "mean"),
    )
    .reset_index()
    .sort_values("mean_persistence_significance_score", ascending=False)
)

channel_order = channel_summary_df["channel"].tolist()

signature_feature_frames = []

for metric in primary_metric_names:
    metric_block = consensus_df[consensus_df["metric"] == metric].set_index("channel")
    metric_frame = pd.DataFrame(index=channel_order)

    metric_frame[f"{metric}_mean_z"] = metric_block.reindex(channel_order)["mean_z_regularized"].to_numpy(dtype=float)
    metric_frame[f"{metric}_neglog10_fisher_q"] = metric_block.reindex(channel_order)["neglog10_fisher_q_greater"].to_numpy(dtype=float)
    metric_frame[f"{metric}_agreement_q05"] = metric_block.reindex(channel_order)["family_agreement_fraction_q05"].to_numpy(dtype=float)

    signature_feature_frames.append(metric_frame)

signature_df = pd.concat(signature_feature_frames, axis=1)
signature_df = signature_df.replace([np.inf, -np.inf], np.nan).fillna(0.0)

signature_X = signature_df.to_numpy(dtype=float)
signature_X = signature_X - np.nanmean(signature_X, axis=0, keepdims=True)
signature_X = signature_X / (np.nanstd(signature_X, axis=0, keepdims=True) + 1e-12)
signature_X = np.nan_to_num(signature_X, nan=0.0, posinf=0.0, neginf=0.0)

if len(channel_order) >= 2:
    signature_distance = squareform(pdist(signature_X, metric="euclidean"))
else:
    signature_distance = np.zeros((len(channel_order), len(channel_order)), dtype=float)

signature_similarity, similarity_sigma = gaussian_similarity_from_distance(signature_distance)

if len(channel_order) >= 3 and np.any(signature_distance > 0):
    Z = linkage(squareform(signature_distance, checks=False), method="average", optimal_ordering=True)
    cluster_order_idx = leaves_list(Z)
else:
    cluster_order_idx = np.arange(len(channel_order))

clustered_channels = [channel_order[i] for i in cluster_order_idx]
clustered_similarity = signature_similarity[np.ix_(cluster_order_idx, cluster_order_idx)]
clustered_distance = signature_distance[np.ix_(cluster_order_idx, cluster_order_idx)]

mds_coords = classical_mds_from_distances(signature_distance, n_components=2)

mds_df = pd.DataFrame({
    "channel": channel_order,
    "region": [channel_summary_df.set_index("channel").loc[ch, "region"] for ch in channel_order],
    "persistence_significance_mds_1": mds_coords[:, 0],
    "persistence_significance_mds_2": mds_coords[:, 1],
    "persistence_significance_score": [channel_summary_df.set_index("channel").loc[ch, "persistence_significance_score"] for ch in channel_order],
    "mean_neglog10_fisher_q": [channel_summary_df.set_index("channel").loc[ch, "mean_neglog10_fisher_q"] for ch in channel_order],
})

family_channel_summary_df = (
    results_df
    .groupby(["family", "family_label", "channel", "region"])
    .agg(
        mean_z_regularized=("z_regularized", "mean"),
        mean_neglog10_q=("neglog10_q_greater", "mean"),
        n_metrics_q_lt_0_05=("q_greater", lambda x: int(np.sum(np.asarray(x, dtype=float) < 0.05))),
        min_q=("q_greater", "min"),
    )
    .reset_index()
)

family_channel_summary_df["family_channel_significance_score"] = (
    0.55 * percentile_score(family_channel_summary_df["mean_neglog10_q"], higher_is_more=True)
    + 0.30 * percentile_score(family_channel_summary_df["mean_z_regularized"], higher_is_more=True)
    + 0.15 * percentile_score(family_channel_summary_df["n_metrics_q_lt_0_05"], higher_is_more=True)
)

tests_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_channel_metric_family_tests.csv")
consensus_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_consensus_metric_tests.csv")
channel_summary_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_channel_summary.csv")
region_summary_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_region_summary.csv")
family_channel_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_family_channel_summary.csv")
signature_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_signature_matrix.csv")
signature_distance_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_signature_distance_matrix.csv")
signature_similarity_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_signature_similarity_matrix.csv")
mds_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_mds_coordinates.csv")
z_info_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_family_metric_z_regularization_info.csv")
npz_path = os.path.join(mode_out_dir, "h1_persistence_significance_outputs.npz")

results_df.to_csv(tests_csv_path, index=False)
consensus_df.to_csv(consensus_csv_path, index=False)
channel_summary_df.to_csv(channel_summary_csv_path, index=False)
region_summary_df.to_csv(region_summary_csv_path, index=False)
family_channel_summary_df.to_csv(family_channel_csv_path, index=False)
signature_df.to_csv(signature_csv_path)
pd.DataFrame(signature_distance, index=channel_order, columns=channel_order).to_csv(signature_distance_csv_path)
pd.DataFrame(signature_similarity, index=channel_order, columns=channel_order).to_csv(signature_similarity_csv_path)
mds_df.to_csv(mds_csv_path, index=False)
family_metric_z_info_df.to_csv(z_info_csv_path, index=False)

np.savez(
    npz_path,
    channels=np.asarray(channel_order),
    primary_metric_names=np.asarray(primary_metric_names),
    signature_matrix=signature_df.to_numpy(dtype=float),
    signature_distance_matrix=signature_distance,
    signature_similarity_matrix=signature_similarity,
    mds_coordinates=mds_coords,
    persistence_significance_score=channel_summary_df.set_index("channel").reindex(channel_order)["persistence_significance_score"].to_numpy(dtype=float),
    mean_neglog10_fisher_q=channel_summary_df.set_index("channel").reindex(channel_order)["mean_neglog10_fisher_q"].to_numpy(dtype=float),
    families_loaded=np.asarray([f["family"] for f in loaded_families]),
    family_labels_loaded=np.asarray([f["label"] for f in loaded_families]),
    similarity_sigma=np.asarray([similarity_sigma]),
)

print("\nPersistence significance tests by channel × metric × surrogate family:")
if HAVE_IPYTHON:
    display(results_df)
else:
    print(results_df)

print("\nConsensus persistence significance by channel × metric:")
if HAVE_IPYTHON:
    display(consensus_df)
else:
    print(consensus_df)

print("\nChannel-level persistence significance summary:")
if HAVE_IPYTHON:
    display(channel_summary_df)
else:
    print(channel_summary_df)

print("\nRegional persistence significance summary:")
if HAVE_IPYTHON:
    display(region_summary_df)
else:
    print(region_summary_df)


fig, ax = plt.subplots(figsize=(13, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title=f"H1 Persistence Significance Score | {FAST_OPTION.upper()} | multi-surrogate consensus",
    xlabel="Channel",
    ylabel="Persistence significance score",
)

rank_df = channel_summary_df.sort_values("persistence_significance_score", ascending=False).reset_index(drop=True)
x = np.arange(len(rank_df))

ax.bar(
    x,
    rank_df["persistence_significance_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.axhline(
    float(np.nanmean(rank_df["persistence_significance_score"])),
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.3,
    alpha=0.85,
    label=f"Mean={np.nanmean(rank_df['persistence_significance_score']):.1f}",
)

for i, row in rank_df.iterrows():
    if i < 8:
        ax.text(
            i,
            row["persistence_significance_score"],
            f"{row['persistence_significance_score']:.0f}",
            ha="center",
            va="bottom",
            color=ACCENT,
            fontsize=8,
        )

ax.set_xticks(x)
ax.set_xticklabels(rank_df["channel"], rotation=90)
ax.set_ylim(0, 105)

legend = ax.legend(loc="best", framealpha=1.0)
style_legend(legend)
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "01_persistence_significance_score_ranking.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=1150)


consensus_neglogq = (
    consensus_df
    .pivot(index="channel", columns="metric", values="neglog10_fisher_q_greater")
    .reindex(index=channel_order, columns=primary_metric_names)
    .to_numpy(dtype=float)
)

consensus_q = (
    consensus_df
    .pivot(index="channel", columns="metric", values="fisher_q_greater")
    .reindex(index=channel_order, columns=primary_metric_names)
    .to_numpy(dtype=float)
)

agreement_counts = (
    consensus_df
    .pivot(index="channel", columns="metric", values="n_family_q_greater_lt_0_05")
    .reindex(index=channel_order, columns=primary_metric_names)
    .to_numpy(dtype=float)
)

available_counts = (
    consensus_df
    .pivot(index="channel", columns="metric", values="n_surrogate_families_available")
    .reindex(index=channel_order, columns=primary_metric_names)
    .to_numpy(dtype=float)
)

vmin, vmax = dynamic_limits(consensus_neglogq, lower_pct=1, upper_pct=98, pad_frac=0.05, lower_bound=0.0)

fig, ax = plt.subplots(figsize=(12.8, 9.2), facecolor=BLACK)
style_ax(
    ax,
    title="Consensus H1 Persistence Significance Heatmap",
    xlabel="H1 metric",
    ylabel="Channel",
)

im = ax.imshow(
    consensus_neglogq,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=vmin,
    vmax=vmax,
)

ax.set_yticks(np.arange(len(channel_order)))
ax.set_yticklabels(channel_order)
ax.set_xticks(np.arange(len(primary_metric_labels)))
ax.set_xticklabels(primary_metric_labels, rotation=45, ha="right")

ax.set_xticks(np.arange(-0.5, len(primary_metric_labels), 1), minor=True)
ax.set_yticks(np.arange(-0.5, len(channel_order), 1), minor=True)
ax.grid(which="minor", color=ACCENT_DIM, linewidth=0.55, alpha=0.35)
ax.tick_params(which="minor", bottom=False, left=False)

for i in range(consensus_neglogq.shape[0]):
    for j in range(consensus_neglogq.shape[1]):
        q_val = consensus_q[i, j]
        agree = agreement_counts[i, j]
        avail = available_counts[i, j]

        if not np.isfinite(q_val):
            continue

        show_text = q_val < 0.10 or i < 6 or i >= consensus_neglogq.shape[0] - 3

        if show_text:
            label = f"{agree:.0f}/{avail:.0f}\nq={q_val:.2g}"
            txt = ax.text(
                j,
                i,
                label,
                ha="center",
                va="center",
                color=ACCENT,
                fontsize=5.3,
                alpha=0.92,
                fontweight="bold",
            )
            txt.set_path_effects([pe.withStroke(linewidth=1.4, foreground=BLACK)])

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "-log10 Fisher FDR q")

force_cyan_text(ax)

fig.text(
    0.09,
    0.035,
    "Cell text: number of surrogate families significant at q<0.05 / available families, plus Fisher-combined FDR q.",
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

plot_path = os.path.join(mode_out_dir, "02_consensus_persistence_significance_heatmap.png")
plt.tight_layout(rect=[0.0, 0.055, 1.0, 1.0])
save_render_close(fig, plot_path, width=1050)


agreement_fraction = (
    consensus_df
    .pivot(index="channel", columns="metric", values="family_agreement_fraction_q05")
    .reindex(index=channel_order, columns=primary_metric_names)
    .to_numpy(dtype=float)
)

fig, ax = plt.subplots(figsize=(12.8, 9.2), facecolor=BLACK)
style_ax(
    ax,
    title="Surrogate-Family Agreement Heatmap",
    xlabel="H1 metric",
    ylabel="Channel",
)

im = ax.imshow(
    agreement_fraction,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0.0,
    vmax=1.0,
)

ax.set_yticks(np.arange(len(channel_order)))
ax.set_yticklabels(channel_order)
ax.set_xticks(np.arange(len(primary_metric_labels)))
ax.set_xticklabels(primary_metric_labels, rotation=45, ha="right")

ax.set_xticks(np.arange(-0.5, len(primary_metric_labels), 1), minor=True)
ax.set_yticks(np.arange(-0.5, len(channel_order), 1), minor=True)
ax.grid(which="minor", color=ACCENT_DIM, linewidth=0.55, alpha=0.35)
ax.tick_params(which="minor", bottom=False, left=False)

for i in range(agreement_fraction.shape[0]):
    for j in range(agreement_fraction.shape[1]):
        agree = agreement_counts[i, j]
        avail = available_counts[i, j]

        if np.isfinite(agree) and np.isfinite(avail):
            txt = ax.text(
                j,
                i,
                f"{agree:.0f}/{avail:.0f}",
                ha="center",
                va="center",
                color=ACCENT,
                fontsize=5.6,
                alpha=0.92,
                fontweight="bold",
            )
            txt.set_path_effects([pe.withStroke(linewidth=1.4, foreground=BLACK)])

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Fraction of surrogate families q<0.05")
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "03_surrogate_family_agreement_heatmap.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=1050)


consensus_z = (
    consensus_df
    .pivot(index="channel", columns="metric", values="mean_z_regularized")
    .reindex(index=channel_order, columns=primary_metric_names)
    .to_numpy(dtype=float)
)

z_relative, z_contrast_info_df = robust_metric_contrast_matrix(consensus_z)
z_relative_clipped = np.clip(z_relative, -3.0, 3.0)

fig, ax = plt.subplots(figsize=(12.8, 9.2), facecolor=BLACK)
style_ax(
    ax,
    title="Consensus H1 Persistence Effect Z-Score Heatmap | fixed contrast",
    xlabel="H1 metric",
    ylabel="Channel",
)

im = ax.imshow(
    z_relative_clipped,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_DIVERGING_CMAP,
    vmin=-3,
    vmax=3,
)

ax.set_yticks(np.arange(len(channel_order)))
ax.set_yticklabels(channel_order)
ax.set_xticks(np.arange(len(primary_metric_labels)))
ax.set_xticklabels(primary_metric_labels, rotation=45, ha="right")

ax.set_xticks(np.arange(-0.5, len(primary_metric_labels), 1), minor=True)
ax.set_yticks(np.arange(-0.5, len(channel_order), 1), minor=True)
ax.grid(which="minor", color=ACCENT_DIM, linewidth=0.55, alpha=0.35)
ax.tick_params(which="minor", bottom=False, left=False)

for i in range(consensus_z.shape[0]):
    for j in range(consensus_z.shape[1]):
        raw_val = consensus_z[i, j]
        rel_val = z_relative[i, j]

        if not np.isfinite(raw_val):
            continue

        show_text = abs(rel_val) >= 1.15 or i < 5 or i >= consensus_z.shape[0] - 3

        if show_text:
            txt = ax.text(
                j,
                i,
                f"{raw_val:+.1f}",
                ha="center",
                va="center",
                color=ACCENT,
                fontsize=5.8,
                alpha=0.95,
                fontweight="bold",
            )
            txt.set_path_effects([pe.withStroke(linewidth=1.4, foreground=BLACK)])

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Within-metric channel contrast")

force_cyan_text(ax)

fig.text(
    0.09,
    0.035,
    "Fixed view: color shows within-metric channel contrast; printed values are mean regularized z-scores across surrogate families.",
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

plot_path = os.path.join(mode_out_dir, "04_consensus_persistence_effect_z_heatmap.png")
plt.tight_layout(rect=[0.0, 0.055, 1.0, 1.0])
save_render_close(fig, plot_path, width=1050)


family_score_matrix = (
    family_channel_summary_df
    .pivot(index="channel", columns="family_label", values="family_channel_significance_score")
    .reindex(index=channel_order)
)

fig, ax = plt.subplots(figsize=(10.8, 9.2), facecolor=BLACK)
style_ax(
    ax,
    title="Persistence Significance Score by Surrogate Family",
    xlabel="Surrogate family",
    ylabel="Channel",
)

family_score_values = family_score_matrix.to_numpy(dtype=float)

im = ax.imshow(
    family_score_values,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
)

ax.set_yticks(np.arange(len(channel_order)))
ax.set_yticklabels(channel_order)
ax.set_xticks(np.arange(len(family_score_matrix.columns)))
ax.set_xticklabels(list(family_score_matrix.columns), rotation=35, ha="right")

for i in range(family_score_values.shape[0]):
    for j in range(family_score_values.shape[1]):
        val = family_score_values[i, j]

        if np.isfinite(val) and (val >= 80 or val <= 20 or i < 5):
            txt = ax.text(
                j,
                i,
                f"{val:.0f}",
                ha="center",
                va="center",
                color=ACCENT,
                fontsize=6.0,
                alpha=0.95,
                fontweight="bold",
            )
            txt.set_path_effects([pe.withStroke(linewidth=1.4, foreground=BLACK)])

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Family-specific significance score")
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "05_surrogate_family_channel_score_heatmap.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=1000)


total_lifetime_delta = (
    results_df[results_df["metric"] == "total_lifetime"]
    .pivot(index="channel", columns="family_label", values="delta")
    .reindex(index=channel_order)
)

total_lifetime_delta_matrix = total_lifetime_delta.to_numpy(dtype=float)
delta_relative, delta_contrast_info_df = robust_metric_contrast_matrix(total_lifetime_delta_matrix)
delta_relative_clipped = np.clip(delta_relative, -3.0, 3.0)

fig, ax = plt.subplots(figsize=(10.8, 9.2), facecolor=BLACK)
style_ax(
    ax,
    title="Observed − Surrogate Mean H1 Total Lifetime",
    xlabel="Surrogate family",
    ylabel="Channel",
)

im = ax.imshow(
    delta_relative_clipped,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_DIVERGING_CMAP,
    vmin=-3,
    vmax=3,
)

ax.set_yticks(np.arange(len(channel_order)))
ax.set_yticklabels(channel_order)
ax.set_xticks(np.arange(len(total_lifetime_delta.columns)))
ax.set_xticklabels(list(total_lifetime_delta.columns), rotation=35, ha="right")

for i in range(total_lifetime_delta_matrix.shape[0]):
    for j in range(total_lifetime_delta_matrix.shape[1]):
        val = total_lifetime_delta_matrix[i, j]
        rel = delta_relative[i, j]

        if np.isfinite(val) and (abs(rel) >= 1.15 or i < 5 or i >= total_lifetime_delta_matrix.shape[0] - 3):
            txt = ax.text(
                j,
                i,
                f"{val:+.2f}",
                ha="center",
                va="center",
                color=ACCENT,
                fontsize=5.8,
                alpha=0.95,
                fontweight="bold",
            )
            txt.set_path_effects([pe.withStroke(linewidth=1.4, foreground=BLACK)])

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Within-family delta contrast")
force_cyan_text(ax)

fig.text(
    0.09,
    0.035,
    "Printed values are raw observed minus surrogate-mean H1 total lifetime. Color is contrast-scaled within each surrogate family.",
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

plot_path = os.path.join(mode_out_dir, "06_total_lifetime_observed_minus_surrogate_heatmap.png")
plt.tight_layout(rect=[0.0, 0.055, 1.0, 1.0])
save_render_close(fig, plot_path, width=1000)


top_tests_df = results_df.sort_values("neglog10_q_greater", ascending=False).head(28).copy()
top_tests_df["label"] = (
    top_tests_df["channel"]
    + " | "
    + top_tests_df["metric_label"]
    + " | "
    + top_tests_df["family_label"]
)

top_tests_df = top_tests_df.sort_values("neglog10_q_greater", ascending=True).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 9.0), facecolor=BLACK)
style_ax(
    ax,
    title="Top Individual Persistence Significance Tests",
    xlabel="-log10 family FDR q",
    ylabel="Channel | metric | surrogate family",
)

y = np.arange(len(top_tests_df))

ax.barh(
    y,
    top_tests_df["neglog10_q_greater"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.8,
    alpha=0.82,
)

ax.set_yticks(y)
ax.set_yticklabels(top_tests_df["label"], fontsize=8)

for yy, (_, row) in zip(y, top_tests_df.iterrows()):
    ax.text(
        row["neglog10_q_greater"] + 0.03,
        yy,
        f"z={row['z_regularized']:+.1f}",
        ha="left",
        va="center",
        color=ACCENT,
        fontsize=7,
    )

force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "07_top_individual_persistence_significance_tests.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=1150)


sim_vmin, sim_vmax = dynamic_limits(
    offdiag_values(clustered_similarity),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
    upper_bound=1.0,
)

fig, ax = plt.subplots(figsize=(9.6, 8.8), facecolor=BLACK)
style_ax(
    ax,
    title="Persistence Significance Signature Similarity Matrix",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    clustered_similarity,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=sim_vmin,
    vmax=sim_vmax,
)

ax.set_xticks(np.arange(len(clustered_channels)))
ax.set_xticklabels(clustered_channels, rotation=90)
ax.set_yticks(np.arange(len(clustered_channels)))
ax.set_yticklabels(clustered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Consensus significance signature similarity")
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "08_persistence_significance_signature_similarity_heatmap.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=950)


dist_vmin, dist_vmax = dynamic_limits(
    offdiag_values(clustered_distance),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

fig, ax = plt.subplots(figsize=(9.6, 8.8), facecolor=BLACK)
style_ax(
    ax,
    title="Persistence Significance Signature Distance Matrix",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    clustered_distance,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=dist_vmin,
    vmax=dist_vmax,
)

ax.set_xticks(np.arange(len(clustered_channels)))
ax.set_xticklabels(clustered_channels, rotation=90)
ax.set_yticks(np.arange(len(clustered_channels)))
ax.set_yticklabels(clustered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Consensus significance signature distance")
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "09_persistence_significance_signature_distance_heatmap.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=950)


fig, ax = plt.subplots(figsize=(8.8, 7.4), facecolor=BLACK)
style_ax(
    ax,
    title="MDS of Persistence Significance Signatures",
    xlabel="Persistence significance MDS 1",
    ylabel="Persistence significance MDS 2",
)

sc = ax.scatter(
    mds_df["persistence_significance_mds_1"],
    mds_df["persistence_significance_mds_2"],
    c=mds_df["persistence_significance_score"],
    s=165,
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
    edgecolor=ACCENT,
    linewidth=1.2,
    alpha=0.95,
)

for _, row in mds_df.iterrows():
    txt = ax.text(
        row["persistence_significance_mds_1"],
        row["persistence_significance_mds_2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.95,
        zorder=5,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

cbar = plt.colorbar(sc, ax=ax, shrink=0.82)
style_colorbar(cbar, "Persistence significance score")
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "10_mds_persistence_significance_signatures.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=950)


fig = plt.figure(figsize=(13.8, 8.2), facecolor=BLACK)
gs = GridSpec(
    1,
    2,
    width_ratios=[1.25, 0.85],
    wspace=0.25,
    figure=fig,
)

ax = fig.add_subplot(gs[0, 0])
rank_ax = fig.add_subplot(gs[0, 1])

style_ax(
    ax,
    title="Scalp Map of H1 Persistence Significance Score",
    grid=False,
)

head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2.2)
ax.add_patch(head)
ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2.2)

scalp_df = channel_summary_df[channel_summary_df["channel"].isin(scalp_xy)].copy()
scalp_df["x"] = scalp_df["channel"].map(lambda ch: scalp_xy[ch][0])
scalp_df["y"] = scalp_df["channel"].map(lambda ch: scalp_xy[ch][1])

scores = scalp_df["persistence_significance_score"].to_numpy(dtype=float)

if np.nanmax(scores) > np.nanmin(scores):
    norm_scores = (scores - np.nanmin(scores)) / (np.nanmax(scores) - np.nanmin(scores) + 1e-12)
else:
    norm_scores = np.full(len(scores), 0.5)

scalp_df["marker_size"] = 90.0 + 520.0 * norm_scores

scatter = ax.scatter(
    scalp_df["x"],
    scalp_df["y"],
    c=scalp_df["persistence_significance_score"],
    s=scalp_df["marker_size"],
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
    edgecolor=ACCENT,
    linewidth=1.25,
    alpha=0.92,
    zorder=4,
)

top_score_channels = scalp_df.sort_values("persistence_significance_score", ascending=False).head(8)["channel"].tolist()

for _, row in scalp_df.iterrows():
    ch = row["channel"]
    x0 = float(row["x"])
    y0 = float(row["y"])
    score = float(row["persistence_significance_score"])

    txt = ax.text(
        x0,
        y0,
        ch,
        ha="center",
        va="center",
        fontsize=8,
        color=ACCENT,
        fontweight="bold",
        zorder=6,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.1, foreground=BLACK)])

    if ch in top_score_channels:
        ax.scatter(
            [x0],
            [y0],
            s=float(row["marker_size"]) * 1.75,
            facecolors="none",
            edgecolors=ACCENT_SOFT,
            linewidth=1.8,
            alpha=0.95,
            zorder=5,
        )

        dx = 0.06 if x0 <= 0 else -0.06
        ha = "left" if x0 <= 0 else "right"

        ann = ax.text(
            x0 + dx,
            y0 + 0.055,
            f"{score:.0f}",
            ha=ha,
            va="center",
            fontsize=8,
            color=ACCENT_SOFT,
            fontweight="bold",
            zorder=7,
        )
        ann.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

ax.text(
    0,
    -1.30,
    "Bigger/brighter = stronger consensus persistence significance across surrogate nulls",
    ha="center",
    va="top",
    color=ACCENT,
    fontsize=9,
)

ax.set_xlim(-1.38, 1.38)
ax.set_ylim(-1.40, 1.36)
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])

cbar = plt.colorbar(scatter, ax=ax, shrink=0.78, pad=0.03)
style_colorbar(cbar, "Persistence significance score")

rank_side_df = scalp_df.sort_values("persistence_significance_score", ascending=True).copy()

style_ax(
    rank_ax,
    title="Persistence Significance Ranking",
    xlabel="Significance score",
    ylabel="Channel",
)

y = np.arange(len(rank_side_df))

rank_ax.barh(
    y,
    rank_side_df["persistence_significance_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.8,
    alpha=0.82,
)

rank_ax.set_yticks(y)
rank_ax.set_yticklabels(rank_side_df["channel"])
rank_ax.set_xlim(0, 105)

rank_ax.axvline(
    float(np.nanmean(rank_side_df["persistence_significance_score"])),
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.2,
    alpha=0.85,
    label=f"Mean={np.nanmean(rank_side_df['persistence_significance_score']):.1f}",
)

for yy, (_, row) in zip(y, rank_side_df.iterrows()):
    score = float(row["persistence_significance_score"])

    rank_ax.text(
        score + 1.2,
        yy,
        f"{score:.0f}",
        va="center",
        ha="left",
        color=ACCENT,
        fontsize=7,
        alpha=0.82,
    )

legend = rank_ax.legend(loc="lower right", framealpha=1.0, fontsize=8)
style_legend(legend)
force_cyan_text(rank_ax)

top_text = "Top consensus: " + ", ".join(
    [
        f"{row.channel}({row.persistence_significance_score:.0f})"
        for row in scalp_df.sort_values("persistence_significance_score", ascending=False).head(5).itertuples()
    ]
)

bottom_text = "Lowest consensus: " + ", ".join(
    [
        f"{row.channel}({row.persistence_significance_score:.0f})"
        for row in scalp_df.sort_values("persistence_significance_score", ascending=False).tail(5).itertuples()
    ]
)

fig.text(
    0.5,
    0.018,
    top_text + "    |    " + bottom_text,
    ha="center",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

plot_path = os.path.join(mode_out_dir, "11_scalp_map_persistence_significance_score.png")
plt.tight_layout(rect=[0.0, 0.045, 1.0, 1.0])
save_render_close(fig, plot_path, width=1150)


fig, ax = plt.subplots(figsize=(12, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="Regional H1 Persistence Significance Score",
    xlabel="Region",
    ylabel="Mean persistence significance score",
)

x = np.arange(len(region_summary_df))

ax.bar(
    x,
    region_summary_df["mean_persistence_significance_score"],
    yerr=region_summary_df["sem_persistence_significance_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
    capsize=7,
    error_kw={"elinewidth": 1.5, "capthick": 1.5, "ecolor": ACCENT},
)

for i, row in region_summary_df.iterrows():
    ax.text(
        i,
        row["mean_persistence_significance_score"] + row["sem_persistence_significance_score"],
        f"{row['mean_persistence_significance_score']:.1f}\nn={int(row['n_channels'])}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=9,
    )

ax.set_xticks(x)
ax.set_xticklabels(region_summary_df["region"], rotation=35, ha="right")
ax.set_ylim(0, 105)
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "12_regional_persistence_significance_score.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=1050)


coverage_rows = []

for family_data in loaded_families:
    family = family_data["family"]
    label = family_data["label"]
    summary_df = family_data["summary_df"]
    long_df = family_data["long_df"]

    elapsed_mean = float(np.nanmean(summary_df["elapsed_sec"])) if "elapsed_sec" in summary_df.columns else np.nan
    elapsed_sum = float(np.nansum(summary_df["elapsed_sec"])) if "elapsed_sec" in summary_df.columns else np.nan

    coverage_rows.append({
        "family": family,
        "family_label": label,
        "n_summary_channels": int(summary_df["channel"].nunique()),
        "n_long_rows": int(len(long_df)),
        "mean_elapsed_sec": elapsed_mean,
        "total_elapsed_sec": elapsed_sum,
    })

coverage_df = pd.DataFrame(coverage_rows)
coverage_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_input_coverage.csv")
coverage_df.to_csv(coverage_csv_path, index=False)

fig, ax = plt.subplots(figsize=(11.5, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="Persistence Significance Input Coverage",
    xlabel="Surrogate family",
    ylabel="Count",
)

x = np.arange(len(coverage_df))
width = 0.36

ax.bar(
    x - width / 2,
    coverage_df["n_summary_channels"],
    width=width,
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
    label="Summary channels",
)

ax.bar(
    x + width / 2,
    coverage_df["n_long_rows"],
    width=width,
    color=BLACK,
    edgecolor=ACCENT_SOFT,
    linewidth=1.4,
    alpha=0.92,
    label="Long-table rows",
)

ax.set_xticks(x)
ax.set_xticklabels(coverage_df["family_label"], rotation=25, ha="right")

for i, row in coverage_df.iterrows():
    ax.text(
        i - width / 2,
        row["n_summary_channels"],
        f"{int(row['n_summary_channels'])}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=9,
    )

    ax.text(
        i + width / 2,
        row["n_long_rows"],
        f"{int(row['n_long_rows'])}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=9,
    )

legend = ax.legend(loc="best", framealpha=1.0)
style_legend(legend)
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "13_surrogate_input_coverage_qc.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=1000)


runtime_available = coverage_df["mean_elapsed_sec"].notna().any()

if runtime_available:
    fig, ax = plt.subplots(figsize=(11.5, 6.2), facecolor=BLACK)
    style_ax(
        ax,
        title="Surrogate PH Runtime Summary",
        xlabel="Surrogate family",
        ylabel="Elapsed seconds",
    )

    x = np.arange(len(coverage_df))

    ax.bar(
        x,
        coverage_df["mean_elapsed_sec"],
        color=CYAN_FILL,
        edgecolor=ACCENT,
        linewidth=0.9,
        alpha=0.82,
        label="Mean per channel",
    )

    for i, row in coverage_df.iterrows():
        if np.isfinite(row["mean_elapsed_sec"]):
            ax.text(
                i,
                row["mean_elapsed_sec"],
                f"{row['mean_elapsed_sec']:.1f}s",
                ha="center",
                va="bottom",
                color=ACCENT,
                fontsize=9,
            )

    ax.set_xticks(x)
    ax.set_xticklabels(coverage_df["family_label"], rotation=25, ha="right")

    legend = ax.legend(loc="best", framealpha=1.0)
    style_legend(legend)
    force_cyan_text(ax)

    plot_path = os.path.join(mode_out_dir, "14_surrogate_runtime_qc.png")
    plt.tight_layout()
    save_render_close(fig, plot_path, width=1000)


log("\n" + "=" * 72)
log("Finished H1 Persistence Significance Testing")
log(f"Surrogate families loaded: {', '.join([f['label'] for f in loaded_families])}")
log(f"Output directory: {mode_out_dir}")
log("Saved CSVs:")
log(f"  {tests_csv_path}")
log(f"  {consensus_csv_path}")
log(f"  {channel_summary_csv_path}")
log(f"  {region_summary_csv_path}")
log(f"  {family_channel_csv_path}")
log(f"  {signature_csv_path}")
log(f"  {signature_similarity_csv_path}")
log(f"  {signature_distance_csv_path}")
log(f"  {mds_csv_path}")
log(f"  {coverage_csv_path}")
log(f"Saved NPZ: {npz_path}")
log("Saved plots:")
log("01_persistence_significance_score_ranking.png")
log("02_consensus_persistence_significance_heatmap.png")
log("03_surrogate_family_agreement_heatmap.png")
log("04_consensus_persistence_effect_z_heatmap.png")
log("05_surrogate_family_channel_score_heatmap.png")
log("06_total_lifetime_observed_minus_surrogate_heatmap.png")
log("07_top_individual_persistence_significance_tests.png")
log("08_persistence_significance_signature_similarity_heatmap.png")
log("09_persistence_significance_signature_distance_heatmap.png")
log("10_mds_persistence_significance_signatures.png")
log("11_scalp_map_persistence_significance_score.png")
log("12_regional_persistence_significance_score.png")
log("13_surrogate_input_coverage_qc.png")
if runtime_available:
    log("14_surrogate_runtime_qc.png")
log("=" * 72)
```

Color, ordering, or ranking can make patterns easier to see but does not create additional statistical evidence beyond the values and uncertainty measures being plotted.</p></div>

### 1. Bootstrap confidence intervals for persistence diagrams

For a persistence statistic \(T(D)\) (here H1 total persistence, maximum lifetime, and persistence entropy), a moving-block bootstrap resamples contiguous time blocks to respect short-range dependence. If \(T^{*(b)}\) is the statistic from bootstrap replicate \(b\), a percentile \((1-\alpha)\) interval is \([Q_{\alpha/2},Q_{1-\alpha/2}]\). The interval quantifies finite-sample uncertainty of the chosen statistic, not uncertainty of every birth/death pair individually.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Bootstrap analysis estimates sampling uncertainty by repeatedly resampling the available observations and recomputing the statistic. The spread of the resampled statistics shows how sensitive the result is to which observations happened to be included. A bootstrap confidence interval is therefore an empirical uncertainty estimate; it does not remove bias in the original estimator and assumes that the resampling scheme adequately represents the data-generating structure.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 1: Bootstrap confidence intervals for persistence diagrams

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_001():

    i = 1
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _moving_block_boot(y,rng,block=128):
        y=np.asarray(y); n=len(y); out=[]
        while sum(map(len,out))<n:
            st=int(rng.integers(0,max(1,n-block+1))); out.append(y[st:st+block])
        return np.concatenate(out)[:n]

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    # Method mathematics and output.
    rng = np.random.default_rng(1000 + i)
    y = _topo_pc1
    vals = []
    for _ in range(_TOPO_SPEED['boot']):
        yy = _moving_block_boot(y, rng, block=max(32, int(0.25 * _topo_fs)))
        vals.append(_feat(_ph_rips(_delay_embed(yy, 3, max(1, int(0.01 * _topo_fs))), 1)))
    A = np.vstack(vals)
    q = np.quantile(A, [0.025, 0.5, 0.975], axis=0)
    out = pd.DataFrame(q, index=['2.5%', 'median', '97.5%'], columns=['H1_total', 'H1_max', 'H1_entropy', 'H1_count'])
    display(out)
    fig, ax = plt.subplots(figsize=(9, 4.4), facecolor=TOPO_BG)
    bp = ax.boxplot([A[:, j] for j in range(A.shape[1])], labels=out.columns, patch_artist=True, medianprops={'color': TOPO_CYAN_SOFT, 'linewidth': 1.6}, whiskerprops={'color': TOPO_ACCENT}, capprops={'color': TOPO_ACCENT}, flierprops={'markeredgecolor': TOPO_ACCENT, 'markerfacecolor': TOPO_BG})
    for box in bp['boxes']:
        box.set(facecolor=TOPO_BG, edgecolor=TOPO_ACCENT, linewidth=1.2)
    ax.set_title('Bootstrap distributions of H1 persistence statistics')
    ax.set_ylabel('bootstrap statistic')
    ax.grid(True, axis='y', color=TOPO_CYAN_DIM, alpha=0.25)
    plt.xticks(rotation=20, ha='right')
    plt.tight_layout()
    plt.show()


_topology_method_001()
del _topology_method_001
```

### 2. Permutation testing for persistence diagrams

A permutation test compares a topological discrepancy \(\Delta\) between two time segments with the null distribution obtained by randomly permuting segment labels. The Monte-Carlo p-value is \(p=(1+\#\{\Delta_b\ge\Delta_{obs}\})/(B+1)\). Diagram discrepancy is summarized here by an explicitly labelled lifetime-spectrum distance rather than being misreported as an exact bottleneck distance.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The bottleneck distance compares two persistence diagrams by matching their birth-death points, with unmatched features allowed to pair with the diagonal. It then looks at the largest matching cost and chooses the matching that makes this worst cost as small as possible. Because it is controlled by the single largest discrepancy, it emphasizes worst-case topological change rather than the sum of many small differences.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> Because the calculation is performed on finite windows or segments, the reported value is a local summary of that selected data interval; a different interval can legitimately produce a different estimate if the EEG is nonstationary. Statistical significance should be read relative to the stated null model and testing procedure; it is strongest when accompanied by an effect size, uncertainty estimate, and correction for the number of comparisons when many tests are run.</p></div>


```python
# Method 2: Permutation testing for persistence diagrams

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_002():

    i = 2
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    # Method mathematics and output.
    rng = np.random.default_rng(1000 + i)
    y = _topo_pc1
    h = len(y) // 2
    obs = np.linalg.norm(_feat(_ph_rips(_delay_embed(y[:h]), 1)) - _feat(_ph_rips(_delay_embed(y[h:]), 1)))
    lab = np.r_[np.zeros(h, int), np.ones(len(y) - h, int)]
    vals = []
    for _ in range(_TOPO_SPEED['sur']):
        rng.shuffle(lab)
        vals.append(np.linalg.norm(_feat(_ph_rips(_delay_embed(y[lab == 0]), 1)) - _feat(_ph_rips(_delay_embed(y[lab == 1]), 1))))
    p = (1 + np.sum(np.asarray(vals) >= obs)) / (len(vals) + 1)
    print('observed discrepancy=', obs, ' permutation p=', p)
    fig, ax = plt.subplots(figsize=(8.5, 4.2), facecolor=TOPO_BG)
    ax.hist(vals, bins=min(20, max(5, len(vals))), color=TOPO_CYAN_FILL, edgecolor=TOPO_ACCENT, alpha=0.7)
    ax.axvline(obs, color=TOPO_CYAN_SOFT, linewidth=2.0, linestyle='--', label=f'observed = {obs:.4g}')
    ax.set_title('Permutation null distribution of topological discrepancy')
    ax.set_xlabel('discrepancy')
    ax.set_ylabel('count')
    ax.grid(True, axis='y', color=TOPO_CYAN_DIM, alpha=0.25)
    ax.legend()
    plt.tight_layout()
    plt.show()


_topology_method_002()
del _topology_method_002
```

### 3. FDR-corrected topological significance testing

For multiple topological tests with p-values \(p_{(1)}\le\cdots\le p_{(m)}\), Benjamini–Hochberg controls the false-discovery rate through \(q_{(i)}=\min_{j\ge i} mp_{(j)}/j\), truncated at 1. Each window is tested against phase-randomized surrogates using H1 total persistence.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> When many hypotheses are tested, some small p-values occur by chance. Benjamini-Hochberg false-discovery-rate control ranks the p-values and chooses a threshold designed to limit the expected fraction of false discoveries among the results called significant. FDR control is less strict than controlling the probability of any false positive, which gives it more power in large exploratory analyses. The correction should be applied to a clearly defined family of related tests.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> Because the calculation is performed on finite windows or segments, the reported value is a local summary of that selected data interval; a different interval can legitimately produce a different estimate if the EEG is nonstationary. Statistical significance should be read relative to the stated null model and testing procedure; it is strongest when accompanied by an effect size, uncertainty estimate, and correction for the number of comparisons when many tests are run.</p></div>


```python
# Method 3: FDR-corrected topological significance testing

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_003():

    i = 3
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _bh(p):
        p=np.asarray(p,float); n=len(p); order=np.argsort(p); q=np.empty(n,float); prev=1.0
        for r in range(n-1,-1,-1):
            i=order[r]; prev=min(prev,p[i]*n/(r+1)); q[i]=prev
        return np.clip(q,0,1)

    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _phase_randomize(y,rng):
        y=np.asarray(y,float); Y=np.fft.rfft(y); mag=np.abs(Y); ph=np.angle(Y)
        if len(ph)>2: ph[1:-1]=rng.uniform(-np.pi,np.pi,len(ph)-2)
        return np.fft.irfft(mag*np.exp(1j*ph),n=len(y))

    def _windows(y,nw=None):
        y=np.asarray(y); nw=int(nw or _TOPO_SPEED['windows']); L=max(300,len(y)//max(4,nw//2)); step=max(1,(len(y)-L)//max(1,nw-1))
        ans=[]
        for st in range(0,max(1,len(y)-L+1),step):
            ans.append((st,st+L,y[st:st+L]))
            if len(ans)>=nw: break
        return ans

    def _window_topo(y,m=3,tau=2):
        rows=[]; diagrams=[]
        for st,en,w in _windows(y):
            X=_delay_embed(w,m,tau,max_points=min(_TOPO_SPEED['points'],70 if FAST_OPTION=='ultra' else _TOPO_SPEED['points']))
            dg=_ph_rips(X,1); rows.append([st/_topo_fs,en/_topo_fs,*_feat(dg)]); diagrams.append(dg)
        return pd.DataFrame(rows,columns=['start_s','end_s','H1_total','H1_max','H1_entropy','H1_count']),diagrams

    # Method mathematics and output.
    rng = np.random.default_rng(1000 + i)
    y = _topo_pc1
    df, dgs = _window_topo(y)
    p = []
    for (_, _, w), dg in zip(_windows(y), dgs):
        obs = _feat(dg)[0]
        null = []
        for _ in range(max(4, _TOPO_SPEED['sur'] // 3)):
            null.append(_feat(_ph_rips(_delay_embed(_phase_randomize(w, rng)), 1))[0])
        p.append((1 + np.sum(np.asarray(null) >= obs)) / (len(null) + 1))
    df['p'] = p
    df['q_BH'] = _bh(p)
    out = df[['start_s', 'H1_total', 'p', 'q_BH']]
    display(out)
    fig, ax = plt.subplots(figsize=(9, 4.2), facecolor=TOPO_BG)
    ax.plot(out['start_s'], out['p'], color=TOPO_CYAN_FILL, linewidth=1.6, label='raw p')
    ax.plot(out['start_s'], out['q_BH'], color=TOPO_CYAN_SOFT, linewidth=1.6, linestyle='--', label='BH q')
    ax.axhline(0.05, color=TOPO_ACCENT, linewidth=1.0, linestyle=':', alpha=0.8)
    ax.set_title('Window-wise topological significance')
    ax.set_xlabel('window start (s)')
    ax.set_ylabel('p / q')
    ax.set_ylim(bottom=0)
    ax.grid(True, color=TOPO_CYAN_DIM, alpha=0.25)
    ax.legend()
    plt.tight_layout()
    plt.show()


_topology_method_003()
del _topology_method_003
```

### 4. Null distribution of H1 total persistence

The null distribution of H1 total persistence is generated from Fourier phase-randomized surrogates. Randomizing phases preserves the global Fourier magnitude spectrum while destroying deterministic phase organization. For finite H1 bars \((b_i,d_i)\), total persistence is \(TP_1=\sum_i(d_i-b_i)\).

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> A surrogate test creates artificial signals that preserve selected simple properties of the original data while deliberately removing the structure being tested. IAAFT surrogates approximately preserve the amplitude distribution and power spectrum, and therefore much of the linear autocorrelation structure, while randomizing nonlinear temporal organization. The observed statistic is then compared with the surrogate distribution. A small surrogate-based p-value means the observed statistic is unusual under that null model; it does not by itself identify the biological mechanism responsible for the difference.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> Statistical significance should be read relative to the stated null model and testing procedure; it is strongest when accompanied by an effect size, uncertainty estimate, and correction for the number of comparisons when many tests are run.</p></div>


```python
# Method 4: Null distribution of H1 total persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_004():

    i = 4
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _phase_randomize(y,rng):
        y=np.asarray(y,float); Y=np.fft.rfft(y); mag=np.abs(Y); ph=np.angle(Y)
        if len(ph)>2: ph[1:-1]=rng.uniform(-np.pi,np.pi,len(ph)-2)
        return np.fft.irfft(mag*np.exp(1j*ph),n=len(y))

    def _plot_line(x,y,title,xlabel='x',ylabel='value'):
        fig,ax=plt.subplots(figsize=(8,3.8),facecolor=TOPO_BG)
        ax.plot(x,y,color=TOPO_ACCENT,lw=1.7)
        ax.set_title(title); ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.28,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )
    _TOPO_DREF = _ph_rips(_TOPO_XREF, 1)

    # Method mathematics and output.
    rng = np.random.default_rng(1000 + i)
    y = _topo_pc1
    ref = _TOPO_DREF
    obs = _feat(ref)[0 if i == 4 else 1]
    vals = []
    for _ in range(_TOPO_SPEED['sur']):
        vals.append(_feat(_ph_rips(_delay_embed(_phase_randomize(y, rng)), 1))[0 if i == 4 else 1])
    _plot_line(np.arange(len(vals)), vals, ('H1 total' if i == 4 else 'maximum lifetime') + ' surrogate null', 'surrogate', 'statistic')
    print('observed=', obs, 'null 95%=', np.quantile(vals, 0.95))


_topology_method_004()
del _topology_method_004
```

### 5. Null distribution of maximum persistence lifetime

Maximum H1 lifetime is \(L_{max}=\max_i(d_i-b_i)\). Phase-randomized surrogates provide a spectrum-preserving null for whether the longest observed loop is unusually persistent.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> A surrogate test creates artificial signals that preserve selected simple properties of the original data while deliberately removing the structure being tested. IAAFT surrogates approximately preserve the amplitude distribution and power spectrum, and therefore much of the linear autocorrelation structure, while randomizing nonlinear temporal organization. The observed statistic is then compared with the surrogate distribution. A small surrogate-based p-value means the observed statistic is unusual under that null model; it does not by itself identify the biological mechanism responsible for the difference.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> Statistical significance should be read relative to the stated null model and testing procedure; it is strongest when accompanied by an effect size, uncertainty estimate, and correction for the number of comparisons when many tests are run.</p></div>

### 6. Repeated-subsample persistence stability

Subsample stability asks whether persistence is robust to point-cloud thinning. For subsets \(X_s\subset X\), the notebook compares their H1 lifetime spectra with the reference diagram. A stable analysis should show small discrepancies as sampling density is reduced.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 6: Repeated-subsample persistence stability

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_006():

    i = 6
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _diag_dist(a,b):
        # Lifetime-spectrum distance: stable summary proxy, not bottleneck/Wasserstein distance.
        la=np.sort(_life(a))[::-1]; lb=np.sort(_life(b))[::-1]; n=max(len(la),len(lb),1)
        aa=np.pad(la,(0,n-len(la))); bb=np.pad(lb,(0,n-len(lb)))
        return float(np.linalg.norm(aa-bb))

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _plot_line(x,y,title,xlabel='x',ylabel='value'):
        fig,ax=plt.subplots(figsize=(8,3.8),facecolor=TOPO_BG)
        ax.plot(x,y,color=TOPO_ACCENT,lw=1.7)
        ax.set_title(title); ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.28,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )
    _TOPO_DREF = _ph_rips(_TOPO_XREF, 1)

    # Method mathematics and output.
    rng = np.random.default_rng(1000 + i)
    ref = _TOPO_DREF
    X = _TOPO_XREF
    fracs = [0.4, 0.55, 0.7, 0.85, 1.0]
    ds = []
    for f in fracs:
        vals = []
        for _ in range(max(3, _TOPO_SPEED['proj'])):
            idx = np.sort(rng.choice(len(X), max(8, int(f * len(X))), replace=False))
            vals.append(_diag_dist(ref[1], _ph_rips(X[idx], 1)[1]))
        ds.append(np.mean(vals))
    _plot_line(fracs, ds, 'Repeated-subsample persistence stability', 'retained fraction', 'H1 lifetime-spectrum distance')


_topology_method_006()
del _topology_method_006
```

### 7. Projection-stability persistence audit

Projection stability tests how much topology changes after random linear projection. A Gaussian projection \(Y=XG/\sqrt{k}\) approximately preserves pairwise distances when \(k\) is sufficiently large (Johnson–Lindenstrauss principle). Persistence summaries are compared before and after projection.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 7: Projection-stability persistence audit

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_007():

    i = 7
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _diag_dist(a,b):
        # Lifetime-spectrum distance: stable summary proxy, not bottleneck/Wasserstein distance.
        la=np.sort(_life(a))[::-1]; lb=np.sort(_life(b))[::-1]; n=max(len(la),len(lb),1)
        aa=np.pad(la,(0,n-len(la))); bb=np.pad(lb,(0,n-len(lb)))
        return float(np.linalg.norm(aa-bb))

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _plot_line(x,y,title,xlabel='x',ylabel='value'):
        fig,ax=plt.subplots(figsize=(8,3.8),facecolor=TOPO_BG)
        ax.plot(x,y,color=TOPO_ACCENT,lw=1.7)
        ax.set_title(title); ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.28,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )
    _TOPO_DREF = _ph_rips(_TOPO_XREF, 1)

    # Method mathematics and output.
    rng = np.random.default_rng(1000 + i)
    ref = _TOPO_DREF
    X = _TOPO_XREF
    dims = [2, 3, 4, 5]
    ds = []
    for k in dims:
        vals = []
        for _ in range(_TOPO_SPEED['proj']):
            G = rng.normal(size=(X.shape[1], k)) / np.sqrt(k)
            vals.append(_diag_dist(ref[1], _ph_rips(X @ G, 1)[1]))
        ds.append(np.mean(vals))
    _plot_line(dims, ds, 'Projection-stability persistence audit', 'projection dimension', 'H1 discrepancy')


_topology_method_007()
del _topology_method_007
```

### 8. UMAP-vs-PCA topology comparison

UMAP constructs a fuzzy k-nearest-neighbour simplicial set and optimizes a low-dimensional representation to match fuzzy memberships. When `umap-learn` is available this cell uses it directly. Otherwise it uses the same fuzzy-neighbour construction followed by spectral coordinates and clearly labels that fallback as **not the full UMAP cross-entropy optimization**. Topology is then compared with PCA.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Principal component analysis rotates the coordinate system so that the first axis captures the greatest sample variance, the second captures the greatest remaining variance subject to being orthogonal to the first, and so on. The directions are eigenvectors of the covariance matrix and their eigenvalues quantify variance along those directions. PCA is therefore a linear dimensionality-reduction method: it can summarize redundant channels or features, but it does not guarantee that the retained components correspond to distinct physiological sources.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 8: UMAP-vs-PCA topology comparison

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_008():

    i = 8
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _umap_or_fuzzy_spectral(X):
        try:
            import umap
            return umap.UMAP(n_components=2,n_neighbors=min(15,len(X)-1),random_state=0).fit_transform(X), 'UMAP'
        except Exception:
            # Explicit fallback: UMAP fuzzy-neighborhood graph followed by spectral coordinates; no SGD cross-entropy step.
            k=min(15,max(2,len(X)-1)); nn=NearestNeighbors(n_neighbors=k+1).fit(X); dist,ind=nn.kneighbors(X)
            rho=dist[:,1]; sigma=np.maximum(np.mean(dist[:,2:],axis=1),TOPO_EPS)
            W=np.zeros((len(X),len(X)))
            for i in range(len(X)):
                for dij,j in zip(dist[i,1:],ind[i,1:]): W[i,j]=math.exp(-max(0.0,dij-rho[i])/sigma[i])
            W=W+W.T-W*W.T
            emb=SpectralEmbedding(n_components=2,affinity='precomputed',random_state=0).fit_transform(W)
            return emb,'fuzzy-kNN spectral fallback (UMAP package unavailable)'

    # Method mathematics and output.
    y = _topo_pc1
    X = _delay_embed(y, 5, max(1, int(0.01 * _topo_fs)))
    pca = PCA(2).fit_transform(X)
    um, label = _umap_or_fuzzy_spectral(X)
    a = _feat(_ph_rips(pca, 1))
    b = _feat(_ph_rips(um, 1))
    out = pd.DataFrame([a, b], index=['PCA', label], columns=['H1_total', 'H1_max', 'entropy', 'count'])
    display(out)
    fig, axs = plt.subplots(2, 2, figsize=(9, 6.5), facecolor=TOPO_BG)
    axs = np.ravel(axs)
    for j, col in enumerate(out.columns):
        axs[j].bar(np.arange(len(out)), out[col].to_numpy(float), color=TOPO_CYAN_FILL, edgecolor=TOPO_ACCENT, alpha=0.75)
        axs[j].set_xticks(np.arange(len(out)))
        axs[j].set_xticklabels(out.index, rotation=20, ha='right')
        axs[j].set_title(col)
        axs[j].grid(True, axis='y', color=TOPO_CYAN_DIM, alpha=0.22)
    fig.suptitle('PCA versus nonlinear projection topology')
    plt.tight_layout()
    plt.show()


_topology_method_008()
del _topology_method_008
```

### 9. Original-embedding-vs-projected topology comparison

Projection distortion is assessed by comparing persistent H1 from the original delay cloud with a PCA projection. Because orthogonal projection can contract distances and destroy loops, agreement is evaluated through persistence summaries rather than assumed.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Principal component analysis rotates the coordinate system so that the first axis captures the greatest sample variance, the second captures the greatest remaining variance subject to being orthogonal to the first, and so on. This often makes the role of each term in the formula much easier to follow.</p></div>

### 10. Persistent homology across embedding dimension

For delay embeddings \(X_m(t)=[x_t,x_{t-\tau},\ldots,x_{t-(m-1)\tau}]\), persistent homology is recomputed as embedding dimension \(m\) varies. Stabilization of H1 summaries suggests that the detected loop structure is not an artifact of one embedding dimension.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Delay embedding turns one measured time series into a sequence of state-space vectors by combining the present sample with delayed copies of the same signal. The idea is that the hidden state of a dynamical system can sometimes be reconstructed from a single observable when enough delayed coordinates are used. The delay controls how far apart in time the coordinates are, while the embedding dimension controls how many delayed coordinates are included. Too small a delay makes coordinates nearly redundant; too large a delay can make them weakly related. Too small an embedding dimension can fold distinct states onto one another.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance. The symbol τ (tau) commonly denotes a time delay, scale, or characteristic time. Here it should be read according to the definition immediately preceding the formula.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 10: Persistent homology across embedding dimension

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_010():

    i = 10
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _plot_line(x,y,title,xlabel='x',ylabel='value'):
        fig,ax=plt.subplots(figsize=(8,3.8),facecolor=TOPO_BG)
        ax.plot(x,y,color=TOPO_ACCENT,lw=1.7)
        ax.set_title(title); ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.28,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method mathematics and output.
    y = _topo_pc1
    ms = list(range(2, _TOPO_SPEED['emb_max'] + 1))
    vals = []
    for m in ms:
        vals.append(_feat(_ph_rips(_delay_embed(y, m, max(1, int(0.01 * _topo_fs))), 1))[0])
    _plot_line(ms, vals, 'Persistent homology across embedding dimension', 'embedding dimension m', 'H1 total persistence')


_topology_method_010()
del _topology_method_010
```

### 11. Persistent homology across delay parameter

Delay sensitivity varies \(\tau\) in the delay map \(X(t)=[x_t,x_{t-\tau},x_{t-2\tau}]\). Too-small delays produce near-diagonal clouds; too-large delays can decorrelate coordinates. Persistence should therefore be interpreted together with this delay sweep.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Persistent homology summarizes how topological features appear and disappear as a distance or filtration threshold increases. In dimension 0 the features are connected components; in dimension 1 they are loops; higher dimensions represent higher-order cavities. Each feature receives a birth scale and a death scale, and its lifetime is the difference between them. Long lifetimes are often treated as more structurally important than very short ones, although noise and sampling density can still affect the diagram.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance. The symbol τ (tau) commonly denotes a time delay, scale, or characteristic time. Here it should be read according to the definition immediately preceding the formula.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>

### 12. Persistent homology across subsampling density

This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 13: Persistent homology across frequency bands

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_013():

    i = 13
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _filter_band(y,lo,hi):
        sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_topo_fs,output='sos'); return signal.sosfiltfilt(sos,y)

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    # Method mathematics and output.
    y = _topo_pc1
    rows = []
    for bn, (lo, hi) in _TOPO_BANDS.items():
        if hi >= _topo_fs / 2:
            continue
        yy = _filter_band(y, lo, hi)
        rows.append([bn, *_feat(_ph_rips(_delay_embed(yy), 1))])
    out = pd.DataFrame(rows, columns=['band', 'H1_total', 'H1_max', 'entropy', 'count'])
    display(out)
    fig, axs = plt.subplots(2, 2, figsize=(9.5, 6.8), facecolor=TOPO_BG)
    axs = np.ravel(axs)
    for j, col in enumerate(['H1_total', 'H1_max', 'entropy', 'count']):
        axs[j].bar(out['band'], out[col].to_numpy(float), color=TOPO_CYAN_FILL, edgecolor=TOPO_ACCENT, alpha=0.75)
        axs[j].set_title(col)
        axs[j].tick_params(axis='x', rotation=25)
        axs[j].grid(True, axis='y', color=TOPO_CYAN_DIM, alpha=0.22)
    fig.suptitle('Persistent-homology statistics across frequency bands')
    plt.tight_layout()
    plt.show()


_topology_method_013()
del _topology_method_013
```

### 14. Betti curves

The Betti curve is \(\beta_k(\epsilon)=\#\{(b_i,d_i): b_i\le\epsilon<d_i\}\). It records the number of k-dimensional homology classes alive at each filtration scale and is computed directly from the persistence diagram.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Persistent homology summarizes how topological features appear and disappear as a distance or filtration threshold increases. In dimension 0 the features are connected components; in dimension 1 they are loops; higher dimensions represent higher-order cavities. Each feature receives a birth scale and a death scale, and its lifetime is the difference between them. Long lifetimes are often treated as more structurally important than very short ones, although noise and sampling density can still affect the diagram.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance. The symbol ε (epsilon) usually represents a small threshold or radius. Changing it changes what the method counts as sufficiently close or similar.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 14: Betti curves

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_014():

    i = 14
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _betti_curve(dgm,epsgrid):
        d=np.asarray(dgm,float).reshape(-1,2)
        if not len(d): return np.zeros_like(epsgrid,float)
        return np.array([np.sum((d[:,0]<=e)&(e<d[:,1])) for e in epsgrid],float)

    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _plot_line(x,y,title,xlabel='x',ylabel='value'):
        fig,ax=plt.subplots(figsize=(8,3.8),facecolor=TOPO_BG)
        ax.plot(x,y,color=TOPO_ACCENT,lw=1.7)
        ax.set_title(title); ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.28,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )
    _TOPO_DREF = _ph_rips(_TOPO_XREF, 1)

    # Method mathematics and output.
    ref = _TOPO_DREF
    finite = np.vstack([_finite(d) for d in ref if len(_finite(d))]) if any((len(_finite(d)) for d in ref)) else np.array([[0, 1.0]])
    e = np.linspace(0, np.max(finite[:, 1]), 120)
    _plot_line(e, _betti_curve(ref[1], e), 'Betti curve H1', 'filtration ε', 'β1(ε)')


_topology_method_014()
del _topology_method_014
```

### 15. Time-resolved Betti curves

Time-resolved Betti curves apply \(\beta_1(t,\epsilon)\) to sliding windows. The resulting time × filtration matrix reveals when loops appear, disappear, or shift in characteristic scale.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Persistent homology summarizes how topological features appear and disappear as a distance or filtration threshold increases. In dimension 0 the features are connected components; in dimension 1 they are loops; higher dimensions represent higher-order cavities. Each feature receives a birth scale and a death scale, and its lifetime is the difference between them. Long lifetimes are often treated as more structurally important than very short ones, although noise and sampling density can still affect the diagram.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance. The symbol ε (epsilon) usually represents a small threshold or radius. Changing it changes what the method counts as sufficiently close or similar.</p><p><b>How to interpret this cell.</b> Because the calculation is performed on finite windows or segments, the reported value is a local summary of that selected data interval; a different interval can legitimately produce a different estimate if the EEG is nonstationary.</p></div>


```python
# Method 15: Time-resolved Betti curves

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_015():

    i = 15
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _betti_curve(dgm,epsgrid):
        d=np.asarray(dgm,float).reshape(-1,2)
        if not len(d): return np.zeros_like(epsgrid,float)
        return np.array([np.sum((d[:,0]<=e)&(e<d[:,1])) for e in epsgrid],float)

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _windows(y,nw=None):
        y=np.asarray(y); nw=int(nw or _TOPO_SPEED['windows']); L=max(300,len(y)//max(4,nw//2)); step=max(1,(len(y)-L)//max(1,nw-1))
        ans=[]
        for st in range(0,max(1,len(y)-L+1),step):
            ans.append((st,st+L,y[st:st+L]))
            if len(ans)>=nw: break
        return ans

    def _window_topo(y,m=3,tau=2):
        rows=[]; diagrams=[]
        for st,en,w in _windows(y):
            X=_delay_embed(w,m,tau,max_points=min(_TOPO_SPEED['points'],70 if FAST_OPTION=='ultra' else _TOPO_SPEED['points']))
            dg=_ph_rips(X,1); rows.append([st/_topo_fs,en/_topo_fs,*_feat(dg)]); diagrams.append(dg)
        return pd.DataFrame(rows,columns=['start_s','end_s','H1_total','H1_max','H1_entropy','H1_count']),diagrams

    # Method mathematics and output.
    y = _topo_pc1
    df, dgs = _window_topo(y)
    mx = max([np.max(_finite(d[1])[:, 1]) if len(_finite(d[1])) else 1 for d in dgs])
    e = np.linspace(0, mx, 60)
    C = np.vstack([_betti_curve(d[1], e) for d in dgs])
    fig, ax = plt.subplots(figsize=(8, 4))
    im = ax.imshow(C.T, origin='lower', aspect='auto', extent=[df.start_s.min(), df.end_s.max(), e[0], e[-1]], cmap=TOPO_CMAP)
    ax.set_xlabel('time (s)')
    ax.set_ylabel('filtration ε')
    ax.set_title('Time-resolved Betti curves' if i == 15 else 'CROCKER plot: β1(t,ε)')
    plt.colorbar(im, ax=ax, label='β1')
    plt.show()


_topology_method_015()
del _topology_method_015
```

### 16. Persistence lifetime density curves

The persistence-lifetime density is the empirical distribution of \(\ell_i=d_i-b_i\). This cell displays a normalized lifetime density rather than treating many short bars and a few long bars as equivalent.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 16: Persistence lifetime density curves

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_016():

    i = 16
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _plot_line(x,y,title,xlabel='x',ylabel='value'):
        fig,ax=plt.subplots(figsize=(8,3.8),facecolor=TOPO_BG)
        ax.plot(x,y,color=TOPO_ACCENT,lw=1.7)
        ax.set_title(title); ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.28,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )
    _TOPO_DREF = _ph_rips(_TOPO_XREF, 1)

    # Method mathematics and output.
    ref = _TOPO_DREF
    L = _life(ref[1])
    bins = np.linspace(0, max(L.max() if len(L) else 1, TOPO_EPS), 30)
    h, b = np.histogram(L, bins=bins, density=True)
    _plot_line((b[:-1] + b[1:]) / 2, h, 'Persistence lifetime density', 'lifetime', 'density')


_topology_method_016()
del _topology_method_016
```

### 17. Persistence entropy curves

Persistence entropy uses normalized lifetimes \(p_i=\ell_i/\sum_j\ell_j\) and \(H=-\sum_i p_i\log p_i\). The time-resolved entropy curve measures how evenly topological persistence is distributed among H1 classes.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 17: Persistence entropy curves

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_017():

    i = 17
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _plot_line(x,y,title,xlabel='x',ylabel='value'):
        fig,ax=plt.subplots(figsize=(8,3.8),facecolor=TOPO_BG)
        ax.plot(x,y,color=TOPO_ACCENT,lw=1.7)
        ax.set_title(title); ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.28,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _windows(y,nw=None):
        y=np.asarray(y); nw=int(nw or _TOPO_SPEED['windows']); L=max(300,len(y)//max(4,nw//2)); step=max(1,(len(y)-L)//max(1,nw-1))
        ans=[]
        for st in range(0,max(1,len(y)-L+1),step):
            ans.append((st,st+L,y[st:st+L]))
            if len(ans)>=nw: break
        return ans

    def _window_topo(y,m=3,tau=2):
        rows=[]; diagrams=[]
        for st,en,w in _windows(y):
            X=_delay_embed(w,m,tau,max_points=min(_TOPO_SPEED['points'],70 if FAST_OPTION=='ultra' else _TOPO_SPEED['points']))
            dg=_ph_rips(X,1); rows.append([st/_topo_fs,en/_topo_fs,*_feat(dg)]); diagrams.append(dg)
        return pd.DataFrame(rows,columns=['start_s','end_s','H1_total','H1_max','H1_entropy','H1_count']),diagrams

    # Method mathematics and output.
    y = _topo_pc1
    df, _ = _window_topo(y)
    _plot_line((df.start_s + df.end_s) / 2, df.H1_entropy, 'Persistence entropy curve', 'time (s)', 'H1 entropy')


_topology_method_017()
del _topology_method_017
```

### 18. CROCKER plots

A CROCKER plot stores Betti numbers over time and filtration scale: \(C(t,\epsilon)=\beta_k(D_t;\epsilon)\). It is a compact topological state portrait, not a persistence diagram itself.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Persistent homology summarizes how topological features appear and disappear as a distance or filtration threshold increases. In dimension 0 the features are connected components; in dimension 1 they are loops; higher dimensions represent higher-order cavities. Each feature receives a birth scale and a death scale, and its lifetime is the difference between them. Long lifetimes are often treated as more structurally important than very short ones, although noise and sampling density can still affect the diagram.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance. The symbol ε (epsilon) usually represents a small threshold or radius. Changing it changes what the method counts as sufficiently close or similar.</p><p><b>How to interpret this cell.</b> The visualization is a way of displaying already-computed numbers. Color, ordering, or ranking can make patterns easier to see but does not create additional statistical evidence beyond the values and uncertainty measures being plotted.</p></div>


```python
# Method 18: CROCKER plots

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_018():

    i = 18
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _betti_curve(dgm,epsgrid):
        d=np.asarray(dgm,float).reshape(-1,2)
        if not len(d): return np.zeros_like(epsgrid,float)
        return np.array([np.sum((d[:,0]<=e)&(e<d[:,1])) for e in epsgrid],float)

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _windows(y,nw=None):
        y=np.asarray(y); nw=int(nw or _TOPO_SPEED['windows']); L=max(300,len(y)//max(4,nw//2)); step=max(1,(len(y)-L)//max(1,nw-1))
        ans=[]
        for st in range(0,max(1,len(y)-L+1),step):
            ans.append((st,st+L,y[st:st+L]))
            if len(ans)>=nw: break
        return ans

    def _window_topo(y,m=3,tau=2):
        rows=[]; diagrams=[]
        for st,en,w in _windows(y):
            X=_delay_embed(w,m,tau,max_points=min(_TOPO_SPEED['points'],70 if FAST_OPTION=='ultra' else _TOPO_SPEED['points']))
            dg=_ph_rips(X,1); rows.append([st/_topo_fs,en/_topo_fs,*_feat(dg)]); diagrams.append(dg)
        return pd.DataFrame(rows,columns=['start_s','end_s','H1_total','H1_max','H1_entropy','H1_count']),diagrams

    # Method mathematics and output.
    y = _topo_pc1
    df, dgs = _window_topo(y)
    mx = max([np.max(_finite(d[1])[:, 1]) if len(_finite(d[1])) else 1 for d in dgs])
    e = np.linspace(0, mx, 60)
    C = np.vstack([_betti_curve(d[1], e) for d in dgs])
    fig, ax = plt.subplots(figsize=(8, 4))
    im = ax.imshow(C.T, origin='lower', aspect='auto', extent=[df.start_s.min(), df.end_s.max(), e[0], e[-1]], cmap=TOPO_CMAP)
    ax.set_xlabel('time (s)')
    ax.set_ylabel('filtration ε')
    ax.set_title('Time-resolved Betti curves' if i == 15 else 'CROCKER plot: β1(t,ε)')
    plt.colorbar(im, ax=ax, label='β1')
    plt.show()


_topology_method_018()
del _topology_method_018
```

### 19. Sliding-window persistent homology

Sliding-window persistent homology computes a persistence diagram independently in each temporal window after applying the same delay-coordinate construction. Window length and overlap therefore define the temporal resolution of the topological trajectory.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Persistent homology summarizes how topological features appear and disappear as a distance or filtration threshold increases. In dimension 0 the features are connected components; in dimension 1 they are loops; higher dimensions represent higher-order cavities. Each feature receives a birth scale and a death scale, and its lifetime is the difference between them. Long lifetimes are often treated as more structurally important than very short ones, although noise and sampling density can still affect the diagram.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> Because the calculation is performed on finite windows or segments, the reported value is a local summary of that selected data interval; a different interval can legitimately produce a different estimate if the EEG is nonstationary.</p></div>


```python
# Method 19: Sliding-window persistent homology

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_019():

    i = 19
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _plot_line(x,y,title,xlabel='x',ylabel='value'):
        fig,ax=plt.subplots(figsize=(8,3.8),facecolor=TOPO_BG)
        ax.plot(x,y,color=TOPO_ACCENT,lw=1.7)
        ax.set_title(title); ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.28,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _windows(y,nw=None):
        y=np.asarray(y); nw=int(nw or _TOPO_SPEED['windows']); L=max(300,len(y)//max(4,nw//2)); step=max(1,(len(y)-L)//max(1,nw-1))
        ans=[]
        for st in range(0,max(1,len(y)-L+1),step):
            ans.append((st,st+L,y[st:st+L]))
            if len(ans)>=nw: break
        return ans

    def _window_topo(y,m=3,tau=2):
        rows=[]; diagrams=[]
        for st,en,w in _windows(y):
            X=_delay_embed(w,m,tau,max_points=min(_TOPO_SPEED['points'],70 if FAST_OPTION=='ultra' else _TOPO_SPEED['points']))
            dg=_ph_rips(X,1); rows.append([st/_topo_fs,en/_topo_fs,*_feat(dg)]); diagrams.append(dg)
        return pd.DataFrame(rows,columns=['start_s','end_s','H1_total','H1_max','H1_entropy','H1_count']),diagrams

    # Method mathematics and output.
    y = _topo_pc1
    df, _ = _window_topo(y)
    display(df)
    _plot_line((df.start_s + df.end_s) / 2, df.H1_total, 'Sliding-window persistent homology', 'time (s)', 'H1 total persistence')


_topology_method_019()
del _topology_method_019
```

### 20. Event-locked persistent homology

Event-locked persistent homology aligns windowed persistence features around event times. If `TOPOLOGY_EVENT_SAMPLES` is supplied those samples are used as true events; otherwise the notebook uses internally detected high-envelope peaks and labels them as demonstration events rather than experimental stimuli.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Persistent homology summarizes how topological features appear and disappear as a distance or filtration threshold increases. In dimension 0 the features are connected components; in dimension 1 they are loops; higher dimensions represent higher-order cavities. Each feature receives a birth scale and a death scale, and its lifetime is the difference between them. Long lifetimes are often treated as more structurally important than very short ones, although noise and sampling density can still affect the diagram.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> Because the calculation is performed on finite windows or segments, the reported value is a local summary of that selected data interval; a different interval can legitimately produce a different estimate if the EEG is nonstationary.</p></div>


```python
# Method 20: Event-locked persistent homology

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_020():

    i = 20
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _filter_band(y,lo,hi):
        sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_topo_fs,output='sos'); return signal.sosfiltfilt(sos,y)

    def _plot_line(x,y,title,xlabel='x',ylabel='value'):
        fig,ax=plt.subplots(figsize=(8,3.8),facecolor=TOPO_BG)
        ax.plot(x,y,color=TOPO_ACCENT,lw=1.7)
        ax.set_title(title); ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.28,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _windows(y,nw=None):
        y=np.asarray(y); nw=int(nw or _TOPO_SPEED['windows']); L=max(300,len(y)//max(4,nw//2)); step=max(1,(len(y)-L)//max(1,nw-1))
        ans=[]
        for st in range(0,max(1,len(y)-L+1),step):
            ans.append((st,st+L,y[st:st+L]))
            if len(ans)>=nw: break
        return ans

    def _window_topo(y,m=3,tau=2):
        rows=[]; diagrams=[]
        for st,en,w in _windows(y):
            X=_delay_embed(w,m,tau,max_points=min(_TOPO_SPEED['points'],70 if FAST_OPTION=='ultra' else _TOPO_SPEED['points']))
            dg=_ph_rips(X,1); rows.append([st/_topo_fs,en/_topo_fs,*_feat(dg)]); diagrams.append(dg)
        return pd.DataFrame(rows,columns=['start_s','end_s','H1_total','H1_max','H1_entropy','H1_count']),diagrams

    # Method mathematics and output.
    y = _topo_pc1
    df, _ = _window_topo(y)
    centers = ((df.start_s + df.end_s) / 2).to_numpy()
    vals = df.H1_total.to_numpy()
    if 'TOPOLOGY_EVENT_SAMPLES' in globals():
        ev = np.asarray(globals()['TOPOLOGY_EVENT_SAMPLES'], int) / _topo_fs
        label = 'supplied events'
    else:
        env = np.abs(signal.hilbert(_filter_band(y, 8, min(13, _topo_fs / 2 - 0.5))))
        pk, _ = signal.find_peaks(env, distance=max(1, int(_topo_fs)), prominence=np.std(env))
        ev = pk / _topo_fs
        label = 'internal alpha-envelope demonstration events'
    rel = np.linspace(-2, 2, 41)
    A = []
    for e0 in ev[:40]:
        A.append(np.interp(e0 + rel, centers, vals, left=np.nan, right=np.nan))
    if A:
        _plot_line(rel, np.nanmean(A, axis=0), f'Event-locked H1 total ({label})', 'time from event (s)', 'H1 total')
    else:
        print('No usable events detected.')


_topology_method_020()
del _topology_method_020
```

### 21. Time-resolved H1 total persistence

For each time window, H1 total persistence is \(TP_1(t)=\sum_i[d_i(t)-b_i(t)]\). This produces a scalar trajectory of loop strength through time.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> Because the calculation is performed on finite windows or segments, the reported value is a local summary of that selected data interval; a different interval can legitimately produce a different estimate if the EEG is nonstationary.</p></div>

### 22. Time-resolved persistence entropy

Time-resolved persistence entropy applies \(H(t)=-\sum_i p_i(t)\log p_i(t)\) to each window, with \(p_i(t)\) the normalized H1 lifetimes in that window.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> Because the calculation is performed on finite windows or segments, the reported value is a local summary of that selected data interval; a different interval can legitimately produce a different estimate if the EEG is nonstationary.</p></div>


```python
# Method 22: Time-resolved persistence entropy

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_022():

    i = 22
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _plot_line(x,y,title,xlabel='x',ylabel='value'):
        fig,ax=plt.subplots(figsize=(8,3.8),facecolor=TOPO_BG)
        ax.plot(x,y,color=TOPO_ACCENT,lw=1.7)
        ax.set_title(title); ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.28,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _windows(y,nw=None):
        y=np.asarray(y); nw=int(nw or _TOPO_SPEED['windows']); L=max(300,len(y)//max(4,nw//2)); step=max(1,(len(y)-L)//max(1,nw-1))
        ans=[]
        for st in range(0,max(1,len(y)-L+1),step):
            ans.append((st,st+L,y[st:st+L]))
            if len(ans)>=nw: break
        return ans

    def _window_topo(y,m=3,tau=2):
        rows=[]; diagrams=[]
        for st,en,w in _windows(y):
            X=_delay_embed(w,m,tau,max_points=min(_TOPO_SPEED['points'],70 if FAST_OPTION=='ultra' else _TOPO_SPEED['points']))
            dg=_ph_rips(X,1); rows.append([st/_topo_fs,en/_topo_fs,*_feat(dg)]); diagrams.append(dg)
        return pd.DataFrame(rows,columns=['start_s','end_s','H1_total','H1_max','H1_entropy','H1_count']),diagrams

    # Method mathematics and output.
    y = _topo_pc1
    df, _ = _window_topo(y)
    col = 'H1_total' if i == 21 else 'H1_entropy'
    _plot_line((df.start_s + df.end_s) / 2, df[col], 'Time-resolved H1 total persistence' if i == 21 else 'Time-resolved persistence entropy', 'time (s)', col)


_topology_method_022()
del _topology_method_022
```

### 23. Time-resolved persistence diagrams

Dynamic persistence diagrams are the ordered family \(\{D_t\}\) from sliding windows. The cell visualizes representative diagrams through time while retaining the full list for downstream tracking.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. ### 24. Topological change-point detection

Topological change-point detection uses the Euclidean jump in the feature vector \(z_t=[TP_1,L_{max},H,N_1]\): \(c_t=\|z_t-z_{t-1}\|_2\). Large robustly standardized jumps flag candidate changes in topology; this is a feature-space detector rather than an exact theorem about discontinuities in the underlying dynamics.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 24: Topological change-point detection

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_024():

    i = 24
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _plot_line(x,y,title,xlabel='x',ylabel='value'):
        fig,ax=plt.subplots(figsize=(8,3.8),facecolor=TOPO_BG)
        ax.plot(x,y,color=TOPO_ACCENT,lw=1.7)
        ax.set_title(title); ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.28,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _windows(y,nw=None):
        y=np.asarray(y); nw=int(nw or _TOPO_SPEED['windows']); L=max(300,len(y)//max(4,nw//2)); step=max(1,(len(y)-L)//max(1,nw-1))
        ans=[]
        for st in range(0,max(1,len(y)-L+1),step):
            ans.append((st,st+L,y[st:st+L]))
            if len(ans)>=nw: break
        return ans

    def _window_topo(y,m=3,tau=2):
        rows=[]; diagrams=[]
        for st,en,w in _windows(y):
            X=_delay_embed(w,m,tau,max_points=min(_TOPO_SPEED['points'],70 if FAST_OPTION=='ultra' else _TOPO_SPEED['points']))
            dg=_ph_rips(X,1); rows.append([st/_topo_fs,en/_topo_fs,*_feat(dg)]); diagrams.append(dg)
        return pd.DataFrame(rows,columns=['start_s','end_s','H1_total','H1_max','H1_entropy','H1_count']),diagrams

    # Method mathematics and output.
    y = _topo_pc1
    df, _ = _window_topo(y)
    Z = df[['H1_total', 'H1_max', 'H1_entropy', 'H1_count']].to_numpy()
    Z = (Z - np.mean(Z, 0)) / (np.std(Z, 0) + TOPO_EPS)
    c = np.r_[0, np.linalg.norm(np.diff(Z, axis=0), axis=1)]
    z = (c - np.median(c)) / (stats.median_abs_deviation(c, scale='normal') + TOPO_EPS)
    _plot_line((df.start_s + df.end_s) / 2, z, 'Topological change-point score', 'time (s)', 'robust z-score')


_topology_method_024()
del _topology_method_024
```

### 25. Topological state segmentation

Topological state segmentation clusters standardized persistence-feature vectors from sliding windows. K-means minimizes \(\sum_t\|z_t-\mu_{c_t}\|^2\), producing recurrent topological regimes rather than clustering raw EEG amplitudes.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values. Because the calculation is performed on finite windows or segments, the reported value is a local summary of that selected data interval; a different interval can legitimately produce a different estimate if the EEG is nonstationary.</p></div>


```python
# Method 25: Topological state segmentation

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_025():

    i = 25
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _windows(y,nw=None):
        y=np.asarray(y); nw=int(nw or _TOPO_SPEED['windows']); L=max(300,len(y)//max(4,nw//2)); step=max(1,(len(y)-L)//max(1,nw-1))
        ans=[]
        for st in range(0,max(1,len(y)-L+1),step):
            ans.append((st,st+L,y[st:st+L]))
            if len(ans)>=nw: break
        return ans

    def _window_topo(y,m=3,tau=2):
        rows=[]; diagrams=[]
        for st,en,w in _windows(y):
            X=_delay_embed(w,m,tau,max_points=min(_TOPO_SPEED['points'],70 if FAST_OPTION=='ultra' else _TOPO_SPEED['points']))
            dg=_ph_rips(X,1); rows.append([st/_topo_fs,en/_topo_fs,*_feat(dg)]); diagrams.append(dg)
        return pd.DataFrame(rows,columns=['start_s','end_s','H1_total','H1_max','H1_entropy','H1_count']),diagrams

    # Method mathematics and output.
    y = _topo_pc1
    df, _ = _window_topo(y)
    Z = df[['H1_total', 'H1_max', 'H1_entropy', 'H1_count']].to_numpy()
    Z = (Z - Z.mean(0)) / (Z.std(0) + TOPO_EPS)
    k = min(4, max(2, len(Z) // 5))
    lab = KMeans(k, random_state=0, n_init=10).fit_predict(Z)
    out = pd.DataFrame({'time_s': (df.start_s + df.end_s) / 2, 'topological_state': lab})
    display(out)
    fig, ax = plt.subplots(figsize=(9, 3.8), facecolor=TOPO_BG)
    ax.step(out['time_s'], out['topological_state'], where='mid', color=TOPO_ACCENT, linewidth=1.7)
    ax.scatter(out['time_s'], out['topological_state'], s=24, color=TOPO_CYAN_FILL, edgecolor=TOPO_ACCENT)
    ax.set_title('Topological state segmentation')
    ax.set_xlabel('time (s)')
    ax.set_ylabel('state')
    ax.grid(True, color=TOPO_CYAN_DIM, alpha=0.22)
    plt.tight_layout()
    plt.show()


_topology_method_025()
del _topology_method_025
```

### 26. Topological recurrence of states

Topological recurrence is measured after state segmentation. For state labels \(s_t\), the transition matrix estimates \(P_{ij}=\Pr(s_{t+1}=j\mid s_t=i)\), while return intervals quantify recurrence times of each topological state.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> Because the calculation is performed on finite windows or segments, the reported value is a local summary of that selected data interval; a different interval can legitimately produce a different estimate if the EEG is nonstationary.</p></div>


```python
# Method 26: Topological recurrence of states

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_026():

    i = 26
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _windows(y,nw=None):
        y=np.asarray(y); nw=int(nw or _TOPO_SPEED['windows']); L=max(300,len(y)//max(4,nw//2)); step=max(1,(len(y)-L)//max(1,nw-1))
        ans=[]
        for st in range(0,max(1,len(y)-L+1),step):
            ans.append((st,st+L,y[st:st+L]))
            if len(ans)>=nw: break
        return ans

    def _window_topo(y,m=3,tau=2):
        rows=[]; diagrams=[]
        for st,en,w in _windows(y):
            X=_delay_embed(w,m,tau,max_points=min(_TOPO_SPEED['points'],70 if FAST_OPTION=='ultra' else _TOPO_SPEED['points']))
            dg=_ph_rips(X,1); rows.append([st/_topo_fs,en/_topo_fs,*_feat(dg)]); diagrams.append(dg)
        return pd.DataFrame(rows,columns=['start_s','end_s','H1_total','H1_max','H1_entropy','H1_count']),diagrams

    # Method mathematics and output.
    y = _topo_pc1
    df, _ = _window_topo(y)
    Z = df[['H1_total', 'H1_max', 'H1_entropy', 'H1_count']].to_numpy()
    Z = (Z - Z.mean(0)) / (Z.std(0) + TOPO_EPS)
    k = min(4, max(2, len(Z) // 5))
    lab = KMeans(k, random_state=0, n_init=10).fit_predict(Z)
    M = np.zeros((k, k), int)
    for a, b in zip(lab[:-1], lab[1:]):
        M[a, b] += 1
    P = M / np.maximum(M.sum(1, keepdims=True), 1)
    out = pd.DataFrame(P, index=[f's{i}' for i in range(k)], columns=[f's{i}' for i in range(k)])
    display(out)
    fig, ax = plt.subplots(figsize=(5.5, 4.8), facecolor=TOPO_BG)
    im = ax.imshow(P, cmap=TOPO_CMAP, vmin=0, vmax=max(TOPO_EPS, float(np.max(P))), aspect='auto')
    ax.set_xticks(range(k))
    ax.set_xticklabels(out.columns)
    ax.set_yticks(range(k))
    ax.set_yticklabels(out.index)
    ax.set_xlabel('next state')
    ax.set_ylabel('current state')
    ax.set_title('Topological-state transition probabilities')
    cb = plt.colorbar(im, ax=ax)
    cb.set_label('transition probability', color=TOPO_ACCENT)
    plt.tight_layout()
    plt.show()


_topology_method_026()
del _topology_method_026
```

### 27. Topology-triggered event detection

Topology-triggered events are peaks in the standardized topological change score, not external stimuli. The event detector uses prominence/separation constraints so neighboring fluctuations are not double-counted.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 27: Topology-triggered event detection

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_027():

    i = 27
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _windows(y,nw=None):
        y=np.asarray(y); nw=int(nw or _TOPO_SPEED['windows']); L=max(300,len(y)//max(4,nw//2)); step=max(1,(len(y)-L)//max(1,nw-1))
        ans=[]
        for st in range(0,max(1,len(y)-L+1),step):
            ans.append((st,st+L,y[st:st+L]))
            if len(ans)>=nw: break
        return ans

    def _window_topo(y,m=3,tau=2):
        rows=[]; diagrams=[]
        for st,en,w in _windows(y):
            X=_delay_embed(w,m,tau,max_points=min(_TOPO_SPEED['points'],70 if FAST_OPTION=='ultra' else _TOPO_SPEED['points']))
            dg=_ph_rips(X,1); rows.append([st/_topo_fs,en/_topo_fs,*_feat(dg)]); diagrams.append(dg)
        return pd.DataFrame(rows,columns=['start_s','end_s','H1_total','H1_max','H1_entropy','H1_count']),diagrams

    # Method mathematics and output.
    y = _topo_pc1
    df, _ = _window_topo(y)
    Z = df[['H1_total', 'H1_max', 'H1_entropy', 'H1_count']].to_numpy()
    Z = (Z - np.mean(Z, 0)) / (np.std(Z, 0) + TOPO_EPS)
    c = np.r_[0, np.linalg.norm(np.diff(Z, axis=0), axis=1)]
    z = (c - np.median(c)) / (stats.median_abs_deviation(c, scale='normal') + TOPO_EPS)
    pk, _ = signal.find_peaks(z, height=2, distance=2)
    out = pd.DataFrame({'time_s': ((df.start_s + df.end_s) / 2).to_numpy()[pk], 'change_z': z[pk]})
    display(out)
    tt = ((df.start_s + df.end_s) / 2).to_numpy()
    fig, ax = plt.subplots(figsize=(9, 4.0), facecolor=TOPO_BG)
    ax.plot(tt, z, color=TOPO_ACCENT, linewidth=1.6)
    ax.scatter(tt[pk], z[pk], s=45, color=TOPO_CYAN_FILL, edgecolor=TOPO_CYAN_SOFT, zorder=3)
    ax.axhline(2, color=TOPO_CYAN_SOFT, linestyle='--', linewidth=1.0)
    ax.set_title('Topology-triggered event detections')
    ax.set_xlabel('time (s)')
    ax.set_ylabel('robust change z')
    ax.grid(True, color=TOPO_CYAN_DIM, alpha=0.22)
    plt.tight_layout()
    plt.show()


_topology_method_027()
del _topology_method_027
```

### 28. Vineyards

A vineyard tracks persistence points across time. Consecutive diagrams are matched by minimum-cost assignment in birth–death coordinates, creating trajectories \((b_i(t),d_i(t))\). This finite matching implements the vineyard idea but should not be confused with a proof of a smooth vineyard theorem for the underlying signal.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 28: Vineyards

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_028():

    i = 28
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _windows(y,nw=None):
        y=np.asarray(y); nw=int(nw or _TOPO_SPEED['windows']); L=max(300,len(y)//max(4,nw//2)); step=max(1,(len(y)-L)//max(1,nw-1))
        ans=[]
        for st in range(0,max(1,len(y)-L+1),step):
            ans.append((st,st+L,y[st:st+L]))
            if len(ans)>=nw: break
        return ans

    def _window_topo(y,m=3,tau=2):
        rows=[]; diagrams=[]
        for st,en,w in _windows(y):
            X=_delay_embed(w,m,tau,max_points=min(_TOPO_SPEED['points'],70 if FAST_OPTION=='ultra' else _TOPO_SPEED['points']))
            dg=_ph_rips(X,1); rows.append([st/_topo_fs,en/_topo_fs,*_feat(dg)]); diagrams.append(dg)
        return pd.DataFrame(rows,columns=['start_s','end_s','H1_total','H1_max','H1_entropy','H1_count']),diagrams

    # Method mathematics and output.
    y = _topo_pc1
    df, dgs = _window_topo(y)
    tracks = []
    prev = None
    for t, dg in enumerate(dgs):
        pts = _finite(dg[1])
        pts = pts[np.argsort(_life(pts))[::-1]][:6] if len(pts) else pts
        if prev is None:
            tracks = [[(t, *p)] for p in pts]
        elif len(prev) and len(pts):
            rr, cc = linear_sum_assignment(cdist(prev, pts))
            used = set()
            for r, c in zip(rr, cc):
                if r < len(tracks):
                    tracks[r].append((t, *pts[c]))
                    used.add(c)
            for c, p in enumerate(pts):
                if c not in used:
                    tracks.append([(t, *p)])
        prev = pts
    fig, ax = plt.subplots(figsize=(8, 4))
    for tr in tracks[:10]:
        arr = np.array(tr, float)
        ax.plot(arr[:, 0], arr[:, 2] - arr[:, 1], lw=1.2, alpha=0.75, color=TOPO_ACCENT)
    ax.set_title('Vineyard: tracked H1 lifetimes')
    ax.set_xlabel('window index')
    ax.set_ylabel('lifetime')
    ax.grid(alpha=0.2)
    plt.show()


_topology_method_028()
del _topology_method_028
```

### 29. Zigzag persistence

Zigzag persistence allows both inclusions and reverse inclusions, unlike ordinary monotone filtrations. This cell implements an **H0 zigzag component tracker** on a sequence of threshold graphs, matching connected components by vertex overlap as edges appear and disappear. It is deliberately labelled H0; it is not a full arbitrary-dimensional zigzag matrix-reduction algorithm.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> A Lyapunov exponent describes how quickly two initially nearby states separate or approach one another as the system evolves. A positive exponent means small differences grow on average, a negative exponent means they shrink, and a value near zero corresponds to a direction with little average exponential change. When several exponents are reported, they describe stretching and contraction along different directions of the reconstructed state space rather than separate EEG frequency bands. The numerical value also has units set by the time step used in the analysis, so comparisons are most meaningful when preprocessing and sampling are consistent.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 29: Zigzag persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_029():

    i = 29
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _windows(y,nw=None):
        y=np.asarray(y); nw=int(nw or _TOPO_SPEED['windows']); L=max(300,len(y)//max(4,nw//2)); step=max(1,(len(y)-L)//max(1,nw-1))
        ans=[]
        for st in range(0,max(1,len(y)-L+1),step):
            ans.append((st,st+L,y[st:st+L]))
            if len(ans)>=nw: break
        return ans

    # Method mathematics and output.
    comps = []
    for st, en, w in _windows(np.arange(len(_topo_x))):
        seg = _topo_x[int(st):int(en)]
        C = np.nan_to_num(np.corrcoef(seg.T))
        A = np.abs(C) > 0.55
        np.fill_diagonal(A, False)
        seen = set()
        cc = []
        for v in range(len(A)):
            if v in seen:
                continue
            stack = [v]
            cur = set([v])
            seen.add(v)
            while stack:
                u = stack.pop()
                for z in np.where(A[u])[0]:
                    z = int(z)
                    if z not in seen:
                        seen.add(z)
                        cur.add(z)
                        stack.append(z)
            cc.append(cur)
        comps.append(cc)
    intervals = []
    active = []
    for t, cc in enumerate(comps):
        new = []
        used = set()
        for birth, lastset in active:
            scores = [len(lastset & c) / max(1, len(lastset | c)) for c in cc]
            j = int(np.argmax(scores)) if scores else -1
            if j >= 0 and scores[j] > 0:
                new.append((birth, cc[j]))
                used.add(j)
            else:
                intervals.append((birth, t))
        for j, c in enumerate(cc):
            if j not in used:
                new.append((t, c))
        active = new
    intervals.extend(((b, len(comps)) for b, _ in active))
    out = pd.DataFrame(intervals, columns=['birth_window', 'death_window'])
    display(out)
    fig, ax = plt.subplots(figsize=(9, 4.8), facecolor=TOPO_BG)
    for r, (b, d) in enumerate(out[['birth_window', 'death_window']].to_numpy(float)):
        ax.plot([b, d], [r, r], color=TOPO_ACCENT, linewidth=1.5)
        ax.scatter([b, d], [r, r], s=16, color=TOPO_CYAN_FILL, edgecolor=TOPO_ACCENT)
    ax.set_title('Zigzag-H0 component intervals')
    ax.set_xlabel('window index')
    ax.set_ylabel('component interval')
    ax.grid(True, color=TOPO_CYAN_DIM, alpha=0.18)
    plt.tight_layout()
    plt.show()


_topology_method_029()
del _topology_method_029
```

### 30. Dynamic persistence diagrams

Dynamic persistence diagrams recompute H0/H1 on a sequence of time windows and summarize diagram-to-diagram changes. This is the diagram-valued analogue of scalar time-resolved persistence features.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> Because the calculation is performed on finite windows or segments, the reported value is a local summary of that selected data interval; a different interval can legitimately produce a different estimate if the EEG is nonstationary.</p></div>


```python
# Method 30: Dynamic persistence diagrams

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_030():

    i = 30
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _windows(y,nw=None):
        y=np.asarray(y); nw=int(nw or _TOPO_SPEED['windows']); L=max(300,len(y)//max(4,nw//2)); step=max(1,(len(y)-L)//max(1,nw-1))
        ans=[]
        for st in range(0,max(1,len(y)-L+1),step):
            ans.append((st,st+L,y[st:st+L]))
            if len(ans)>=nw: break
        return ans

    def _window_topo(y,m=3,tau=2):
        rows=[]; diagrams=[]
        for st,en,w in _windows(y):
            X=_delay_embed(w,m,tau,max_points=min(_TOPO_SPEED['points'],70 if FAST_OPTION=='ultra' else _TOPO_SPEED['points']))
            dg=_ph_rips(X,1); rows.append([st/_topo_fs,en/_topo_fs,*_feat(dg)]); diagrams.append(dg)
        return pd.DataFrame(rows,columns=['start_s','end_s','H1_total','H1_max','H1_entropy','H1_count']),diagrams

    # Method mathematics and output.
    y = _topo_pc1
    df, dgs = _window_topo(y)
    out = pd.DataFrame({'time_s': (df.start_s + df.end_s) / 2, 'diagram_H1_bars': [len(_finite(d[1])) for d in dgs], 'H1_total': df.H1_total})
    display(out)
    fig, ax = plt.subplots(figsize=(9, 4.2), facecolor=TOPO_BG)
    ax2 = ax.twinx()
    ax.plot(out['time_s'], out['H1_total'], color=TOPO_ACCENT, linewidth=1.7, label='H1 total')
    ax2.plot(out['time_s'], out['diagram_H1_bars'], color=TOPO_CYAN_SOFT, linewidth=1.3, linestyle='--', label='finite H1 bars')
    ax.set_title('Dynamic persistence-diagram summary')
    ax.set_xlabel('time (s)')
    ax.set_ylabel('H1 total')
    ax2.set_ylabel('finite H1 bars', color=TOPO_CYAN_SOFT)
    ax.grid(True, color=TOPO_CYAN_DIM, alpha=0.22)
    lines = ax.get_lines() + ax2.get_lines()
    ax.legend(lines, [ln.get_label() for ln in lines], loc='best')
    plt.tight_layout()
    plt.show()


_topology_method_030()
del _topology_method_030
```

### 31. Dynamic graph persistent homology

Dynamic graph persistent homology builds a functional-connectivity graph in each time window. Edge dissimilarities \(d_{ij}=1-|r_{ij}|\) define a weighted flag/Rips filtration on channels, from which H0/H1 persistence is computed over time.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Persistent homology summarizes how topological features appear and disappear as a distance or filtration threshold increases. In dimension 0 the features are connected components; in dimension 1 they are loops; higher dimensions represent higher-order cavities. Each feature receives a birth scale and a death scale, and its lifetime is the difference between them. Long lifetimes are often treated as more structurally important than very short ones, although noise and sampling density can still affect the diagram.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values. Because the calculation is performed on finite windows or segments, the reported value is a local summary of that selected data interval; a different interval can legitimately produce a different estimate if the EEG is nonstationary.</p></div>


```python
# Method 31: Dynamic graph persistent homology

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_031():

    i = 31
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    def _windows(y,nw=None):
        y=np.asarray(y); nw=int(nw or _TOPO_SPEED['windows']); L=max(300,len(y)//max(4,nw//2)); step=max(1,(len(y)-L)//max(1,nw-1))
        ans=[]
        for st in range(0,max(1,len(y)-L+1),step):
            ans.append((st,st+L,y[st:st+L]))
            if len(ans)>=nw: break
        return ans

    # Method mathematics and output.
    rows = []
    nw = _windows(np.arange(len(_topo_x)))
    for st, en, _ in nw:
        seg = _topo_x[int(st):int(en)]
        C = np.nan_to_num(np.corrcoef(seg.T))
        D = 1 - np.abs(C)
        np.fill_diagonal(D, 0)
        H = np.eye(len(D)) - np.ones_like(D) / len(D)
        B = -0.5 * H @ D ** 2 @ H
        ev, U = np.linalg.eigh(B)
        idx = np.argsort(ev)[::-1]
        pos = ev[idx] > 1e-09
        q = min(5, np.sum(pos))
        X = U[:, idx[:q]] * np.sqrt(np.maximum(ev[idx[:q]], 0))
        dg = _ph_rips(X, 1)
        rows.append([st / _topo_fs, *_feat(dg)])
    out = pd.DataFrame(rows, columns=['time_s', 'H1_total', 'H1_max', 'entropy', 'count'])
    display(out)
    fig, axs = plt.subplots(2, 2, figsize=(9.5, 6.6), sharex=True, facecolor=TOPO_BG)
    axs = np.ravel(axs)
    for ax, col in zip(axs, ['H1_total', 'H1_max', 'entropy', 'count']):
        ax.plot(out['time_s'], out[col], color=TOPO_ACCENT, linewidth=1.6)
        ax.set_title(col)
        ax.grid(True, color=TOPO_CYAN_DIM, alpha=0.22)
    axs[2].set_xlabel('time (s)')
    axs[3].set_xlabel('time (s)')
    fig.suptitle('Dynamic graph persistent homology')
    plt.tight_layout()
    plt.show()


_topology_method_031()
del _topology_method_031
```

### 32. Alpha complex persistence

For a planar Delaunay triangulation, the alpha complex is the subcomplex selected by an empty-ball radius threshold. This cell uses the Delaunay triangulation of the first two PCA coordinates and assigns each simplex a circumradius-style alpha scale, then reduces the filtered complex over \(\mathbb Z_2\). The implementation is explicitly a finite 2-D alpha-complex construction.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Principal component analysis rotates the coordinate system so that the first axis captures the greatest sample variance, the second captures the greatest remaining variance subject to being orthogonal to the first, and so on. The directions are eigenvectors of the covariance matrix and their eigenvalues quantify variance along those directions. PCA is therefore a linear dimensionality-reduction method: it can summarize redundant channels or features, but it does not guarantee that the retained components correspond to distinct physiological sources.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 32: Alpha complex persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_032():

    i = 32
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _triangle_meb_radius(a,b,c):
        # exact Euclidean minimum-enclosing-ball radius for three points
        ds=np.array([np.linalg.norm(a-b),np.linalg.norm(a-c),np.linalg.norm(b-c)],float); L=np.max(ds)
        ss=np.sort(ds)
        if ss[2]**2 >= ss[0]**2+ss[1]**2-1e-12: return float(L/2)
        area=np.linalg.norm(np.cross(np.r_[b-a,0] if a.size==2 else b-a, np.r_[c-a,0] if a.size==2 else c-a))/2
        if area<=1e-12:return float(L/2)
        return float(np.prod(ds)/(4*area))

    def _delaunay_cells(X,alpha_style=False):
        Z=PCA(2).fit_transform(np.asarray(X,float)); tri=Delaunay(Z); n=len(Z); cells=[]; edge_val={}
        for i in range(n): cells.append((('v',i),0,0,[]))
        tvals={}
        for simp in tri.simplices:
            i,j,k=sorted(map(int,simp)); r=_triangle_meb_radius(Z[i],Z[j],Z[k]); tvals[(i,j,k)]=r
            for a,b in [(i,j),(i,k),(j,k)]: edge_val[(a,b)]=min(edge_val.get((a,b),np.inf),r)
        for (i,j),rv in edge_val.items():
            half=np.linalg.norm(Z[i]-Z[j])/2
            val=min(rv,half) if alpha_style else rv
            cells.append((('e',i,j),val,1,[('v',i),('v',j)]))
        for (i,j,k),r in tvals.items(): cells.append((('t',i,j,k),r,2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1),Z,tri

    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )

    # Method mathematics and output.
    X = _TOPO_XREF
    dg, Z, tr = _delaunay_cells(X, alpha_style=True)
    _plot_diag(dg[1], 'Alpha-complex H1 persistence (2-D PCA Delaunay)')


_topology_method_032()
del _topology_method_032
```

### 33. Čech complex persistence

The Čech complex at radius \(r\) contains a simplex when the closed radius-\(r\) balls centered at all of its vertices have nonempty common intersection. For H1 it is enough to include vertices, edges at \(\|x_i-x_j\|/2\), and triangles at the exact three-point minimum-enclosing-ball radius.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 33: Čech complex persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_033():

    i = 33
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _triangle_meb_radius(a,b,c):
        # exact Euclidean minimum-enclosing-ball radius for three points
        ds=np.array([np.linalg.norm(a-b),np.linalg.norm(a-c),np.linalg.norm(b-c)],float); L=np.max(ds)
        ss=np.sort(ds)
        if ss[2]**2 >= ss[0]**2+ss[1]**2-1e-12: return float(L/2)
        area=np.linalg.norm(np.cross(np.r_[b-a,0] if a.size==2 else b-a, np.r_[c-a,0] if a.size==2 else c-a))/2
        if area<=1e-12:return float(L/2)
        return float(np.prod(ds)/(4*area))

    def _cech_ph(X):
        X=np.asarray(X,float); n=len(X); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        for i in range(n):
            for j in range(i+1,n): cells.append((('e',i,j),np.linalg.norm(X[i]-X[j])/2,1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                for k in range(j+1,n):
                    r=_triangle_meb_radius(X[i],X[j],X[k]); cells.append((('t',i,j,k),r,2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )

    # Method mathematics and output.
    X = _TOPO_XREF
    Z = PCA(2).fit_transform(X)
    dg = _cech_ph(Z)
    _plot_diag(dg[1], 'Čech H1 persistence')


_topology_method_033()
del _topology_method_033
```

### 34. Witness complex persistence

A witness complex uses a small landmark set \(L\) and the remaining observations as witnesses. A landmark edge is admitted when some witness is simultaneously close to both landmarks; higher simplices are filled by the flag rule. This reduces the complex size while retaining geometry sampled by the full cloud.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> A Lyapunov exponent describes how quickly two initially nearby states separate or approach one another as the system evolves. A positive exponent means small differences grow on average, a negative exponent means they shrink, and a value near zero corresponds to a direction with little average exponential change. When several exponents are reported, they describe stretching and contraction along different directions of the reconstructed state space rather than separate EEG frequency bands. The numerical value also has units set by the time step used in the analysis, so comparisons are most meaningful when preprocessing and sampling are consistent.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 34: Witness complex persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_034():

    i = 34
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )

    # Method mathematics and output.
    X = _TOPO_XREF
    L = X[np.linspace(0, len(X) - 1, min(14, len(X))).astype(int)]
    W = X
    DL = cdist(W, L)
    n = len(L)
    D = np.full((n, n), np.inf)
    np.fill_diagonal(D, 0)
    nearest = np.min(DL, axis=1)
    for a in range(n):
        for b in range(a + 1, n):
            vals = np.maximum(DL[:, a], DL[:, b]) - (nearest if i == 34 else 0)
            D[a, b] = D[b, a] = max(0, float(np.min(vals)))
    dg = _flag_ph(D)
    _plot_diag(dg[1], ('Witness' if i == 34 else 'Lazy witness') + ' H1 persistence')


_topology_method_034()
del _topology_method_034
```

### 35. Lazy witness complex persistence

The lazy witness complex builds only witness-supported landmark edges and then fills every clique as a simplex. Its filtration is therefore the flag complex of a witness-weighted graph; it is cheaper than constructing all higher-order witness conditions explicitly.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 35: Lazy witness complex persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_035():

    i = 35
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )

    # Method mathematics and output.
    X = _TOPO_XREF
    L = X[np.linspace(0, len(X) - 1, min(14, len(X))).astype(int)]
    W = X
    DL = cdist(W, L)
    n = len(L)
    D = np.full((n, n), np.inf)
    np.fill_diagonal(D, 0)
    nearest = np.min(DL, axis=1)
    for a in range(n):
        for b in range(a + 1, n):
            vals = np.maximum(DL[:, a], DL[:, b]) - (nearest if i == 34 else 0)
            D[a, b] = D[b, a] = max(0, float(np.min(vals)))
    dg = _flag_ph(D)
    _plot_diag(dg[1], ('Witness' if i == 34 else 'Lazy witness') + ' H1 persistence')


_topology_method_035()
del _topology_method_035
```

### 36. Delaunay complex persistence

Delaunay persistence filters the planar Delaunay complex by simplex circumradius. Because only Delaunay simplices are retained, the complex is far sparser than the full Rips complex and emphasizes empty-circle geometry.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 36: Delaunay complex persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_036():

    i = 36
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _triangle_meb_radius(a,b,c):
        # exact Euclidean minimum-enclosing-ball radius for three points
        ds=np.array([np.linalg.norm(a-b),np.linalg.norm(a-c),np.linalg.norm(b-c)],float); L=np.max(ds)
        ss=np.sort(ds)
        if ss[2]**2 >= ss[0]**2+ss[1]**2-1e-12: return float(L/2)
        area=np.linalg.norm(np.cross(np.r_[b-a,0] if a.size==2 else b-a, np.r_[c-a,0] if a.size==2 else c-a))/2
        if area<=1e-12:return float(L/2)
        return float(np.prod(ds)/(4*area))

    def _delaunay_cells(X,alpha_style=False):
        Z=PCA(2).fit_transform(np.asarray(X,float)); tri=Delaunay(Z); n=len(Z); cells=[]; edge_val={}
        for i in range(n): cells.append((('v',i),0,0,[]))
        tvals={}
        for simp in tri.simplices:
            i,j,k=sorted(map(int,simp)); r=_triangle_meb_radius(Z[i],Z[j],Z[k]); tvals[(i,j,k)]=r
            for a,b in [(i,j),(i,k),(j,k)]: edge_val[(a,b)]=min(edge_val.get((a,b),np.inf),r)
        for (i,j),rv in edge_val.items():
            half=np.linalg.norm(Z[i]-Z[j])/2
            val=min(rv,half) if alpha_style else rv
            cells.append((('e',i,j),val,1,[('v',i),('v',j)]))
        for (i,j,k),r in tvals.items(): cells.append((('t',i,j,k),r,2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1),Z,tri

    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )

    # Method mathematics and output.
    X = _TOPO_XREF
    dg, _, _ = _delaunay_cells(X, alpha_style=False)
    _plot_diag(dg[1], 'Delaunay-complex H1 persistence')


_topology_method_036()
del _topology_method_036
```

### 37. Cubical persistence

Cubical persistence treats a scalar image as a cubical cell complex. In a lower-star filtration each vertex enters at its scalar value, each edge/square at the maximum value of its incident vertices. Matrix reduction over \(\mathbb Z_2\) then yields H0/H1 persistence.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 37: Cubical persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_037():

    i = 37
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _cubical_ph(A,upper=False):
        A=np.asarray(A,float); A=(-A if upper else A); nr,nc=A.shape; cells=[]
        # lower-star filtration on grid vertices; edge/square value=max of incident vertices
        for r in range(nr):
            for c in range(nc): cells.append((('v',r,c),float(A[r,c]),0,[]))
        for r in range(nr):
            for c in range(nc-1): cells.append((('eh',r,c),float(max(A[r,c],A[r,c+1])),1,[('v',r,c),('v',r,c+1)]))
        for r in range(nr-1):
            for c in range(nc): cells.append((('ev',r,c),float(max(A[r,c],A[r+1,c])),1,[('v',r,c),('v',r+1,c)]))
        for r in range(nr-1):
            for c in range(nc-1):
                val=float(np.max(A[r:r+2,c:c+2])); bd=[('eh',r,c),('eh',r+1,c),('ev',r,c),('ev',r,c+1)]
                cells.append((('sq',r,c),val,2,bd))
        dg=_ph_cells(cells,1)
        if upper:
            # transform filtration coordinates back to original superlevel threshold convention
            for d in range(len(dg)):
                if len(dg[d]): dg[d]=-dg[d][:,::-1]
        return dg

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    def _signal_image(y,nr=24,nc=32):
        n=nr*nc; z=signal.resample(np.asarray(y,float),n); return z.reshape(nr,nc)

    # Method mathematics and output.
    A = _signal_image(_topo_pc1, 18, 24)
    upper = i in (39, 41)
    dg = _cubical_ph(A, upper=upper)
    _plot_diag(dg[1], {37: 'Cubical', 38: 'Lower-star', 39: 'Upper-star', 40: 'Sublevel-set', 41: 'Superlevel-set'}[i] + ' H1 persistence')


_topology_method_037()
del _topology_method_037
```

### 38. Lower-star filtration

### 41. Superlevel-set persistence

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 41: Superlevel-set persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_041():

    i = 41
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _cubical_ph(A,upper=False):
        A=np.asarray(A,float); A=(-A if upper else A); nr,nc=A.shape; cells=[]
        # lower-star filtration on grid vertices; edge/square value=max of incident vertices
        for r in range(nr):
            for c in range(nc): cells.append((('v',r,c),float(A[r,c]),0,[]))
        for r in range(nr):
            for c in range(nc-1): cells.append((('eh',r,c),float(max(A[r,c],A[r,c+1])),1,[('v',r,c),('v',r,c+1)]))
        for r in range(nr-1):
            for c in range(nc): cells.append((('ev',r,c),float(max(A[r,c],A[r+1,c])),1,[('v',r,c),('v',r+1,c)]))
        for r in range(nr-1):
            for c in range(nc-1):
                val=float(np.max(A[r:r+2,c:c+2])); bd=[('eh',r,c),('eh',r+1,c),('ev',r,c),('ev',r,c+1)]
                cells.append((('sq',r,c),val,2,bd))
        dg=_ph_cells(cells,1)
        if upper:
            # transform filtration coordinates back to original superlevel threshold convention
            for d in range(len(dg)):
                if len(dg[d]): dg[d]=-dg[d][:,::-1]
        return dg

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    def _signal_image(y,nr=24,nc=32):
        n=nr*nc; z=signal.resample(np.asarray(y,float),n); return z.reshape(nr,nc)

    # Method mathematics and output.
    A = _signal_image(_topo_pc1, 18, 24)
    upper = i in (39, 41)
    dg = _cubical_ph(A, upper=upper)
    _plot_diag(dg[1], {37: 'Cubical', 38: 'Lower-star', 39: 'Upper-star', 40: 'Sublevel-set', 41: 'Superlevel-set'}[i] + ' H1 persistence')


_topology_method_041()
del _topology_method_041
```

### 42. Distance-to-measure persistence

Distance-to-measure (DTM) replaces raw distance by a robust local scale. For each sample point, \(\mathrm{DTM}_k(x)=\sqrt{k^{-1}\sum_{j=1}^k d(x,X_{(j)})^2}\). These values become vertex weights in a weighted Rips filtration, reducing sensitivity to isolated outliers.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 42: Distance-to-measure persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_042():

    i = 42
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _weighted_rips_ph(X,weights=None,density_scale=1.0):
        X=np.asarray(X,float); D=squareform(pdist(X)); n=len(X)
        if weights is None: weights=np.zeros(n)
        W=np.asarray(weights,float); E=np.zeros_like(D)
        for i in range(n):
            for j in range(i+1,n): E[i,j]=E[j,i]=max(D[i,j]/2,W[i],W[j])*density_scale
        return _flag_ph(E)

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )

    # Method mathematics and output.
    X = _TOPO_XREF
    D = squareform(pdist(X))
    k = min(6, len(X) - 1)
    nn = np.sort(D, axis=1)[:, 1:k + 1]
    w = np.sqrt(np.mean(nn ** 2, axis=1))
    dg = _weighted_rips_ph(X, w)
    _plot_diag(dg[1], {42: 'DTM-weighted', 43: 'Density-weighted', 44: 'Weighted Vietoris–Rips'}[i] + ' H1 persistence')


_topology_method_042()
del _topology_method_042
```

### 43. Density-weighted Rips persistence

Density-weighted Rips persistence assigns larger filtration penalties to points in low-density regions. Local kNN radius is used as an inverse-density scale and combined with pairwise distance so sparse outliers do not create early simplices.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 43: Density-weighted Rips persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_043():

    i = 43
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _weighted_rips_ph(X,weights=None,density_scale=1.0):
        X=np.asarray(X,float); D=squareform(pdist(X)); n=len(X)
        if weights is None: weights=np.zeros(n)
        W=np.asarray(weights,float); E=np.zeros_like(D)
        for i in range(n):
            for j in range(i+1,n): E[i,j]=E[j,i]=max(D[i,j]/2,W[i],W[j])*density_scale
        return _flag_ph(E)

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )

    # Method mathematics and output.
    X = _TOPO_XREF
    D = squareform(pdist(X))
    k = min(6, len(X) - 1)
    nn = np.sort(D, axis=1)[:, 1:k + 1]
    w = np.sqrt(np.mean(nn ** 2, axis=1))
    w = w / (np.median(w) + TOPO_EPS) * np.median(D[D > 0]) * 0.15
    dg = _weighted_rips_ph(X, w)
    _plot_diag(dg[1], {42: 'DTM-weighted', 43: 'Density-weighted', 44: 'Weighted Vietoris–Rips'}[i] + ' H1 persistence')


_topology_method_043()
del _topology_method_043
```

### 44. Weighted Vietoris-Rips persistence

Weighted Vietoris–Rips persistence augments the ordinary edge scale with vertex weights. Here an edge \((i,j)\) enters at \(\max(d_{ij}/2,w_i,w_j)\), and triangles enter at the maximum of their three edge scales.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Persistent homology summarizes how topological features appear and disappear as a distance or filtration threshold increases. In dimension 0 the features are connected components; in dimension 1 they are loops; higher dimensions represent higher-order cavities. Each feature receives a birth scale and a death scale, and its lifetime is the difference between them. Long lifetimes are often treated as more structurally important than very short ones, although noise and sampling density can still affect the diagram.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 44: Weighted Vietoris-Rips persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_044():

    i = 44
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _weighted_rips_ph(X,weights=None,density_scale=1.0):
        X=np.asarray(X,float); D=squareform(pdist(X)); n=len(X)
        if weights is None: weights=np.zeros(n)
        W=np.asarray(weights,float); E=np.zeros_like(D)
        for i in range(n):
            for j in range(i+1,n): E[i,j]=E[j,i]=max(D[i,j]/2,W[i],W[j])*density_scale
        return _flag_ph(E)

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )

    # Method mathematics and output.
    X = _TOPO_XREF
    D = squareform(pdist(X))
    k = min(6, len(X) - 1)
    nn = np.sort(D, axis=1)[:, 1:k + 1]
    w = np.sqrt(np.mean(nn ** 2, axis=1))
    w = 0.25 * w
    dg = _weighted_rips_ph(X, w)
    _plot_diag(dg[1], {42: 'DTM-weighted', 43: 'Density-weighted', 44: 'Weighted Vietoris–Rips'}[i] + ' H1 persistence')


_topology_method_044()
del _topology_method_044
```

### 45. kNN graph filtration

The kNN graph filtration retains only local-neighbour edges. Each retained edge carries its Euclidean length and the flag complex is filtered by edge length, which suppresses long shortcuts present in the complete Rips graph.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 45: kNN graph filtration

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_045():

    i = 45
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _knn_distance(X,k=6):
        X=np.asarray(X,float); D=squareform(pdist(X)); n=len(X); M=np.full_like(D,np.inf); np.fill_diagonal(M,0)
        for i in range(n):
            idx=np.argsort(D[i])[1:min(k+1,n)]; M[i,idx]=D[i,idx]
        M=np.minimum(M,M.T); return M

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )

    # Method mathematics and output.
    X = _TOPO_XREF
    D = _knn_distance(X, k=min(6, len(X) - 1))
    dg = _flag_ph(D)
    _plot_diag(dg[1], ('kNN graph' if i == 45 else 'Graph-distance') + ' H1 persistence')


_topology_method_045()
del _topology_method_045
```

### 46. Graph-distance filtration

Graph-distance filtration first constructs a sparse kNN graph, computes shortest-path geodesic distances, and then applies a flag/Rips filtration to those graph distances. It probes topology of the sampled manifold metric rather than direct Euclidean chords.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 46: Graph-distance filtration

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_046():

    i = 46
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _graph_geodesic_matrix(W):
        G=nx.Graph(); n=len(W); G.add_nodes_from(range(n))
        for i in range(n):
            for j in range(i+1,n):
                if np.isfinite(W[i,j]) and W[i,j]>0: G.add_edge(i,j,weight=float(W[i,j]))
        D=np.full((n,n),np.inf); np.fill_diagonal(D,0)
        for i,dd in nx.all_pairs_dijkstra_path_length(G,weight='weight'):
            for j,v in dd.items(): D[i,j]=v
        return D

    def _knn_distance(X,k=6):
        X=np.asarray(X,float); D=squareform(pdist(X)); n=len(X); M=np.full_like(D,np.inf); np.fill_diagonal(M,0)
        for i in range(n):
            idx=np.argsort(D[i])[1:min(k+1,n)]; M[i,idx]=D[i,idx]
        M=np.minimum(M,M.T); return M

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )

    # Method mathematics and output.
    X = _TOPO_XREF
    D = _knn_distance(X, k=min(6, len(X) - 1))
    D = _graph_geodesic_matrix(D)
    dg = _flag_ph(D)
    _plot_diag(dg[1], ('kNN graph' if i == 45 else 'Graph-distance') + ' H1 persistence')


_topology_method_046()
del _topology_method_046
```

### 47. Correlation-network clique-complex persistence

Correlation-network clique persistence uses dissimilarity \(d_{ij}=1-|r_{ij}|\). As the threshold increases, correlation edges appear and every clique is filled as a simplex, giving the flag-complex persistence of the functional network.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> A Lyapunov exponent describes how quickly two initially nearby states separate or approach one another as the system evolves. A positive exponent means small differences grow on average, a negative exponent means they shrink, and a value near zero corresponds to a direction with little average exponential change. When several exponents are reported, they describe stretching and contraction along different directions of the reconstructed state space rather than separate EEG frequency bands. The numerical value also has units set by the time step used in the analysis, so comparisons are most meaningful when preprocessing and sampling are consistent.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 47: Correlation-network clique-complex persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_047():

    i = 47
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _filter_band(y,lo,hi):
        sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_topo_fs,output='sos'); return signal.sosfiltfilt(sos,y)

    def _fc_mats(seg):
        seg=np.asarray(seg,float); C=np.nan_to_num(np.corrcoef(seg.T));
        # coherence in alpha band
        n=seg.shape[1]; COH=np.eye(n); PLV=np.eye(n); PLI=np.eye(n); MI=np.eye(n)
        lo,hi=8,min(13,_topo_fs/2-.5); fil=np.column_stack([_filter_band(seg[:,j],lo,hi) for j in range(n)]) if hi>lo else seg
        ph=np.angle(signal.hilbert(fil,axis=0))
        # discretized MI on z-scored amplitudes
        bins=np.linspace(-2.5,2.5,9)
        disc=np.clip(np.digitize(seg,bins),0,len(bins))
        for i in range(n):
            for j in range(i+1,n):
                f,co=signal.coherence(seg[:,i],seg[:,j],fs=_topo_fs,nperseg=min(256,len(seg))); m=(f>=lo)&(f<=hi); cv=float(np.nanmean(co[m])) if np.any(m) else 0
                d=ph[:,i]-ph[:,j]; pv=abs(np.mean(np.exp(1j*d))); pli=abs(np.mean(np.sign(np.sin(d))))
                mi=mutual_info_score(disc[:,i],disc[:,j]);
                for M,v in [(COH,cv),(PLV,pv),(PLI,pli),(MI,mi)]: M[i,j]=M[j,i]=v
        if np.max(MI)>0: MI=MI/np.max(MI)
        return C,COH,PLV,PLI,MI

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method mathematics and output.
    nch = min(_topo_x.shape[1], 20 if FAST_OPTION == 'full' else 14 if FAST_OPTION == 'balanced' else 10 if FAST_OPTION == 'fast' else 8)
    seg = _topo_x[:, :nch]
    C, COH, PLV, PLI, MI = _fc_mats(seg)
    def diss(S, mask=None):
        D = 1 - np.clip(np.abs(S), 0, 1)
        np.fill_diagonal(D, 0)
        if mask is not None:
            D = np.where(mask, D, np.inf)
            np.fill_diagonal(D, 0)
        return D
    D = diss(C)
    dg = _flag_ph(D)
    _plot_diag(dg[1], f'{i}. ' + [''][0] + 'network H1 persistence')


_topology_method_047()
del _topology_method_047
```

### 48. Coherence-network clique-complex persistence

Coherence-network clique persistence replaces correlation with magnitude-squared coherence in the alpha band, using \(d_{ij}=1-C_{ij}\). Because coherence is symmetric, this filtration is undirected.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Coherence is a frequency-specific measure of linear association between two signals. It compares the magnitude of their cross-spectrum with the power spectra of the individual signals. Values near one mean that a stable linear relationship at that frequency explains a large fraction of the spectral activity, while values near zero indicate weak consistency. High coherence can arise from common input, volume conduction, or shared reference effects, so it should not automatically be interpreted as direct connectivity.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. ### 49. PLV-network clique-complex persistence

PLV-network clique persistence uses \(\mathrm{PLV}_{ij}=|\langle e^{i(\phi_i-\phi_j)}\rangle|\) and dissimilarity \(1-\mathrm{PLV}_{ij}\), then forms the flag complex of the resulting weighted graph.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Phase-locking value measures how consistent the phase difference between two signals is across samples, trials, or windows. Each phase difference is converted to a unit complex number on the circle and these vectors are averaged. If phase differences are widely spread, they cancel and the magnitude approaches zero. PLV ignores amplitude and does not establish causal influence.</p><p><b>How to read the mathematics.</b> An exponential describes multiplicative growth or decay. In dynamical systems, a positive exponent corresponds to growth with time and a negative exponent to decay. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance. Expectation or angle-bracket notation denotes an average with respect to a probability distribution or ensemble.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>

### 50. Phase-lag-index clique-complex persistence

Phase-lag-index persistence uses \(\mathrm{PLI}_{ij}=|\langle\operatorname{sign}(\sin(\phi_i-\phi_j))\rangle|\). The flag filtration is based on \(1-\mathrm{PLI}\), emphasizing consistent nonzero phase leads/lags.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance. Expectation or angle-bracket notation denotes an average with respect to a probability distribution or ensemble. Greek phase symbols such as φ or θ represent angular position on a cycle. Because phase wraps around, differences are interpreted modulo one full cycle.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 50: Phase-lag-index clique-complex persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_050():

    i = 50
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _filter_band(y,lo,hi):
        sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_topo_fs,output='sos'); return signal.sosfiltfilt(sos,y)

    def _fc_mats(seg):
        seg=np.asarray(seg,float); C=np.nan_to_num(np.corrcoef(seg.T));
        # coherence in alpha band
        n=seg.shape[1]; COH=np.eye(n); PLV=np.eye(n); PLI=np.eye(n); MI=np.eye(n)
        lo,hi=8,min(13,_topo_fs/2-.5); fil=np.column_stack([_filter_band(seg[:,j],lo,hi) for j in range(n)]) if hi>lo else seg
        ph=np.angle(signal.hilbert(fil,axis=0))
        # discretized MI on z-scored amplitudes
        bins=np.linspace(-2.5,2.5,9)
        disc=np.clip(np.digitize(seg,bins),0,len(bins))
        for i in range(n):
            for j in range(i+1,n):
                f,co=signal.coherence(seg[:,i],seg[:,j],fs=_topo_fs,nperseg=min(256,len(seg))); m=(f>=lo)&(f<=hi); cv=float(np.nanmean(co[m])) if np.any(m) else 0
                d=ph[:,i]-ph[:,j]; pv=abs(np.mean(np.exp(1j*d))); pli=abs(np.mean(np.sign(np.sin(d))))
                mi=mutual_info_score(disc[:,i],disc[:,j]);
                for M,v in [(COH,cv),(PLV,pv),(PLI,pli),(MI,mi)]: M[i,j]=M[j,i]=v
        if np.max(MI)>0: MI=MI/np.max(MI)
        return C,COH,PLV,PLI,MI

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method mathematics and output.
    nch = min(_topo_x.shape[1], 20 if FAST_OPTION == 'full' else 14 if FAST_OPTION == 'balanced' else 10 if FAST_OPTION == 'fast' else 8)
    seg = _topo_x[:, :nch]
    C, COH, PLV, PLI, MI = _fc_mats(seg)
    def diss(S, mask=None):
        D = 1 - np.clip(np.abs(S), 0, 1)
        np.fill_diagonal(D, 0)
        if mask is not None:
            D = np.where(mask, D, np.inf)
            np.fill_diagonal(D, 0)
        return D
    D = diss(PLI)
    dg = _flag_ph(D)
    _plot_diag(dg[1], f'{i}. ' + [''][0] + 'network H1 persistence')


_topology_method_050()
del _topology_method_050
```

### 51. Mutual-information network clique-complex persistence

Mutual-information network persistence discretizes amplitudes, estimates pairwise mutual information, rescales it to \([0,1]\), and filters the clique complex by \(1-\mathrm{MI}_{ij}\). This captures nonlinear statistical dependence beyond correlation.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Mutual information measures how much knowing one variable reduces uncertainty about another. Unlike linear correlation, it can capture nonlinear statistical dependence. In delay-selection problems, mutual information between a signal and a delayed copy is examined as the delay changes; a local minimum is often used as a compromise between coordinates that are too redundant and coordinates that are almost unrelated. The estimate is sensitive to finite sample size and to how probability densities are estimated.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 51: Mutual-information network clique-complex persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_051():

    i = 51
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _filter_band(y,lo,hi):
        sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_topo_fs,output='sos'); return signal.sosfiltfilt(sos,y)

    def _fc_mats(seg):
        seg=np.asarray(seg,float); C=np.nan_to_num(np.corrcoef(seg.T));
        # coherence in alpha band
        n=seg.shape[1]; COH=np.eye(n); PLV=np.eye(n); PLI=np.eye(n); MI=np.eye(n)
        lo,hi=8,min(13,_topo_fs/2-.5); fil=np.column_stack([_filter_band(seg[:,j],lo,hi) for j in range(n)]) if hi>lo else seg
        ph=np.angle(signal.hilbert(fil,axis=0))
        # discretized MI on z-scored amplitudes
        bins=np.linspace(-2.5,2.5,9)
        disc=np.clip(np.digitize(seg,bins),0,len(bins))
        for i in range(n):
            for j in range(i+1,n):
                f,co=signal.coherence(seg[:,i],seg[:,j],fs=_topo_fs,nperseg=min(256,len(seg))); m=(f>=lo)&(f<=hi); cv=float(np.nanmean(co[m])) if np.any(m) else 0
                d=ph[:,i]-ph[:,j]; pv=abs(np.mean(np.exp(1j*d))); pli=abs(np.mean(np.sign(np.sin(d))))
                mi=mutual_info_score(disc[:,i],disc[:,j]);
                for M,v in [(COH,cv),(PLV,pv),(PLI,pli),(MI,mi)]: M[i,j]=M[j,i]=v
        if np.max(MI)>0: MI=MI/np.max(MI)
        return C,COH,PLV,PLI,MI

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method mathematics and output.
    nch = min(_topo_x.shape[1], 20 if FAST_OPTION == 'full' else 14 if FAST_OPTION == 'balanced' else 10 if FAST_OPTION == 'fast' else 8)
    seg = _topo_x[:, :nch]
    C, COH, PLV, PLI, MI = _fc_mats(seg)
    def diss(S, mask=None):
        D = 1 - np.clip(np.abs(S), 0, 1)
        np.fill_diagonal(D, 0)
        if mask is not None:
            D = np.where(mask, D, np.inf)
            np.fill_diagonal(D, 0)
        return D
    D = diss(MI)
    dg = _flag_ph(D)
    _plot_diag(dg[1], f'{i}. ' + [''][0] + 'network H1 persistence')


_topology_method_051()
del _topology_method_051
```

### 52. Signed correlation filtration

A signed-correlation filtration keeps sign information instead of replacing \(r\) by \(|r|\). Positive and negative edges are analyzed as separate weighted subgraphs because they encode different relation types.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 52: Signed correlation filtration

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_052():

    i = 52
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _filter_band(y,lo,hi):
        sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_topo_fs,output='sos'); return signal.sosfiltfilt(sos,y)

    def _fc_mats(seg):
        seg=np.asarray(seg,float); C=np.nan_to_num(np.corrcoef(seg.T));
        # coherence in alpha band
        n=seg.shape[1]; COH=np.eye(n); PLV=np.eye(n); PLI=np.eye(n); MI=np.eye(n)
        lo,hi=8,min(13,_topo_fs/2-.5); fil=np.column_stack([_filter_band(seg[:,j],lo,hi) for j in range(n)]) if hi>lo else seg
        ph=np.angle(signal.hilbert(fil,axis=0))
        # discretized MI on z-scored amplitudes
        bins=np.linspace(-2.5,2.5,9)
        disc=np.clip(np.digitize(seg,bins),0,len(bins))
        for i in range(n):
            for j in range(i+1,n):
                f,co=signal.coherence(seg[:,i],seg[:,j],fs=_topo_fs,nperseg=min(256,len(seg))); m=(f>=lo)&(f<=hi); cv=float(np.nanmean(co[m])) if np.any(m) else 0
                d=ph[:,i]-ph[:,j]; pv=abs(np.mean(np.exp(1j*d))); pli=abs(np.mean(np.sign(np.sin(d))))
                mi=mutual_info_score(disc[:,i],disc[:,j]);
                for M,v in [(COH,cv),(PLV,pv),(PLI,pli),(MI,mi)]: M[i,j]=M[j,i]=v
        if np.max(MI)>0: MI=MI/np.max(MI)
        return C,COH,PLV,PLI,MI

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method mathematics and output.
    nch = min(_topo_x.shape[1], 20 if FAST_OPTION == 'full' else 14 if FAST_OPTION == 'balanced' else 10 if FAST_OPTION == 'fast' else 8)
    seg = _topo_x[:, :nch]
    C, COH, PLV, PLI, MI = _fc_mats(seg)
    def diss(S, mask=None):
        D = 1 - np.clip(np.abs(S), 0, 1)
        np.fill_diagonal(D, 0)
        if mask is not None:
            D = np.where(mask, D, np.inf)
            np.fill_diagonal(D, 0)
        return D
    pos = _flag_ph(diss(np.maximum(C, 0), C > 0))
    neg = _flag_ph(diss(np.maximum(-C, 0), C < 0))
    out = pd.DataFrame([_feat(pos), _feat(neg)], index=['positive', 'negative'], columns=['H1_total', 'H1_max', 'entropy', 'count'])
    display(out)
    fig, axs = plt.subplots(2, 2, figsize=(9, 6.5), facecolor=TOPO_BG)
    axs = np.ravel(axs)
    for ax, col in zip(axs, out.columns):
        ax.bar(np.arange(len(out)), out[col].to_numpy(float), color=TOPO_CYAN_FILL, edgecolor=TOPO_ACCENT, alpha=0.75)
        ax.set_xticks(np.arange(len(out)))
        ax.set_xticklabels(out.index)
        ax.set_title(col)
        ax.grid(True, axis='y', color=TOPO_CYAN_DIM, alpha=0.22)
    fig.suptitle('Signed-correlation filtration topology')
    plt.tight_layout()
    plt.show()
    return
    dg = _flag_ph(D)
    _plot_diag(dg[1], f'{i}. ' + [''][0] + 'network H1 persistence')


_topology_method_052()
del _topology_method_052
```

### 53. Positive-correlation clique persistence

Positive-correlation clique persistence uses only \(r_{ij}>0\), with stronger positive correlations entering earlier through dissimilarity \(1-r_{ij}\).

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 53: Positive-correlation clique persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_053():

    i = 53
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _filter_band(y,lo,hi):
        sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_topo_fs,output='sos'); return signal.sosfiltfilt(sos,y)

    def _fc_mats(seg):
        seg=np.asarray(seg,float); C=np.nan_to_num(np.corrcoef(seg.T));
        # coherence in alpha band
        n=seg.shape[1]; COH=np.eye(n); PLV=np.eye(n); PLI=np.eye(n); MI=np.eye(n)
        lo,hi=8,min(13,_topo_fs/2-.5); fil=np.column_stack([_filter_band(seg[:,j],lo,hi) for j in range(n)]) if hi>lo else seg
        ph=np.angle(signal.hilbert(fil,axis=0))
        # discretized MI on z-scored amplitudes
        bins=np.linspace(-2.5,2.5,9)
        disc=np.clip(np.digitize(seg,bins),0,len(bins))
        for i in range(n):
            for j in range(i+1,n):
                f,co=signal.coherence(seg[:,i],seg[:,j],fs=_topo_fs,nperseg=min(256,len(seg))); m=(f>=lo)&(f<=hi); cv=float(np.nanmean(co[m])) if np.any(m) else 0
                d=ph[:,i]-ph[:,j]; pv=abs(np.mean(np.exp(1j*d))); pli=abs(np.mean(np.sign(np.sin(d))))
                mi=mutual_info_score(disc[:,i],disc[:,j]);
                for M,v in [(COH,cv),(PLV,pv),(PLI,pli),(MI,mi)]: M[i,j]=M[j,i]=v
        if np.max(MI)>0: MI=MI/np.max(MI)
        return C,COH,PLV,PLI,MI

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method mathematics and output.
    nch = min(_topo_x.shape[1], 20 if FAST_OPTION == 'full' else 14 if FAST_OPTION == 'balanced' else 10 if FAST_OPTION == 'fast' else 8)
    seg = _topo_x[:, :nch]
    C, COH, PLV, PLI, MI = _fc_mats(seg)
    def diss(S, mask=None):
        D = 1 - np.clip(np.abs(S), 0, 1)
        np.fill_diagonal(D, 0)
        if mask is not None:
            D = np.where(mask, D, np.inf)
            np.fill_diagonal(D, 0)
        return D
    D = diss(np.maximum(C, 0), C > 0)
    dg = _flag_ph(D)
    _plot_diag(dg[1], f'{i}. ' + [''][0] + 'network H1 persistence')


_topology_method_053()
del _topology_method_053
```

### 55. MST-plus-edge filtration

The MST-plus-edge filtration starts from the minimum spanning tree, which guarantees connectivity without cycles, then adds remaining edges in increasing dissimilarity order. H1 births directly mark redundant connections beyond the tree.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 55: MST-plus-edge filtration

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_055():

    i = 55
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _filter_band(y,lo,hi):
        sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_topo_fs,output='sos'); return signal.sosfiltfilt(sos,y)

    def _fc_mats(seg):
        seg=np.asarray(seg,float); C=np.nan_to_num(np.corrcoef(seg.T));
        # coherence in alpha band
        n=seg.shape[1]; COH=np.eye(n); PLV=np.eye(n); PLI=np.eye(n); MI=np.eye(n)
        lo,hi=8,min(13,_topo_fs/2-.5); fil=np.column_stack([_filter_band(seg[:,j],lo,hi) for j in range(n)]) if hi>lo else seg
        ph=np.angle(signal.hilbert(fil,axis=0))
        # discretized MI on z-scored amplitudes
        bins=np.linspace(-2.5,2.5,9)
        disc=np.clip(np.digitize(seg,bins),0,len(bins))
        for i in range(n):
            for j in range(i+1,n):
                f,co=signal.coherence(seg[:,i],seg[:,j],fs=_topo_fs,nperseg=min(256,len(seg))); m=(f>=lo)&(f<=hi); cv=float(np.nanmean(co[m])) if np.any(m) else 0
                d=ph[:,i]-ph[:,j]; pv=abs(np.mean(np.exp(1j*d))); pli=abs(np.mean(np.sign(np.sin(d))))
                mi=mutual_info_score(disc[:,i],disc[:,j]);
                for M,v in [(COH,cv),(PLV,pv),(PLI,pli),(MI,mi)]: M[i,j]=M[j,i]=v
        if np.max(MI)>0: MI=MI/np.max(MI)
        return C,COH,PLV,PLI,MI

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method mathematics and output.
    nch = min(_topo_x.shape[1], 20 if FAST_OPTION == 'full' else 14 if FAST_OPTION == 'balanced' else 10 if FAST_OPTION == 'fast' else 8)
    seg = _topo_x[:, :nch]
    C, COH, PLV, PLI, MI = _fc_mats(seg)
    def diss(S, mask=None):
        D = 1 - np.clip(np.abs(S), 0, 1)
        np.fill_diagonal(D, 0)
        if mask is not None:
            D = np.where(mask, D, np.inf)
            np.fill_diagonal(D, 0)
        return D
    base = diss(C)
    G = nx.Graph()
    G.add_nodes_from(range(nch))
    for a in range(nch):
        for b in range(a + 1, nch):
            G.add_edge(a, b, weight=float(base[a, b]))
    T = nx.minimum_spanning_tree(G, weight='weight')
    D = np.full_like(base, np.inf)
    np.fill_diagonal(D, 0)
    for a, b, d in T.edges(data=True):
        D[a, b] = D[b, a] = d['weight']
    for a in range(nch):
        for b in range(a + 1, nch):
            if not np.isfinite(D[a, b]):
                D[a, b] = D[b, a] = base[a, b] + 1e-09
    dg = _flag_ph(D)
    _plot_diag(dg[1], f'{i}. ' + [''][0] + 'network H1 persistence')


_topology_method_055()
del _topology_method_055
```

### 56. Functional-connectivity flag-complex persistence

For an undirected functional graph, the flag complex (clique complex) fills every complete subgraph with a simplex. Persistence of this filtration distinguishes transient graph cycles from loops that survive many connectivity thresholds.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 56: Functional-connectivity flag-complex persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_056():

    i = 56
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _filter_band(y,lo,hi):
        sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_topo_fs,output='sos'); return signal.sosfiltfilt(sos,y)

    def _fc_mats(seg):
        seg=np.asarray(seg,float); C=np.nan_to_num(np.corrcoef(seg.T));
        # coherence in alpha band
        n=seg.shape[1]; COH=np.eye(n); PLV=np.eye(n); PLI=np.eye(n); MI=np.eye(n)
        lo,hi=8,min(13,_topo_fs/2-.5); fil=np.column_stack([_filter_band(seg[:,j],lo,hi) for j in range(n)]) if hi>lo else seg
        ph=np.angle(signal.hilbert(fil,axis=0))
        # discretized MI on z-scored amplitudes
        bins=np.linspace(-2.5,2.5,9)
        disc=np.clip(np.digitize(seg,bins),0,len(bins))
        for i in range(n):
            for j in range(i+1,n):
                f,co=signal.coherence(seg[:,i],seg[:,j],fs=_topo_fs,nperseg=min(256,len(seg))); m=(f>=lo)&(f<=hi); cv=float(np.nanmean(co[m])) if np.any(m) else 0
                d=ph[:,i]-ph[:,j]; pv=abs(np.mean(np.exp(1j*d))); pli=abs(np.mean(np.sign(np.sin(d))))
                mi=mutual_info_score(disc[:,i],disc[:,j]);
                for M,v in [(COH,cv),(PLV,pv),(PLI,pli),(MI,mi)]: M[i,j]=M[j,i]=v
        if np.max(MI)>0: MI=MI/np.max(MI)
        return C,COH,PLV,PLI,MI

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method mathematics and output.
    nch = min(_topo_x.shape[1], 20 if FAST_OPTION == 'full' else 14 if FAST_OPTION == 'balanced' else 10 if FAST_OPTION == 'fast' else 8)
    seg = _topo_x[:, :nch]
    C, COH, PLV, PLI, MI = _fc_mats(seg)
    def diss(S, mask=None):
        D = 1 - np.clip(np.abs(S), 0, 1)
        np.fill_diagonal(D, 0)
        if mask is not None:
            D = np.where(mask, D, np.inf)
            np.fill_diagonal(D, 0)
        return D
    D = diss(C)
    dg = _flag_ph(D)
    _plot_diag(dg[1], f'{i}. ' + [''][0] + 'network H1 persistence')


_topology_method_056()
del _topology_method_056
```

### 57. Band-specific network persistence

Band-specific network persistence recomputes correlation networks after band-pass filtering into canonical frequency bands, then compares flag-complex H1 summaries across bands.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 57: Band-specific network persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_057():

    i = 57
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _filter_band(y,lo,hi):
        sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_topo_fs,output='sos'); return signal.sosfiltfilt(sos,y)

    def _fc_mats(seg):
        seg=np.asarray(seg,float); C=np.nan_to_num(np.corrcoef(seg.T));
        # coherence in alpha band
        n=seg.shape[1]; COH=np.eye(n); PLV=np.eye(n); PLI=np.eye(n); MI=np.eye(n)
        lo,hi=8,min(13,_topo_fs/2-.5); fil=np.column_stack([_filter_band(seg[:,j],lo,hi) for j in range(n)]) if hi>lo else seg
        ph=np.angle(signal.hilbert(fil,axis=0))
        # discretized MI on z-scored amplitudes
        bins=np.linspace(-2.5,2.5,9)
        disc=np.clip(np.digitize(seg,bins),0,len(bins))
        for i in range(n):
            for j in range(i+1,n):
                f,co=signal.coherence(seg[:,i],seg[:,j],fs=_topo_fs,nperseg=min(256,len(seg))); m=(f>=lo)&(f<=hi); cv=float(np.nanmean(co[m])) if np.any(m) else 0
                d=ph[:,i]-ph[:,j]; pv=abs(np.mean(np.exp(1j*d))); pli=abs(np.mean(np.sign(np.sin(d))))
                mi=mutual_info_score(disc[:,i],disc[:,j]);
                for M,v in [(COH,cv),(PLV,pv),(PLI,pli),(MI,mi)]: M[i,j]=M[j,i]=v
        if np.max(MI)>0: MI=MI/np.max(MI)
        return C,COH,PLV,PLI,MI

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method mathematics and output.
    nch = min(_topo_x.shape[1], 20 if FAST_OPTION == 'full' else 14 if FAST_OPTION == 'balanced' else 10 if FAST_OPTION == 'fast' else 8)
    seg = _topo_x[:, :nch]
    C, COH, PLV, PLI, MI = _fc_mats(seg)
    def diss(S, mask=None):
        D = 1 - np.clip(np.abs(S), 0, 1)
        np.fill_diagonal(D, 0)
        if mask is not None:
            D = np.where(mask, D, np.inf)
            np.fill_diagonal(D, 0)
        return D
    rows = []
    for bn, (lo, hi) in _TOPO_BANDS.items():
        if hi >= _topo_fs / 2:
            continue
        F = np.column_stack([_filter_band(seg[:, j], lo, hi) for j in range(nch)])
        R = np.nan_to_num(np.corrcoef(F.T))
        rows.append([bn, *_feat(_flag_ph(diss(R)))])
    out = pd.DataFrame(rows, columns=['band', 'H1_total', 'H1_max', 'entropy', 'count'])
    display(out)
    fig, axs = plt.subplots(2, 2, figsize=(9.5, 6.8), facecolor=TOPO_BG)
    axs = np.ravel(axs)
    for ax, col in zip(axs, ['H1_total', 'H1_max', 'entropy', 'count']):
        ax.bar(out['band'], out[col].to_numpy(float), color=TOPO_CYAN_FILL, edgecolor=TOPO_ACCENT, alpha=0.75)
        ax.set_title(col)
        ax.tick_params(axis='x', rotation=25)
        ax.grid(True, axis='y', color=TOPO_CYAN_DIM, alpha=0.22)
    fig.suptitle('Band-specific network persistence')
    plt.tight_layout()
    plt.show()
    return
    dg = _flag_ph(D)
    _plot_diag(dg[1], f'{i}. ' + [''][0] + 'network H1 persistence')


_topology_method_057()
del _topology_method_057
```

### 58. Threshold-filtration Betti curves

Threshold-filtration Betti curves evaluate \(\beta_0(\epsilon)\) and \(\beta_1(\epsilon)\) of a weighted functional-network flag complex over a dense threshold grid.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Persistent homology summarizes how topological features appear and disappear as a distance or filtration threshold increases. In dimension 0 the features are connected components; in dimension 1 they are loops; higher dimensions represent higher-order cavities. Each feature receives a birth scale and a death scale, and its lifetime is the difference between them. Long lifetimes are often treated as more structurally important than very short ones, although noise and sampling density can still affect the diagram.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance. The symbol ε (epsilon) usually represents a small threshold or radius. Changing it changes what the method counts as sufficiently close or similar.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 58: Threshold-filtration Betti curves

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_058():

    i = 58
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _betti_curve(dgm,epsgrid):
        d=np.asarray(dgm,float).reshape(-1,2)
        if not len(d): return np.zeros_like(epsgrid,float)
        return np.array([np.sum((d[:,0]<=e)&(e<d[:,1])) for e in epsgrid],float)

    def _filter_band(y,lo,hi):
        sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_topo_fs,output='sos'); return signal.sosfiltfilt(sos,y)

    def _fc_mats(seg):
        seg=np.asarray(seg,float); C=np.nan_to_num(np.corrcoef(seg.T));
        # coherence in alpha band
        n=seg.shape[1]; COH=np.eye(n); PLV=np.eye(n); PLI=np.eye(n); MI=np.eye(n)
        lo,hi=8,min(13,_topo_fs/2-.5); fil=np.column_stack([_filter_band(seg[:,j],lo,hi) for j in range(n)]) if hi>lo else seg
        ph=np.angle(signal.hilbert(fil,axis=0))
        # discretized MI on z-scored amplitudes
        bins=np.linspace(-2.5,2.5,9)
        disc=np.clip(np.digitize(seg,bins),0,len(bins))
        for i in range(n):
            for j in range(i+1,n):
                f,co=signal.coherence(seg[:,i],seg[:,j],fs=_topo_fs,nperseg=min(256,len(seg))); m=(f>=lo)&(f<=hi); cv=float(np.nanmean(co[m])) if np.any(m) else 0
                d=ph[:,i]-ph[:,j]; pv=abs(np.mean(np.exp(1j*d))); pli=abs(np.mean(np.sign(np.sin(d))))
                mi=mutual_info_score(disc[:,i],disc[:,j]);
                for M,v in [(COH,cv),(PLV,pv),(PLI,pli),(MI,mi)]: M[i,j]=M[j,i]=v
        if np.max(MI)>0: MI=MI/np.max(MI)
        return C,COH,PLV,PLI,MI

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    def _plot_line(x,y,title,xlabel='x',ylabel='value'):
        fig,ax=plt.subplots(figsize=(8,3.8),facecolor=TOPO_BG)
        ax.plot(x,y,color=TOPO_ACCENT,lw=1.7)
        ax.set_title(title); ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.28,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method mathematics and output.
    nch = min(_topo_x.shape[1], 20 if FAST_OPTION == 'full' else 14 if FAST_OPTION == 'balanced' else 10 if FAST_OPTION == 'fast' else 8)
    seg = _topo_x[:, :nch]
    C, COH, PLV, PLI, MI = _fc_mats(seg)
    def diss(S, mask=None):
        D = 1 - np.clip(np.abs(S), 0, 1)
        np.fill_diagonal(D, 0)
        if mask is not None:
            D = np.where(mask, D, np.inf)
            np.fill_diagonal(D, 0)
        return D
    D = diss(C)
    dg = _flag_ph(D)
    fin = np.hstack([_finite(d).ravel() for d in dg if len(_finite(d))])
    mx = float(np.max(fin)) if len(fin) else 1
    e = np.linspace(0, mx, 80)
    _plot_line(e, _betti_curve(dg[1], e), 'Threshold-filtration β1 curve', 'threshold', 'β1')
    return
    dg = _flag_ph(D)
    _plot_diag(dg[1], f'{i}. ' + [''][0] + 'network H1 persistence')


_topology_method_058()
del _topology_method_058
```

### 59. Persistent graph cycle localization

Persistent graph-cycle localization maps long-lived H1 behavior back to graph cycle bases near the corresponding filtration scales. Because homology generators are not unique, the localized cycle is a representative, not a canonical loop.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 59: Persistent graph cycle localization

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_059():

    i = 59
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _filter_band(y,lo,hi):
        sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_topo_fs,output='sos'); return signal.sosfiltfilt(sos,y)

    def _fc_mats(seg):
        seg=np.asarray(seg,float); C=np.nan_to_num(np.corrcoef(seg.T));
        # coherence in alpha band
        n=seg.shape[1]; COH=np.eye(n); PLV=np.eye(n); PLI=np.eye(n); MI=np.eye(n)
        lo,hi=8,min(13,_topo_fs/2-.5); fil=np.column_stack([_filter_band(seg[:,j],lo,hi) for j in range(n)]) if hi>lo else seg
        ph=np.angle(signal.hilbert(fil,axis=0))
        # discretized MI on z-scored amplitudes
        bins=np.linspace(-2.5,2.5,9)
        disc=np.clip(np.digitize(seg,bins),0,len(bins))
        for i in range(n):
            for j in range(i+1,n):
                f,co=signal.coherence(seg[:,i],seg[:,j],fs=_topo_fs,nperseg=min(256,len(seg))); m=(f>=lo)&(f<=hi); cv=float(np.nanmean(co[m])) if np.any(m) else 0
                d=ph[:,i]-ph[:,j]; pv=abs(np.mean(np.exp(1j*d))); pli=abs(np.mean(np.sign(np.sin(d))))
                mi=mutual_info_score(disc[:,i],disc[:,j]);
                for M,v in [(COH,cv),(PLV,pv),(PLI,pli),(MI,mi)]: M[i,j]=M[j,i]=v
        if np.max(MI)>0: MI=MI/np.max(MI)
        return C,COH,PLV,PLI,MI

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method mathematics and output.
    nch = min(_topo_x.shape[1], 20 if FAST_OPTION == 'full' else 14 if FAST_OPTION == 'balanced' else 10 if FAST_OPTION == 'fast' else 8)
    seg = _topo_x[:, :nch]
    C, COH, PLV, PLI, MI = _fc_mats(seg)
    def diss(S, mask=None):
        D = 1 - np.clip(np.abs(S), 0, 1)
        np.fill_diagonal(D, 0)
        if mask is not None:
            D = np.where(mask, D, np.inf)
            np.fill_diagonal(D, 0)
        return D
    D = diss(C)
    dg = _flag_ph(D)
    h = _finite(dg[1])
    th = float(np.median(h)) if len(h) else float(np.quantile(D[np.isfinite(D) & (D > 0)], 0.4))
    A = (D <= th) & np.isfinite(D)
    G = nx.from_numpy_array(A.astype(int))
    cycles = nx.cycle_basis(G)
    out = pd.DataFrame({'representative_cycle': [tuple(c) for c in cycles], 'length': [len(c) for c in cycles]})
    display(out)
    if len(out):
        fig, ax = plt.subplots(figsize=(9, 4.2), facecolor=TOPO_BG)
        ax.bar(np.arange(len(out)), out['length'], color=TOPO_CYAN_FILL, edgecolor=TOPO_ACCENT, alpha=0.75)
        ax.set_title('Persistent graph-cycle localization')
        ax.set_xlabel('cycle index')
        ax.set_ylabel('cycle length')
        ax.grid(True, axis='y', color=TOPO_CYAN_DIM, alpha=0.22)
        plt.tight_layout()
        plt.show()
    return
    counts = np.zeros(nch, int)
    for cyc in cycles:
        for v in cyc:
            counts[v] += 1
    regs = {'left': [j for j, n in enumerate(_topo_names[:nch]) if n.endswith(('1', '3', '5', '7'))], 'right': [j for j, n in enumerate(_topo_names[:nch]) if n.endswith(('2', '4', '6', '8'))], 'midline': [j for j, n in enumerate(_topo_names[:nch]) if n.endswith('z')]}
    out = pd.DataFrame({'region': list(regs), 'cycle_participation': [int(sum((counts[j] for j in js))) for js in regs.values()]})
    display(out)
    fig, ax = plt.subplots(figsize=(7, 4.2), facecolor=TOPO_BG)
    ax.bar(out['region'], out['cycle_participation'], color=TOPO_CYAN_FILL, edgecolor=TOPO_ACCENT, alpha=0.75)
    ax.set_title('Regional loop participation')
    ax.set_ylabel('cycle participation')
    ax.grid(True, axis='y', color=TOPO_CYAN_DIM, alpha=0.22)
    plt.tight_layout()
    plt.show()
    return
    dg = _flag_ph(D)
    _plot_diag(dg[1], f'{i}. ' + [''][0] + 'network H1 persistence')


_topology_method_059()
del _topology_method_059
```

### 60. Regional loop participation

Regional loop participation counts how often electrodes from anatomical regions appear in representative graph cycles over persistence-relevant thresholds. The statistic is descriptive because representative cycles can change within the same homology class.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. ### 61. Topological hubness

Topological hubness counts node participation in representative persistent-cycle bases, weighted by cycle persistence scale. It differs from degree centrality: a node can have many edges yet participate in few independent cycles.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 61: Topological hubness

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_061():

    i = 61
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _filter_band(y,lo,hi):
        sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_topo_fs,output='sos'); return signal.sosfiltfilt(sos,y)

    def _fc_mats(seg):
        seg=np.asarray(seg,float); C=np.nan_to_num(np.corrcoef(seg.T));
        # coherence in alpha band
        n=seg.shape[1]; COH=np.eye(n); PLV=np.eye(n); PLI=np.eye(n); MI=np.eye(n)
        lo,hi=8,min(13,_topo_fs/2-.5); fil=np.column_stack([_filter_band(seg[:,j],lo,hi) for j in range(n)]) if hi>lo else seg
        ph=np.angle(signal.hilbert(fil,axis=0))
        # discretized MI on z-scored amplitudes
        bins=np.linspace(-2.5,2.5,9)
        disc=np.clip(np.digitize(seg,bins),0,len(bins))
        for i in range(n):
            for j in range(i+1,n):
                f,co=signal.coherence(seg[:,i],seg[:,j],fs=_topo_fs,nperseg=min(256,len(seg))); m=(f>=lo)&(f<=hi); cv=float(np.nanmean(co[m])) if np.any(m) else 0
                d=ph[:,i]-ph[:,j]; pv=abs(np.mean(np.exp(1j*d))); pli=abs(np.mean(np.sign(np.sin(d))))
                mi=mutual_info_score(disc[:,i],disc[:,j]);
                for M,v in [(COH,cv),(PLV,pv),(PLI,pli),(MI,mi)]: M[i,j]=M[j,i]=v
        if np.max(MI)>0: MI=MI/np.max(MI)
        return C,COH,PLV,PLI,MI

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method mathematics and output.
    nch = min(_topo_x.shape[1], 20 if FAST_OPTION == 'full' else 14 if FAST_OPTION == 'balanced' else 10 if FAST_OPTION == 'fast' else 8)
    seg = _topo_x[:, :nch]
    C, COH, PLV, PLI, MI = _fc_mats(seg)
    def diss(S, mask=None):
        D = 1 - np.clip(np.abs(S), 0, 1)
        np.fill_diagonal(D, 0)
        if mask is not None:
            D = np.where(mask, D, np.inf)
            np.fill_diagonal(D, 0)
        return D
    D = diss(C)
    dg = _flag_ph(D)
    h = _finite(dg[1])
    th = float(np.median(h)) if len(h) else float(np.quantile(D[np.isfinite(D) & (D > 0)], 0.4))
    A = (D <= th) & np.isfinite(D)
    G = nx.from_numpy_array(A.astype(int))
    cycles = nx.cycle_basis(G)
    counts = np.zeros(nch, int)
    for cyc in cycles:
        for v in cyc:
            counts[v] += 1
    out = pd.DataFrame({'channel': _topo_names[:nch], 'cycle_hubness': counts}).sort_values('cycle_hubness', ascending=False)
    display(out)
    fig, ax = plt.subplots(figsize=(9, 4.5), facecolor=TOPO_BG)
    ax.bar(out['channel'], out['cycle_hubness'], color=TOPO_CYAN_FILL, edgecolor=TOPO_ACCENT, alpha=0.75)
    ax.set_title('Topological hubness by channel')
    ax.set_ylabel('cycle participation count')
    ax.tick_params(axis='x', rotation=55)
    ax.grid(True, axis='y', color=TOPO_CYAN_DIM, alpha=0.22)
    plt.tight_layout()
    plt.show()
    return
    regs = {'left': [j for j, n in enumerate(_topo_names[:nch]) if n.endswith(('1', '3', '5', '7'))], 'right': [j for j, n in enumerate(_topo_names[:nch]) if n.endswith(('2', '4', '6', '8'))], 'midline': [j for j, n in enumerate(_topo_names[:nch]) if n.endswith('z')]}
    out = pd.DataFrame({'region': list(regs), 'cycle_participation': [int(sum((counts[j] for j in js))) for js in regs.values()]})
    display(out)
    fig, ax = plt.subplots(figsize=(7, 4.2), facecolor=TOPO_BG)
    ax.bar(out['region'], out['cycle_participation'], color=TOPO_CYAN_FILL, edgecolor=TOPO_ACCENT, alpha=0.75)
    ax.set_title('Regional loop participation')
    ax.set_ylabel('cycle participation')
    ax.grid(True, axis='y', color=TOPO_CYAN_DIM, alpha=0.22)
    plt.tight_layout()
    plt.show()
    return
    dg = _flag_ph(D)
    _plot_diag(dg[1], f'{i}. ' + [''][0] + 'network H1 persistence')


_topology_method_061()
del _topology_method_061
```

### 62. Topology-preserving graph simplification

Topology-preserving graph simplification removes high-dissimilarity edges greedily while preserving the graph's H1 rank \(\beta_1=E-V+C\) at a selected threshold. This is a graph-level simplification criterion, not a guarantee that the full persistence diagram is unchanged.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> The visualization is a way of displaying already-computed numbers. Color, ordering, or ranking can make patterns easier to see but does not create additional statistical evidence beyond the values and uncertainty measures being plotted.</p></div>


```python
# Method 62: Topology-preserving graph simplification

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_062():

    i = 62
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _filter_band(y,lo,hi):
        sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_topo_fs,output='sos'); return signal.sosfiltfilt(sos,y)

    def _fc_mats(seg):
        seg=np.asarray(seg,float); C=np.nan_to_num(np.corrcoef(seg.T));
        # coherence in alpha band
        n=seg.shape[1]; COH=np.eye(n); PLV=np.eye(n); PLI=np.eye(n); MI=np.eye(n)
        lo,hi=8,min(13,_topo_fs/2-.5); fil=np.column_stack([_filter_band(seg[:,j],lo,hi) for j in range(n)]) if hi>lo else seg
        ph=np.angle(signal.hilbert(fil,axis=0))
        # discretized MI on z-scored amplitudes
        bins=np.linspace(-2.5,2.5,9)
        disc=np.clip(np.digitize(seg,bins),0,len(bins))
        for i in range(n):
            for j in range(i+1,n):
                f,co=signal.coherence(seg[:,i],seg[:,j],fs=_topo_fs,nperseg=min(256,len(seg))); m=(f>=lo)&(f<=hi); cv=float(np.nanmean(co[m])) if np.any(m) else 0
                d=ph[:,i]-ph[:,j]; pv=abs(np.mean(np.exp(1j*d))); pli=abs(np.mean(np.sign(np.sin(d))))
                mi=mutual_info_score(disc[:,i],disc[:,j]);
                for M,v in [(COH,cv),(PLV,pv),(PLI,pli),(MI,mi)]: M[i,j]=M[j,i]=v
        if np.max(MI)>0: MI=MI/np.max(MI)
        return C,COH,PLV,PLI,MI

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method mathematics and output.
    nch = min(_topo_x.shape[1], 20 if FAST_OPTION == 'full' else 14 if FAST_OPTION == 'balanced' else 10 if FAST_OPTION == 'fast' else 8)
    seg = _topo_x[:, :nch]
    C, COH, PLV, PLI, MI = _fc_mats(seg)
    def diss(S, mask=None):
        D = 1 - np.clip(np.abs(S), 0, 1)
        np.fill_diagonal(D, 0)
        if mask is not None:
            D = np.where(mask, D, np.inf)
            np.fill_diagonal(D, 0)
        return D
    D = diss(C)
    th = float(np.quantile(D[np.isfinite(D) & (D > 0)], 0.45))
    A = (D <= th) & np.isfinite(D)
    G = nx.from_numpy_array(A.astype(int))
    beta = lambda H: H.number_of_edges() - H.number_of_nodes() + nx.number_connected_components(H)
    b0 = beta(G)
    H = G.copy()
    edges = sorted(H.edges(), key=lambda e: D[e[0], e[1]], reverse=True)
    for e in edges:
        H.remove_edge(*e)
        if beta(H) != b0:
            H.add_edge(*e)
    print('edges original=', G.number_of_edges(), ' simplified=', H.number_of_edges(), ' beta1=', b0)
    fig, ax = plt.subplots(figsize=(6.8, 4.2), facecolor=TOPO_BG)
    vals = [G.number_of_edges(), H.number_of_edges()]
    ax.bar(['original edges', 'simplified edges'], vals, color=TOPO_CYAN_FILL, edgecolor=TOPO_ACCENT, alpha=0.75)
    ax.axhline(b0, color=TOPO_CYAN_SOFT, linestyle='--', linewidth=1.4, label=f'β1 = {b0}')
    ax.set_title('Topology-preserving graph simplification')
    ax.set_ylabel('count')
    ax.grid(True, axis='y', color=TOPO_CYAN_DIM, alpha=0.22)
    ax.legend()
    plt.tight_layout()
    plt.show()
    return
    dg = _flag_ph(D)
    _plot_diag(dg[1], f'{i}. ' + [''][0] + 'network H1 persistence')


_topology_method_062()
del _topology_method_062
```

### 63. Clique-complex Euler curves

For a flag complex truncated at triangles, the Euler characteristic is \(\chi=V-E+F\), where \(F\) is the number of 3-cliques. Evaluating \(\chi(\epsilon)\) over the clique filtration produces an Euler curve.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance. The symbol ε (epsilon) usually represents a small threshold or radius. Changing it changes what the method counts as sufficiently close or similar.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 63: Clique-complex Euler curves

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_063():

    i = 63
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _euler_flag_curve(D,grid):
        D=np.asarray(D,float); n=len(D); vals=[]
        for e in grid:
            A=(D<=e)&np.isfinite(D); np.fill_diagonal(A,False); V=n; E=int(np.sum(np.triu(A,1))); T=0
            for i in range(n):
                for j in range(i+1,n):
                    if A[i,j]: T+=int(np.sum(A[i,j+1:]&A[j,j+1:]))
            vals.append(V-E+T)
        return np.asarray(vals)

    def _filter_band(y,lo,hi):
        sos=signal.butter(4,[lo,hi],btype='bandpass',fs=_topo_fs,output='sos'); return signal.sosfiltfilt(sos,y)

    def _fc_mats(seg):
        seg=np.asarray(seg,float); C=np.nan_to_num(np.corrcoef(seg.T));
        # coherence in alpha band
        n=seg.shape[1]; COH=np.eye(n); PLV=np.eye(n); PLI=np.eye(n); MI=np.eye(n)
        lo,hi=8,min(13,_topo_fs/2-.5); fil=np.column_stack([_filter_band(seg[:,j],lo,hi) for j in range(n)]) if hi>lo else seg
        ph=np.angle(signal.hilbert(fil,axis=0))
        # discretized MI on z-scored amplitudes
        bins=np.linspace(-2.5,2.5,9)
        disc=np.clip(np.digitize(seg,bins),0,len(bins))
        for i in range(n):
            for j in range(i+1,n):
                f,co=signal.coherence(seg[:,i],seg[:,j],fs=_topo_fs,nperseg=min(256,len(seg))); m=(f>=lo)&(f<=hi); cv=float(np.nanmean(co[m])) if np.any(m) else 0
                d=ph[:,i]-ph[:,j]; pv=abs(np.mean(np.exp(1j*d))); pli=abs(np.mean(np.sign(np.sin(d))))
                mi=mutual_info_score(disc[:,i],disc[:,j]);
                for M,v in [(COH,cv),(PLV,pv),(PLI,pli),(MI,mi)]: M[i,j]=M[j,i]=v
        if np.max(MI)>0: MI=MI/np.max(MI)
        return C,COH,PLV,PLI,MI

    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _flag_ph(D):
        D=np.asarray(D,float); n=len(D); cells=[]
        for i in range(n): cells.append((('v',i),0,0,[]))
        finite=np.isfinite(D)
        for i in range(n):
            for j in range(i+1,n):
                if finite[i,j]: cells.append((('e',i,j),float(D[i,j]),1,[('v',i),('v',j)]))
        for i in range(n):
            for j in range(i+1,n):
                if not finite[i,j]: continue
                for k in range(j+1,n):
                    if finite[i,k] and finite[j,k]: cells.append((('t',i,j,k),float(max(D[i,j],D[i,k],D[j,k])),2,[('e',i,j),('e',i,k),('e',j,k)]))
        return _ph_cells(cells,1)

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    def _plot_line(x,y,title,xlabel='x',ylabel='value'):
        fig,ax=plt.subplots(figsize=(8,3.8),facecolor=TOPO_BG)
        ax.plot(x,y,color=TOPO_ACCENT,lw=1.7)
        ax.set_title(title); ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.28,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method mathematics and output.
    nch = min(_topo_x.shape[1], 20 if FAST_OPTION == 'full' else 14 if FAST_OPTION == 'balanced' else 10 if FAST_OPTION == 'fast' else 8)
    seg = _topo_x[:, :nch]
    C, COH, PLV, PLI, MI = _fc_mats(seg)
    def diss(S, mask=None):
        D = 1 - np.clip(np.abs(S), 0, 1)
        np.fill_diagonal(D, 0)
        if mask is not None:
            D = np.where(mask, D, np.inf)
            np.fill_diagonal(D, 0)
        return D
    D = diss(C)
    vals = D[np.isfinite(D) & (D > 0)]
    grid = np.linspace(vals.min(), vals.max(), 80)
    chi = _euler_flag_curve(D, grid)
    _plot_line(grid, chi, 'Clique-complex Euler curve' if i == 63 else 'Graph-filtration Euler characteristic curve', 'threshold ε', 'χ')
    return
    dg = _flag_ph(D)
    _plot_diag(dg[1], f'{i}. ' + [''][0] + 'network H1 persistence')


_topology_method_063()
del _topology_method_063
```

### 65. True H2 persistence with tetrahedra

For a Vietoris–Rips filtration, an H2 class is a closed triangulated surface modulo boundaries. Correct H2 persistence therefore requires **tetrahedra**: triangles can create 2-cycles, while 3-simplices kill them. The filtration value of a simplex \(\sigma\) is \(\max_{i,j\in\sigma} d(x_i,x_j)\), and the boundary matrix is reduced over \(\mathbb Z_2\).

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Persistent homology summarizes how topological features appear and disappear as a distance or filtration threshold increases. In dimension 0 the features are connected components; in dimension 1 they are loops; higher dimensions represent higher-order cavities. Each feature receives a birth scale and a death scale, and its lifetime is the difference between them. Long lifetimes are often treated as more structurally important than very short ones, although noise and sampling density can still affect the diagram.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 65: True H2 persistence with tetrahedra

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_065():

    i = 65
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _ph_simplices_nd(ss, max_hom_dim):
        index={simp:i for i,(simp,_,_) in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0:
                col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=max_hom_dim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else:
                births.add(j)
        paired=set(low_to_col)
        dg=[]
        for d in range(max_hom_dim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dg.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dg

    def _rips_simplices_nd(X, max_hom_dim=2, max_points=None):
        X=np.asarray(X,float)
        if max_points is None: max_points=_TOPO_HD_POINTS
        if len(X)>max_points:
            X=X[np.linspace(0,len(X)-1,max_points).astype(int)]
        n=len(X); D=squareform(pdist(X)) if n>1 else np.zeros((n,n))
        max_simplex_dim=int(max_hom_dim)+1  # cofaces one dimension higher kill H_k
        ss=[]
        for r in range(1,max_simplex_dim+2):
            dim=r-1
            for simp in combinations(range(n),r):
                filt=0.0 if r==1 else float(max(D[a,b] for a,b in combinations(simp,2)))
                ss.append((tuple(simp),filt,dim))
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss, X

    def _ph_rips_nd(X,max_hom_dim=2,max_points=None):
        ss,Xs=_rips_simplices_nd(X,max_hom_dim,max_points)
        return _ph_simplices_nd(ss,max_hom_dim), Xs

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method-local derived quantities required by this method.
    _TOPO_XREF = _delay_embed(
        _topo_pc1,
        m=3,
        tau=max(1, int(round(0.01 * _topo_fs))),
    )

    # Method mathematics and output.
    dg, X = _ph_rips_nd(_TOPO_XREF, 2)
    _plot_diag(dg[2], 'True H2 Rips persistence (tetrahedra included)')
    return


_topology_method_065()
del _topology_method_065
```

### 66. H3 persistence

H3 persistence requires one dimension more again: tetrahedral 3-cycles are killed by 4-simplices. The reducer therefore contains simplices through dimension four and reports \(H_3=\ker \partial_3/\operatorname{im}\partial_4\). The speed modes reduce the sampled point count because the number of 4-simplices grows as \(\binom{n}{5}\).

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> A Lyapunov exponent describes how quickly two initially nearby states separate or approach one another as the system evolves. A positive exponent means small differences grow on average, a negative exponent means they shrink, and a value near zero corresponds to a direction with little average exponential change. When several exponents are reported, they describe stretching and contraction along different directions of the reconstructed state space rather than separate EEG frequency bands. The numerical value also has units set by the time step used in the analysis, so comparisons are most meaningful when preprocessing and sampling are consistent.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance. A derivative measures how quickly one quantity changes when another changes. Partial derivatives are used when a function depends on several variables.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>

### 67. High-dimensional Rips persistence

High-dimensional Rips persistence applies the same metric filtration to a higher-dimensional delay embedding and computes several homology dimensions simultaneously. For each \(k\), the notebook reports finite-class count, total persistence \(\sum_i(d_i-b_i)\), and maximum lifetime \(\max_i(d_i-b_i)\).

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Delay embedding turns one measured time series into a sequence of state-space vectors by combining the present sample with delayed copies of the same signal. The idea is that the hidden state of a dynamical system can sometimes be reconstructed from a single observable when enough delayed coordinates are used. The delay controls how far apart in time the coordinates are, while the embedding dimension controls how many delayed coordinates are included. Too small a delay makes coordinates nearly redundant; too large a delay can make them weakly related. Too small an embedding dimension can fold distinct states onto one another.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 67: High-dimensional Rips persistence

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_067():

    i = 67
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _ph_simplices_nd(ss, max_hom_dim):
        index={simp:i for i,(simp,_,_) in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0:
                col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=max_hom_dim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else:
                births.add(j)
        paired=set(low_to_col)
        dg=[]
        for d in range(max_hom_dim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dg.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dg

    def _rips_simplices_nd(X, max_hom_dim=2, max_points=None):
        X=np.asarray(X,float)
        if max_points is None: max_points=_TOPO_HD_POINTS
        if len(X)>max_points:
            X=X[np.linspace(0,len(X)-1,max_points).astype(int)]
        n=len(X); D=squareform(pdist(X)) if n>1 else np.zeros((n,n))
        max_simplex_dim=int(max_hom_dim)+1  # cofaces one dimension higher kill H_k
        ss=[]
        for r in range(1,max_simplex_dim+2):
            dim=r-1
            for simp in combinations(range(n),r):
                filt=0.0 if r==1 else float(max(D[a,b] for a,b in combinations(simp,2)))
                ss.append((tuple(simp),filt,dim))
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss, X

    def _ph_rips_nd(X,max_hom_dim=2,max_points=None):
        ss,Xs=_rips_simplices_nd(X,max_hom_dim,max_points)
        return _ph_simplices_nd(ss,max_hom_dim), Xs

    # Method mathematics and output.
    X = _delay_embed(_topo_pc1, m=6, tau=max(1, int(0.008 * _topo_fs)), max_points=_TOPO_HD_POINTS)
    dg, _ = _ph_rips_nd(X, 3)
    rows = []
    for d in range(4):
        L = _life(dg[d])
        rows.append([d, len(L), L.sum(), L.max() if len(L) else 0.0])
    out = pd.DataFrame(rows, columns=['homology_dim', 'finite_classes', 'total_persistence', 'max_lifetime'])
    display(out)
    fig, axs = plt.subplots(1, 3, figsize=(12, 3.8), facecolor=TOPO_BG)
    for ax, col in zip(axs, ['finite_classes', 'total_persistence', 'max_lifetime']):
        ax.bar(out['homology_dim'].astype(str), out[col].to_numpy(float), color=TOPO_CYAN_FILL, edgecolor=TOPO_ACCENT, alpha=0.75)
        ax.set_title(col)
        ax.set_xlabel('homology dimension')
        ax.grid(True, axis='y', color=TOPO_CYAN_DIM, alpha=0.22)
    fig.suptitle('High-dimensional Rips persistence')
    plt.tight_layout()
    plt.show()
    return


_topology_method_067()
del _topology_method_067
```

### 68. Persistent homology on 3D embeddings

A 3-D delay embedding maps \(x_t\) to \((x_t,x_{t-\tau},x_{t-2\tau})\). Persistent \(H_0,H_1,H_2\) then quantify connected components, loops, and enclosed void-like 2-cycles of the sampled trajectory across Rips scale.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Persistent homology summarizes how topological features appear and disappear as a distance or filtration threshold increases. In dimension 0 the features are connected components; in dimension 1 they are loops; higher dimensions represent higher-order cavities. Each feature receives a birth scale and a death scale, and its lifetime is the difference between them. Long lifetimes are often treated as more structurally important than very short ones, although noise and sampling density can still affect the diagram.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance. The symbol τ (tau) commonly denotes a time delay, scale, or characteristic time. Here it should be read according to the definition immediately preceding the formula.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 68: Persistent homology on 3D embeddings

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_068():

    i = 68
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _ph_simplices_nd(ss, max_hom_dim):
        index={simp:i for i,(simp,_,_) in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0:
                col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=max_hom_dim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else:
                births.add(j)
        paired=set(low_to_col)
        dg=[]
        for d in range(max_hom_dim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dg.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dg

    def _rips_simplices_nd(X, max_hom_dim=2, max_points=None):
        X=np.asarray(X,float)
        if max_points is None: max_points=_TOPO_HD_POINTS
        if len(X)>max_points:
            X=X[np.linspace(0,len(X)-1,max_points).astype(int)]
        n=len(X); D=squareform(pdist(X)) if n>1 else np.zeros((n,n))
        max_simplex_dim=int(max_hom_dim)+1  # cofaces one dimension higher kill H_k
        ss=[]
        for r in range(1,max_simplex_dim+2):
            dim=r-1
            for simp in combinations(range(n),r):
                filt=0.0 if r==1 else float(max(D[a,b] for a,b in combinations(simp,2)))
                ss.append((tuple(simp),filt,dim))
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss, X

    def _ph_rips_nd(X,max_hom_dim=2,max_points=None):
        ss,Xs=_rips_simplices_nd(X,max_hom_dim,max_points)
        return _ph_simplices_nd(ss,max_hom_dim), Xs

    # Method mathematics and output.
    X = _delay_embed(_topo_pc1, m=3, tau=max(1, int(0.01 * _topo_fs)), max_points=_TOPO_HD_POINTS)
    dg, _ = _ph_rips_nd(X, 2)
    out = pd.DataFrame({'H': ['H0', 'H1', 'H2'], 'finite': [len(_life(d)) for d in dg], 'total': [float(_life(d).sum()) for d in dg]})
    display(out)
    fig, axs = plt.subplots(1, 2, figsize=(8.5, 3.8), facecolor=TOPO_BG)
    for ax, col in zip(axs, ['finite', 'total']):
        ax.bar(out['H'], out[col], color=TOPO_CYAN_FILL, edgecolor=TOPO_ACCENT, alpha=0.75)
        ax.set_title(col)
        ax.grid(True, axis='y', color=TOPO_CYAN_DIM, alpha=0.22)
    fig.suptitle('Persistent homology on a 3D embedding')
    plt.tight_layout()
    plt.show()
    return


_topology_method_068()
del _topology_method_068
```

### 69. Persistent homology on 4D-to-10D embeddings

Embedding dimension is varied from 4 through 10 while keeping the delay rule fixed. Comparing H1 and H2 total persistence tests whether higher-dimensional reconstructions reveal stable topology rather than topology created solely by a low-dimensional projection.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Delay embedding turns one measured time series into a sequence of state-space vectors by combining the present sample with delayed copies of the same signal. The idea is that the hidden state of a dynamical system can sometimes be reconstructed from a single observable when enough delayed coordinates are used. The delay controls how far apart in time the coordinates are, while the embedding dimension controls how many delayed coordinates are included. Too small a delay makes coordinates nearly redundant; too large a delay can make them weakly related. Too small an embedding dimension can fold distinct states onto one another.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 69: Persistent homology on 4D-to-10D embeddings

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_069():

    i = 69
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _ph_simplices_nd(ss, max_hom_dim):
        index={simp:i for i,(simp,_,_) in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0:
                col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=max_hom_dim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else:
                births.add(j)
        paired=set(low_to_col)
        dg=[]
        for d in range(max_hom_dim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dg.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dg

    def _rips_simplices_nd(X, max_hom_dim=2, max_points=None):
        X=np.asarray(X,float)
        if max_points is None: max_points=_TOPO_HD_POINTS
        if len(X)>max_points:
            X=X[np.linspace(0,len(X)-1,max_points).astype(int)]
        n=len(X); D=squareform(pdist(X)) if n>1 else np.zeros((n,n))
        max_simplex_dim=int(max_hom_dim)+1  # cofaces one dimension higher kill H_k
        ss=[]
        for r in range(1,max_simplex_dim+2):
            dim=r-1
            for simp in combinations(range(n),r):
                filt=0.0 if r==1 else float(max(D[a,b] for a,b in combinations(simp,2)))
                ss.append((tuple(simp),filt,dim))
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss, X

    def _ph_rips_nd(X,max_hom_dim=2,max_points=None):
        ss,Xs=_rips_simplices_nd(X,max_hom_dim,max_points)
        return _ph_simplices_nd(ss,max_hom_dim), Xs

    # Method mathematics and output.
    rows = []
    for m in range(4, 11):
        X = _delay_embed(_topo_pc1, m=m, tau=max(1, int(0.008 * _topo_fs)), max_points=_TOPO_HD_POINTS)
        dg, _ = _ph_rips_nd(X, 2)
        rows.append([m, float(_life(dg[1]).sum()), float(_life(dg[2]).sum())])
    out = pd.DataFrame(rows, columns=['embedding_dim', 'H1_total', 'H2_total'])
    display(out)
    fig, ax = plt.subplots(figsize=(8.5, 4.1), facecolor=TOPO_BG)
    ax.plot(out['embedding_dim'], out['H1_total'], color=TOPO_ACCENT, linewidth=1.7, marker='o', label='H1 total')
    ax.plot(out['embedding_dim'], out['H2_total'], color=TOPO_CYAN_SOFT, linewidth=1.5, marker='s', linestyle='--', label='H2 total')
    ax.set_title('Persistent homology across 4D–10D embeddings')
    ax.set_xlabel('embedding dimension')
    ax.set_ylabel('total persistence')
    ax.grid(True, color=TOPO_CYAN_DIM, alpha=0.22)
    ax.legend()
    plt.tight_layout()
    plt.show()
    return


_topology_method_069()
del _topology_method_069
```

### 70. Persistent homology across embedding dimension and delay jointly

The joint audit evaluates a persistence statistic on a grid of embedding dimension \(m\) and delay \(\tau\): \(T(m,\tau)=\sum_i(d_i-b_i)\) for H1. Stable plateaus indicate reconstruction choices for which the inferred loop structure is comparatively insensitive to either parameter.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Delay embedding turns one measured time series into a sequence of state-space vectors by combining the present sample with delayed copies of the same signal. The idea is that the hidden state of a dynamical system can sometimes be reconstructed from a single observable when enough delayed coordinates are used. The delay controls how far apart in time the coordinates are, while the embedding dimension controls how many delayed coordinates are included. Too small a delay makes coordinates nearly redundant; too large a delay can make them weakly related. Too small an embedding dimension can fold distinct states onto one another.</p><p><b>How to read the mathematics.</b> The summation symbol means to add the indicated quantity over the stated index range; dividing such a sum by the number of terms produces an average. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance. The symbol τ (tau) commonly denotes a time delay, scale, or characteristic time. Here it should be read according to the definition immediately preceding the formula.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 70: Persistent homology across embedding dimension and delay jointly

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_070():

    i = 70
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _imshow(A,title,xlabel='',ylabel=''):
        fig,ax=plt.subplots(figsize=(7,4)); im=ax.imshow(A,aspect='auto',origin='lower',cmap=TOPO_CMAP); ax.set_title(title); ax.set_xlabel(xlabel); ax.set_ylabel(ylabel); plt.colorbar(im,ax=ax); plt.show()

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    # Method mathematics and output.
    dims = range(2, min(7, _TOPO_SPEED['emb_max'] + 1))
    taus = [1, max(2, int(0.005 * _topo_fs)), max(3, int(0.01 * _topo_fs))]
    M = np.zeros((len(list(dims)), len(taus)))
    for a, m in enumerate(dims):
        for b, tau in enumerate(taus):
            M[a, b] = _life(_ph_rips(_delay_embed(_topo_pc1, m, tau), 1)[1]).sum()
    _imshow(M, 'H1 total persistence across embedding dimension × delay', 'delay index', 'embedding dimension')
    return


_topology_method_070()
del _topology_method_070
```

### 71. Projection distortion topology audit

Projection distortion is quantified by comparing persistence after PCA projection with a 3-D reference projection. The displayed diagram discrepancy is a **lifetime-spectrum Euclidean proxy**, not bottleneck distance: sorted lifetime vectors are zero-padded and compared in \(\ell_2\). Variance retained is reported separately.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The bottleneck distance compares two persistence diagrams by matching their birth-death points, with unmatched features allowed to pair with the diagonal. It then looks at the largest matching cost and chooses the matching that makes this worst cost as small as possible. Because it is controlled by the single largest discrepancy, it emphasizes worst-case topological change rather than the sum of many small differences.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 71: Projection distortion topology audit

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_071():

    i = 71
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _diag_dist(a,b):
        # Lifetime-spectrum distance: stable summary proxy, not bottleneck/Wasserstein distance.
        la=np.sort(_life(a))[::-1]; lb=np.sort(_life(b))[::-1]; n=max(len(la),len(lb),1)
        aa=np.pad(la,(0,n-len(la))); bb=np.pad(lb,(0,n-len(lb)))
        return float(np.linalg.norm(aa-bb))

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    # Method mathematics and output.
    X = _delay_embed(_topo_pc1, m=7, tau=max(1, int(0.008 * _topo_fs)), max_points=max(_TOPO_SPEED['points'], _TOPO_HD_POINTS))
    ref = _ph_rips(PCA(3).fit_transform(X), 1)[1]
    rows = []
    for d in [2, 3, 4, 5]:
        Z = PCA(min(d, X.shape[1])).fit_transform(X)
        h = _ph_rips(Z, 1)[1]
        rows.append([d, _diag_dist(ref, h), PCA(min(d, X.shape[1])).fit(X).explained_variance_ratio_.sum()])
    out = pd.DataFrame(rows, columns=['projection_dim', 'H1_lifetime_distance_to_3D', 'variance_retained'])
    display(out)
    fig, ax = plt.subplots(figsize=(8.5, 4.1), facecolor=TOPO_BG)
    ax2 = ax.twinx()
    ax.plot(out['projection_dim'], out['H1_lifetime_distance_to_3D'], color=TOPO_ACCENT, marker='o', linewidth=1.7, label='H1 distance')
    ax2.plot(out['projection_dim'], out['variance_retained'], color=TOPO_CYAN_SOFT, marker='s', linestyle='--', linewidth=1.4, label='variance retained')
    ax.set_title('Projection-distortion topology audit')
    ax.set_xlabel('projection dimension')
    ax.set_ylabel('H1 distance')
    ax2.set_ylabel('variance retained', color=TOPO_CYAN_SOFT)
    ax.grid(True, color=TOPO_CYAN_DIM, alpha=0.22)
    lines = ax.get_lines() + ax2.get_lines()
    ax.legend(lines, [ln.get_label() for ln in lines], loc='best')
    plt.tight_layout()
    plt.show()
    return


_topology_method_071()
del _topology_method_071
```

### 72. Topological saturation across embeddings

Topological saturation across embeddings uses increments \(\|z_m-z_{m-1}\|_2\) of the persistence feature vector \(z=[TP_1,L_{\max},H_{pers},N_1]\). Small successive increments are evidence of empirical saturation; they are not a proof of a Takens embedding dimension.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> Delay embedding turns one measured time series into a sequence of state-space vectors by combining the present sample with delayed copies of the same signal. The idea is that the hidden state of a dynamical system can sometimes be reconstructed from a single observable when enough delayed coordinates are used. The delay controls how far apart in time the coordinates are, while the embedding dimension controls how many delayed coordinates are included. Too small a delay makes coordinates nearly redundant; too large a delay can make them weakly related. Too small an embedding dimension can fold distinct states onto one another.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 72: Topological saturation across embeddings

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_072():

    i = 72
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    # Method mathematics and output.
    rows = []
    prev = None
    for m in range(2, min(9, _TOPO_SPEED['emb_max'] + 2)):
        dg = _ph_rips(_delay_embed(_topo_pc1, m, max(1, int(0.008 * _topo_fs))), 1)
        f = _feat(dg)
        rows.append([m, np.nan if prev is None else float(np.linalg.norm(f - prev))])
        prev = f
    out = pd.DataFrame(rows, columns=['embedding_dim', 'topological_increment'])
    display(out)
    fig, ax = plt.subplots(figsize=(8.5, 4.0), facecolor=TOPO_BG)
    ax.plot(out['embedding_dim'], out['topological_increment'], color=TOPO_ACCENT, marker='o', linewidth=1.7)
    ax.set_title('Topological saturation across embeddings')
    ax.set_xlabel('embedding dimension')
    ax.set_ylabel('topological increment')
    ax.grid(True, color=TOPO_CYAN_DIM, alpha=0.22)
    plt.tight_layout()
    plt.show()
    return


_topology_method_072()
del _topology_method_072
```

### 73. Topological saturation across delays

Topological saturation across delays applies the same feature-increment diagnostic to successive delay values. A flat region means the chosen persistence summaries are relatively insensitive to further delay changes; it does not imply that all topological invariants are delay-independent.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 73: Topological saturation across delays

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_073():

    i = 73
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _delay_embed(y,m=3,tau=2,max_points=None):
        y=np.asarray(y,float).ravel(); m=int(m); tau=int(tau)
        n=len(y)-(m-1)*tau
        if n<3: return np.empty((0,m))
        X=np.column_stack([y[j*tau:j*tau+n] for j in range(m)])
        if max_points is None: max_points=_TOPO_SPEED['points']
        if len(X)>max_points:
            idx=np.linspace(0,len(X)-1,max_points).astype(int); X=X[idx]
        X=(X-X.mean(0,keepdims=True))/(X.std(0,keepdims=True)+TOPO_EPS)
        return X

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _life(dgm):
        d=_finite(dgm)
        if not len(d): return np.empty(0)
        L=np.maximum(d[:,1]-d[:,0],0.0)
        scale=max(1.0,float(np.nanmax(np.abs(d))))
        return L[L > 1e-12*scale]

    def _pentropy(dgm):
        L=_life(dgm); s=L.sum()
        if s<=TOPO_EPS:return 0.0
        p=L/s; return float(-(p*np.log(p+TOPO_EPS)).sum())

    def _feat(dgms):
        h1=dgms[1] if len(dgms)>1 else np.empty((0,2)); L=_life(h1)
        return np.array([L.sum(), L.max() if len(L) else 0.0, _pentropy(h1), float(len(L))])

    def _simplices_rips(X,maxdim=1):
        X=np.asarray(X,float); n=len(X)
        if n<2: return [((i,),0.0,0) for i in range(n)]
        D=squareform(pdist(X))
        ss=[((i,),0.0,0) for i in range(n)]
        for i in range(n):
            for j in range(i+1,n): ss.append(((i,j),float(D[i,j]),1))
        if maxdim>=1:
            # triangles kill H1 classes; needed even if only H1 is reported
            for i in range(n):
                for j in range(i+1,n):
                    dij=D[i,j]
                    for k in range(j+1,n): ss.append(((i,j,k),float(max(dij,D[i,k],D[j,k])),2))
        # faces precede cofaces at equal filtration value
        ss.sort(key=lambda z:(z[1],z[2],z[0]))
        return ss

    def _ph_rips(X,maxdim=1):
        ss=_simplices_rips(X,maxdim=maxdim)
        index={s[0]:i for i,s in enumerate(ss)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(simp,filt,dim) in enumerate(ss):
            if dim==0: col=set()
            else:
                col={index[simp[:q]+simp[q+1:]] for q in range(len(simp))}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j
                bdim=ss[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(ss[low][1]),float(filt)))
            else: births.add(j)
        paired_births=set(low_to_col.keys())
        dgms=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired_births):
                if ss[j][2]==d: arr.append([float(ss[j][1]),np.inf])
            dgms.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2),float))
        return dgms

    # Method mathematics and output.
    taus = np.unique(np.maximum(1, (np.linspace(0.002, 0.03, 8) * _topo_fs).astype(int)))
    rows = []
    prev = None
    for tau in taus:
        f = _feat(_ph_rips(_delay_embed(_topo_pc1, 3, int(tau)), 1))
        rows.append([tau / _topo_fs, np.nan if prev is None else float(np.linalg.norm(f - prev))])
        prev = f
    out = pd.DataFrame(rows, columns=['delay_s', 'topological_increment'])
    display(out)
    fig, ax = plt.subplots(figsize=(8.5, 4.0), facecolor=TOPO_BG)
    ax.plot(out['delay_s'], out['topological_increment'], color=TOPO_ACCENT, marker='o', linewidth=1.7)
    ax.set_title('Topological saturation across delays')
    ax.set_xlabel('delay (s)')
    ax.set_ylabel('topological increment')
    ax.grid(True, color=TOPO_CYAN_DIM, alpha=0.22)
    plt.tight_layout()
    plt.show()
    return


_topology_method_073()
del _topology_method_073
```

### 74. Cubical persistence on spectrograms

A spectrogram is treated as a scalar field on a 2-D cubical grid. In a lower-star filtration, a vertex enters at its scalar value, an edge at the maximum of its endpoint values, and a square at the maximum of its four vertices. Matrix reduction then gives cubical H0/H1 persistence.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 74: Cubical persistence on spectrograms

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_074():

    i = 74
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _cubical_ph(A,upper=False):
        A=np.asarray(A,float); A=(-A if upper else A); nr,nc=A.shape; cells=[]
        # lower-star filtration on grid vertices; edge/square value=max of incident vertices
        for r in range(nr):
            for c in range(nc): cells.append((('v',r,c),float(A[r,c]),0,[]))
        for r in range(nr):
            for c in range(nc-1): cells.append((('eh',r,c),float(max(A[r,c],A[r,c+1])),1,[('v',r,c),('v',r,c+1)]))
        for r in range(nr-1):
            for c in range(nc): cells.append((('ev',r,c),float(max(A[r,c],A[r+1,c])),1,[('v',r,c),('v',r+1,c)]))
        for r in range(nr-1):
            for c in range(nc-1):
                val=float(np.max(A[r:r+2,c:c+2])); bd=[('eh',r,c),('eh',r+1,c),('ev',r,c),('ev',r,c+1)]
                cells.append((('sq',r,c),val,2,bd))
        dg=_ph_cells(cells,1)
        if upper:
            # transform filtration coordinates back to original superlevel threshold convention
            for d in range(len(dg)):
                if len(dg[d]): dg[d]=-dg[d][:,::-1]
        return dg

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    def _tf_spectrogram(y):
        y=np.asarray(y,float)
        nr,nc=_TOPO_IMG
        nper=min(len(y), 512 if FAST_OPTION=='full' else 384 if FAST_OPTION=='balanced' else 256 if FAST_OPTION=='fast' else 192)
        nover=nper//2
        f,t,S=signal.spectrogram(y,fs=_topo_fs,nperseg=nper,noverlap=nover,scaling='density',mode='psd')
        m=(f>=1)&(f<=min(45,_topo_fs/2-.5)); A=np.log10(S[m]+TOPO_EPS)
        return f[m],t,signal.resample(signal.resample(A,nr,axis=0),nc,axis=1)

    # Method-local derived quantities required by this method.
    _TOPO_SPEC_F, _TOPO_SPEC_T, _TOPO_SPEC = _tf_spectrogram(_topo_pc1)

    # Method mathematics and output.
    dg = _cubical_ph(_TOPO_SPEC)
    _plot_diag(dg[1], 'Cubical persistence on log-spectrogram')
    return


_topology_method_074()
del _topology_method_074
```

### 75. Cubical persistence on wavelet scalograms

The Morlet scalogram is a time–scale scalar field. The same cubical lower-star rule is applied to log wavelet energy, so persistence measures connected basins and loop-like level-set structure in the scale–time plane rather than the topology of the raw 1-D trace.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> The main reading strategy is to separate three things: the measured EEG samples, the mathematical representation constructed from those samples, and the final summary statistic. The statistic is not a direct physical quantity measured by the electrodes; it is the result of applying the stated assumptions and transformations to the recorded signal. When comparing channels or conditions, keep the preprocessing, sampling rate, window length, and estimator settings fixed so that numerical differences are interpretable rather than artifacts of different analysis choices.</p><p><b>How to read the mathematics.</b> A logarithm converts multiplicative scaling into additive differences. Its base changes the numerical units but not the ordering of values when used consistently. Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> A useful practical check is dimensional consistency: ask what each input quantity represents, what units it has, and whether the final statistic is dimensional or dimensionless. This often makes the role of each term in the formula much easier to follow.</p></div>


```python
# Method 75: Cubical persistence on wavelet scalograms

# Self-contained mathematical implementation. Shared catalogue cells provide only constants/imports/raw input.

def _topology_method_075():

    i = 75
    # Method-local preprocessing: no mathematical state is reused from another catalogue method.
    _topo_local_raw = np.asarray(_topo_raw[:_TOPO_SPEED['max_samples']], float)
    _topo_local_raw = signal.detrend(_topo_local_raw, axis=0, type='constant')
    _topo_down = int(_TOPO_SPEED['down'])
    _topo_x = (
        signal.resample_poly(_topo_local_raw, up=1, down=_topo_down, axis=0)
        if _topo_down > 1 else _topo_local_raw.copy()
    )
    _topo_fs = _topo_fs0 / _topo_down
    _topo_x = (
        _topo_x - np.nanmean(_topo_x, axis=0, keepdims=True)
    ) / (
        np.nanstd(_topo_x, axis=0, keepdims=True) + TOPO_EPS
    )
    _topo_names = _TOPO_NAMES[:_topo_x.shape[1]]
    _topo_pc1 = PCA(n_components=1).fit_transform(_topo_x).ravel()

    # Method-local mathematical definitions required by this cell only.
    def _ph_cells(cells,maxdim=2):
        """Persistent homology over Z2 for an arbitrary finite filtered cell complex.
        cells: list of (cell_id, filtration, dimension, boundary_ids)."""
        cells=sorted(cells,key=lambda z:(float(z[1]),int(z[2]),str(z[0])))
        index={z[0]:i for i,z in enumerate(cells)}
        R=[]; low_to_col={}; births=set(); pairs=[]
        for j,(cid,filt,dim,boundary) in enumerate(cells):
            col={index[b] for b in boundary if b in index}
            while col:
                low=max(col)
                if low in low_to_col: col ^= R[low_to_col[low]]
                else: break
            R.append(set(col))
            if col:
                low=max(col); low_to_col[low]=j; bdim=cells[low][2]
                if bdim<=maxdim: pairs.append((bdim,float(cells[low][1]),float(filt)))
            else: births.add(j)
        paired=set(low_to_col)
        out=[]
        for d in range(maxdim+1):
            arr=[[b,de] for dd,b,de in pairs if dd==d]
            for j in sorted(births-paired):
                if cells[j][2]==d: arr.append([float(cells[j][1]),np.inf])
            out.append(np.asarray(arr,float).reshape(-1,2) if arr else np.empty((0,2)))
        return out

    def _cubical_ph(A,upper=False):
        A=np.asarray(A,float); A=(-A if upper else A); nr,nc=A.shape; cells=[]
        # lower-star filtration on grid vertices; edge/square value=max of incident vertices
        for r in range(nr):
            for c in range(nc): cells.append((('v',r,c),float(A[r,c]),0,[]))
        for r in range(nr):
            for c in range(nc-1): cells.append((('eh',r,c),float(max(A[r,c],A[r,c+1])),1,[('v',r,c),('v',r,c+1)]))
        for r in range(nr-1):
            for c in range(nc): cells.append((('ev',r,c),float(max(A[r,c],A[r+1,c])),1,[('v',r,c),('v',r+1,c)]))
        for r in range(nr-1):
            for c in range(nc-1):
                val=float(np.max(A[r:r+2,c:c+2])); bd=[('eh',r,c),('eh',r+1,c),('ev',r,c),('ev',r,c+1)]
                cells.append((('sq',r,c),val,2,bd))
        dg=_ph_cells(cells,1)
        if upper:
            # transform filtration coordinates back to original superlevel threshold convention
            for d in range(len(dg)):
                if len(dg[d]): dg[d]=-dg[d][:,::-1]
        return dg

    def _cwt_scalogram(y):
        y=np.asarray(y,float); nr,nc=_TOPO_IMG
        freqs=np.geomspace(2,min(40,_topo_fs/2-.5),nr)
        out=[]
        for f in freqs:
            s=max(2.0,6*_topo_fs/(2*np.pi*f)); half=int(min(len(y)//4,max(8,4*s)))
            tt=np.arange(-half,half+1); wav=np.exp(2j*np.pi*f*tt/_topo_fs)*np.exp(-(tt**2)/(2*s*s)); wav/=np.sqrt(np.sum(np.abs(wav)**2)+TOPO_EPS)
            out.append(np.abs(signal.fftconvolve(y,wav,mode='same'))**2)
        A=np.log1p(np.asarray(out)); A=signal.resample(A,nc,axis=1)
        return freqs,np.linspace(0,len(y)/_topo_fs,nc),A

    def _finite(dgm):
        d=np.asarray(dgm,float).reshape(-1,2)
        return d[np.isfinite(d[:,1])] if len(d) else d

    def _plot_diag(dgm,title):
        d=_finite(dgm); fig,ax=plt.subplots(figsize=(4.8,4.2),facecolor=TOPO_BG)
        if len(d):
            mx=max(TOPO_EPS,float(np.max(d)))
            ax.scatter(d[:,0],d[:,1],s=22,facecolors=TOPO_BG,edgecolors=TOPO_ACCENT,linewidths=1.1)
            ax.plot([0,mx],[0,mx],ls='--',color=TOPO_CYAN_SOFT,alpha=.65)
        ax.set_title(title); ax.set_xlabel('birth'); ax.set_ylabel('death')
        ax.grid(True,color=TOPO_CYAN_DIM,alpha=.22,linewidth=.8)
        for spine in ax.spines.values(): spine.set_color(TOPO_ACCENT)
        plt.tight_layout(); plt.show()

    # Method-local derived quantities required by this method.
    _TOPO_CWT_F, _TOPO_CWT_T, _TOPO_CWT = _cwt_scalogram(_topo_pc1)

    # Method mathematics and output.
    dg = _cubical_ph(_TOPO_CWT)
    _plot_diag(dg[1], 'Cubical persistence on Morlet scalogram')
    return


_topology_method_075()
del _topology_method_075
```

### 76. Cubical persistence on channel-time matrices

The multichannel signal is viewed as a channel × time scalar image after controlled resampling. Cubical persistence studies sublevel topology of this rectangular field. Channel order is therefore part of the representation and must be interpreted as an analysis convention, not physical scalp adjacency.

<div style="font-size:13px;font-family:'Times New Roman',Times,serif;background-color:#202020;color:#D8D8D8;padding:14px 16px;border-radius:8px;margin:12px 10px;line-height:1.6;border-left:3px solid #777;"><p><b>Core idea.</b> A Lyapunov exponent describes how quickly two initially nearby states separate or approach one another as the system evolves. A positive exponent means small differences grow on average, a negative exponent means they shrink, and a value near zero corresponds to a direction with little average exponential change. When several exponents are reported, they describe stretching and contraction along different directions of the reconstructed state space rather than separate EEG frequency bands. The numerical value also has units set by the time step used in the analysis, so comparisons are most meaningful when preprocessing and sampling are consistent.</p><p><b>How to read the mathematics.</b> Double vertical bars normally denote a vector or matrix norm, which is a measure of size or distance.</p><p><b>How to interpret this cell.</b> For EEG, differences between channels should be interpreted as differences in the analyzed signal representation, not automatically as differences in underlying neural sources, because referencing, volume conduction, noise, and preprocessing can influence channel-level values.</p></div>
